```python
A = [4, 7, 1, 8, 3, 6, 2]
n = len(A)
for i in range(n // 2):
    temp = A[i]
    A[i] = A[n - 1 - i]
    A[n - 1 - i] = temp
for i in range(1, n):
    if A[i] < A[i - 1]:
        A[i] = A[i] + A[i - 1]
print(A[n - 1] - A[0], A[3])
```

## Question 9

### Letter Tree

The letters of a word are inserted one at a time, from left to right, into a binary search tree that starts empty.
The first letter is the root.
Each later letter starts at the root and moves down: to the left child if it is less than or equal to the letter in the node, and to the right child if it is greater.
So a letter equal to one already in the tree goes to the **left**, as if it were smaller.
The letter is placed at the first empty spot it reaches.

- The **depth** of a node is the number of steps from the root down to it; the root has depth 0.
- A **leaf** is a node with no children. A tree of one node has one leaf, the root.
- The **internal path length** is the sum of the depths of all the nodes.

The input is one line: a word of capital letters.
Print one line: the internal path length, the number of leaves, and the greatest depth of any node, separated by single spaces.

### Constraints

- The word has between 1 and 30 letters, each from `A` to `Z`.
- Letters may repeat.

### Sample Input

```text
CONTEST
```

### Sample Output

```text
15 2 4
```
