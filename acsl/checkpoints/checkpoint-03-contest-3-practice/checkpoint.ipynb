```python
A = [4, 7, 1, 8, 3, 6, 2]
n = len(A)
for i in range(n // 2):
    temp = A[i]
    A[i] = A[n - 1 - i]
    A[n - 1 - i] = temp
for i in range(1, n):
    if A[i] < A[i - 1]:
        A[i] = A[i] + A[i - 1]
print(A[n - 1] - A[0], A[3])
```

This program is written in ACSL pseudocode.

```text
FOR r = 1 TO 4
   FOR c = 1 TO 4
      A(r, c) = (r * c + r) % 5
   NEXT c
NEXT r
s = 0
FOR k = 1 TO 4
   s = s + A(k, k) - A(k, 5 - k)
NEXT k
OUTPUT s
```

### The Counting Line

`N` players, numbered 1 to `N`, stand in a **queue**, with player 1 at the front and player `N` at the back.
The game goes in rounds until the queue is empty.
In each round, the player at the front leaves the front and joins the back of the queue, `K − 1` times over; then the player who is now at the front leaves the game.

The input is one line: `N` and `K`, separated by one space.
Print one line: the numbers of the players in the order they leave the game, separated by single spaces.

### Constraints

- `1 <= N <= 500`
- `1 <= K <= 500`
- `K` may be larger than the number of players left in the queue.

### Sample Input

```text
7 3
```

### Sample Output

```text
3 6 2 7 5 1 4
```

In the first round, players 1 and 2 move to the back, making the queue `3 4 5 6 7 1 2`, and player 3 leaves.
In the second round, players 4 and 5 move to the back, making it `6 7 1 2 4 5`, and player 6 leaves.
