```text
input a, b
c = a % b
if c > 2 && !(a > 50) then
   d = a - b * c
else
   if c == 0 || b > 6 then
      d = a + b
   else
      d = 2 * c - a
   end if
end if
if d < 0 then
   d = -d
end if
output d
```

### All the Constructs

This program is written in ACSL pseudocode.

- `A` is a 2D array: `A(r, c)` is the entry in row `r`, column `c`.
- Strings are indexed from 0, and `S[p:q]` is the ACSL substring from position `p` **through** position `q` (both included), as in Unit 3.
- `/` is real division, so it can give a decimal.
- `int(x)` is the greatest integer less than or equal to `x`, `abs(x)` is the absolute value and `sqrt(x)` is the square root.

```text
S = "COMPUTERS"
for r = 1 to 3
   for c = 1 to 3
      A(r, c) = r * c - 2
   next c
next r
t = 0
for k = 1 to 3 step 2
   t = t + A(k, 4 - k)
next k
x = int((t - 7) / 2)
y = int(sqrt(A(3, 3) + 2))
while abs(x) < 10
   x = x * y
end while
p = abs(x) % 4
w = S[p:p + 2]
output w
```

### Base Palindromes

A number is a **base-B palindrome** when its digits in base B read the same forwards and backwards.
For example, 300 in base 7 is 606₇, which is a palindrome, but 300 in base 10 is not.

The input is one line: a decimal whole number `N`, a space, then a base `B`.
Write `N` in base `B`, using the digits `0`–`9` and then the capital letters `A`–`F` for the digit values 10–15.
Print one line: the base-`B` digits of `N`, one space, then `YES` if those digits form a palindrome or `NO` if they do not.

### Constraints

- `1 <= N <= 1000000`
- `2 <= B <= 16`
- The base-`B` digits have no leading zeros.

### Sample Input

```text
300 7
```

### Sample Output

```text
606 YES
```
