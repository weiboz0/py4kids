### All the Constructs

This program is written in ACSL pseudocode.

- `A` is a 2D array: `A(r, c)` is the entry in row `r`, column `c`.
- Strings are indexed from 0, and `S[p:q]` is the ACSL substring from position `p` **through** position `q` (both included), as in Unit 3.
- `/` is real division, so it can give a decimal.
- `int(x)` is the greatest integer less than or equal to `x`, `abs(x)` is the absolute value and `sqrt(x)` is the square root.

```text
S = "COMPUTERS"
for r = 1 to 3
   for c = 1 to 3
      A(r, c) = r * c - 2
   next c
next r
t = 0
for k = 1 to 3 step 2
   t = t + A(k, 4 - k)
next k
x = int((t - 7) / 2)
y = int(sqrt(A(3, 3) + 2))
while abs(x) < 10
   x = x * y
end while
p = abs(x) % 4
w = S[p:p + 2]
output w
```

For example, 26 is 222₃ in base 3, and 400 is 1111₇ in base 7 (343 + 49 + 7 + 1).

The input is one line: a decimal whole number `N`.
Find the **smallest** base `B`, with `B` at least 2, in which all the digits of `N` are the same.
Print one line: the base `B`, the value of that repeated digit, and the number of digits, separated by single spaces.
Write the base and the digit's value as decimal numbers, since the base can be larger than 16.

### Constraints

- `1 <= N <= 100000`

### Sample Input

```text
400
```

### Sample Output

```text
7 1 4
```
