## Question 1

Hex to octal goes through binary: write each hex digit as 4 bits, then regroup the bits in threes from the right.

- `3` is `0011`, `A` is `1010`, `C` is `1100`, so 3AC₁₆ is `0011 1010 1100`.
- Regroup from the right in threes: `001 110 101 100`.
- Each group is one octal digit: `1 6 5 4`.

Check in decimal: 3AC₁₆ = 3 × 256 + 10 × 16 + 12 = 940, and 1654₈ = 512 + 6 × 64 + 5 × 8 + 4 = 940.

**Answer:** `1654`

```python
DIGITS = "0123456789ABCDEF"


def to_base(n, base):
    text = ""
    while n > 0:
        text = DIGITS[n % base] + text
        n = n // base
    return text


assert str(to_base(int("3AC", 16), 8)) == "1654"
```

## Question 2

- 2C7₁₆ = 2 × 256 + 12 × 16 + 7 = 512 + 192 + 7 = 711.

- Read the remainders from the last one up: `DE3`.

You can also multiply in hex directly: 7 × 5 = 35 = 2 × 16 + 3, write `3` carry 2; C × 5 + 2 = 62 = 3 × 16 + 14, write `E` carry 3; 2 × 5 + 3 = 13, write `D`.

**Answer:** `DE3`

```python
DIGITS = "0123456789ABCDEF"


def to_base(n, base):
    text = ""
    while n > 0:
        text = DIGITS[n % base] + text
        n = n // base
    return text


assert str(to_base(int("2C7", 16) * 5, 16)) == "DE3"
```

## Question 3

Work down until `x ≤ 12`, then add the values back up.

| call | rule | value |
|---|---|---|
| f(23) | 23 > 12 | f(19) + 23 |
| f(19) | 19 > 12 | f(15) + 19 |
| f(15) | 15 > 12 | f(11) + 15 |
| f(11) | 11 ≤ 12 | (11 mod 5) + 3 = 1 + 3 = 4 |

Now fill back up: f(15) = 4 + 15 = 19, f(19) = 19 + 19 = 38, f(23) = 38 + 23 = 61.

**Answer:** `61`

```python
def f(x):
    if x > 12:
        return f(x - 4) + x
    return x % 5 + 3


assert str(f(23)) == "61"
```

## Question 4

Each call reaches back two and three places, so build the values from the bottom up and no call is worked out twice.

- The base case `2 × n − 1` covers every `n` below 4: h(1) = 1, h(2) = 3, h(3) = 5.
- h(4) = h(2) + h(1) + 1 = 3 + 1 + 1 = 5
- h(5) = h(3) + h(2) + 1 = 5 + 3 + 1 = 9
- h(6) = h(4) + h(3) + 1 = 5 + 5 + 1 = 11
- h(7) = h(5) + h(4) + 1 = 9 + 5 + 1 = 15
- h(8) = h(6) + h(5) + 1 = 11 + 9 + 1 = 21
- h(9) = h(7) + h(6) + 1 = 15 + 11 + 1 = 27

h(4) reaches h(1), which is still a base case (1 < 4), so no value below h(1) is ever needed.

**Answer:** `27`

```python
def h(n):
    if n >= 4:
        return h(n - 2) + h(n - 3) + 1
    return 2 * n - 1


assert str(h(9)) == "27"
```

## Question 8

**Notice:** try the bases in order from 2, and stop at the first one that works, so the base printed is the smallest.
For each base, the last digit of `N` is `N % B`; divide by `B` again and again, checking that every remainder equals that last digit and counting the digits.
The loop always stops: at the latest, base `N − 1` writes a number from 3 up as `11`, and base `N + 1` writes 1 or 2 as a single digit.
The fixtures include `1` (`2 1 1`, one digit in base 2), `2` (`3 2 1`: `10₂` fails, so the answer is the one-digit base 3), `3` (`2 1 2`, `11₂`), a digit larger than 9 (`300` is `FF` in base 19, so `19 15 2`), `2730` (`222222₄`, six digits), `99999` (`99999₁₀`), and the slow cases `99991` and `100000`, whose smallest base is near `N` itself (`99990 1 2` and `399 250 2`).

```python
n = int(input())
base = 2
found = False
while not found:
    digit = n % base
    rest = n
    count = 0
    same = True
    while rest > 0:
        if rest % base != digit:
            same = False
        rest = rest // base
        count = count + 1
    if same:
        found = True
    else:
        base = base + 1
print(str(base) + " " + str(digit) + " " + str(count))
```
