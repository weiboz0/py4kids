## Question 1

Hex to octal goes through binary: write each hex digit as 4 bits, then regroup the bits in threes from the right.

- `3` is `0011`, `A` is `1010`, `C` is `1100`, so 3AC₁₆ is `0011 1010 1100`.
- Regroup from the right in threes: `001 110 101 100`.
- Each group is one octal digit: `1 6 5 4`.

Check in decimal: 3AC₁₆ = 3 × 256 + 10 × 16 + 12 = 940, and 1654₈ = 512 + 6 × 64 + 5 × 8 + 4 = 940.

**Answer:** `1654`

```python
DIGITS = "0123456789ABCDEF"


def to_base(n, base):
    text = ""
    while n > 0:
        text = DIGITS[n % base] + text
        n = n // base
    return text


assert str(to_base(int("3AC", 16), 8)) == "1654"
```

## Question 2

Convert to decimal, multiply, and convert back.

- 2C7₁₆ = 2 × 256 + 12 × 16 + 7 = 512 + 192 + 7 = 711.
- 711 × 5 = 3555.
- 3555 ÷ 16 = 222 remainder 3; 222 ÷ 16 = 13 remainder 14 (`E`); 13 ÷ 16 = 0 remainder 13 (`D`).
- Read the remainders from the last one up: `DE3`.

You can also multiply in hex directly: 7 × 5 = 35 = 2 × 16 + 3, write `3` carry 2; C × 5 + 2 = 62 = 3 × 16 + 14, write `E` carry 3; 2 × 5 + 3 = 13, write `D`.

**Answer:** `DE3`

```python
DIGITS = "0123456789ABCDEF"


def to_base(n, base):
    text = ""
    while n > 0:
        text = DIGITS[n % base] + text
        n = n // base
    return text


assert str(to_base(int("2C7", 16) * 5, 16)) == "DE3"
```

## Question 3

Work down until `x ≤ 12`, then add the values back up.

| call | rule | value |
|---|---|---|
| f(23) | 23 > 12 | f(19) + 23 |
| f(19) | 19 > 12 | f(15) + 19 |
| f(15) | 15 > 12 | f(11) + 15 |
| f(11) | 11 ≤ 12 | (11 mod 5) + 3 = 1 + 3 = 4 |

Now fill back up: f(15) = 4 + 15 = 19, f(19) = 19 + 19 = 38, f(23) = 38 + 23 = 61.

**Answer:** `61`

```python
def f(x):
    if x > 12:
        return f(x - 4) + x
    return x % 5 + 3


assert str(f(23)) == "61"
```

## Question 4

Build the values from the bottom up, so no call is worked out twice.

- g(0) = 1, g(1) = 2, g(2) = 3 (the base case, `n + 1`).
- g(3) = g(2) + g(0) = 3 + 1 = 4
- g(4) = g(3) + g(1) = 4 + 2 = 6
- g(5) = g(4) + g(2) = 6 + 3 = 9
- g(6) = g(5) + g(3) = 9 + 4 = 13
- g(7) = g(6) + g(4) = 13 + 6 = 19

**Answer:** `19`

```python
def g(n):
    if n >= 3:
        return g(n - 1) + g(n - 3)
    return n + 1


assert str(g(7)) == "19"
```

## Question 8

**Notice:** build the base-`B` digits by repeated division (each remainder goes on the **front**), then compare the digits from both ends toward the middle.
`N` is at least 1, so there is always at least one digit.
The fixtures include a one-digit number (`1 2`, always a palindrome), a number whose base-`B` form ends in `0` (`10 2` is `1010`), hex letters (`255 16` is `FF`), and `1231 10`, which catches a program that compares only the first and last digits.

```python
DIGITS = "0123456789ABCDEF"

parts = input().split()
n = int(parts[0])
base = int(parts[1])
text = ""
while n > 0:
    text = DIGITS[n % base] + text
    n = n // base
answer = "YES"
left = 0
right = len(text) - 1
while left < right:
    if text[left] != text[right]:
        answer = "NO"
    left = left + 1
    right = right - 1
print(text + " " + answer)
```
