## Question 1

Put in every parenthesis first, then move each operator to the end of its pair.

- `↑` comes before `*` and `/`, and `*` and `/` go left to right, so the fully parenthesised form is `(A - (((B + C) * (D ↑ E)) / F))`.
- `(B + C)` becomes `B C +`, and `D ↑ E` becomes `D E ↑`.
- Their product is `B C + D E ↑ *`, and dividing by `F` makes it `B C + D E ↑ * F /`.
- The last operator is the `-` after `A`, so it goes at the very end.

Check: the operands stay in their original order `A B C D E F`.

**Answer:** `A B C + D E ↑ * F / -`

```python
import sys

sys.path.insert(0, "../../units/unit-04-prefix-infix-postfix/assets/verify")
import pip_eval

assert str(pip_eval.to_postfix("A - (B + C) * D ↑ E / F")) == "A B C + D E ↑ * F / -"
```

## Question 2

## Question 3

Do the operators in precedence order: `NOT`, then the shift and the circulate, then `AND`, then `OR`.

- `NOT 10110` = `01001`.
- `RSHIFT-1 01101` = `00110` (the last bit falls off, a 0 comes in on the left).
- `LCIRC-2 11010` = `01011` (the first two bits `11` wrap round to the end).
- `AND` next: `00110 AND 01011` = `00010`.
- `OR` last: `01001 OR 00010` = `01011`.

**Answer:** `01011`

```python
import sys

sys.path.insert(0, "../../units/unit-05-bit-string-flicking/assets/verify")
import bsf_eval

assert str(bsf_eval.evaluate("NOT 10110 OR RSHIFT-1 01101 AND LCIRC-2 11010")) == "01011"
```

## Question 4

Work one bit position at a time, `x OR 10100 = 10111`.

- Positions 1 and 3 (where `10100` has a 1): the result is 1 whatever `x` has, so `x` is free there.
- Position 2 (the result is 0): `x` must be 0.
- Positions 4 and 5 (the result is 1 but `10100` has 0): `x` must be 1.

So `x` is `*0*11`, where each `*` can be 0 or 1: four strings.
From smallest to largest: `00011`, `00111`, `10011`, `10111`.

**Answer:** `00011, 00111, 10011, 10111`

```python
import sys

sys.path.insert(0, "../../units/unit-05-bit-string-flicking/assets/verify")
import bsf_eval

assert str(", ".join(bsf_eval.solve("x OR 10100", "10111", 5))) == "00011, 00111, 10011, 10111"
```

## Question 7

Work out the variables first, then call the function.

- `(SETQ A 7)` makes `A` = 7.
- `(SETQ B (SUB A 12))` makes `B` = 7 - 12 = -5.
- `(F A B)` sets `X` = 7 and `Y` = -5 inside `F`.
- `(ADD (SQUARE X) Y (MULT 2 X))` = 49 + (-5) + 14 = 58.
- `(DIV 58 4)` = 14.5, which is not whole, so it is written as a decimal.

**Answer:** `14.5`

```python
import sys

sys.path.insert(0, "../../units/unit-07-lisp/assets/verify")
import lisp_eval

assert str(lisp_eval.run("""(SETQ A 7)
(SETQ B (SUB A 12))
(DEFUN F (X Y) (DIV (ADD (SQUARE X) Y (MULT 2 X)) 4))
(F A B)""")) == "14.5"
```

## Question 8

`L` has three elements: `(RED (BLUE GREEN))`, `(4 5 6)` and `YELLOW`.

- `(CADDR L)` is the third element: `YELLOW`.
- `(CAR L)` is `(RED (BLUE GREEN))`, and its `CADR` (second element) is `(BLUE GREEN)`.
- `(CONS YELLOW (BLUE GREEN))` puts `YELLOW` on the front: `(YELLOW BLUE GREEN)`.
- `REVERSE` turns the top-level list around: `(GREEN BLUE YELLOW)`.

**Answer:** `(GREEN BLUE YELLOW)`

```python
import sys

sys.path.insert(0, "../../units/unit-07-lisp/assets/verify")
import lisp_eval

assert str(lisp_eval.run("""(SETQ L '((RED (BLUE GREEN)) (4 5 6) YELLOW))
(REVERSE (CONS (CADDR L) (CADR (CAR L))))""")) == "(GREEN BLUE YELLOW)"
```

## Question 9

**Notice:** read the expression one character at a time, keeping a stack of operators and a stack of finished pieces (each a list with a `top` count, as in the units).
An operand is printed straight into the postfix and pushed as a piece; an operator is pushed on the operator stack; `(` does nothing.
Each `)` closes exactly one operator: pop it, add it to the postfix, pop the right piece and then the left piece, and push the prefix piece `op left right`.
This is the brackets-first method: a `)` arrives just after an operator's two operands are complete, so the postfix gets every operator right after its operands, and the one piece left at the end is the whole prefix.
The fixtures include a single operand with no brackets (`A`, printed on both lines), one operator on two digits (`(9-4)`), a chain bracketed from the left (`(((A-B)-C)-D)`) and one bracketed from the right (`(A^(B^(C^D)))`), mixed digits and letters, and a random expression close to the 400-character limit.

```python
text = input().split()[0]
operators = []
top = 0
pieces = []
count = 0
postfix = ""
for ch in text:
    if ch == "+" or ch == "-" or ch == "*" or ch == "/" or ch == "^":
        if top == len(operators):
            operators.append(ch)
        else:
            operators[top] = ch
        top = top + 1
    elif ch == ")":
        top = top - 1
        op = operators[top]
        postfix = postfix + " " + op
        right = pieces[count - 1]
        left = pieces[count - 2]
        count = count - 2
        piece = op + " " + left + " " + right
        pieces[count] = piece
        count = count + 1
    elif ch != "(":
        postfix = postfix + " " + ch
        if count == len(pieces):
            pieces.append(ch)
        else:
            pieces[count] = ch
        count = count + 1
print(postfix[1:])
print(pieces[0])
```
