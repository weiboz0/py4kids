## Question 6

_Division: Junior._

### Counting Down by Fives

This program is written in ACSL pseudocode.
ACSL papers also write keywords in lowercase, as here.
`%` is the remainder, `/` is real division, and `int(x)` is the greatest integer less than or equal to `x`.
A `for` loop includes its end value when the counter reaches it exactly.

```text
input n
c = 0
m = 0
for k = n to 1 step -5
   x = k
   s = 0
   while x > 0
      s = s + x % 10
      x = int(x / 10)
   end while
   if s > m then
      m = s
   end if
   c = c + s
next k
output c - m
```

What value does the program output when the input value is 47?

## Question 9

### Deepest Stack

A postfix expression can be evaluated with a stack.
Read the tokens from left to right.
A number is pushed onto the stack.
An operator pops the top two values, applies the operator with the lower value on the left and the top value on the right, and pushes the result.
When the tokens run out, the one value left on the stack is the value of the expression.

The input is one line: a postfix expression whose tokens are separated by single spaces.
Each operand is a whole number from 0 to 99, and each operator is `+`, `-` or `*`.
Print one line: the value of the expression, one space, then the largest number of values that were on the stack at any moment.

### Constraints

- The expression has between 1 and 39 tokens.
- The expression is valid: every operator finds two values on the stack, and exactly one value is left at the end.
- The value of the expression, and every value on the stack, is between -1000000 and 1000000.

### Sample Input

```text
6 2 3 + 4 * + 5 -
```

### Sample Output

```text
21 3
```
