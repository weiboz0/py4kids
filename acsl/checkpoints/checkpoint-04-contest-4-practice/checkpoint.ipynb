This program is in **Python**, so Python's slicing rules apply.
`%` is the remainder.

```python
A = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
S = "ZEBRA"
T = ""
for j in range(len(S)):
    k = 0
    while A[k] != S[j]:
        k = k + 1
    T = T + A[(k + j) % 26]
print(T, T[1:4])
```

## Question 9

_Division: Junior and above._

### What Kind of Route?

The vertices of a **directed** graph are capital letters, and the edge `AB` goes from `A` to `B`.
A **route** is a list of vertices written together, such as `ABCA`.
Using the words of Unit 12, every route is exactly one of these kinds:

- `NOT A PATH`: somewhere in the route, a vertex is not joined to the next one by an edge going that way.
- `CYCLE`: a path that ends at the vertex where it started, with no other vertex appearing twice. In a directed graph a cycle may have just two vertices: `ABA` is a cycle when `AB` and `BA` are both edges.
- `SIMPLE PATH`: a path in which no vertex appears twice.
- `PATH`: any other path, which visits some vertex more than once.

The first line of input holds the vertex letters, in alphabetical order, run together (such as `ABCDE`).
The second line holds the edges, separated by single spaces, each written as two letters.
The third line holds `R`, the number of routes.
Each of the next `R` lines holds one route.

Print one line for each route, in order: its kind, written exactly as above.

### Constraints

- There are between 2 and 26 vertices: different capital letters, in alphabetical order. They need not start at `A`.
- There are between 1 and 200 edges, and every edge joins two vertices of the first line.
- No edge joins a vertex to itself, and no edge appears twice. `AB` and `BA` may both appear: they are two different edges.
- `1 <= R <= 400`, and every route has between 2 and 30 letters, each a vertex of the first line.

### Sample Input

```text
ABCDE
AB BC CA CD DB DE EA
6
ABCA
DEAB
ABCDBC
ACDE
ABCABCA
CDEABC
```

### Sample Output

```text
CYCLE
SIMPLE PATH
PATH
NOT A PATH
PATH
CYCLE
```

`ABCDBC` follows the edges but visits `B` and `C` twice, so it is a path but not a simple path.
`ACDE` fails at its first step: the edge between `A` and `C` is `CA`, which goes the other way.
`ABCABCA` ends where it started, but it goes round the triangle twice, so it is a path, not a cycle.
