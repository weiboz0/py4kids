## Exercise 4

**Notice:** strip the brackets, split on spaces, and choose the rule by the first word, remembering that each function has a word name and a symbol name.
`/` gives a decimal such as `-3.0` even when the quotient is whole, so the program prints `int(quotient)` when the quotient equals its whole part.
The fixtures include `(DIV 12 -4)` (a whole negative quotient that must print `-3`, not `-3.0`), `(/ -7 8)` (a negative decimal), `(DIV 0 -5)`, `(EXP 7 0)` (which is `1`), `(EXP -3 5)`, and ten arguments to `ADD`.

```python
line = input()
inside = line[1:len(line) - 1]
parts = inside.split()
name = parts[0]
numbers = []
for k in range(1, len(parts)):
    numbers.append(int(parts[k]))
if name == "ADD" or name == "+":
    result = 0
    for value in numbers:
        result = result + value
    print(result)
elif name == "MULT" or name == "*":
    result = 1
    for value in numbers:
        result = result * value
    print(result)
elif name == "SUB" or name == "-":
    print(numbers[0] - numbers[1])
elif name == "DIV" or name == "/":
    quotient = numbers[0] / numbers[1]
    if quotient == int(quotient):
        print(int(quotient))
    else:
        print(quotient)
elif name == "SQUARE":
    print(numbers[0] * numbers[0])
elif name == "EXP":
    result = 1
    for k in range(numbers[1]):
        result = result * numbers[0]
    print(result)
```

## Exercise 13

**Notice:** `(REVERSE (CDR L))` drops the first atom **before** reversing, so for `(A B C D)` it drops `A`, not `D`.
Build the reversed list with a loop that counts down and `append`s.
For a one-atom list the `CDR` is empty, and `show` prints it as `NIL`; the `(SOLO)` fixture checks this.

```python
def show(items):
    if len(items) == 0:
        return "NIL"
    text = "("
    for k in range(len(items)):
        if k > 0:
            text = text + " "
        text = text + items[k]
    return text + ")"


def cdr(items):
    rest = []
    for k in range(1, len(items)):
        rest.append(items[k])
    return rest


def reverse(items):
    result = []
    for k in range(len(items) - 1, -1, -1):
        result.append(items[k])
    return result


line = input()
atoms = line[1:len(line) - 1].split()
print(show(reverse(atoms)))
print(show(reverse(cdr(atoms))))
```

## Exercise 14

**Notice:** the second line is the first atom of `L`, then `X`, then the rest of `L`: `X` goes in second place.
The helper functions `car`, `cdr` and `cons` are the ones from the lesson.
The one-atom fixture `(A)` checks that an empty `CDR` still works, and the number fixture checks that atoms are kept as text.

```python
def show(items):
    if len(items) == 0:
        return "NIL"
    text = "("
    for k in range(len(items)):
        if k > 0:
            text = text + " "
        text = text + items[k]
    return text + ")"


def car(items):
    return items[0]


def cdr(items):
    rest = []
    for k in range(1, len(items)):
        rest.append(items[k])
    return rest


def cons(first, items):
    result = [first]
    for item in items:
        result.append(item)
    return result


x = input().strip()
line = input()
atoms = line[1:len(line) - 1].split()
print(show(cons(x, atoms)))
print(show(cons(car(atoms), cons(x, cdr(atoms)))))
```

## Exercise 18

`(CAR (REVERSE L))` is the last item of `L`, and `(REVERSE (CDR (REVERSE L)))` is `L` without its last item.
So `ROT` moves the last item to the front.

## Exercise 20

**Notice:** the letters between `C` and `R` are at positions `1` to `len(name) - 2`, and they are applied from the right end, so the loop counts down.
A `D` replaces the list by its `CDR`; an `A` takes the first atom, and the value becomes an atom.
Doing the steps in the wrong order passes the sample `CDDR` but fails `CADR` on `(1 2 3)`, which must print `2`.
The fixtures also cover `CAR` and `CDR` on a one-atom list (`ONLY` and `NIL`) and `CDDR` on `(P Q)`.

```python
def show(items):
    if len(items) == 0:
        return "NIL"
    text = "("
    for k in range(len(items)):
        if k > 0:
            text = text + " "
        text = text + items[k]
    return text + ")"


def cdr(items):
    rest = []
    for k in range(1, len(items)):
        rest.append(items[k])
    return rest


name = input().strip()
line = input()
items = line[1:len(line) - 1].split()
is_list = True
atom = ""
for k in range(len(name) - 2, 0, -1):
    if name[k] == "A":
        atom = items[0]
        is_list = False
    else:
        items = cdr(items)
if is_list:
    print(show(items))
else:
    print(atom)
```
