# Unit 7 — LISP Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work out the value with pencil and paper and write one exact answer.
  - Write a number as a whole number when it is whole (`-3`, not `-3.0`) and as a decimal otherwise (`7.5`).
  - Write a list with single spaces and round brackets, in the same capital letters as the question: `(A (B C) D)`.
  - Write the empty list, and "no", as `NIL`; write "yes" as `true`.
  - Leave out the quote mark: the value of `'(A B)` is `(A B)`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.
  Build new lists with loops and `append`, as in the lesson.

LISP is on the Intermediate and Senior papers only, so each exercise is marked Intermediate or Senior: the lowest division it belongs to.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 4

_Division: Intermediate and above._

### One-Call Calculator

The input is one line holding a single LISP statement whose arguments are all whole numbers, such as `(DIV 54 4)`.
The function is one of these:

- `ADD` or `+`, and `MULT` or `*`, with 2 to 10 arguments
- `SUB` or `-`, and `DIV` or `/`, with exactly 2 arguments
- `SQUARE`, with 1 argument
- `EXP`, with 2 arguments `a n`, giving `a` to the power `n`

Print the value of the statement.
Print a whole number without a decimal point; print any other value as a decimal.

### Constraints

- The line starts with `(`, ends with `)`, and has exactly one space between items.
- Every argument is between -1000 and 1000.
- For `EXP`, `n` is between 0 and 10, and the value is between -1000000000 and 1000000000.
- For `DIV` and `/`, the second argument is not 0, and the quotient is either a whole number or a decimal with at most 3 digits after the point.

### Sample Input

```text
(DIV 54 4)
```

### Sample Output

```text
13.5
```

The input `(- 6 9)` prints `-3`, and `(DIV 12 -4)` prints `-3` as well.

- The list has between 1 and 20 atoms.

## Exercise 20

_Division: Senior and above._

### Shortcuts in Python

The first line of input holds the name of one of these functions: `CAR`, `CDR`, `CADR`, `CDDR` or `CADDR`.
The second line holds a LISP list of atoms, `L`, such as `(P Q R S)`.
Print the value of the function applied to `L`.

The value is either an atom, which you print as it is, or a list, which you print the LISP way, with round brackets and single spaces.
Print an empty list as `NIL`.

Work through the letters between the `C` and the `R` from right to left, as LISP does, rather than writing a separate case for each name.

### Constraints

- `L` has between 1 and 20 atoms.
- Each atom is a word of capital letters or a whole number, with no brackets inside it.
- The second line starts with `(`, ends with `)`, and has exactly one space between atoms.
- The list is always long enough: every `A` step is taken on a list with at least one atom.

### Sample Input

```text
CDDR
(P Q R S)
```

### Sample Output

```text
(R S)
```

For the same list, `CADDR` prints `R`, and for the list `(P Q)`, `CDDR` prints `NIL`.
