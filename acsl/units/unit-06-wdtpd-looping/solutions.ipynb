## Exercise 7

**Predict:** from `n = 6` the values are `6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1`.
That is `8` steps, and the highest value is `16`, so the output is `8 16`, matching the sample.

**Verify:** `WHILE n != 1` becomes `while n != 1:`, and ACSL's `int(n / 2)` becomes `n // 2` (the same as `math.floor(n / 2)` for whole numbers).
The `IF … ELSE … END IF` becomes `if` / `else`, and the peak check is a separate `if` after it.

**Notice:** the fixtures include `1`, where the `WHILE` test is false at once, so the output is `0 1` (the peak counts the starting value); `2`, whose peak is its own start; `27`, which climbs to `9232`; and `871`, the number up to `1000` that takes the most steps (`178`, peak `190996`).

```python
n = int(input())
steps = 0
peak = n
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    steps = steps + 1
    if n > peak:
        peak = n
print(steps, peak)
```

## Exercise 11

**Predict:** with `n = 7` and `m = 20`, `k` counts down `7, 6, 5, 4, 3, 2, 1`, and `20 % k` is `6, 2, 0, 0, 2, 0, 0`.
The first value `6` (at `k = 7`) is never beaten, so the output is `6 7`, matching the sample.

**Verify:** `FOR k = n TO 1 STEP -1` counts down and includes `1`, so it becomes `range(n, 0, -1)`.

**Notice:** `best` starts at `-1`, so the first remainder always replaces it, even `0`.
The `5 60` fixture has every remainder `0` and must print `0 5`; starting `best` at `0` would leave `at` as `0`.
The `10 7` fixture has the remainder `7` at `k = 10, 9, 8`; because the test is `>`, the **first** one, `10`, is kept.
`1 1` is the smallest case (`0 1`), and `100 1000` is the largest (`90 91`).

```python
parts = input().split()
n = int(parts[0])
m = int(parts[1])
best = -1
at = 0
for k in range(n, 0, -1):
    r = m % k
    if r > best:
        best = r
        at = k
print(best, at)
```

## Exercise 17

**Predict:** with `n = 6` and `m = 4`, the pairs `i < j` whose product is a multiple of 4 are
`(1, 4)`, `(2, 4)`, `(2, 6)`, `(3, 4)`, `(4, 5)` and `(4, 6)`: six pairs, so the output is `6`, matching the sample.

**Verify:** `FOR i = 1 TO n - 1` becomes `range(1, n)`, and `FOR j = i + 1 TO n` becomes `range(i + 1, n + 1)`, each one past the ACSL end.

**Notice:** the fixtures include `2 2` (the only pair `(1, 2)` counts, output `1`: a loop that stops one short of `n` prints `0`), `2 1000` (`0`), `200 1` (every pair counts: `200 * 199 / 2 = 19900`), and the largest cases `200 7` and `200 1000`.

```python
parts = input().split()
n = int(parts[0])
m = int(parts[1])
count = 0
for i in range(1, n):
    for j in range(i + 1, n + 1):
        if (i * j) % m == 0:
            count = count + 1
print(count)
```
