### Up and Down to One (predict, then verify)
This ACSL program starts from a whole number `n`.
While `n` is not `1`, it halves `n` when `n` is even, and otherwise replaces `n` with `3 * n + 1`.
It counts the steps and remembers the highest value `n` reaches, counting the starting value.

```text
INPUT n
steps = 0
peak = n
WHILE n != 1
   IF n % 2 == 0 THEN
      n = int(n / 2)
   ELSE
      n = 3 * n + 1
   END IF
   steps = steps + 1
   IF n > peak THEN
      peak = n
   END IF
END WHILE
OUTPUT steps, peak
```

1. **Predict:** trace the program by hand on the sample input and write down what it outputs, before you look at the sample output.
2. **Verify:** translate the program into Python and run it on the sample input.

The input is one line holding the whole number `n`.
Print the number of steps and the peak, separated by one space.

### Constraints

- `1 <= n <= 1000`

### Sample Input

```text
6
```

### Sample Output

```text
8 16
```
