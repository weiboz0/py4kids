## Exercise 6

**Predict:** with `hours = 7` and `day = 3`, the day is not `6` or `7`, `7 <= 2` and `7 <= 5` are false, so `fee = 9 + (7 - 5) * 4 = 9 + 8 = 17`. `17 > 20` is false, so the output is `17`, matching the sample.

**Verify:** `||` becomes `or` and `ELSE IF` becomes `elif`.
The `fee > 20` test is a separate `IF` after the chain, so it is a separate `if` in Python, not another `elif`.

**Notice:** the fixtures test every branch: the sample, a weekend day with many hours (`24 6` gives `5`, not the capped weekday fee), exactly `2` hours (free), exactly `5` hours (`9`), and `8 2`, the first whole number of hours whose fee (`21`) goes over the cap and must print `20`.

```python
parts = input().split()
hours = int(parts[0])
day = int(parts[1])
if day == 6 or day == 7:
    fee = 5
elif hours <= 2:
    fee = 0
elif hours <= 5:
    fee = (hours - 2) * 3
else:
    fee = 9 + (hours - 5) * 4
if fee > 20:
    fee = 20
print(fee)
```

## Exercise 16

**Predict:** with `n = 23`:

- `int(sqrt(23)) = 4`, take away `16`: `n = 7`, `count = 1`.
- `int(sqrt(7)) = 2`, take away `4`: `n = 3`, `count = 2`.
- `int(sqrt(3)) = 1`, take away `1` three times: `n = 0`, `count = 5`.

The output is `5`, matching the sample.

**Verify:** ACSL's `int(sqrt(n))` becomes `math.floor(math.sqrt(n))`, and `r ^ 2` becomes `r ** 2`.

**Notice:** the fixtures sit on both sides of perfect squares: `16`, `25` and `10000` are squares and take one step, while `15`, `24` and `9999` sit just below a square and need several (`9999` goes `9801`, `196`, `1`, `1`, giving `4`). The smallest input, `1`, prints `1`.

```python
import math

n = int(input())
count = 0
while n > 0:
    r = math.floor(math.sqrt(n))
    n = n - r ** 2
    count = count + 1
print(count)
```
