# Unit 15 — Assembly Language Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  - The value in a location is a whole number, such as `-9`.
  - For "what is printed?", write every printed number in order, on one line, separated by single spaces: `13 8 3`.
  - Some questions give **options** labelled `A`, `B`, `C`, ….
    Write the labels of every option that qualifies, in alphabetical order, separated by a comma and a space: `A, C`.
    If no option qualifies, write `NONE`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.

The rules are the ones in the lesson.
`ACC` starts at 0, every `DC` line sets its location before the program starts, and `DIV` keeps the whole-number part of `ACC` ÷ `LOC`, rounding **toward zero** (`-7 DIV 2` is `-3`).
In every exercise, every number stays between −999,999 and 999,999, and no program divides by zero.

Assembly language is on the Intermediate and Senior papers only, so each exercise is marked Intermediate or Senior: the lowest division it belongs to.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 4

_Division: Intermediate and above._

### Straight-Line Translator

Write a Python program that does exactly what this assembly program does.

```text
      READ  A
      READ  B
      LOAD  A
      MULT  A
      SUB   B
      DIV   =3
      STORE C
      PRINT C
      LOAD  B
      SUB   A
      DIV   =4
      STORE D
      PRINT D
      END
```

The input is two whole numbers, `A` and then `B`, one per line.
Print the two numbers the assembly program prints, one per line, in the same order.
Remember that `DIV` rounds toward zero, which Python's `//` does not do for negative numbers.

### Constraints

- `A` and `B` are between −100 and 100.

### Sample Input

```text
4
9
```

### Sample Output

```text
2
1
```

With `A` = 3 and `B` = −7, the second number printed is `-2`: `ACC` is −10 before `DIV =4`, and −10 ÷ 4 = −2.5 rounds toward zero.

## Exercise 11

_Division: Intermediate and above._

### Accumulator Calculator

Write a program that runs a short assembly program that uses **only** immediate data, and prints `ACC` after every line.

The first line of input holds a whole number `n`.
Each of the next `n` lines holds one instruction: an opcode (`LOAD`, `ADD`, `SUB`, `MULT` or `DIV`), then one or more spaces, then `=` and a whole number, such as `MULT  =-6`.
There are no labels.
`ACC` starts at 0.
After each instruction, print the value of `ACC` on its own line.

### Constraints

- `n` is between 1 and 20.
- Every number in the input, and every value of `ACC`, is between −999,999 and 999,999.
- No `DIV` has `=0`.

### Sample Input

```text
5
LOAD  =7
MULT  =-6
ADD   =3
DIV   =8
SUB   =10
```

### Sample Output

```text
7
-42
-39
-4
-14
```

`-39 DIV 8` is `-4`, because −39 ÷ 8 = −4.875 rounds toward zero.
If the first instruction is `DIV =3`, the program prints `0`, since `ACC` starts at 0.

## Exercise 17

_Division: Senior and above._

### Assembly Interpreter

Extend the interpreter of Lesson 3 so that it runs every ACSL opcode: `DC`, `LOAD`, `STORE`, `ADD`, `SUB`, `MULT`, `DIV`, `BE`, `BG`, `BL`, `BU`, `READ`, `PRINT` and `END`, with immediate data `=value` for `LOAD`, `ADD`, `SUB`, `MULT` and `DIV`.

The input is:

1. one line with a whole number `n`, the number of program lines
2. the `n` lines of the program, in the column layout of the lesson (words separated by one or more spaces; a line with no label starts with spaces)
3. one line with a whole number `m`, the number of input numbers
4. `m` lines, each holding one input number, which the program's `READ` lines read in order

Run the program from its first line until it reaches `END`, and print each number a `PRINT` line prints, on its own line.

The rules are the lesson's:

- `ACC` starts at 0.
- A line's first word is its label exactly when it is not an opcode.
- Every `DC` line sets its location before the program starts, even if it comes after `END`; when the program reaches a `DC` line, it just moves on.
- A `STORE` or `READ` into a label with no `DC` line makes that location.
- `DIV` keeps the whole-number part of `ACC` ÷ `LOC`, rounding toward zero.

### Constraints

- `n` is between 2 and 60, and `m` is between 0 and 20.
- Labels start with a letter, go on with letters and digits, and are never opcodes. Each label is on at most one line, and every branch goes to a label that exists.
- A location is always set (by `DC`, `STORE` or `READ`) before it is read, and the program never reads more input numbers than there are.
- Every value stays between −999,999 and 999,999, no `DIV` divides by zero, and the program reaches `END` within 10,000 lines run.
- The program prints at least one number.

### Sample Input

This is ACSL's `N!` program with a `PRINT` added, run with the input 5.

```text
11
      READ  X
      LOAD  X
TOP   SUB   =1
      BE    DONE
      STORE A
      MULT  X
      STORE X
      LOAD  A
      BU    TOP
DONE  PRINT X
      END
1
5
```

### Sample Output

```text
120
```
