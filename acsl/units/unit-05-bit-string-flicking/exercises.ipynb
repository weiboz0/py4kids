# Unit 5 — Bit-String Flicking Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  - Write every bit string at its **full length**, including 0s on the left: `00110`, not `110`.
  - When a question asks for all the values of x, list them in **increasing order**, separated by a comma and a space: `0011, 1011`.
  - When it asks **how many**, write just the number, such as `4`.
  - Use `*` for a free bit only when the question asks for one pattern.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.

Each exercise is marked with the lowest ACSL division it belongs to.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 13

_Division: Junior and above._

### The Cue Sheet

A stage light board shows a bit string.
The cue sheet lists commands, and the board carries them out one at a time, from the top of the list down.
Each command is one of:

- `NOT`
- `AND s`, `OR s` or `XOR s`, where `s` is a bit string the same length as the board
- `LSHIFT-k`, `RSHIFT-k`, `LCIRC-k` or `RCIRC-k`, where `k` is a whole number

The first line of input holds the board.
The second line holds `N`, the number of commands.
Each of the next `N` lines holds one command.
After each command, print the board on its own line.

### Constraints

- The board has 1 to 30 bits.
- `1 <= N <= 20`
- `1 <= k <= 100`. A shift by the length or more leaves all zeros; a circulate count may be bigger than the length.

### Sample Input

```text
10110010
4
RCIRC-3
AND 11110000
NOT
XOR 00001111
```

### Sample Output

```text
01010110
01010000
10101111
10100000
```

## Exercise 14

_Division: Junior and above._

### Unary Chain

The input is one line: zero or more one-string operators, then a bit string, all separated by single spaces.
Each operator is `NOT`, `LSHIFT-k`, `RSHIFT-k`, `LCIRC-k` or `RCIRC-k`.
Evaluate the expression by ACSL's rules (one-string operators go from right to left) and print the result at full length.

### Constraints

- The bit string has 1 to 30 bits.
- There are at most 10 operators.
- `1 <= k <= 100`. A shift by the length or more leaves all zeros; a circulate count may be bigger than the length.

### Sample Input

```text
RCIRC-9 NOT LSHIFT-1 0110101
```

### Sample Output

```text
0100101
```

`LSHIFT-1 0110101` = `1101010`, then `NOT` gives `0010101`, and `RCIRC-9` on 7 bits is `RCIRC-2`, giving `0100101`.

## Exercise 19

_Division: Intermediate and above._

### Find Every x

The first line of input is a chain of one-string operators ending in the letter `x`, separated by single spaces, such as `RSHIFT-1 LCIRC-2 x`.
Each operator is `NOT`, `LSHIFT-k`, `RSHIFT-k`, `LCIRC-k` or `RCIRC-k`, and they apply from right to left.
The second line is a bit string `T`.

Find every bit string x of the same length as `T` for which the chain applied to x gives `T`.
Print them one per line, in increasing order.
If there are none, print `NONE`.

### Constraints

- `T` has 1 to 12 bits.
- There are 1 to 10 operators.
- `1 <= k <= 100`

### Sample Input

```text
LSHIFT-1 RCIRC-2 x
0110
```

### Sample Output

```text
1100
1110
```

With x = `abcd`, `RCIRC-2 x` = `cdab` and `LSHIFT-1` of that is `dab0`, so d = 0, a = 1, b = 1, and c is free.

## Exercise 21

_Division: Senior and above._

### The Whole Expression

The input is one line holding an ACSL bit-string expression with **no parentheses**, its parts separated by single spaces.
It uses bit strings, the one-string operators `NOT`, `LSHIFT-k`, `RSHIFT-k`, `LCIRC-k` and `RCIRC-k`, and the two-string operators `AND`, `XOR` and `OR`.
Every bit string in the expression has the same length.

Evaluate it with ACSL's order of operations and print the result.

### Constraints

- Each bit string has 1 to 30 bits.
- There are 1 to 10 bit strings.
- `1 <= k <= 100`

### Sample Input

```text
LSHIFT-1 01101 XOR RCIRC-2 11011 AND NOT 10101 OR 00010
```

### Sample Output

```text
10010
```

The one-string operators give `11010 XOR 11110 AND 01010 OR 00010`; `AND` next gives `11010 XOR 01010 OR 00010`; then `XOR` gives `10000 OR 00010`, and `OR` gives `10010`.
