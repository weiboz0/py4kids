## Exercise 7

**Notice:** one list serves both structures. For a stack, `top` counts the items and a `POP` reads `items[top - 1]`; for a queue, `head` marks the front and a `POP` reads `items[head]`. The values are kept as text, so `-4`, `12` and `DOG` print back exactly as read. The fixtures that pop an empty structure first (`POP` before any `PUSH`), pop twice in a row past the end, and refill a stack after popping (`PUSH 4` after two `POP`s writes over an old slot) catch a program that forgets `NIL` or loses track of the top.

````python
kind = input()
n = int(input())

items = []
top = 0
head = 0
for step in range(n):
    parts = input().split()
    if parts[0] == "PUSH":
        if kind == "STACK":
            if top == len(items):
                items.append(parts[1])
            else:
                items[top] = parts[1]
            top = top + 1
        else:
            items.append(parts[1])
    elif kind == "STACK":
        if top == 0:
            print("NIL")
        else:
            top = top - 1
            print(items[top])
    else:
        if head == len(items):
            print("NIL")
        else:
            print(items[head])
            head = head + 1
````

## Exercise 8

**Notice:** `insert` walks down from the root with a `while` loop and counts one level for each step, so it can return the new node's depth. Use `<=` to go left: the `AAAA` fixture (depths `0 1 2 3`) and `MISSISSIPPI` catch a program that sends equal letters right. A one-letter word has depth `0` and internal path length `0`.

````python
def insert(key, left, right, letter):
    key.append(letter)
    left.append(-1)
    right.append(-1)
    new = len(key) - 1
    depth = 0
    node = 0
    placed = new == 0
    while not placed:
        depth = depth + 1
        if letter <= key[node]:
            if left[node] == -1:
                left[node] = new
                placed = True
            else:
                node = left[node]
        else:
            if right[node] == -1:
                right[node] = new
                placed = True
            else:
                node = right[node]
    return depth


word = input()
key = []
left = []
right = []
line = ""
total = 0
for letter in word:
    depth = insert(key, left, right, letter)
    total = total + depth
    if line == "":
        line = str(depth)
    else:
        line = line + " " + str(depth)
print(line)
print(total)
````

## Exercise 16

**Notice:** one `insert` (with `<=` for left) and three recursive traversals that differ only in where the node's own letter goes: before both subtrees (preorder), between them (inorder) or after them (postorder). An empty subtree (`-1`) gives the empty string. The `DCBA` fixture builds a tree that is all left children, and `SEESAW` and `PROGRAM` hold duplicates, which must go left.

````python
def insert(key, left, right, letter):
    key.append(letter)
    left.append(-1)
    right.append(-1)
    new = len(key) - 1
    node = 0
    placed = new == 0
    while not placed:
        if letter <= key[node]:
            if left[node] == -1:
                left[node] = new
                placed = True
            else:
                node = left[node]
        else:
            if right[node] == -1:
                right[node] = new
                placed = True
            else:
                node = right[node]


def inorder(node, key, left, right):
    if node == -1:
        return ""
    return inorder(left[node], key, left, right) + key[node] + inorder(right[node], key, left, right)


def preorder(node, key, left, right):
    if node == -1:
        return ""
    return key[node] + preorder(left[node], key, left, right) + preorder(right[node], key, left, right)


def postorder(node, key, left, right):
    if node == -1:
        return ""
    return postorder(left[node], key, left, right) + postorder(right[node], key, left, right) + key[node]


word = input()
key = []
left = []
right = []
for letter in word:
    insert(key, left, right, letter)
print(inorder(0, key, left, right))
print(preorder(0, key, left, right))
print(postorder(0, key, left, right))
````

## Exercise 17

**Notice:** the heap is a list with an unused slot 0, so the parent of position `i` is `i // 2`. The swap test is **strict** (`>` for `MIN`, `<` for `MAX`), so a number equal to its parent stays put; the `5 5 3 3 1 1 1` fixture catches a program that swaps on ties. Row `k` holds positions `2^k` to `2^(k+1) - 1`, so each row starts at `start` and the next at `2 * start`; the last row may be short. The numbers may be negative, so read them with `int`.

````python
def out_of_order(parent, child, kind):
    if kind == "MIN":
        return parent > child
    return parent < child


kind = input()
heap = [0]
for text in input().split():
    heap.append(int(text))
    i = len(heap) - 1
    while i > 1 and out_of_order(heap[i // 2], heap[i], kind):
        heap[i // 2], heap[i] = heap[i], heap[i // 2]
        i = i // 2

start = 1
while start < len(heap):
    row = str(heap[start])
    for i in range(start + 1, min(2 * start, len(heap))):
        row = row + " " + str(heap[i])
    print(row)
    start = 2 * start
````

## Exercise 21

**Notice:** find the node and its parent with a `while` loop from the root (the letter appears exactly once, so `<` and `>` lead straight to it). Choose the replacement: the other child when one side is empty (`-1` when both are), else the left child, after hanging the old right subtree off the right-most node of the left child's tree. Then point the parent at the replacement, or make it the root when the deleted node was the root. The fixtures delete the root with two children (`KEYBOARD`/`K`, `SANDWICH`/`S`), with one child (`TRAIN`/`T`, `BRAVE`/`B`), a leaf (`AB`/`B`), and nodes with one or two children deeper down.

````python
def insert(key, left, right, letter):
    key.append(letter)
    left.append(-1)
    right.append(-1)
    new = len(key) - 1
    node = 0
    placed = new == 0
    while not placed:
        if letter <= key[node]:
            if left[node] == -1:
                left[node] = new
                placed = True
            else:
                node = left[node]
        else:
            if right[node] == -1:
                right[node] = new
                placed = True
            else:
                node = right[node]


def preorder(node, key, left, right):
    if node == -1:
        return ""
    return key[node] + preorder(left[node], key, left, right) + preorder(right[node], key, left, right)


word = input()
target = input()
key = []
left = []
right = []
for letter in word:
    insert(key, left, right, letter)

parent = -1
node = 0
while key[node] != target:
    parent = node
    if target < key[node]:
        node = left[node]
    else:
        node = right[node]

if left[node] == -1:
    replacement = right[node]
elif right[node] == -1:
    replacement = left[node]
else:
    replacement = left[node]
    last = replacement
    while right[last] != -1:
        last = right[last]
    right[last] = right[node]

root = 0
if parent == -1:
    root = replacement
elif left[parent] == node:
    left[parent] = replacement
else:
    right[parent] = replacement

print(preorder(root, key, left, right))
````
