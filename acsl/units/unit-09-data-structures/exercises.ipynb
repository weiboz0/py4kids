# Unit 9 — Data Structures Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  - A popped value is a number or a letter, and `NIL` when the stack or queue was empty.
  - A heap row or a traversal is written as its letters **run together** (`RORN`), or as its numbers separated by single spaces (`12 18 25`).
  - A position in a heap is counted row by row, left to right, with the root at position 1.
  - A depth, a path length or a count is just a whole number, such as `12`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.

The ACSL rules of this unit hold in every exercise: a key equal to one already in a BST goes to the **left**, the root has depth **0**, a "heap" is a **min-heap** unless the question says max-heap, and a heap is built by inserting the items **one at a time**, in the order given.

Each exercise is marked with the lowest ACSL division it belongs to.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 7

_Division: Junior and above._

### The Returns Desk

The first line of input is `STACK` or `QUEUE`, telling you which structure to use.
The second line is a whole number n, the number of commands.
Each of the next n lines is one command on that structure, which starts out empty:

- `PUSH v`, where `v` is a number or a word, puts `v` in;
- `POP` takes an item out, by the rules of the structure.

For every `POP`, print the value taken out, on its own line.
If the structure is empty, print `NIL` instead.

### Constraints

- 1 ≤ n ≤ 100, and at least one command is a `POP`.
- Each `v` is a whole number or a word of capital letters, with no spaces.

### Sample Input

```text
STACK
7
PUSH 12
PUSH 5
POP
PUSH 9
POP
POP
POP
```

### Sample Output

```text
5
9
12
NIL
```

The fourth `POP` finds the stack empty.
With `QUEUE` on the first line, the same commands print `12`, `5`, `9` and `NIL`.

## Exercise 21

_Division: Senior and above._

### Pruning the Tree

The first line of input is a word of capital letters.
The second line is one capital letter that appears in the word **exactly once**.

Insert the letters of the word, in order, into an empty binary search tree (a letter equal to one already in the tree goes to the left).
Then delete the given letter by ACSL's rule:

- a node with no children is removed;
- a node with one child is replaced by that child;
- a node with two children is replaced by its left child, and its right subtree becomes the right child of the right-most node of the left child's tree.

Print the **preorder** traversal of the new tree, with its letters run together.

### Constraints

- The word has between 2 and 30 letters.
- The letter to delete may be the root.

### Sample Input

```text
KEYBOARD
K
```

### Sample Output

```text
EBADYOR
```

`K` is the root, with left child `E` and right child `Y`.
`E` becomes the root, and since `E` has no right child, the `Y` subtree becomes the right child of `E`.
