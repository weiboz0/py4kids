### Vowel Squeeze (predict, then verify)
This ACSL program reads a word in capital letters.
It builds a new word from the letters that are **not** vowels (`A`, `E`, `I`, `O`, `U`), in their order, and counts the vowels it leaves out.

```text
INPUT S
T = ""
v = 0
FOR j = 0 TO len(S) - 1
   c = S[j]
   IF c == "A" || c == "E" || c == "I" || c == "O" || c == "U" THEN
      v = v + 1
   ELSE
      T = T + c
   END IF
NEXT j
OUTPUT T, v
```

### Longest Run (predict, then verify)
A **run** is a group of equal letters next to each other: `BOOKKEEPER` has the runs `B`, `OO`, `KK`, `EE`, `P`, `E` and `R`.
This ACSL program reads a word and finds its longest run.
When several runs are equally long, it keeps the **first** one.

```text
INPUT S
best = 1
bestc = S[0]
run = 1
FOR j = 1 TO len(S) - 1
   IF S[j] == S[j - 1] THEN
      run = run + 1
   ELSE
      run = 1
   END IF
   IF run > best THEN
      best = run
      bestc = S[j]
   END IF
NEXT j
OUTPUT bestc, best
```

1. **Predict:** trace the program by hand on the sample input and write down what it outputs, before you look at the sample output.

The input is one line holding the word.

### Sample Input

```text
BOOKKEEPER
```

```text
O 2
```

### Palindrome Sentences (predict, then verify)
A sentence is a **palindrome** when its letters read the same forwards and backwards once the spaces are taken out: `NEVER ODD OR EVEN` is one.
This ACSL program reads a line, keeps its letters, reverses them and compares.

```text
INPUT S
T = ""
FOR j = 0 TO len(S) - 1
   IF S[j] != " " THEN
      T = T + S[j]
   END IF
NEXT j
R = ""
FOR j = 0 TO len(T) - 1
   R = T[j] + R
NEXT j
IF T == R THEN
   OUTPUT "YES", len(T)
ELSE
   OUTPUT "NO", len(T)
END IF
```

1. **Predict:** trace the program by hand on the sample input and write down what it outputs, before you look at the sample output.

- the line holds at least one letter
- spaces may appear anywhere, including at the start or the end of the line, and next to each other

### The Perfect Shuffle
The program cuts the word into two halves and deals them out like cards, one from each half in turn, until the word comes back to how it started.

```text
O = "ABCDEFGHIJ"
S = O
k = 0
first = ""
done = 0
WHILE done == 0
   h = int(len(S) / 2)
   L = S[:h]
   R = S[h:]
   S = ""
   FOR j = 0 TO h - 1
      S = S + R[j] + L[j]
   NEXT j
   k = k + 1
   IF k == 1 THEN
      first = S
   END IF
   IF S == O THEN
      done = 1
   END IF
END WHILE
OUTPUT k, first
```
