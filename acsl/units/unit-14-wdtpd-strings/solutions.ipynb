# Unit 14 — What Does This Program Do? (Strings) Solutions

Worked answers for the short-answer exercises, and reference programs for the predict-then-verify exercises.
Each programming solution is also saved as a runnable program in the `assets` folder (for example `assets/ex8.py`); run it with `python assets/ex8.py < input.txt`.
Each short-answer program is copied into the `assets` folder too (for example `assets/trace1.py`), translated into Python where it was written in ACSL pseudocode, and a check cell runs it to confirm the answer.
In the translations, `FOR` becomes a `range` that ends one past the ACSL end, and ACSL's `int(x)` becomes `math.floor(x)` (here `//`, since every length is positive).
ACSL's substrings become Python slices like this: `S[:n]` stays `S[:n]` (the first `n` characters); `S[n:]`, the **last** `n` characters, becomes `S[len(S) - n:]`; and `S[a:b]`, positions `a` **through** `b`, becomes `S[a:b + 1]`.

## Exercise 8

**Predict:** in `PROGRAMMING` the vowels are `O`, `A` and `I`, so `v = 3`, and the other letters, in order, make `PRGRMMNG`.
The output is `PRGRMMNG 3`, matching the sample.

**Verify:** `FOR j = 0 TO len(S) - 1` includes its end, so it becomes `range(len(S))`, and `||` becomes `or`.
The two outputs are printed with `print(T, v)`, which puts one space between them.

**Notice:** the fixtures include a word with no vowels (`RHYTHM`, output `RHYTHM 0`), a one-letter word, a word that is mostly vowels (`QUEUE`, output `Q 4`), and words with `Y`, which is not a vowel.

```python
S = input()
T = ""
v = 0
for j in range(len(S)):
    c = S[j]
    if c == "A" or c == "E" or c == "I" or c == "O" or c == "U":
        v = v + 1
    else:
        T = T + c
print(T, v)
```

## Exercise 13

**Predict:** in `BOOKKEEPER`, `run` reaches `2` first at the second `O`, and `2 > 1`, so `bestc = O`, `best = 2`.
The runs `KK` and `EE` also reach `2`, but `2 > 2` is false, so the first run is kept. The output is `O 2`, matching the sample.

**Verify:** `FOR j = 1 TO len(S) - 1` becomes `range(1, len(S))`.

**Notice:** the fixtures include a one-letter word (the loop runs zero times, output `Z 1`), a word with no repeats (`ABC`, output `A 1`), a longest run at the very end (`AAABBBB`, output `B 4`), a word that is one run, and ties (`CCAABB`, output `C 2`).

```python
S = input()
best = 1
bestc = S[0]
run = 1
for j in range(1, len(S)):
    if S[j] == S[j - 1]:
        run = run + 1
    else:
        run = 1
    if run > best:
        best = run
        bestc = S[j]
print(bestc, best)
```

## Exercise 17

**Predict:** `NEVER ODD OR EVEN` without its spaces is `NEVERODDOREVEN`, which has 14 letters and reads the same backwards.
The output is `YES 14`, matching the sample.

**Verify:** `!=` stays `!=`; building `R` by putting each letter in front reverses `T`.
Python's `input()` keeps the spaces at the start and end of the line, and the program drops them with the others.

**Notice:** the fixtures include a single letter, two different letters (`AB`, output `NO 2`), spaces at the start and the end (`  STEP ON NO PETS `, output `YES 12`), two spaces together (`NO  ON`), and a sentence that is a palindrome only if the spaces are ignored (`TOP SPOT`, output `YES 7`).

```python
S = input()
T = ""
for j in range(len(S)):
    if S[j] != " ":
        T = T + S[j]
R = ""
for j in range(len(T)):
    R = T[j] + R
if T == R:
    print("YES", len(T))
else:
    print("NO", len(T))
```
