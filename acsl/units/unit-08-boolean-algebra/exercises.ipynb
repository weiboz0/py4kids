# Unit 8 — Boolean Algebra Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  - A statement question is answered `TRUE` or `FALSE`.
  - A value is `1` or `0`; a "how many" question is answered with just the number, such as `4`.
  - A truth-table question asks for the **result column**: its digits from top to bottom as one string, with the rows in counting-up order (`00`, `01`, `10`, `11` for `A`, `B`), such as `1101`.
  - Ordered pairs and triples are written with no spaces inside, in counting-up order, separated by a comma and a space: `(0,1), (1,1)`. Write `NONE` if there are none.
  - A simplify question asks for a **sum of products**, with the fewest terms and then the fewest letters: letters in alphabetical order inside each term, and terms in order as Lesson 2 describes (`A * B + ~A * C`). Write `1` or `0` if it is always TRUE or always FALSE.
  - An option question is answered with the labels of every correct option, in order, separated by a comma and a space, such as `A, C`, or `NONE`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.

Each exercise is marked with the lowest ACSL division it belongs to.
Do every exercise at or below your division.
**Exercises 1 to 6 are an Elementary mock test:** six questions in 30 minutes, with no programming, using only what Lesson 1 teaches.
Exercises 7 to 9 are extra Elementary practice.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 25

_Division: Intermediate and above._

### Order of Operations

The input is one line holding a Boolean expression over the variables `A`, `B`, `C` and `D`, with no brackets.
Each **literal** is one of the four letters, or a letter with a `~` in front and no space (`~D`).
The operators are `*` (AND), `+` (OR), and XOR and XNOR, which are typed as the words `XOR` and `XNOR` because `⊕` and `⊙` are not on the keyboard.
Every literal and operator is separated from the next by one space, and literals and operators alternate.

Use ACSL's order of operations: first `~`, then `*`, then `XOR` and `XNOR` (on the same level, left to right), and last `+`.
Print the number of ordered quadruples `(A,B,C,D)` that make the expression TRUE.

### Constraints

- The expression has between 1 and 40 literals.
- An expression does not have to use all four letters; a letter it leaves out can be 0 or 1 freely.

### Sample Input

```text
A * B XOR C + ~D
```

### Sample Output

```text
12
```

This is `(A * B) ⊕ C + ~D`.
It is TRUE on the 8 quadruples with `D = 0`, and on 4 of the 8 with `D = 1`.

### Same or Different?

```text
A XOR B
A * ~B + ~A * B
```

```text
EQUIVALENT
```
