# Unit 1 — Computer Number Systems Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  Write a number in a base without the small base subscript: if the question asks for base 8, the answer to "45 in base 8" is `55`.
  Write hex digits as capital letters (`2D`, not `2d`), leave out leading zeros, and write fractions in lowest terms, such as `5/8`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.

Each exercise is marked with the lowest ACSL division it belongs to.
Do every exercise at or below your division.
**Exercises 1 to 9 are an Elementary practice test:** no programming, only what Lesson 1 teaches.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 15

_Division: Junior and above._

### Digits in Another Base

The input is one line holding two integers, `N` and `B`.
Print `N` written in base `B`, then one space, then the **sum of its digits** as a base-10 number.
Digits above 9 are the letters `A` to `F`, and the letter's value (A = 10, …, F = 15) counts in the sum.

### Constraints

- `0 <= N <= 1000000`
- `2 <= B <= 16`

### Sample Input

```text
500 7
```

### Sample Output

```text
1313 8
```

500 is 1313₇, and 1 + 3 + 1 + 3 = 8.

## Exercise 16

_Division: Junior and above._

### The Biggest Card

Each card holds a number and the base it is written in.
Find the card with the largest value.

The first line holds `N`, the number of cards.
Each of the next `N` lines holds one card: the digits of the number (hex digits as capital letters), a space, and its base `B`.
Print the largest value, in base 10.

### Constraints

- `1 <= N <= 50`
- `2 <= B <= 16`, and every digit on a card is a digit of that card's base.
- Every value is between 0 and 1000000 in base 10.

### Sample Input

```text
3
101101 2
55 8
2F 16
```

### Sample Output

```text
47
```

The cards are worth 45, 45 and 47.

## Exercise 17

_Division: Intermediate and above._

### Double Palindromes

A **palindrome** reads the same forwards and backwards, such as 121 or 1001.
Some numbers are palindromes both in base 10 and in base 2: 9 is `9` in base 10 and `1001` in base 2.

The input is one line holding two integers, `A` and `B`.
Print how many whole numbers from `A` to `B` (including both) are palindromes in base 10 **and** in base 2.

### Constraints

- `1 <= A <= B <= 100000`

### Sample Input

```text
1 10
```

### Sample Output

```text
5
```

They are 1, 3, 5, 7 and 9 (in base 2: `1`, `11`, `101`, `111` and `1001`).
