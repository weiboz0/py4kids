## Exercise 11

Divide by 7 again and again, writing down the remainders.

- 500 ÷ 7 = 71 remainder 3
- 71 ÷ 7 = 10 remainder 1
- 10 ÷ 7 = 1 remainder 3
- 1 ÷ 7 = 0 remainder 1

Read the remainders from the bottom up: 1313.
Check: 1 × 343 + 3 × 49 + 1 × 7 + 3 = 343 + 147 + 7 + 3 = 500.

**Answer:** `1313`

````python
def to_base(n, base):
    text = ""
    while n > 0:
        text = "0123456789ABCDEF"[n % base] + text
        n = n // base
    return text

assert str(to_base(500, 7)) == "1313"
````

## Exercise 12

Subtract column by column from the right; a borrow is worth 16.

```text
  C04A
- 3B7F
------
  84CB
```

- Ones: A − F is 10 − 15, too small. Borrow: 10 + 16 = 26, and 26 − 15 = 11 = B.
- Sixteens: the 4 became 3, and 3 − 7 is too small. The 0 to its left has nothing to lend, so the borrow comes from the C (next step). Now 3 + 16 = 19, and 19 − 7 = 12 = C.
- 256s: the 0 had to lend, so borrow from C: the 0 becomes 16, then lends 1 to become 15. 15 − B = 15 − 11 = 4.
- 4096s: C became B, and B − 3 = 11 − 3 = 8.

Check in base 10: C04A₁₆ = 49226, 3B7F₁₆ = 15231, and 49226 − 15231 = 33995 = 84CB₁₆.

**Answer:** `84CB`

````python
assert str(format(0xC04A - 0x3B7F, "X")) == "84CB"
````

## Exercise 15

**Notice:** use repeated division by `B`, adding up each remainder as it is found; the remainder is the digit's value, so a letter digit like `F` adds 15.
`N = 0` is a special case: the loop never runs, so the program must print `0 0` itself.
The fixtures include `0 16` (→ `0 0`) and `255 16` (→ `FF 30`, which catches a sum that skips letter digits).

````python
DIGITS = "0123456789ABCDEF"

parts = input().split()
n = int(parts[0])
base = int(parts[1])
text = ""
total = 0
if n == 0:
    text = "0"
while n > 0:
    digit = n % base
    text = DIGITS[digit] + text
    total = total + digit
    n = n // base
print(text, total)
````

## Exercise 16

**Notice:** each card is changed to base 10 with `int(digits, base)` before comparing.
Comparing the digit strings themselves is wrong: in the fixture `FF 16`, `377 8`, `1000 10` the values are 255, 255 and 1000, so the answer is `1000`.
The fixture with the single card `0 2` catches a largest-so-far that is printed without ever being set.

````python
n = int(input())
best = -1
for i in range(n):
    parts = input().split()
    value = int(parts[0], int(parts[1]))
    if value > best:
        best = value
print(best)
````

## Exercise 17

**Notice:** try every number from `A` to `B`, checking its base-10 digits and its base-2 digits for being a palindrome, by comparing position `i` with position `len - 1 - i`.
Both ends are included, so the fixtures `33 33` and `585 585` (585 is `1001001001` in base 2) must print `1`.
The full range `1 100000` gives `18`.

````python
DIGITS = "0123456789ABCDEF"


def to_base(n, base):
    if n == 0:
        return "0"
    text = ""
    while n > 0:
        text = DIGITS[n % base] + text
        n = n // base
    return text


def is_palindrome(text):
    last = len(text) - 1
    for i in range(len(text)):
        if text[i] != text[last - i]:
            return False
    return True


parts = input().split()
a = int(parts[0])
b = int(parts[1])
total = 0
for number in range(a, b + 1):
    if is_palindrome(str(number)):
        if is_palindrome(to_base(number, 2)):
            total = total + 1
print(total)
````

## Exercise 20

The places after the point in base 2 are 1/2, 1/4, 1/8, 1/16.
In sixteenths they are 8/16, 4/16, 2/16 and 1/16.

11/16 = 8/16 + 2/16 + 1/16, so the 1/2, 1/8 and 1/16 places hold a 1 and the 1/4 place holds a 0.

11/16 = 0.1011₂.
Check: 1/2 + 1/8 + 1/16 = 8/16 + 2/16 + 1/16 = 11/16.

**Answer:** `0.1011`

````python
from fractions import Fraction


def binary_fraction(x):
    text = "0."
    while x != 0:
        x = x * 2
        if x >= 1:
            text = text + "1"
            x = x - 1
        else:
            text = text + "0"
    return text

assert str(binary_fraction(Fraction(11, 16))) == "0.1011"
````
