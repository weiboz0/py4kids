### Leaders From the Right (predict, then verify)
A **leader** is a value that is strictly bigger than every value to its right.
The last value is always a leader, because nothing is to its right.
This ACSL program scans the array from right to left, counting the leaders and keeping the biggest value seen so far.

```text
INPUT n
FOR i = 1 TO n
   INPUT A(i)
NEXT i
count = 1
best = A(n)
FOR i = n - 1 TO 1 STEP -1
   IF A(i) > best THEN
      count = count + 1
      best = A(i)
   END IF
NEXT i
OUTPUT count, best
```

1. **Predict:** trace the program by hand on the sample input and write down what it outputs, before you look at the sample output.
2. **Verify:** translate the program into Python and run it on the sample input.

The first input line holds `n`; the second line holds the `n` numbers, separated by spaces.
Print the number of leaders and the largest value, separated by one space.

### Constraints

- `1 <= n <= 100`
- each value is a whole number from `-1000` to `1000`

### Sample Input

```text
8
5 9 3 9 4 7 2 3
```

### Sample Output

```text
3 9
```

### Frame and Middle (predict, then verify)
A grid of whole numbers has a **frame** (the boxes in its first row, last row, first column or last column) and a **middle** (every other box).
This ACSL program adds up the frame and the middle separately.

```text
INPUT rows, cols
FOR r = 1 TO rows
   FOR c = 1 TO cols
      INPUT A(r, c)
   NEXT c
NEXT r
edge = 0
inner = 0
FOR r = 1 TO rows
   FOR c = 1 TO cols
      IF r == 1 || r == rows || c == 1 || c == cols THEN
         edge = edge + A(r, c)
      ELSE
         inner = inner + A(r, c)
      END IF
   NEXT c
NEXT r
OUTPUT edge, inner
```

### Pascal's Triangle (predict, then verify)
In **Pascal's triangle**, each row starts and ends with `1`, and every other number is the sum of the two numbers just above it.
This ACSL program builds rows `0` to `n` in a 2D array `P`, where `P(r, c)` is the number in row `r` at place `c` (both counted from `0`).
It outputs `P(n, k)` and the sum of row `n`.

```text
INPUT n, k
FOR r = 0 TO n
   FOR c = 0 TO n
      P(r, c) = 0
   NEXT c
NEXT r
P(0, 0) = 1
FOR r = 1 TO n
   P(r, 0) = 1
   FOR c = 1 TO r
      P(r, c) = P(r - 1, c - 1) + P(r - 1, c)
   NEXT c
NEXT r
s = 0
FOR c = 0 TO n
   s = s + P(n, c)
NEXT c
OUTPUT P(n, k), s
```

1. **Predict:** trace the program by hand on the sample input and write down what it outputs, before you look at the sample output.
2. **Verify:** translate the program into Python and run it on the sample input.

The input is one line holding `n` and `k`, separated by one space.

### Constraints

- `0 <= k <= n <= 30`

### Sample Input

```text
5 2
```
