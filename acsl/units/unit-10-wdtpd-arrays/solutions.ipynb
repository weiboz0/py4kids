# Unit 10 — What Does This Program Do? (Arrays) Solutions

Worked answers for the short-answer exercises, and reference programs for the predict-then-verify exercises.
Each programming solution is also saved as a runnable program in the `assets` folder (for example `assets/ex8.py`); run it with `python assets/ex8.py < input.txt`.
Each short-answer program is copied into the `assets` folder too (for example `assets/trace1.py`), translated into Python where it was written in ACSL pseudocode, and a check cell runs it to confirm the answer.
In the translations, `FOR` becomes a `range` that ends one past the ACSL end in the direction of travel, and ACSL's `int(x)` becomes `math.floor(x)`, which rounds down even below zero.
An ACSL array `A(i)` becomes a Python list `A[i]`; when the program starts at position `1`, the list gets an unused slot `0` in front, so the positions match.
A 2D array `A(r, c)` becomes a list of rows, `A[r][c]`.

## Exercise 8

**Predict:** with the sample `5 9 3 9 4 7 2 3`, `best` starts at `A(8) = 3` and the scan goes left.
`2` is not a leader; `7 > 3` is (`count = 2`, `best = 7`); `4` no; `9 > 7` is a leader (`count = 3`, `best = 9`); `3` no; `9 > 9` no; `5` no.
The output is `3 9`, matching the sample.

**Verify:** the list gets an unused slot `0` in front, so `A[i]` matches ACSL's `A(i)`.
`FOR i = n - 1 TO 1 STEP -1` counts down and includes `1`, so it becomes `range(n - 1, 0, -1)`.

**Notice:** the fixtures include `n = 1` (the loop runs zero times, output `1 42`), a decreasing list (every value is a leader), an increasing list (only the last), all values equal (the test is strict `>`, so only the last counts: `1 7`), and all negative values (`best` must start at `A(n)`, not at `0`).

````python
n = int(input())
parts = input().split()
A = [0]
for piece in parts:
    A.append(int(piece))
count = 1
best = A[n]
for i in range(n - 1, 0, -1):
    if A[i] > best:
        count = count + 1
        best = A[i]
print(count, best)
````

## Exercise 14

**Predict:** in the sample grid, the only middle boxes are `6` and `7` (row 2, columns 2 and 3), so `inner = 13`.
The whole grid adds up to `51`, so `edge = 51 - 13 = 38`. The output is `38 13`, matching the sample.

**Verify:** each row list gets an unused slot `0` in front, and an unused row `0` goes on top, so `A[r][c]` matches ACSL's `A(r, c)`.
The `||` in the `IF` becomes `or`.

**Notice:** the fixtures include one box (`1 1`), a single row and a single column (every box is on the frame, so `inner` is `0`), a `2 2` grid (no middle), and a `4 4` grid with negative values on the frame.

````python
first = input().split()
rows = int(first[0])
cols = int(first[1])
A = []
row = []
for c in range(cols + 1):
    row.append(0)
A.append(row)
for r in range(rows):
    row = [0]
    for piece in input().split():
        row.append(int(piece))
    A.append(row)
edge = 0
inner = 0
for r in range(1, rows + 1):
    for c in range(1, cols + 1):
        if r == 1 or r == rows or c == 1 or c == cols:
            edge = edge + A[r][c]
        else:
            inner = inner + A[r][c]
print(edge, inner)
````

## Exercise 17

**Predict:** with `n = 5`, the rows are `1`; `1 1`; `1 2 1`; `1 3 3 1`; `1 4 6 4 1`; `1 5 10 10 5 1`.
`P(5, 2) = 10`, and row `5` adds up to `32`. The output is `10 32`, matching the sample.

**Verify:** the rows and places start at `0`, so no unused slot is needed: `P[r][c]` is `P(r, c)`.
The array is filled with `0`s first, so `P[r - 1][c]` reads `0` when `c = r` (past the end of the row above).

**Notice:** the fixtures include `0 0` (only `P(0, 0)`, output `1 1`), the two ends of a row (`6 0` and `6 6`, both `1 64`), and the largest case `30 15`, whose row sum `1073741824` is `2` to the power `30`.

````python
first = input().split()
n = int(first[0])
k = int(first[1])
P = []
for r in range(n + 1):
    row = []
    for c in range(n + 1):
        row.append(0)
    P.append(row)
P[0][0] = 1
for r in range(1, n + 1):
    P[r][0] = 1
    for c in range(1, r + 1):
        P[r][c] = P[r - 1][c - 1] + P[r - 1][c]
s = 0
for c in range(n + 1):
    s = s + P[n][c]
print(P[n][k], s)
````
