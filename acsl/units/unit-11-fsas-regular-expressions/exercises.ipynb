# Unit 11 — FSAs and Regular Expressions Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  - Most questions give **options** labelled `A`, `B`, `C`, ….
    Write the labels of every option that qualifies, in alphabetical order, separated by a comma and a space: `A, C, D`.
    If no option qualifies, write `NONE`.
  - Write a state exactly as the table names it (`q1`), and a count as a whole number (`6`).
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.
  Do not use Python's `re` module: the point is to write the checking yourself.

Every FSA is given as a transition table: `→` marks the initial state, `(final)` marks each final state, and `—` means there is no transition (a string that needs it is rejected).
In a regular expression, `|` and `U` both mean union, and `λ` is the empty string.

FSAs and regular expressions are on the Intermediate and Senior papers only, so each exercise is marked Intermediate or Senior: the lowest division it belongs to.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 4

_Division: Intermediate and above._

### FSA Simulator

Write a program that reads an FSA and some strings, and says which strings the FSA accepts.

The input is:

1. one line with the name of the initial state
2. one line with the names of the final states, separated by single spaces
3. one line with a whole number `t`, the number of transitions
4. `t` lines, each of the form `state symbol next`: in state `state`, reading `symbol`, move to state `next`
5. one line with a whole number `n`, the number of strings to test
6. `n` lines, each holding one string to test; a line holding just `-` stands for the empty string `λ`

For each of the `n` strings, in order, print `ACCEPT` if the FSA accepts it and `REJECT` if not.
If the FSA needs a transition that is not in the list, the string is rejected.

Store the transitions in a dict whose keys are `(state, symbol)` tuples, as in the lesson.

### Constraints

- There are between 1 and 20 states, and each state name is a word of letters and digits.
- Every symbol is a single small letter or digit.
- `t` is between 1 and 100, and no `(state, symbol)` pair appears twice.
- `n` is between 1 and 20, and each string has between 0 and 30 symbols (written `-` when it has 0).
- A string may contain a symbol that no transition uses.

### Sample Input

```text
A
C
6
A a B
A b A
B a B
B b C
C a B
C b A
4
aab
ba
abab
-
```

### Sample Output

```text
ACCEPT
REJECT
ACCEPT
REJECT
```

The last string is empty, so the FSA stays in `A`, which is not final.

## Exercise 12

_Division: Intermediate and above._

### Pattern Checker

A **pattern** is a line of small letters, where each letter may be followed by one of these marks:

- `?`: the letter appears zero times or once
- `+`: the letter appears one or more times
- `*`: the letter appears zero or more times

A letter with no mark appears exactly once.
A word **matches** the pattern when the whole word can be read as the pattern, from its first character to its last.
For example, `ab?c+d*` matches `acc` and `abcd`, but not `abd` (it needs at least one `c`).

The first line of input is the pattern.
The second line holds a whole number `n`.
Each of the next `n` lines holds one word; a line holding just `-` stands for the empty word `λ`.
For each word, in order, print `ACCEPT` if it matches the pattern and `REJECT` if not.

Write the matching yourself, with a recursive function like the one in Lesson 3.

### Constraints

- The pattern has between 1 and 20 characters. It starts with a letter, and every mark comes straight after a letter.
- `n` is between 1 and 20.
- Each word has between 0 and 20 small letters (written `-` when it has 0).

### Sample Input

```text
ab?c+d*
6
acc
abcd
abd
abbc
acdd
ab
```

### Sample Output

```text
ACCEPT
ACCEPT
REJECT
REJECT
ACCEPT
REJECT
```

For the pattern `a*b?`, the empty word `-` prints `ACCEPT`; for the pattern `a+`, it prints `REJECT`.

## Exercise 17

_Division: Senior and above._

### Full Pattern Checker

Extend your Pattern Checker to the extended syntax of Lesson 3.
A pattern is now a line of **items**, and each item may be followed by one mark `?`, `+` or `*`.
An item is one of:

- a single letter or digit, which matches that character
- `.`, which matches any one character
- a class in square brackets, which matches one character: `[abc]` matches any character listed; `[^abc]` matches any character **not** listed; and inside a class, `x-y` stands for every character from `x` to `y`, as in `[a-d]` or `[A-Z0-9]`

The input is the same as in the Pattern Checker: the pattern, then `n`, then `n` words, with `-` standing for the empty word.
For each word, in order, print `ACCEPT` if it matches the pattern and `REJECT` if not.

### Constraints

- The pattern has between 1 and 30 characters. A class holds letters, digits and ranges `x-y` (with `x` before `y`), and may start with `^`; it is never empty.
- `n` is between 1 and 20.
- Each word has between 0 and 20 letters (small or capital) and digits (written `-` when it has 0).

### Sample Input

```text
Hi?g+h+[^a-ceiou]
7
Highb
HigghhhC
Hihh
Hiihgh
Highd
HgggggghX
Hgh
```

### Sample Output

```text
REJECT
ACCEPT
REJECT
REJECT
ACCEPT
ACCEPT
REJECT
```

Two characters can be compared with `<=` just like numbers: `"a" <= "c"` is `True`.
