# Unit 4 — Prefix/Infix/Postfix Notation Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  Write a prefix or postfix expression with **one space between tokens** (`+ 3 * 4 2`), write powers as `↑`, and never change the order of the operands.
  Write a value as a whole number when it is whole (`12`, not `12.0`), and otherwise as a decimal, such as `13.5`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.
  In program input and output the power sign is typed as `^`.

Each exercise is marked with the lowest ACSL division it belongs to.
Do every exercise at or below your division.
**Exercises 1 to 9 are an Elementary practice test:** no programming, only what Lesson 1 teaches.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 16

_Division: Junior and above._

### The Stack Calculator

The input is one line holding a postfix expression.
Its tokens are separated by single spaces.
Each operand is a whole number, and each operator is one of `+`, `-`, `*`, `/` and `^` (the power `↑`).
Print the value of the expression.

### Constraints

- The expression has between 1 and 40 tokens, and it is a correct postfix expression.
- Every operand is a whole number from 0 to 100.
- Every division divides exactly, every power has an exponent from 0 to 5, and every value along the way is a whole number between -1000000 and 1000000.

### Sample Input

```text
6 8 4 2 - / +
```

### Sample Output

```text
10
```

This is `6 + 8 / (4 - 2)`.
Print the value as a whole number: `10`, not `10.0`.

## Exercise 21

_Division: Intermediate and above._

### The Decimal Calculator

The first line of input gives values to some letters: each letter is followed by its value, all separated by single spaces, such as `A 7 B 2`.
The second line holds a postfix expression, with its tokens separated by single spaces.
Each operand is one of those letters or a whole number, and each operator is one of `+`, `-`, `*`, `/` and `^` (the power `↑`).
A division does **not** have to come out whole.

Print the value of the expression, with each letter replaced by its value: as a whole number if it is whole, and otherwise as a decimal, the way Python prints it (such as `15.5` or `-0.125`).

### Constraints

- The first line gives values to between 1 and 10 different capital letters, and every value is a whole number from 0 to 100.
- The expression has between 1 and 40 tokens, and it is a correct postfix expression.
- Every number operand is a whole number from 0 to 100, and every letter in the expression has a value on the first line.
- Every value along the way is a multiple of 1/8 between -1000000 and 1000000, and never 0 when it is divided by.
- Every power has a whole-number exponent from 0 to 5.

### Sample Input

```text
A 7 B 2
A B / 3 4 * +
```

### Sample Output

```text
15.5
```

With `A = 7` and `B = 2`, this is 7 / 2 + 3 × 4: 3.5 + 12 is 15.5.

## Exercise 23

_Division: Senior and above._

### Prefix to Postfix

The input is one line holding a **prefix** expression.
Its tokens are separated by single spaces.
Each operand is a single capital letter or a whole number, and each operator is one of `+`, `-`, `*`, `/` and `^`.
Print the same expression in **postfix**, with its tokens separated by single spaces.

### Constraints

- The expression has between 1 and 60 tokens, and it is a correct prefix expression.
- Every number is a whole number from 0 to 1000.

### Sample Input

```text
* - A / B C + D ^ E 2
```

### Sample Output

```text
A B C / - D E 2 ^ + *
```

This is `(A - B / C) * (D + E ↑ 2)`.
