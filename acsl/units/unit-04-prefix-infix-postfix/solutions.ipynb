## Exercise 16

**Notice:** a postfix evaluator with the stack idiom: an operand is pushed; an operator takes the top value as its **right** side and the one under it as its **left** side. Every division is exact, so `//` keeps the value a whole number and it prints as `10`, not `10.0`. The `10 3 -`, `2 3 ^` and `3 11 - 2 /` fixtures catch a program that swaps the two sides, and the one-token fixture `7` must print `7`.

```python
OPERATORS = ["+", "-", "*", "/", "^"]


def apply(op, left, right):
    if op == "+":
        return left + right
    if op == "-":
        return left - right
    if op == "*":
        return left * right
    if op == "/":
        return left // right
    return left ** right


stack = []
top = 0
for token in input().split():
    if token in OPERATORS:
        right = stack[top - 1]
        left = stack[top - 2]
        top = top - 2
        value = apply(token, left, right)
    else:
        value = int(token)
    if top == len(stack):
        stack.append(value)
    else:
        stack[top] = value
    top = top + 1
print(stack[top - 1])
```

## Exercise 17

**Notice:** scan the prefix tokens from **right to left** with the same stack. Now an operator takes the top value as its **left** side and the one under it as its **right** side. The `- 10 3`, `^ 2 3` and `/ 12 4` fixtures catch a program that keeps the postfix order of sides.

```python
OPERATORS = ["+", "-", "*", "/", "^"]


def apply(op, left, right):
    if op == "+":
        return left + right
    if op == "-":
        return left - right
    if op == "*":
        return left * right
    if op == "/":
        return left // right
    return left ** right


tokens = input().split()
stack = []
top = 0
for i in range(len(tokens) - 1, -1, -1):
    token = tokens[i]
    if token in OPERATORS:
        left = stack[top - 1]
        right = stack[top - 2]
        top = top - 2
        value = apply(token, left, right)
    else:
        value = int(token)
    if top == len(stack):
        stack.append(value)
    else:
        stack[top] = value
    top = top + 1
print(stack[top - 1])
```

## Exercise 21

**Notice:** read the first line two items at a time, keeping the letters and their values in two parallel lists, then run the stack from Exercise 16 with true division `/`. An operand that is `in` the list of letters is replaced by its value, found by walking along the list; any other operand is a number. A value can now be a decimal such as `3.5`, or a whole number stored as a decimal such as `5.0`, so `show` prints `int(value)` when the value is whole and the decimal otherwise. The `N 2 / 2 *` fixture must print `5`, not `5.0`; `P Q - 8 /` must print `-1`; `K` alone prints its value `9`; and in `X 1` with `1 8 /` the letter is given a value but never used.

```python
OPERATORS = ["+", "-", "*", "/", "^"]


def apply(op, left, right):
    if op == "+":
        return left + right
    if op == "-":
        return left - right
    if op == "*":
        return left * right
    if op == "/":
        return left / right
    return left ** right


def show(value):
    if value == int(value):
        return str(int(value))
    return str(value)


pairs = input().split()
letters = []
values = []
for i in range(0, len(pairs), 2):
    letters.append(pairs[i])
    values.append(int(pairs[i + 1]))

stack = []
top = 0
for token in input().split():
    if token in OPERATORS:
        right = stack[top - 1]
        left = stack[top - 2]
        top = top - 2
        value = apply(token, left, right)
    elif token in letters:
        k = 0
        while letters[k] != token:
            k = k + 1
        value = values[k]
    else:
        value = int(token)
    if top == len(stack):
        stack.append(value)
    else:
        stack[top] = value
    top = top + 1
print(show(stack[top - 1]))
```

## Exercise 23

**Notice:** scan the prefix tokens right to left with a stack of **strings**. An operand is pushed as it is; an operator takes the top string as its left side and the next as its right side, and pushes `left + " " + right + " " + operator`. The `- 100 B` fixture catches swapped sides, and the one-token fixture `A` must print `A`.

```python
OPERATORS = ["+", "-", "*", "/", "^"]

tokens = input().split()
stack = []
top = 0
for i in range(len(tokens) - 1, -1, -1):
    token = tokens[i]
    if token in OPERATORS:
        left = stack[top - 1]
        right = stack[top - 2]
        top = top - 2
        item = left + " " + right + " " + token
    else:
        item = token
    if top == len(stack):
        stack.append(item)
    else:
        stack[top] = item
    top = top + 1
print(stack[top - 1])
```
