## Exercise 1

The input line is `17 5`, so `parts` is `['17', '5']`, `a` is `17` and `b` is `5`.

- `a // b` is whole-number division: 5 goes into 17 three whole times, so it is `3`.
- `a % b` is the remainder: `17 - 3 * 5 = 2`, so it is `2`.
- `a - b * 2` does the multiplication first: `5 * 2 = 10`, then `17 - 10 = 7`.

`print` puts one space between its three values.

**Answer:** `3 2 7`

```python
import subprocess
import sys

result = subprocess.run([sys.executable, "assets/trace1.py"], input='17 5\n', capture_output=True, text=True)
output = result.stdout.rstrip("\n")
assert str(output) == '3 2 7'
```

## Exercise 5

**Notice:** a sentinel loop reads until the `0`; `value % 2 == 0` tells even from odd. The fixtures with only one kind of number catch a counter mix-up.

```python
evens = 0
odds = 0
value = int(input())
while value != 0:
    if value % 2 == 0:
        evens = evens + 1
    else:
        odds = odds + 1
    value = int(input())
print(evens, odds)
```

## Exercise 7

**Notice:** "at least 80" is `>= 80`, so `Cy 80` makes the list. Build the answer as a string of names; if it is still empty at the end, print `NONE`.

```python
n = int(input())
names = ""
for i in range(n):
    parts = input().split()
    name = parts[0]
    score = int(parts[1])
    if score >= 80:
        if names == "":
            names = name
        else:
            names = names + " " + name
if names == "":
    print("NONE")
else:
    print(names)
```

## Exercise 8

**Notice:** each line is its own small sentinel list: walk along the line's parts until the `0`. The fixture with `100 10 1000` catches a program that stops at the digit 0 inside a number instead of at the value `0`.

```python
k = int(input())
answer = ""
for i in range(k):
    parts = input().split()
    total = 0
    position = 0
    value = int(parts[position])
    while value != 0:
        total = total + value
        position = position + 1
        value = int(parts[position])
    if answer == "":
        answer = str(total)
    else:
        answer = answer + " " + str(total)
print(answer)
```

## Exercise 9

Start with `a = 3`, `b = 8`, `c = 5`.

In a tuple assignment, Python works out **every value on the right first**, using the old values, and only then stores them.

1. `a, b = b, a`: the right side is `(8, 3)`, so now `a = 8`, `b = 3`, and `c` is still `5`.
2. `b, c = c, b + a`: the right side uses the values from step 1: `c` is `5`, and `b + a` is `3 + 8 = 11`. So now `b = 5` and `c = 11`.

The trap is to update `b` to `5` first and then compute `b + a = 5 + 8 = 13`; the right side was finished before anything changed.

**Answer:** `8 5 11`

```python
import subprocess
import sys

result = subprocess.run([sys.executable, "assets/trace9.py"], input='3 8 5\n', capture_output=True, text=True)
output = result.stdout.rstrip("\n")
assert str(output) == '8 5 11'
```

## Exercise 10

**Notice:** store each point as a tuple, and keep the best point and its distance. Only a **strictly** larger distance replaces the best, so the first of several tied points wins, as the tie fixture checks.

```python
n = int(input())
points = []
for i in range(n):
    parts = input().split()
    points.append((int(parts[0]), int(parts[1])))
best = points[0]
best_distance = abs(best[0]) + abs(best[1])
for point in points:
    x, y = point
    distance = abs(x) + abs(y)
    if distance > best_distance:
        best = point
        best_distance = distance
print(best[0], best[1])
```

## Exercise 11

**Notice:** complete search over every pair `i < j`, counting `values[i] > values[j]`. Equal values are not out of order (the `2 2 2 2` fixture must print `0`).

```python
n = int(input())
parts = input().split()
values = []
for i in range(n):
    values.append(int(parts[i]))
count = 0
for i in range(n):
    for j in range(i + 1, n):
        if values[i] > values[j]:
            count = count + 1
print(count)
```

## Exercise 12

The values are `3 7 5 5 2 8`, at positions 1 to 6.
The loops try every pair of positions `i < j` and look for pairs that add to `10`.

- `3 + 7 = 10`: positions 1 and 2. `count` becomes 1, `last` becomes `(1, 2)`.
- `3` with anything else: `3 + 5`, `3 + 5`, `3 + 2`, `3 + 8` are not 10.

`last` is replaced each time, so it keeps the **last** pair found. The program adds 1 to `i` and `j` so the positions count from 1.

## Exercise 13

**Notice:** try every pair and keep the largest product. Start `best` at the first pair's product, not at `0`: with `-5 4` the only (and so best) product is `-20`.

```python
n = int(input())
parts = input().split()
values = []
for i in range(n):
    values.append(int(parts[i]))
best = values[0] * values[1]
for i in range(n):
    for j in range(i + 1, n):
        product = values[i] * values[j]
        if product > best:
            best = product
print(best)
```

## Exercise 14

**Notice:** try every count `x` of `A` coins and `y` of `B` coins; the rest must then be paid in `C` coins, which works when the rest is not negative and divides evenly by `C`. That is two loops instead of three, fast enough for `T = 300` with `A = 1`. `T = 0` has one way (no coins).

```python
parts = input().split()
a = int(parts[0])
b = int(parts[1])
c = int(parts[2])
t = int(parts[3])
ways = 0
for x in range(t // a + 1):
    for y in range(t // b + 1):
        rest = t - a * x - b * y
        if rest >= 0:
            if rest % c == 0:
                ways = ways + 1
print(ways)
```

## Exercise 15

**Notice:** complete search over every group `i < j < k`, checking all three triangle rules. The `10 3 4` fixture catches a program that checks only the third stick against the other two.

```python
n = int(input())
parts = input().split()
sticks = []
for i in range(n):
    sticks.append(int(parts[i]))
count = 0
largest = 0
for i in range(n):
    for j in range(i + 1, n):
        for k in range(j + 1, n):
            a = sticks[i]
            b = sticks[j]
            c = sticks[k]
            if a < b + c and b < a + c and c < a + b:
                count = count + 1
                perimeter = a + b + c
                if perimeter > largest:
                    largest = perimeter
print(count, largest)
```
