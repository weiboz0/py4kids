## Exercise 6

_Division: Junior and above._

### Trace the Sentinel Loop

```python
total = 0
count = 0
value = int(input())
while value != 0:
    if value > count:
        total = total + value
    count = count + 1
    value = int(input())
print(count, total)
```

### Input

```text
5
1
2
9
0
```

## Exercise 12

_Division: Intermediate and above._

### Count the Tens

```python
values = []
for part in input().split():
    values.append(int(part))
count = 0
last = (0, 0)
for i in range(len(values)):
    for j in range(i + 1, len(values)):
        if values[i] + values[j] == 10:
            count = count + 1
            last = (i + 1, j + 1)
print(count, last[0], last[1])
```

### Input

```text
3 7 5 5 2 8
```

### Making Change

A shop has coins worth `A`, `B` and `C` cents, and plenty of each.

A way is a choice of how many coins of each value to use: whole numbers `x`, `y` and `z`, each 0 or more, with `A*x + B*y + C*z` equal to `T`.
Two ways are different if any of the three counts is different.

The input is one line holding `A B C T`.
Print the number of ways.

### Constraints

- `1 <= A, B, C <= 50`
- `0 <= T <= 300`

### Sample Input

```text
5
```

The five ways, as `(x, y, z)`, are `(6, 0, 0)`, `(4, 1, 0)`, `(2, 2, 0)`, `(0, 3, 0)` and `(1, 0, 1)`.
