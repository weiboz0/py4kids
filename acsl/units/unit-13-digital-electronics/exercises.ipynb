# Unit 13 — Digital Electronics Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** work it out with pencil and paper and write one exact answer.
  - An output is `1` or `0`; a "how many" question is answered with just the number, such as `5`.
  - A result column is the output's digits from top to bottom, with the rows in counting-up order, such as `1101`.
  - Tuples list the inputs in the order of the `INPUTS` line, with no spaces inside, in counting-up order, separated by a comma and a space: `(0,1,0), (1,1,1)`. Write `NONE` if there are none.
  - A simplify question asks for a **sum of products**, with the fewest terms and then the fewest letters, written as in Unit 8 (`A * B + ~A * C`), or `1` or `0` if the circuit is always TRUE or always FALSE.
  - An option question is answered with the labels of every correct option, in order, separated by a comma and a space, such as `A, C`, or `NONE`. The labels are just names for the options; they have nothing to do with the inputs `A`, `B`, `C` and `D`.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Save it as a `.py` file and run it from a terminal, typing the sample input or feeding it a file.

Every circuit is a netlist, as in Lesson 1: the `INPUTS` line, then one gate per line, and the last gate is the output.
Every input on the `INPUTS` line counts, even one that no gate uses.

Each exercise is marked with the lowest ACSL division it belongs to.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

### Constraints

- Each netlist follows the rules of Exercise 11, and the two `INPUTS` lines are identical.
- The two netlists may use the same gate names.

### Sample Input

```text
4
INPUTS A B
p = NOR(A, B)
q = NOR(A, p)
r = NOR(B, p)
out = NOR(q, r)
1
INPUTS A B
out = XNOR(A, B)
```
