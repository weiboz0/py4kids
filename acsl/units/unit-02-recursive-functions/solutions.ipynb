## Exercise 1

Work down to the base case, then back up.

| Call | Case | Value |
|---|---|---|
| `f(11)` | `11 > 4` | `f(9) + 3` |
| `f(9)` | `9 > 4` | `f(7) + 3` |
| `f(7)` | `7 > 4` | `f(5) + 3` |
| `f(5)` | `5 > 4` | `f(3) + 3` |
| `f(3)` | `3 <= 4` | `3 * 3 - 1 = 8` |

Going back up: `f(5) = 11`, `f(7) = 14`, `f(9) = 17`, `f(11) = 20`.

**Answer:** `20`

```python
def f(x):
    if x > 4:
        return f(x - 2) + 3
    return x * x - 1


assert str(f(11)) == "20"
```

## Exercise 2

Check which line applies at every call: first whether `x < 5`, then whether `x mod 3` is `0`.

| Call | Case | Value |
|---|---|---|
| `f(10)` | `10 mod 3 = 1` | `2 * f(9)` |
| `f(9)` | `9 mod 3 = 0` | `f(7) + 9` |
| `f(7)` | `7 mod 3 = 1` | `2 * f(6)` |
| `f(6)` | `6 mod 3 = 0` | `f(4) + 6` |
| `f(4)` | `4 < 5` | `4 + 1 = 5` |

Going back up: `f(6) = 5 + 6 = 11`, `f(7) = 2 * 11 = 22`, `f(9) = 22 + 9 = 31`, `f(10) = 2 * 31 = 62`.

**Answer:** `62`

```python
def f(x):
    if x < 5:
        return x + 1
    if x % 3 == 0:
        return f(x - 2) + x
    return 2 * f(x - 1)


assert str(f(10)) == "62"
```

## Exercise 3

Both arguments change at every call: `x` drops by `y`, and `y` grows by 1.

| Call | Case | Value |
|---|---|---|
| `f(20, 2)` | `20 > 2` | `f(18, 3) + 2` |
| `f(18, 3)` | `18 > 3` | `f(15, 4) + 2` |
| `f(15, 4)` | `15 > 4` | `f(11, 5) + 2` |
| `f(11, 5)` | `11 > 5` | `f(6, 6) + 2` |
| `f(6, 6)` | `6 <= 6` | `6 * 6 = 36` |

Four calls each add `2` on the way back up: `36 + 8 = 44`.

**Answer:** `44`

```python
def f(x, y):
    if x > y:
        return f(x - y, y + 1) + 2
    return x * y


assert str(f(20, 2)) == "44"
```

## Exercise 4

Each call makes two more calls, so work **upward** from the base cases instead of drawing the whole tree.

- Base cases: `f(0) = 1`, `f(1) = 2`, `f(2) = 3`.
- `f(3) = f(2) + f(1) - 1 = 3 + 2 - 1 = 4`
- `f(4) = 4 + 3 - 1 = 6`
- `f(5) = 6 + 4 - 1 = 9`
- `f(6) = 9 + 6 - 1 = 14`
- `f(7) = 14 + 9 - 1 = 22`

**Answer:** `22`

```python
def f(x):
    if x > 2:
        return f(x - 1) + f(x - 2) - 1
    return x + 1


assert str(f(7)) == "22"
```

## Exercise 5

**Notice:** the test is `x > 8`, so `g(8)` is already a base case (`23`) while `g(9) = g(4) + 9 = 20`; the fixtures for `8` and `9` catch a program that uses `>=`.
A negative `x` goes straight to the base case, and `x = 2000` takes 400 calls, well inside Python's recursion limit.

```python
def g(x):
    if x > 8:
        return g(x - 5) + x
    return 3 * x - 1


x = int(input())
print(g(x))
```

## Exercise 6

Each call divides by 3 (rounding down) and adds the remainder.

| Call | `floor(x / 3)` | `x mod 3` | Value |
|---|---|---|---|
| `f(50)` | `16` | `2` | `f(16) + 2` |
| `f(16)` | `5` | `1` | `f(5) + 1` |
| `f(5)` | `1` | `2` | `f(1) + 2` |
| `f(1)` | | | `1` (since `1 < 3`) |

Going back up: `f(5) = 3`, `f(16) = 4`, `f(50) = 6`.
(This adds up the digits of 50 written in base 3: `50 = 1212` in base 3, and `1 + 2 + 1 + 2 = 6`.)

**Answer:** `6`

```python
def f(x):
    if x >= 3:
        return f(x // 3) + x % 3
    return x


assert str(f(50)) == "6"
```

## Exercise 8

**Notice:** there are three base cases, `t(0) = 0`, `t(1) = 1` and `t(2) = 2`, all covered by the single line `n <= 2`; fixtures for `0`, `2` and `3` check them.
`t(20)` makes a few hundred thousand calls, which still finishes in well under a second.

```python
def t(n):
    if n > 2:
        return t(n - 1) + t(n - 2) + t(n - 3)
    return n


n = int(input())
print(t(n))
```

## Exercise 9

**Notice:** check the base case (`x <= 0 or y <= 0`) **first**, then `x > y`.
The `1 1` fixture goes `f(1, 1)` to `f(0, -2) = -2`, so it prints `-1`; a program that forgets that `x + y` can be negative, or tests the cases in the wrong order, fails it.

```python
def f(x, y):
    if x <= 0 or y <= 0:
        return x + y
    if x > y:
        return f(x - 2, y - 1) + 3
    return f(x - 1, y - 3) + x


parts = input().split()
x = int(parts[0])
y = int(parts[1])
print(f(x, y))
```

## Exercise 10

`f` and `g` call each other, so keep one table and write down which function each row is.

| Call | Case | Value |
|---|---|---|
| `f(9)` | `9 > 0` | `g(8) + 2` |
| `g(8)` | `8 > 1` | `2 * f(6)` |
| `f(6)` | `6 > 0` | `g(5) + 2` |
| `g(5)` | `5 > 1` | `2 * f(3)` |
| `f(3)` | `3 > 0` | `g(2) + 2` |
| `g(2)` | `2 > 1` | `2 * f(0)` |
| `f(0)` | `0 <= 0` | `1` |

Going back up: `g(2) = 2`, `f(3) = 4`, `g(5) = 8`, `f(6) = 10`, `g(8) = 20`, `f(9) = 22`.

**Answer:** `22`

```python
def f(x):
    if x > 0:
        return g(x - 1) + 2
    return 1


def g(x):
    if x > 1:
        return 2 * f(x - 2)
    return x + 3


assert str(f(9)) == "22"
```

## Exercise 11

**Notice:** `floor(x / 2)` is `x // 2` in Python.
Both `0` and `1` are base cases of `p` (they print `1`), and `p(2) = q(1) + 2 = (p(0) - 1) + 2 = 2`.
Each pair of calls halves `x`, so even `x = 1000000` needs only about 40 calls.

```python
def p(x):
    if x > 1:
        return q(x // 2) + x
    return 1


def q(x):
    if x > 0:
        return p(x - 1) - 1
    return 0


x = int(input())
print(p(x))
```

## Exercise 12

Work with `abs(x)` at each step: the argument of the next call is `abs(x) - 7`, which can be negative again.

| Call | `abs(x)` | Case | Value |
|---|---|---|---|
| `f(-20)` | `20` | `20 > 5` | `f(13) + 2` |
| `f(13)` | `13` | `13 > 5` | `f(6) + 2` |
| `f(6)` | `6` | `6 > 5` | `f(-1) + 2` |
| `f(-1)` | `1` | `1 <= 5` | `-1 + 4 = 3` |

Note the base case uses `x`, not `abs(x)`: `f(-1) = -1 + 4 = 3`.
Going back up: `f(6) = 5`, `f(13) = 7`, `f(-20) = 9`.

**Answer:** `9`

```python
def f(x):
    if abs(x) > 5:
        return f(abs(x) - 7) + 2
    return x + 4


assert str(f(-20)) == "9"
```

## Exercise 13

Let `C(x)` be the number of calls made to work out `f(x)`, counting the call itself.

- A base case (`x <= 1`) makes no more calls: `C(0) = 1` and `C(1) = 1`.
- Otherwise `f(x)` is one call plus all the calls for `f(x - 1)` and for `f(x - 2)`: `C(x) = 1 + C(x - 1) + C(x - 2)`.

Working upward:

| `x` | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|---|
| `C(x)` | 1 | 1 | 3 | 5 | 9 | 15 | 25 |

**Answer:** `25`

```python
calls = 0


def f(x):
    global calls
    calls = calls + 1
    if x > 1:
        return f(x - 1) + f(x - 2)
    return x


f(6)
assert str(calls) == "25"
```

## Exercise 14

**Notice:** read `N` first, then one number per line, and print one answer per line.
Odd `x` becomes `x - 3`, which is even, so the next step halves it; even `x = 1000000000` needs only about 60 calls.
The fixtures `10` (the smallest recursive case, `f(5) + 1 = 6`) and `11` (odd: `f(8) + 2 = 10`) check the boundaries.

```python
def f(x):
    if x < 10:
        return x
    if x % 2 == 0:
        return f(x // 2) + 1
    return f(x - 3) + 2


n = int(input())
for i in range(n):
    x = int(input())
    print(f(x))
```

## Exercise 15

Fill a table with a row for each `x` and a column for each `y`.
Row `x = 0` and column `y = 0` are all `1`.
Every other cell is the sum of three cells: the one **above** it (`x - 1`), the one to its **left** (`y - 1`), and the one **diagonally up-left**.

| | `y = 0` | `y = 1` | `y = 2` | `y = 3` | `y = 4` |
|---|---|---|---|---|---|
| `x = 0` | 1 | 1 | 1 | 1 | 1 |
| `x = 1` | 1 | 3 | 5 | 7 | 9 |
| `x = 2` | 1 | 5 | 13 | 25 | 41 |
| `x = 3` | 1 | 7 | 25 | 63 | 129 |

For example `f(3, 4) = f(2, 4) + f(3, 3) + f(2, 3) = 41 + 63 + 25 = 129`.

**Answer:** `129`

```python
def f(x, y):
    if x > 0 and y > 0:
        return f(x - 1, y) + f(x, y - 1) + f(x - 1, y - 1)
    return 1


assert str(f(3, 4)) == "129"
```

## Exercise 16

**Notice:** each call returns a pair.
A base case (`x < 3`) is worth `x + 1` and counts as **one** call; a recursive case adds the two values, and its call count is the two children's counts **plus 1** for itself.
The fixtures `0` and `2` (base cases, `1 1` and `3 1`) and `3` (`4 3`) catch a program that forgets to count the call itself.

```python
def f(x):
    if x >= 3:
        value1, calls1 = f(x - 1)
        value2, calls2 = f(x - 3)
        return (value1 + value2, calls1 + calls2 + 1)
    return (x + 1, 1)


x = int(input())
value, calls = f(x)
print(value, calls)
```
