# Unit 2 — Recursive Functions Exercises

There are two kinds of exercises here, just as on an ACSL contest.

- **Short-answer:** a recursive definition (or a short program) is given.
  Work out the value by hand with a call table, and write down **one exact answer**.
  Do not type it into Python until you have committed to an answer; working it by hand is the skill the contest tests.
- **Programming:** write a complete program that reads the input with `input()` and prints exactly the output described.
  Every programming exercise gives you the definition: your job is to turn it into a recursive Python function.
  Save it as a `.py` file and run it from a terminal, for example `python my_solution.py < my_input.txt`.
  Print nothing extra.

The definitions use the notation from the lesson: `*` for times, `x mod 3` for the remainder, `floor(x / 2)` for division rounded down, and `abs(x)` for the size of `x` without its sign.

Each exercise is marked with the lowest ACSL division it belongs to: Junior, Intermediate or Senior.
Do every exercise at or below your division.
Exercises marked *Challenge* are extra practice for when you want more.

## Exercise 5

_Division: Junior and above._

### Implement the Definition

Write a program that reads one whole number `x` and prints `g(x)`, where

```text
g(x) = { g(x - 5) + x      if x > 8
       { 3 * x - 1         if x <= 8
```

Write `g` as a recursive Python function.

### Input

One line holding the integer `x`.

### Output

One line holding `g(x)`.

### Constraints

- `-100 <= x <= 2000`

### Sample Input

```text
20
```

### Sample Output

```text
59
```

The calls go `g(20)`, `g(15)`, `g(10)`, `g(5)`; `g(5) = 14`, so `g(10) = 24`, `g(15) = 39` and `g(20) = 59`.

## Exercise 9

_Division: Junior and above._

### Two Numbers Meet

Write a program that reads two whole numbers `x` and `y` and prints `f(x, y)`, where

```text
f(x, y) = { x + y                  if x <= 0 or y <= 0
          { f(x - 2, y - 1) + 3    if x > 0 and y > 0 and x > y
          { f(x - 1, y - 3) + x    if x > 0 and y > 0 and x <= y
```

Write `f` as a recursive Python function with two parameters.

### Input

One line holding `x` and `y`, separated by one space.

### Output

One line holding `f(x, y)`.

### Constraints

- `1 <= x <= 100`
- `1 <= y <= 100`

### Sample Input

```text
10 6
```

### Sample Output

```text
14
```

The calls go `f(10, 6)`, `f(8, 5)`, `f(6, 4)`, `f(4, 3)`, `f(2, 2)`, `f(1, -1)`.
`f(1, -1) = 0`, so `f(2, 2) = 0 + 2 = 2`, and each call above it adds `3`, giving `14`.

## Exercise 14

_Division: Intermediate and above._

### Many Questions

The function `f` is defined for whole numbers `x >= 0`:

```text
f(x) = { x                     if x < 10
       { f(floor(x / 2)) + 1   if x >= 10 and x mod 2 = 0
       { f(x - 3) + 2          if x >= 10 and x mod 2 = 1
```

The first line of the input holds `N`, the number of questions.
Each of the next `N` lines holds one whole number `x`.
For each `x`, in order, print `f(x)` on its own line.

### Constraints

- `1 <= N <= 100`
- `0 <= x <= 1000000000`

### Sample Input

```text
3
7
40
99
```

### Sample Output

```text
7
8
12
```

For `40`: `f(40) = f(20) + 1`, `f(20) = f(10) + 1`, `f(10) = f(5) + 1`, and `f(5) = 5`, so `f(40) = 8`.

### Value and Calls

The function `f` is defined for whole numbers `x >= 0`:

```text
f(x) = { f(x - 1) + f(x - 3)     if x >= 3
       { x + 1                   if x < 3
```

Read one whole number `x`.
Print two numbers separated by one space: the value `f(x)`, and the total number of calls to `f` made when `f(x)` is worked out by following the definition exactly, counting the first call `f(x)` itself.

Write `f` as a recursive function that returns a tuple `(value, calls)`.

### Constraints

- `0 <= x <= 25`

### Sample Input

```text
6
```

### Sample Output

```text
13 11
```

`f(6)` calls `f(5)` and `f(3)`; `f(5)` calls `f(4)` and `f(2)`; and so on down to calls with `x < 3`, which make no further calls.
