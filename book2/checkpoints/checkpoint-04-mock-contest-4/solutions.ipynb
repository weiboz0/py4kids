## Question 1 — Count the Islands

```python
def solve(data):
    lines = data.split("\n")
    header = lines[0].split()
    rows = int(header[0])
    cols = int(header[1])
    grid = []
    r = 0
    while r < rows:
        grid.append(lines[1 + r])
        r = r + 1
    visited = set()

    def fill(cr, cc):
        visited.add((cr, cc))
        neighbours = [(cr - 1, cc), (cr + 1, cc), (cr, cc - 1), (cr, cc + 1)]
        for spot in neighbours:
            nr = spot[0]
            nc = spot[1]
            if 0 <= nr and nr < rows and 0 <= nc and nc < cols:
                if grid[nr][nc] == "#" and (nr, nc) not in visited:
                    fill(nr, nc)

    count = 0
    r = 0
    while r < rows:
        c = 0
        while c < cols:
            if grid[r][c] == "#" and (r, c) not in visited:
                count = count + 1
                fill(r, c)
            c = c + 1
        r = r + 1
    return str(count)


assert solve("3 4\n#.#.\n##..\n...#") == "3"
assert solve("1 1\n.") == "0"
assert solve("2 2\n#.\n.#") == "2"
assert solve("1 5\n#.#.#") == "3"
```

## Question 2 — Fewest Steps

```python
from collections import deque


def solve(data):
    lines = data.split("\n")
    header = lines[0].split()
    rows = int(header[0])
    cols = int(header[1])
    grid = []
    r = 0
    while r < rows:
        grid.append(lines[1 + r])
        r = r + 1
    start = (0, 0)
    target = (0, 0)
    r = 0
    while r < rows:
        c = 0
        while c < cols:
            if grid[r][c] == "S":
                start = (r, c)
            if grid[r][c] == "T":
                target = (r, c)
            c = c + 1
        r = r + 1
    queue = deque()
    queue.append((start[0], start[1], 0))
    visited = set()
    visited.add(start)
    while len(queue) > 0:
        item = queue.popleft()
        cr = item[0]
        cc = item[1]
        dist = item[2]
        if (cr, cc) == target:
            return str(dist)
        neighbours = [(cr - 1, cc), (cr + 1, cc), (cr, cc - 1), (cr, cc + 1)]
        for spot in neighbours:
            nr = spot[0]
            nc = spot[1]
            if 0 <= nr and nr < rows and 0 <= nc and nc < cols:
                if grid[nr][nc] != "#" and (nr, nc) not in visited:
                    visited.add((nr, nc))
                    queue.append((nr, nc, dist + 1))
    return "-1"


assert solve("1 3\nS.T") == "2"
assert solve("2 3\nS#T\n...") == "4"
assert solve("2 2\nS#\n#T") == "-1"
assert solve("4 4\nT.S.\n.#..\n....\n.###") == "2"
```

## Question 3 — One Network?

```python
def solve(data):
    lines = data.split("\n")
    header = lines[0].split()
    n = int(header[0])
    m = int(header[1])
    adj = {}
    node = 1
    while node <= n:
        adj[node] = []
        node = node + 1
    i = 0
    while i < m:
        parts = lines[1 + i].split()
        u = int(parts[0])
        v = int(parts[1])
        adj[u].append(v)
        adj[v].append(u)
        i = i + 1
    visited = set()

    def walk(u, seen):
        seen.add(u)
        for nb in adj[u]:
            if nb not in seen:
                walk(nb, seen)

    walk(1, visited)
    if len(visited) == n:
        return "YES"
    return "NO"


assert solve("3 2\n1 2\n2 3") == "YES"
assert solve("3 1\n1 2") == "NO"
assert solve("1 0") == "YES"
assert solve("4 2\n1 2\n3 4") == "NO"
assert solve("2 1\n2 1") == "YES"
```

## Question 4 — Exact Budget Pair

```python
def solve(data):
    tokens = data.split()
    n = int(tokens[0])
    target = int(tokens[1])
    prices = []
    i = 0
    while i < n:
        prices.append(int(tokens[2 + i]))
        i = i + 1
    prices = sorted(prices)
    lo = 0
    hi = n - 1
    while lo < hi:
        total = prices[lo] + prices[hi]
        if total == target:
            return "YES"
        if total < target:
            lo = lo + 1
        else:
            hi = hi - 1
    return "NO"


assert solve("4 10\n1 8 3 7") == "YES"
assert solve("4 10\n1 2 3 4") == "NO"
assert solve("2 9\n4 5") == "YES"
assert solve("5 6\n1 1 2 4 5") == "YES"
assert solve("3 8\n1 4 6") == "NO"
```

## Question 5 — Longest Affordable Streak

```python
def solve(data):
    tokens = data.split()
    n = int(tokens[0])
    budget = int(tokens[1])
    cost = []
    i = 0
    while i < n:
        cost.append(int(tokens[2 + i]))
        i = i + 1
    left = 0
    window = 0
    best = 0
    right = 0
    while right < n:
        window = window + cost[right]
        while window > budget:
            window = window - cost[left]
            left = left + 1
        span = right - left + 1
        if span > best:
            best = span
        right = right + 1
    return str(best)


assert solve("5 7\n2 3 1 2 4") == "3"
assert solve("3 5\n6 6 6") == "0"
assert solve("4 10\n1 2 3 4") == "4"
assert solve("5 5\n5 1 1 1 5") == "3"
assert solve("5 9\n2 2 3 3 1") == "4"
```

## Question 6 — Astronomical Power Modulo M

```python
def solve(data):
    tokens = data.split()
    base = int(tokens[0])
    modulus = int(tokens[1])
    exponent = int(tokens[2])
    answer = 1 % modulus
    current = base % modulus
    while exponent > 0:
        if exponent % 2 == 1:
            answer = answer * current % modulus
        current = current * current % modulus
        exponent = exponent // 2
    return str(answer)


assert solve("2 5 3") == "3"
assert solve("20 7 1") == "6"
assert solve("9 1 0") == "0"
assert solve("3 11 0") == "1"
assert solve("2 5 4") == "1"
assert solve("7 13 1000000000000000000") == "9"
```

## Question 7 — Trace the Traversal

```python
from collections import deque


def solve(data):
    lines = data.split("\n")
    header = lines[0].split()
    n = int(header[0])
    m = int(header[1])
    start = int(header[2])
    adj = {}
    node = 1
    while node <= n:
        adj[node] = []
        node = node + 1
    i = 0
    while i < m:
        parts = lines[1 + i].split()
        u = int(parts[0])
        v = int(parts[1])
        adj[u].append(v)
        i = i + 1
    visited = {start}
    queue = deque()
    queue.append(start)
    order = ""
    while len(queue) > 0:
        u = queue.popleft()
        if len(order) > 0:
            order = order + " "
        order = order + str(u)
        for nb in adj[u]:
            if nb not in visited:
                visited.add(nb)
                queue.append(nb)
    return order


assert solve("3 2 1\n1 2\n1 3") == "1 2 3"
assert solve("4 3 1\n1 2\n1 3\n3 4") == "1 2 3 4"
assert solve("4 4 1\n1 2\n1 3\n2 4\n3 4") == "1 2 3 4"
```
