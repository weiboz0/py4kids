## Question 6

### Astronomical Power Modulo M

Given a base `A`, a modulus `M`, and an exponent `E`, return `A` to the power `E`, modulo `M`.

Use repeated squaring and reduce modulo `M` after every multiplication and every squaring; computing the
unreduced power first is infeasible for the enormous exponent. **Do not use a built-in modular-power
shortcut** (no three-argument `pow`). Your algorithm must use O(log E) modular multiplications.

Input: `A M E` on one line. Return the result (between `0` and `M - 1`) as an integer string.

**Constraints:** `0 <= A <= 10^18`, `1 <= M <= 10^18`, `0 <= E <= 10^18`.

**Sample Input**

```text
7 13 1000000000000000000
```

**Sample Output**

```text
9
```

### Trace the Traversal

Trace this exact breadth-first routine and predict what it prints. Neighbours are visited in the order the
edges are given.

```text
seed visited with the single start node          # visited = {start}
put start on the back of a FIFO queue
while the queue is not empty:
    take a node u from the FRONT of the queue
    print u
    for each out-neighbour v of u, in input order:
        if v is not visited:
            mark v visited and put it on the BACK of the queue
```

ids in the order the routine prints them, space-separated.

**Constraints:** `1 <= N <= 30`, `0 <= M <= 200`.

**Sample Input**

```text
3 2 1
1 2
1 3
```

```text
1 2 3
```
