## Question 1

```python
def solve(data: str) -> str:
    parts = data.split()
    value_count = int(parts[0])
    values = []
    used = []
    i = 0
    while i < value_count:
        values.append(int(parts[i + 1]))
        used.append(False)
        i = i + 1

    def count_arrangements(depth, previous_index):
        if depth == value_count:
            return 1
        count = 0
        i = 0
        while i < value_count:
            allowed = previous_index == -1
            if previous_index != -1:
                allowed = abs(values[i] - values[previous_index]) != 1
            if used[i] == False and allowed:
                used[i] = True
                count = count + count_arrangements(depth + 1, i)
                used[i] = False
            i = i + 1
        return count

    return str(count_arrangements(0, -1))

assert solve("3\n1 2 4\n") == "2"
assert solve("2\n1 2\n") == "0"
assert solve("4\n1 2 3 10\n") == "4"
```

## Question 2

```python
from collections import deque

def solve(data: str) -> str:
    tokens = data.split()
    stack = deque()
    i = 0
    while i < len(tokens):
        token = tokens[i]
        if token == "+" or token == "-" or token == "*":
            right = stack.popleft()
            left = stack.popleft()
            if token == "+":
                value = left + right
            elif token == "-":
                value = left - right
            else:
                value = left * right
            stack.appendleft(value)
        else:
            stack.appendleft(int(token))
        i = i + 1
    return str(stack.popleft())

assert solve("8 3 - 2 * 7 +\n") == "17"
assert solve("5 3 -\n") == "2"
assert solve("9 2 - 4 * 3 +\n") == "31"
assert solve("-6\n") == "-6"
```

## Question 3

```python
def solve(data: str) -> str:
    number = int(data.strip())
    if number == 0:
        return "0"
    digits = ""
    while number > 0:
        digit = str(number % 2)
        digits = digit + digits
        number = number // 2
    return digits

assert solve("13\n") == "1101"
assert solve("0\n") == "0"
assert solve("8\n") == "1000"
assert solve("21\n") == "10101"
```

## Question 4

```python
def solve(data: str) -> str:
    parts = data.split()
    item_count = int(parts[0])
    target = int(parts[1])
    weights = []
    i = 0
    while i < item_count:
        weights.append(int(parts[i + 2]))
        i = i + 1

    matching = 0
    for mask in range(1 << item_count):
        total = 0
        i = 0
        while i < item_count:
            if mask & (1 << i):
                total = total + weights[i]
            i = i + 1
        if total == target:
            matching = matching + 1
    return str(matching)

assert solve("6 73\n1 3 7 15 31 63\n") == "1"
assert solve("3 0\n2 4 8\n") == "1"
assert solve("3 14\n2 4 8\n") == "1"
assert solve("5 10\n1 3 7 15 31\n") == "1"
```

## Question 5

```python
def solve(data: str) -> str:
    parts = data.split()
    value_count = int(parts[0])

    def gcd(a, b):
        while b != 0:
            a, b = b, a % b
        return a

    common = int(parts[1])
    multiple = int(parts[1])
    i = 1
    while i < value_count:
        value = int(parts[i + 1])
        common = gcd(common, value)
        pair_gcd = gcd(multiple, value)
        multiple = multiple // pair_gcd * value
        i = i + 1
    result = str(common)
    result = result + " "
    result = result + str(multiple)
    return result

assert solve("4\n12 18 24 30\n").split() == "6 360".split()
assert solve("3\n5 7 11\n").split() == "1 385".split()
assert solve("4\n4 8 16 32\n").split() == "4 32".split()
assert solve("3\n12 12 12\n").split() == "12 12".split()
```

## Question 6

```python
def solve(data: str) -> str:
    limit = int(data.strip())
    is_prime = []
    i = 0
    while i <= limit:
        is_prime.append(True)
        i = i + 1
    is_prime[0] = False
    is_prime[1] = False

    prime = 2
    while prime * prime <= limit:
        if is_prime[prime]:
            multiple = prime * prime
            while multiple <= limit:
                is_prime[multiple] = False
                multiple = multiple + prime
        prime = prime + 1

    count = 0
    i = 2
    while i <= limit:
        if is_prime[i]:
            count = count + 1
        i = i + 1
    return str(count)

assert solve("49\n") == "15"
assert solve("2\n") == "1"
assert solve("10\n") == "4"
```

## Question 7

```python
def solve(data: str) -> str:
    parts = data.split()
    node_count = int(parts[0])
    root = int(parts[1])
    values = []
    left = []
    right = []
    i = 0
    while i < node_count:
        values.append(int(parts[i * 3 + 2]))
        left.append(int(parts[i * 3 + 3]))
        right.append(int(parts[i * 3 + 4]))
        i = i + 1

    result = ""

    def walk(child):
        nonlocal result
        if child == -1:
            return
        if result != "":
            result = result + " "
        result = result + str(values[child])
        walk(left[child])
        walk(right[child])

    walk(root)
    return result

assert solve("4 2\n10 1 3\n30 -1 -1\n40 0 -1\n20 -1 -1\n").split() == "40 10 30 20".split()
assert solve("1 0\n42 -1 -1\n") == "42"
assert solve("4 0\n9 1 -1\n7 2 -1\n5 3 -1\n3 -1 -1\n").split() == "9 7 5 3".split()
assert solve("5 0\n50 1 2\n80 3 -1\n10 -1 4\n70 -1 -1\n20 -1 -1\n").split() == "50 80 70 10 20".split()
```
