## Question 3

### Best Trading Gain

The first line is `N`, and the second line contains `N` daily prices in time order. Choose one day to buy and a later day to sell, and return the greatest possible gain `sell price - buy price`; return `0` if every later price is lower than or equal to every possible earlier buying price. ### Constraints

`2 <= N <= 200000`.
