## Exercise 1

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    q = int(parts[1])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 2]))
        i = i + 1
    values = sorted(values)
    output = ""
    i = 0
    while i < q:
        target = int(parts[n + i + 2])
        lo = 0
        hi = n - 1
        answer = "NO"
        while lo <= hi:
            mid = (lo + hi) // 2
            if values[mid] == target:
                answer = "YES"
                lo = hi + 1
            elif values[mid] < target:
                lo = mid + 1
            else:
                hi = mid - 1
        if i > 0:
            output = output + "\n"
        output = output + answer
        i = i + 1
    return output

assert solve("5 4\n11 2 7 4 19\n5 2 12 19\n").split() == "NO\nYES\nNO\nYES\n".split()
assert solve("4 4\n9 5 1 7\n1 9 6 10\n").split() == "YES\nYES\nNO\nNO\n".split()
```

## Exercise 2

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    q = int(parts[1])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 2]))
        i = i + 1
    values = sorted(values)

    def lower_bound(target):
        lo = 0
        hi = n
        while lo < hi:
            mid = (lo + hi) // 2
            if values[mid] < target:
                lo = mid + 1
            else:
                hi = mid
        return lo

    def upper_bound(target):
        lo = 0
        hi = n
        while lo < hi:
            mid = (lo + hi) // 2
            if values[mid] <= target:
                lo = mid + 1
            else:
                hi = mid
        return lo

    output = ""
    i = 0
    while i < q:
        target = int(parts[n + i + 2])
        count = upper_bound(target) - lower_bound(target)
        if i > 0:
            output = output + "\n"
        output = output + str(count)
        i = i + 1
    return output

assert solve("8 4\n3 9 1 9 6 9 4 9\n3 5 10 9\n").split() == "1\n0\n0\n4\n".split()
assert solve("5 3\n8 2 8 1 8\n1 3 8\n").split() == "1\n0\n3\n".split()
```

## Exercise 4

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    target = int(parts[1])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 2]))
        i = i + 1
    values = sorted(values)
    answer = "NO"
    i = 0
    while i < n - 1 and answer == "NO":
        complement = target - values[i]
        lo = i + 1
        hi = n - 1
        while lo <= hi:
            mid = (lo + hi) // 2
            if values[mid] == complement:
                answer = "YES"
                lo = hi + 1
            elif values[mid] < complement:
                lo = mid + 1
            else:
                hi = mid - 1
        i = i + 1
    return answer

assert solve("5 23\n1 8 12 4 19\n").split() == "YES\n".split()
assert solve("3 10\n5 2 9\n").split() == "NO\n".split()
assert solve("4 10\n8 5 1 5\n").split() == "YES\n".split()
assert solve("4 5\n10 3 2 20\n").split() == "YES\n".split()
assert solve("4 100\n1 2 40 60\n").split() == "YES\n".split()
```

## Exercise 5

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 1]))
        i = i + 1
    values = sorted(values)
    best = abs(values[1] - values[0])
    i = 2
    while i < n:
        gap = values[i] - values[i - 1]
        if gap < best:
            best = gap
        i = i + 1
    return str(best)

assert solve("6\n2 8 13 19 30 31\n").split() == "1\n".split()
assert solve("3\n10 -5 9\n").split() == "1\n".split()
assert solve("2\n100 -100\n").split() == "200\n".split()
```

## Exercise 6

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    target = int(parts[1])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 2]))
        i = i + 1
    answer = "NO"
    i = 0
    while i < n - 2 and answer == "NO":
        j = i + 1
        while j < n - 1 and answer == "NO":
            k = j + 1
            while k < n and answer == "NO":
                if values[i] + values[j] + values[k] == target:
                    answer = "YES"
                k = k + 1
            j = j + 1
        i = i + 1
    return answer

assert solve("7 59\n2 5 9 14 20 31 40\n").split() == "YES\n".split()
assert solve("3 12\n2 5 9\n").split() == "NO\n".split()
assert solve("4 16\n20 2 5 9\n").split() == "YES\n".split()
assert solve("3 13\n2 5 9\n").split() == "NO\n".split()
```

## Exercise 7

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    allowed_days = int(parts[1])
    weights = []
    i = 0
    while i < n:
        weights.append(int(parts[i + 2]))
        i = i + 1
    lo = max(weights)
    hi = sum(weights)
    while lo < hi:
        mid = (lo + hi) // 2
        days = 1
        load = 0
        i = 0
        while i < n:
            if load + weights[i] > mid:
                days = days + 1
                load = weights[i]
            else:
                load = load + weights[i]
            i = i + 1
        if days <= allowed_days:
            hi = mid
        else:
            lo = mid + 1
    return str(lo)

assert solve("5 3\n4 2 7 3 9\n").split() == "10\n".split()
assert solve("1 1\n9\n").split() == "9\n".split()
assert solve("3 2\n5 1 1\n").split() == "5\n".split()
assert solve("3 1\n2 3 7\n").split() == "12\n".split()
assert solve("3 2\n1 1 9\n").split() == "9\n".split()
```

## Exercise 8

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    q = int(parts[1])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 2]))
        i = i + 1
    values = sorted(values)

    def lower_bound(target):
        lo = 0
        hi = n
        while lo < hi:
            mid = (lo + hi) // 2
            if values[mid] < target:
                lo = mid + 1
            else:
                hi = mid
        return lo

    def upper_bound(target):
        lo = 0
        hi = n
        while lo < hi:
            mid = (lo + hi) // 2
            if values[mid] <= target:
                lo = mid + 1
            else:
                hi = mid
        return lo

    output = ""
    i = 0
    while i < q:
        left = int(parts[n + i * 2 + 2])
        right = int(parts[n + i * 2 + 3])
        count = upper_bound(right) - lower_bound(left)
        if i > 0:
            output = output + "\n"
        output = output + str(count)
        i = i + 1
    return output

assert solve("5 4\n10 2 7 15 20\n1 6\n7 15\n11 14\n16 20\n").split() == "1\n3\n0\n1\n".split()
assert solve("4 4\n10 1 7 4\n1 1\n10 10\n5 6\n1 10\n").split() == "1\n1\n0\n4\n".split()
assert solve("2 1\n1 5\n7 9\n").split() == "0\n".split()
```

## Exercise 9

```python
def solve(data: str) -> str:
    parts = data.split()
    target_sum = int(parts[0])
    target_product = int(parts[1])
    answer = "NONE"
    first = 1
    while first <= 9 and answer == "NONE":
        second = 0
        while second <= 9 and answer == "NONE":
            third = 0
            while third <= 9 and answer == "NONE":
                digit_sum = first + second + third
                digit_product = first * second * third
                if digit_sum == target_sum and digit_product == target_product:
                    answer = str(first * 100 + second * 10 + third)
                third = third + 1
            second = second + 1
        first = first + 1
    return answer

assert solve("12 28\n").split() == "147\n".split()
assert solve("1 0\n").split() == "100\n".split()
assert solve("27 729\n").split() == "999\n".split()
assert solve("2 9\n").split() == "NONE\n".split()
```
