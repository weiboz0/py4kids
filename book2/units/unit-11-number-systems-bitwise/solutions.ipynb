## Exercise 1

```python
def solve(data: str) -> str:
    number = int(data.split()[0])
    if number == 0:
        return "0"
    result = ""
    while number > 0:
        digit = number % 2
        result = str(digit) + result
        number = number // 2
    return result

assert solve("42\n") == "101010"
assert solve("0\n") == "0"
assert solve("1\n") == "1"
assert solve("8\n") == "1000"
assert solve("5\n") == "101"
```

## Exercise 2

```python
def solve(data: str) -> str:
    parts = data.split()
    number = int(parts[0])
    hexadecimal = parts[1]
    digits = "0123456789ABCDEF"

    converted = ""
    if number == 0:
        converted = "0"
    while number > 0:
        digit = number % 16
        converted = digits[digit] + converted
        number = number // 16

    decimal = 0
    i = 0
    while i < len(hexadecimal):
        digit = 0
        while digits[digit] != hexadecimal[i]:
            digit = digit + 1
        decimal = decimal * 16 + digit
        i = i + 1
    return converted + "\n" + str(decimal)

assert solve("4095 2AF\n").split() == "FFF 687".split()
assert solve("0 0\n").split() == "0 0".split()
assert solve("16 10\n").split() == "10 16".split()
assert solve("255 FF\n").split() == "FF 255".split()
assert solve("4095 2AF\n") == "FFF\n687"
```

## Exercise 3

```python
def solve(data: str) -> str:
    parts = data.split()
    width = int(parts[0])
    state = int(parts[1])
    command_count = int(parts[2])
    full_mask = (1 << width) - 1
    part_index = 3
    command_index = 0
    result = ""
    while command_index < command_count:
        command = parts[part_index]
        if command == "NOT":
            state = ~state & full_mask
            reported = state
            part_index = part_index + 1
        else:
            bit_index = int(parts[part_index + 1])
            bit = 1 << bit_index
            if command == "TEST":
                if state & bit:
                    reported = 1
                else:
                    reported = 0
            elif command == "SET":
                state = state | bit
                reported = state
            elif command == "CLEAR":
                state = state & (full_mask ^ bit)
                reported = state
            else:
                state = state ^ bit
                reported = state
            part_index = part_index + 2
        if command_index > 0:
            result = result + "\n"
        result = result + str(reported)
        command_index = command_index + 1
    return result

assert solve("4 5 5\nTEST 0\nSET 3\nCLEAR 2\nFLIP 1\nNOT\n").split() == "1 13 9 11 4".split()
assert solve("3 5 4\nTEST 0\nSET 1\nCLEAR 2\nFLIP 0\n").split() == "1 7 3 2".split()
assert solve("3 0 1\nNOT\n") == "7"
assert solve("3 7 1\nNOT\n") == "0"
```

## Exercise 4

```python
def solve(data: str) -> str:
    parts = data.split()
    signal_count = int(parts[0])
    lone_code = 0
    i = 0
    while i < signal_count:
        lone_code = lone_code ^ int(parts[i + 1])
        i = i + 1
    return str(lone_code)

assert solve("7\n12 4 12 9 4 6 9\n") == "6"
assert solve("1\n37\n") == "37"
assert solve("5\n8 0 8 11 11\n") == "0"
```

## Exercise 5

```python
def solve(data: str) -> str:
    parts = data.split()
    item_count = int(parts[0])
    target = int(parts[1])
    weights = []
    i = 0
    while i < item_count:
        weights.append(int(parts[i + 2]))
        i = i + 1

    matching_count = 0
    for mask in range(1 << item_count):
        total = 0
        i = 0
        while i < item_count:
            if mask & (1 << i):
                total = total + weights[i]
            i = i + 1
        if total == target:
            matching_count = matching_count + 1
    return str(matching_count)

assert solve("4 5\n8 1 4 10\n") == "1"
assert solve("4 15\n8 1 4 2\n") == "1"
assert solve("3 0\n2 4 8\n") == "1"
assert solve("3 14\n2 4 8\n") == "1"
assert solve("4 6\n1 2 4 20\n") == "1"
```

## Exercise 6

```python
def solve(data: str) -> str:
    parts = data.split()
    pair_count = int(parts[0])
    result = ""
    i = 0
    while i < pair_count:
        first = int(parts[i * 2 + 1])
        second = int(parts[i * 2 + 2])
        a = first
        b = second
        while b:
            a, b = b, a % b
        greatest_common_divisor = a
        if first == 0 or second == 0:
            least_common_multiple = 0
        else:
            least_common_multiple = first // greatest_common_divisor * second
        if i > 0:
            result = result + "\n"
        result = result + str(greatest_common_divisor)
        result = result + " "
        result = result + str(least_common_multiple)
        i = i + 1
    return result

assert solve("4\n7 9\n12 12\n5 20\n18 24\n").split() == "1 63 12 12 5 20 6 72".split()
assert solve("2\n0 15\n21 0\n").split() == "15 0 21 0".split()
assert solve("1\n7 9\n") == "1 63"
```

## Exercise 7

```python
def solve(data: str) -> str:
    limit = int(data.split()[0])
    is_prime = []
    i = 0
    while i <= limit:
        is_prime.append(True)
        i = i + 1
    if limit >= 0:
        is_prime[0] = False
    if limit >= 1:
        is_prime[1] = False

    p = 2
    while p * p <= limit:
        if is_prime[p]:
            multiple = p * p
            while multiple <= limit:
                is_prime[multiple] = False
                multiple = multiple + p
        p = p + 1

    prime_count = 0
    i = 2
    while i <= limit:
        if is_prime[i]:
            prime_count = prime_count + 1
        i = i + 1
    return str(prime_count)

assert solve("49\n") == "15"
assert solve("0\n") == "0"
assert solve("1\n") == "0"
assert solve("2\n") == "1"
```

## Exercise 8

```python
def solve(data: str) -> str:
    parts = data.split()
    base = int(parts[0])
    modulus = int(parts[1])
    exponent = int(parts[2])
    result = 1 % modulus
    base = base % modulus
    while exponent > 0:
        if exponent % 2 == 1:
            result = result * base % modulus
        base = base * base % modulus
        exponent = exponent // 2
    return str(result % modulus)

assert solve("7 13 1000000000000000000\n") == "9"
assert solve("2 5 3\n") == "3"
assert solve("20 7 1\n") == "6"
assert solve("9 1 0\n") == "0"
assert solve("3 11 0\n") == "1"
```

## Exercise 9

```python
def solve(data: str) -> str:
    parts = data.split()
    width = int(parts[0])
    hexadecimal = parts[1]
    digits = "0123456789ABCDEF"
    value = 0
    i = 0
    while i < len(hexadecimal):
        digit = 0
        while digits[digit] != hexadecimal[i]:
            digit = digit + 1
        value = value * 16 + digit
        i = i + 1

    complemented = ~value & ((1 << width) - 1)
    converted = ""
    if complemented == 0:
        converted = "0"
    while complemented > 0:
        digit = complemented % 16
        converted = digits[digit] + converted
        complemented = complemented // 16

    required_digits = width // 4
    while len(converted) < required_digits:
        converted = "0" + converted
    return converted

assert solve("8 A5\n") == "5A"
assert solve("4 0\n") == "F"
assert solve("8 FF\n") == "00"
assert solve("12 F0F\n") == "0F0"
```

## Exercise 10

```python
def solve(data: str) -> str:
    target_index = int(data.split()[0])
    limit = 200000
    is_prime = []
    i = 0
    while i <= limit:
        is_prime.append(True)
        i = i + 1
    is_prime[0] = False
    is_prime[1] = False

    p = 2
    while p * p <= limit:
        if is_prime[p]:
            multiple = p * p
            while multiple <= limit:
                is_prime[multiple] = False
                multiple = multiple + p
        p = p + 1

    prime_count = 0
    candidate = 2
    while candidate <= limit:
        if is_prime[candidate]:
            prime_count = prime_count + 1
            if prime_count == target_index:
                return str(candidate)
        candidate = candidate + 1
    return ""

assert solve("15\n") == "47"
assert solve("1\n") == "2"
assert solve("16\n") == "53"
```
