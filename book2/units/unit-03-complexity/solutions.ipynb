## Exercise 1

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    target = int(parts[1])
    qualifying = 0
    i = 0
    while i < n:
        score = int(parts[i + 2])
        if score >= target:
            qualifying = qualifying + 1
        i = i + 1
    return str(qualifying)

assert solve("6 80\n72 91 80 64 79 95\n").split() == "3\n".split()
assert solve("4 10\n1 2 3 10\n").split() == "1\n".split()
```

## Exercise 2

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    goal = int(parts[1])
    total = 0
    answer = -1
    i = 0
    while i < n:
        total = total + int(parts[i + 2])
        if answer == -1 and total >= goal:
            answer = i + 1
        i = i + 1
    return str(answer)

assert solve("5 25\n4 6 3 5 7\n").split() == "5\n".split()
assert solve("4 10\n2 2 2 4\n").split() == "4\n".split()
```

## Exercise 3

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    target = int(parts[1])
    current = 0
    best = 0
    i = 0
    while i < n:
        score = int(parts[i + 2])
        if score >= target:
            current = current + 1
            if current > best:
                best = current
        else:
            current = 0
        i = i + 1
    return str(best)

assert solve("8 10\n10 12 4 11 3 14 15 16\n").split() == "3\n".split()
assert solve("5 7\n1 1 7 7 7\n").split() == "3\n".split()
```

## Exercise 4

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    frequencies = {}
    best_value = int(parts[1])
    best_frequency = 0
    i = 0
    while i < n:
        value = int(parts[i + 1])
        frequencies[value] = frequencies.get(value, 0) + 1
        frequency = frequencies[value]
        if frequency > best_frequency or frequency == best_frequency and value < best_value:
            best_value = value
            best_frequency = frequency
        i = i + 1
    return str(best_value)

assert solve("7\n8 3 8 5 3 8 3\n").split() == "3\n".split()
assert solve("5\n2 1 2 1 1\n").split() == "1\n".split()
assert solve("4\n7 5 7 5\n").split() == "5\n".split()
```

## Exercise 5

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    target = int(parts[1])
    seen = {}
    found = False
    i = 0
    while i < n:
        value = int(parts[i + 2])
        needed = target - value
        if seen.get(needed, 0) > 0:
            found = True
        seen[value] = seen.get(value, 0) + 1
        i = i + 1
    if found:
        return "YES"
    return "NO"

assert solve("6 17\n4 12 8 2 20 5\n").split() == "YES\n".split()
assert solve("5 9\n1 2 3 4 8\n").split() == "YES\n".split()
assert solve("4 20\n1 2 3 4\n").split() == "NO\n".split()
```

## Exercise 6

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    values = []
    i = 0
    while i < n:
        values.append(int(parts[i + 1]))
        i = i + 1
    values.sort()
    best = values[1] - values[0]
    i = 2
    while i < n:
        gap = values[i] - values[i - 1]
        if gap < best:
            best = gap
        i = i + 1
    return str(best)

assert solve("6\n40 3 22 15 60 21\n").split() == "1\n".split()
assert solve("5\n1 20 40 70 19\n").split() == "1\n".split()
```

## Exercise 7

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    seen = {}
    answer = -1
    i = 0
    while i < n:
        value = int(parts[i + 1])
        if answer == -1 and seen.get(value, 0) > 0:
            answer = value
        seen[value] = 1
        i = i + 1
    return str(answer)

assert solve("7\n9 2 14 6 3 11 2\n").split() == "2\n".split()
assert solve("5\n4 5 6 7 4\n").split() == "4\n".split()
assert solve("4\n1 2 3 4\n").split() == "-1\n".split()
```

## Exercise 8

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    limit = int(parts[1])
    positions = []
    i = 0
    while i < n:
        positions.append(int(parts[i + 2]))
        i = i + 1
    positions.sort()
    best = -1
    i = 1
    while i < n:
        gap = positions[i] - positions[i - 1]
        if gap <= limit and gap > best:
            best = gap
        i = i + 1
    return str(best)

assert solve("7 12\n40 3 18 70 90 55 81\n").split() == "11\n".split()
assert solve("5 10\n0 30 60 90 40\n").split() == "10\n".split()
assert solve("3 4\n0 10 20\n").split() == "-1\n".split()
```

## Exercise 9

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    first = int(parts[1])
    current = first
    best = first
    i = 1
    while i < n:
        value = int(parts[i + 1])
        if current + value < value:
            current = value
        else:
            current = current + value
        if current > best:
            best = current
        i = i + 1
    return str(best)

assert solve("6\n-5 2 -8 1 4 10\n").split() == "15\n".split()
assert solve("4\n-10 5 -2 20\n").split() == "23\n".split()
assert solve("3\n-8 -2 -5\n").split() == "-2\n".split()
```
