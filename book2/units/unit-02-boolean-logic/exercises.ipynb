# Unit 2 — Boolean Logic & Algebra Exercises

Each problem is a program that **reads the whole input from stdin** and **prints the exact output** (real contest style — no `solve()` wrapper): `import sys; data = sys.stdin.read()`, parse it, decide, print. Write each program in a `.py` file and run it with piped input, e.g. `python my_solution.py < case.txt`.

## Exercise 1

### Access Gate

The input contains three flags: `member guest banned`. Each flag is `1` for true or `0` for false. Print `GRANTED` exactly when the person is a member or guest, and is not banned. Otherwise, print `DENIED`.

### Constraints

- Each flag is `0` or `1`.
- The flags appear in the order `member guest banned`.

### Sample Input

```text
0 1 0
```

### Sample Output

```text
GRANTED
```

## Exercise 2

### De Morgan in Action

The input contains Boolean flags `a b`, written as `1` or `0`. Compute `not (a and not b)`. De Morgan's law says this is exactly the same as `(not a) or b`, so you may use whichever form is easier to read — both give the same answer for every row. Print `TRUE` or `FALSE`.

### Constraints

- `a` and `b` are each `0` or `1`.

### Sample Input

```text
1 0
```

### Sample Output

```text
FALSE
```


## Exercise 3

### At Least One Signal

The first value is `N`, followed by `N` flags. Print `YES` if at least one flag is `1`; otherwise, print `NO`. Scan the flags with a manual loop and a Boolean flag variable.

This sample intentionally puts the decisive `1` in the last position. A loop that stops one position early will incorrectly print `NO`.

### Constraints

- `1 <= N <= 1000`
- Every flag is `0` or `1`.

### Sample Input

```text
5
0 0 0 0 1
```

### Sample Output

```text
YES
```

## Exercise 5

### Majority with a Dissenter

The first value is `N`, followed by `N` answers, each written as `X` or `O`. Print `YES` if more than half of the answers are `X` and at least one answer is `O`. Otherwise, print `NO`.

### Constraints

- Every answer is exactly `X` or `O`.

### Sample Input

```text
5
X X O X X
```

### Sample Output

```text
YES
```

## Exercise 6

### Club Vote

The first two values are `N Q`: the number of members present and the number required for a quorum. They are followed by `N` votes, each `YES` or `NO`. Print `PASS` if at least `Q` members are present and strictly more than half of the votes are `YES`; otherwise, print `FAIL`.

### Constraints

- `1 <= N <= 1000`
- `1 <= Q <= 1000`
- Every vote is exactly `YES` or `NO`.

### Sample Input

```text
7 6
YES NO YES YES NO YES YES
```

### Sample Output

```text
PASS
```

## Exercise 7

### Skip Safe Numbers

The first value is `N`, followed by `N` tokens. Each token is either the word `SKIP` or an integer. Print `YES` if at least one integer token is positive; otherwise, print `NO`.

For each token, first check that it is not `SKIP`, and only then convert it with `int()`. Combine those two checks with short-circuiting so the word `SKIP` is never converted to an integer. Use a manual loop and a Boolean flag variable to remember whether a positive integer was found.

### Constraints

- `1 <= N <= 1000`
- Every numeric token is between -10000 and 10000.
- At least one token may be `SKIP`.

### Sample Input

```text
5
SKIP -3 0 SKIP 8
```

### Sample Output

```text
YES
```

## Exercise 8

Challenge (stretch)

### Robot Lab Rule Engine

The input contains five flags in this order: `student mentor badge banned emergency`. Each flag is `1` for true or `0` for false. Print `OPEN` when the person is not banned and either `(student or mentor) and badge` is true, or when `emergency and mentor` is true. Otherwise, print `CLOSED`. Use parentheses to make the complete rule clear.

The emergency rule stands on its own: if there is an emergency and a mentor is present, the lab opens even if that mentor is banned.

### Constraints

- Each flag is `0` or `1`.
- The flags always appear in the stated order.

### Sample Input

```text
1 0 1 0 0
```

### Sample Output

```text
OPEN
```


## Exercise 9

Challenge (stretch)

### Truth-Table Row Evaluator

The input contains a rule name followed by Boolean flags `a b`, written as `1` or `0`. The rule is one of `A_AND_NOT_B`, `NOT_A_OR_B`, `A_OR_B`, or `A_AND_B`. Evaluate that rule for the given truth-table row without using automatic expression evaluation. Print `TRUE` or `FALSE`.

### Constraints

- The rule name is one of the four names listed above.
- `a` and `b` are each `0` or `1`.

### Sample Input

```text
A_AND_NOT_B 1 0
```

### Sample Output

```text
TRUE
```
