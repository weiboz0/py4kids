## Exercise 1

```python
def solve(data: str) -> str:
    values = data.split()
    member = int(values[0]) == 1
    guest = int(values[1]) == 1
    banned = int(values[2]) == 1
    if (member or guest) and not banned:
        return "GRANTED"
    else:
        return "DENIED"

assert solve("0 1 0").split() == "GRANTED".split()
assert solve("1 0 1").split() == "DENIED".split()
```

## Exercise 2

```python
def solve(data: str) -> str:
    values = data.split()
    a = int(values[0]) == 1
    b = int(values[1]) == 1
    result = not (a and not b)
    if result:
        return "TRUE"
    else:
        return "FALSE"

assert solve("1 0").split() == "FALSE".split()
assert solve("0 0").split() == "TRUE".split()
assert solve("1 1").split() == "TRUE".split()
assert solve("0 1").split() == "TRUE".split()
```

## Exercise 3

```python
def solve(data: str) -> str:
    values = data.split()
    n = int(values[0])
    found_one = False
    for i in range(n):
        if int(values[i + 1]) == 1:
            found_one = True
    if found_one:
        return "YES"
    else:
        return "NO"

assert solve("5\n0 0 0 0 1").split() == "YES".split()
assert solve("4\n0 0 0 0").split() == "NO".split()
```

## Exercise 4

```python
def solve(data: str) -> str:
    values = data.split()
    n = int(values[0])
    every_check_passed = True
    for i in range(n):
        if int(values[i + 1]) == 0:
            every_check_passed = False
    if every_check_passed:
        return "YES"
    else:
        return "NO"

assert solve("4\n1 1 0 1").split() == "NO".split()
assert solve("3\n1 1 0").split() == "NO".split()
assert solve("3\n1 1 1").split() == "YES".split()
```

## Exercise 5

```python
def solve(data: str) -> str:
    values = data.split()
    n = int(values[0])
    x_count = 0
    has_dissenter = False
    for i in range(n):
        if values[i + 1] == "X":
            x_count = x_count + 1
        else:
            has_dissenter = True
    if x_count * 2 > n and has_dissenter:
        return "YES"
    else:
        return "NO"

assert solve("5\nX X O X X").split() == "YES".split()
assert solve("3\nX X O").split() == "YES".split()
assert solve("3\nX X X").split() == "NO".split()
assert solve("4\nX X O O").split() == "NO".split()
```

## Exercise 6

```python
def solve(data: str) -> str:
    values = data.split()
    n = int(values[0])
    quorum = int(values[1])
    yes_count = 0
    for i in range(n):
        if values[i + 2] == "YES":
            yes_count = yes_count + 1
    if n >= quorum and yes_count * 2 > n:
        return "PASS"
    else:
        return "FAIL"

assert solve("7 6\nYES NO YES YES NO YES YES").split() == "PASS".split()
assert solve("3 3\nYES NO YES").split() == "PASS".split()
assert solve("3 4\nYES YES YES").split() == "FAIL".split()
assert solve("4 4\nYES YES NO NO").split() == "FAIL".split()
```

## Exercise 7

```python
def solve(data: str) -> str:
    values = data.split()
    n = int(values[0])
    found_positive = False
    for i in range(n):
        token = values[i + 1]
        if token != "SKIP" and int(token) > 0:
            found_positive = True
    if found_positive:
        return "YES"
    else:
        return "NO"

assert solve("5\nSKIP -3 0 SKIP 8").split() == "YES".split()
assert solve("4\nSKIP -2 0 SKIP").split() == "NO".split()
```

## Exercise 8

```python
def solve(data: str) -> str:
    values = data.split()
    student = int(values[0]) == 1
    mentor = int(values[1]) == 1
    badge = int(values[2]) == 1
    banned = int(values[3]) == 1
    emergency = int(values[4]) == 1
    if ((not banned) and ((student or mentor) and badge)) or (emergency and mentor):
        return "OPEN"
    else:
        return "CLOSED"

assert solve("1 0 1 0 0").split() == "OPEN".split()
assert solve("0 1 0 1 1").split() == "OPEN".split()
assert solve("1 0 1 1 0").split() == "CLOSED".split()
```

## Exercise 9

```python
def solve(data: str) -> str:
    values = data.split()
    rule = values[0]
    a = int(values[1]) == 1
    b = int(values[2]) == 1
    if rule == "A_AND_NOT_B":
        result = a and not b
    elif rule == "NOT_A_OR_B":
        result = (not a) or b
    elif rule == "A_OR_B":
        result = a or b
    else:
        result = a and b
    if result:
        return "TRUE"
    else:
        return "FALSE"

assert solve("A_AND_NOT_B 1 0").split() == "TRUE".split()
assert solve("A_AND_NOT_B 1 1").split() == "FALSE".split()
assert solve("NOT_A_OR_B 0 0").split() == "TRUE".split()
assert solve("A_OR_B 0 0").split() == "FALSE".split()
assert solve("A_AND_B 1 1").split() == "TRUE".split()
assert solve("A_OR_B 1 0").split() == "TRUE".split()
assert solve("A_AND_B 1 0").split() == "FALSE".split()
assert solve("NOT_A_OR_B 1 0").split() == "FALSE".split()
```
