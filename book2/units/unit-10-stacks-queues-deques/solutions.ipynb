## Exercise 1

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    bracket_count = int(parts[0])
    brackets = ""
    if bracket_count > 0:
        brackets = parts[1]
    matching_open = {")": "(", "]": "[", "}": "{"}
    stack = deque()
    i = 0
    while i < bracket_count:
        bracket = brackets[i]
        if bracket == "(" or bracket == "[" or bracket == "{":
            stack.appendleft(bracket)
        else:
            if len(stack) == 0:
                return "NO"
            if stack[0] != matching_open[bracket]:
                return "NO"
            stack.popleft()
        i = i + 1
    if len(stack) == 0:
        return "YES"
    return "NO"

assert solve("6\n([{}])\n") == "YES"
assert solve("0\n") == "YES"
assert solve("1\n)\n") == "NO"
assert solve("3\n([]\n") == "NO"
assert solve("2\n(]\n") == "NO"
```

## Exercise 2

```python
from collections import deque

def solve(data: str) -> str:
    tokens = data.split()
    stack = deque()
    i = 0
    while i < len(tokens):
        token = tokens[i]
        if token == "+" or token == "-" or token == "*":
            right = stack[0]
            stack.popleft()
            left = stack[0]
            stack.popleft()
            if token == "+":
                value = left + right
            elif token == "-":
                value = left - right
            else:
                value = left * right
            stack.appendleft(value)
        else:
            stack.appendleft(int(token))
        i = i + 1
    return str(stack[0])

assert solve("8 3 - 2 *\n") == "10"
assert solve("8 3 - 2 * 5 +\n") == "15"
assert solve("2 3 -\n") == "-1"
assert solve("5\n") == "5"
assert solve("2 3 + 4 *\n") == "20"
assert solve("-2 3 * 4 +\n") == "-2"
```

## Exercise 3

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    command_count = int(parts[0])
    document = deque()
    part_index = 1
    command_index = 0
    while command_index < command_count:
        command = parts[part_index]
        if command == "TYPE":
            document.appendleft(parts[part_index + 1])
            part_index = part_index + 2
        else:
            document.popleft()
            part_index = part_index + 1
        command_index = command_index + 1
    if len(document) == 0:
        return "EMPTY"
    answer = deque()
    while len(document) > 0:
        answer.appendleft(document.popleft())
    result = ""
    while len(answer) > 0:
        result = result + answer.popleft()
    return result

assert solve("8\nTYPE C\nTYPE O\nTYPE D\nUNDO\nTYPE O\nTYPE D\nTYPE E\nUNDO\n") == "COOD"
assert solve("3\nTYPE A\nTYPE B\nUNDO\n") == "A"
assert solve("2\nTYPE X\nUNDO\n") == "EMPTY"
```

## Exercise 4

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    signal_count = int(parts[0])
    signals = []
    answers = []
    i = 0
    while i < signal_count:
        signals.append(int(parts[i + 1]))
        answers.append(-1)
        i = i + 1
    stack = deque()
    i = 0
    while i < signal_count:
        while len(stack) > 0 and signals[stack[0]] < signals[i]:
            weaker_index = stack[0]
            stack.popleft()
            answers[weaker_index] = signals[i]
        stack.appendleft(i)
        i = i + 1
    answer_parts = []
    i = 0
    while i < signal_count:
        answer_parts.append(str(answers[i]))
        i = i + 1
    result = ""
    j = 0
    while j < len(answer_parts):
        if j > 0:
            result = result + " "
        result = result + answer_parts[j]
        j = j + 1
    return result

assert solve("8\n4 2 7 7 1 3 8 5\n").split() == "7 7 8 8 3 8 -1 -1".split()
assert solve("2\n5 5\n").split() == "-1 -1".split()
assert solve("5\n5 4 3 2 6\n").split() == "6 6 6 6 -1".split()
assert solve("3\n-3 -2 -1\n").split() == "-2 -1 -1".split()
```

## Exercise 5

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    command_count = int(parts[0])
    line = deque()
    served = []
    part_index = 1
    command_index = 0
    while command_index < command_count:
        command = parts[part_index]
        if command == "ARRIVE":
            line.append(parts[part_index + 1])
            part_index = part_index + 2
        else:
            served.append(line.popleft())
            part_index = part_index + 1
        command_index = command_index + 1
    result = ""
    j = 0
    while j < len(served):
        if j > 0:
            result = result + "\n"
        result = result + served[j]
        j = j + 1
    return result

assert solve("6\nARRIVE Maya\nARRIVE Leo\nSERVE\nARRIVE Noor\nSERVE\nSERVE\n").split() == "Maya Leo Noor".split()
assert solve("3\nARRIVE A\nARRIVE B\nSERVE\n") == "A"
assert solve("1\nARRIVE Solo\n") == ""
```

## Exercise 6

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    command_count = int(parts[0])
    input_stack = deque()
    output_stack = deque()
    dequeued = []
    part_index = 1
    command_index = 0
    while command_index < command_count:
        command = parts[part_index]
        if command == "ENQUEUE":
            input_stack.appendleft(int(parts[part_index + 1]))
            part_index = part_index + 2
        else:
            if len(output_stack) == 0:
                while len(input_stack) > 0:
                    value = input_stack[0]
                    input_stack.popleft()
                    output_stack.appendleft(value)
            dequeued.append(str(output_stack[0]))
            output_stack.popleft()
            part_index = part_index + 1
        command_index = command_index + 1
    result = ""
    j = 0
    while j < len(dequeued):
        if j > 0:
            result = result + "\n"
        result = result + dequeued[j]
        j = j + 1
    return result

assert solve("7\nENQUEUE 10\nENQUEUE 20\nDEQUEUE\nENQUEUE 30\nENQUEUE 40\nDEQUEUE\nDEQUEUE\n").split() == "10 20 30".split()
assert solve("3\nENQUEUE 1\nENQUEUE 2\nDEQUEUE\n") == "1"
assert solve("6\nENQUEUE -1\nENQUEUE -2\nDEQUEUE\nENQUEUE -3\nDEQUEUE\nDEQUEUE\n").split() == "-1 -2 -3".split()
```

## Exercise 7

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    card_count = int(parts[0])
    cards = deque()
    i = 0
    while i < card_count:
        cards.append(int(parts[i + 1]))
        i = i + 1
    reversed_cards = deque()
    while len(cards) > 0:
        reversed_cards.appendleft(cards.popleft())
    answer_parts = []
    while len(reversed_cards) > 0:
        answer_parts.append(str(reversed_cards.popleft()))
    result = ""
    j = 0
    while j < len(answer_parts):
        if j > 0:
            result = result + " "
        result = result + answer_parts[j]
        j = j + 1
    return result

assert solve("7\n10 20 30 40 50 60 70\n").split() == "70 60 50 40 30 20 10".split()
assert solve("1\n42\n") == "42"
assert solve("3\n-1 0 9\n").split() == "9 0 -1".split()
```

## Exercise 8

```python
from collections import deque

def solve(data: str) -> str:
    text = data.split()[0]
    stack = deque()
    i = 0
    while i < len(text):
        character = text[i]
        if len(stack) > 0 and stack[0] == character:
            stack.popleft()
        else:
            stack.appendleft(character)
        i = i + 1
    if len(stack) == 0:
        return "EMPTY"
    answer = deque()
    while len(stack) > 0:
        answer.appendleft(stack.popleft())
    result = ""
    while len(answer) > 0:
        result = result + answer.popleft()
    return result

assert solve("ABCCBADEED\n") == "EMPTY"
assert solve("ABC\n") == "ABC"
assert solve("ABBA\n") == "EMPTY"
assert solve("ABBAC\n") == "C"
assert solve("AAA\n") == "A"
```

## Exercise 9

```python
from collections import deque

def solve(data: str) -> str:
    parts = data.split()
    car_count = int(parts[0])
    desired = []
    i = 0
    while i < car_count:
        desired.append(int(parts[i + 1]))
        i = i + 1
    side_track = deque()
    next_arrival = 1
    i = 0
    while i < car_count:
        wanted = desired[i]
        while next_arrival <= car_count:
            if len(side_track) > 0 and side_track[0] == wanted:
                break
            side_track.appendleft(next_arrival)
            next_arrival = next_arrival + 1
        if len(side_track) == 0 or side_track[0] != wanted:
            return "NO"
        side_track.popleft()
        i = i + 1
    return "YES"

assert solve("5\n2 1 4 3 5\n") == "YES"
assert solve("3\n3 1 2\n") == "NO"
assert solve("3\n2 1 3\n") == "YES"
assert solve("1\n1\n") == "YES"
assert solve("5\n1 2 5 3 4\n") == "NO"
```
