# Unit 4 — Sets, Tuples & Sorting

Two contest rosters may repeat the same students.
How can we quickly count distinct students, find who appears on both rosters, and rank their results?
Sets, tuples, and custom sorting give us the tools.

## Sets Remember Distinct Values

A set stores each value at most once.
A set literal uses braces, such as `{"Ava", "Bo", "Cy"}`.
An empty set is written as `set()`, because `{}` makes an empty dictionary.
Use `.add(value)` to add a value and `.discard(value)` to remove a value safely.

```python
def solve(data: str) -> str:
    names = data.split()
    distinct = {"warmup"}
    distinct.discard("warmup")
    for name in names:
        distinct.add(name)
    return str(len(distinct))

assert solve("Ava Bo Ava Cy Bo Dia") == "4"
```

## Fast Membership

The question `value in a_set` takes O(1) time on average.
That makes a set useful when a large problem asks many membership questions.

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    roster = set()
    for position in range(n):
        roster.add(tokens[position + 1])
    target = tokens[n + 1]
    if target in roster:
        return "YES"
    return "NO"

assert solve("5 Ava Bo Cy Dia Eli Dia") == "YES"
```

## Union, Intersection, and Difference

A **union** contains values from either set.
An **intersection** contains values that are in both sets.
We can build either result with a loop and `.add()`.
A **difference** contains values from the first set that are not in the second, and Python writes that as `first - second`.

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    m = int(tokens[1])
    first = set()
    second = set()
    for position in range(n):
        first.add(tokens[position + 2])
    for position in range(m):
        second.add(tokens[position + n + 2])
    either = set()
    for name in first:
        either.add(name)
    for name in second:
        either.add(name)
    return str(len(either))

assert solve("3 3 Ava Bo Cy Cy Dia Eli") == "5"
```

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    m = int(tokens[1])
    first = set()
    second = set()
    for position in range(n):
        first.add(tokens[position + 2])
    for position in range(m):
        second.add(tokens[position + n + 2])
    both = set()
    for name in first:
        if name in second:
            both.add(name)
    return str(len(both))

assert solve("4 4 Ava Bo Cy Dia Eli Ava Cy Dia") == "3"
```

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    m = int(tokens[1])
    first = set()
    second = set()
    for position in range(n):
        first.add(tokens[position + 2])
    for position in range(m):
        second.add(tokens[position + n + 2])
    only_first = first - second
    return str(len(only_first))

assert solve("5 3 Ava Bo Cy Dia Eli Bo Cy Dia") == "2"
```

## Tuples Keep Records Together

A tuple such as `("Ng", "Ava")` keeps related values together in a fixed record.
A tuple can also be a composite dictionary key when one field is not enough.
When a loop visits tuple records, unpacking gives each field a clear name.

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    people = {}
    for position in range(n):
        surname = tokens[1 + position * 2]
        first_name = tokens[2 + position * 2]
        person = (surname, first_name)
        people[person] = True
    return str(len(people))

assert solve("4 Ng Ava Ng Bo Ng Ava Fox Cy") == "3"
```

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    rows = []
    for position in range(n):
        name = tokens[1 + position * 2]
        score = int(tokens[2 + position * 2])
        rows.append((name, score))
    total = 0
    for name, score in rows:
        total = total + score
    return str(total)

assert solve("3 Ava 8 Bo 11 Cy 6") == "25"
```

## Sort by a Chosen Field

`sorted(rows, key=by_score)` makes a new list ordered by the value returned from `by_score`.
The key function has a name, so we can read exactly what controls the order.

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    rows = []
    for position in range(n):
        rows.append((tokens[1 + position * 2], int(tokens[2 + position * 2])))

    def by_score(record):
        return record[1]

    ordered = sorted(rows, key=by_score)
    name, score = ordered[0]
    return name + " " + str(score)

assert solve("4 Ava 18 Bo 9 Cy 14 Dia 11") == "Bo 9"
```

For descending order, return a negative number from the key function.
A larger score then produces a smaller key and moves toward the front.

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    rows = []
    for position in range(n):
        rows.append((tokens[1 + position * 2], int(tokens[2 + position * 2])))

    def highest_score_first(record):
        return -record[1]

    ordered = sorted(rows, key=highest_score_first)
    name, score = ordered[0]
    return name + " " + str(score)

assert solve("4 Ava 18 Bo 9 Cy 24 Dia 11") == "Cy 24"
```

## Compound Sort Keys

Sometimes two rules decide the order.
The tuple key `(-score, name)` sorts higher scores first, then sorts tied names alphabetically.

```python
def solve(data: str) -> str:
    tokens = data.split()
    n = int(tokens[0])
    rows = []
    for position in range(n):
        rows.append((tokens[1 + position * 2], int(tokens[2 + position * 2])))

    def leaderboard_key(record):
        name, score = record
        return (-score, name)

    ordered = sorted(rows, key=leaderboard_key)
    first_name, first_score = ordered[0]
    return first_name + " " + str(first_score)

assert solve("4 Zoe 20 Ava 20 Bo 15 Cy 9") == "Ava 20"
```
