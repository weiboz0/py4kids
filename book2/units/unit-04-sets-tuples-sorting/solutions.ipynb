# Unit 4 — Sets, Tuples & Sorting Solutions

Reference solutions for all nine exercises.

## Exercise 1

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    colors = {parts[1]}
    i = 1
    while i < n:
        colors.add(parts[i + 1])
        i = i + 1
    return str(len(colors))

assert solve("7\nred blue red green blue yellow purple\n").split() == "5\n".split()
assert solve("4\nred red red red\n").split() == "1\n".split()
assert solve("3\nred red gold\n").split() == "2\n".split()
```

## Exercise 2

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    m = int(parts[1])
    robotics = set()
    i = 0
    while i < n:
        robotics.add(parts[i + 2])
        i = i + 1
    both = set()
    i = 0
    while i < m:
        name = parts[n + i + 2]
        if name in robotics:
            both.add(name)
        i = i + 1
    return str(len(both))

assert solve("4 4\nAva Ben Cy Dia\nZoe Ava Cy Dia\n").split() == "3\n".split()
assert solve("1 1\nSolo\nSolo\n").split() == "1\n".split()
assert solve("2 2\nAva Ben\nCy Dia\n").split() == "0\n".split()
```

## Exercise 3

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    m = int(parts[1])
    first = set()
    i = 0
    while i < n:
        first.add(parts[i + 2])
        i = i + 1
    second = set()
    i = 0
    while i < m:
        second.add(parts[n + i + 2])
        i = i + 1
    only_first = first - second
    return str(len(only_first))

assert solve("5 4\nAva Ben Cy Dia Ivy\nBen Cy Zoe Dia\n").split() == "2\n".split()
assert solve("3 3\nA B C\nA B C\n").split() == "0\n".split()
assert solve("3 1\nA B Last\nA\n").split() == "2\n".split()
```

## Exercise 4

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    seen = set()
    answer = "NONE"
    i = 0
    while i < n:
        code = parts[i + 1]
        if answer == "NONE" and code in seen:
            answer = code
        seen.add(code)
        i = i + 1
    return answer

assert solve("7\nQ2 W5 E8 R1 T4 Y7 W5\n").split() == "W5\n".split()
assert solve("1\nSOLO\n").split() == "NONE\n".split()
assert solve("4\nA B B A\n").split() == "B\n".split()
```

## Exercise 5

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    contestants = []
    i = 0
    while i < n:
        name = parts[i * 2 + 1]
        score = int(parts[i * 2 + 2])
        contestant = (name, score)
        contestants.append(contestant)
        i = i + 1

    def leaderboard_key(contestant):
        name, score = contestant
        return (-score, name)

    ranked = sorted(contestants, key=leaderboard_key)
    output = ""
    i = 0
    while i < n:
        name, score = ranked[i]
        if i > 0:
            output = output + "\n"
        output = output + name + " " + str(score)
        i = i + 1
    return output

assert solve("5\nMia 90\nZoe 100\nAva 100\nNoah 80\nLiam 70\n").split() == "Ava 100\nZoe 100\nMia 90\nNoah 80\nLiam 70\n".split()
assert solve("3\nZoe 10\nAna 10\nMia 1\n").split() == "Ana 10\nZoe 10\nMia 1\n".split()
```

## Exercise 6

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    k = int(parts[1])
    players = []
    i = 0
    while i < n:
        name = parts[i * 2 + 2]
        score = int(parts[i * 2 + 3])
        player = (name, score)
        players.append(player)
        i = i + 1

    def ranking_key(player):
        name, score = player
        return (-score, name)

    ranked = sorted(players, key=ranking_key)
    name, score = ranked[k - 1]
    return name + " " + str(score)

assert solve("6 4\nAna 95\nBo 88\nCy 91\nDi 88\nEli 70\nFay 90\n").split() == "Bo 88\n".split()
assert solve("4 2\nMia 90\nAva 100\nLiam 80\nZoe 100\n").split() == "Zoe 100\n".split()
assert solve("3 2\nZoe 100\nAva 100\nMia 90\n").split() == "Zoe 100\n".split()
```

## Exercise 7

```python
def solve(data: str) -> str:
    parts = data.split()
    b = int(parts[0])
    n = int(parts[1])
    banned = set()
    i = 0
    while i < b:
        banned.add(parts[i + 2])
        i = i + 1
    submitted = set()
    i = 0
    while i < n:
        submitted.add(parts[b + i + 2])
        i = i + 1
    for code in banned:
        submitted.discard(code)
    return str(len(submitted))

assert solve("3 8\nred gray black\nred blue green blue gray yellow green purple\n").split() == "4\n".split()
assert solve("1 2\nbad\nsolo bad\n").split() == "1\n".split()
assert solve("1 4\nbad\nsolo solo bad new\n").split() == "2\n".split()
```

## Exercise 8

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    waiting = {}
    pairs = 0
    i = 0
    while i < n:
        grade = int(parts[i * 2 + 1])
        color = parts[i * 2 + 2]
        group = (grade, color)
        waiting[group] = waiting.get(group, 0) + 1
        if waiting[group] == 2:
            pairs = pairs + 1
            waiting[group] = 0
        i = i + 1
    return str(pairs)

assert solve("7\n7 red\n8 blue\n7 red\n8 green\n8 blue\n7 gold\n7 gold\n").split() == "3\n".split()
assert solve("1\n7 red\n").split() == "0\n".split()
assert solve("3\n7 red\n7 red\n7 red\n").split() == "1\n".split()
```

## Exercise 9

```python
def solve(data: str) -> str:
    parts = data.split()
    b = int(parts[0])
    n = int(parts[1])
    k = int(parts[2])
    banned = {parts[3]}
    i = 1
    while i < b:
        banned.add(parts[i + 3])
        i = i + 1
    best_scores = {}
    i = 0
    while i < n:
        name = parts[b + i * 2 + 3]
        score = int(parts[b + i * 2 + 4])
        if name not in banned:
            if name not in best_scores or score > best_scores[name]:
                best_scores[name] = score
        i = i + 1
    contestants = []
    for name in best_scores:
        contestant = (name, best_scores[name])
        contestants.append(contestant)

    def finalist_key(contestant):
        name, score = contestant
        return (-score, name)

    ranked = sorted(contestants, key=finalist_key)
    output = ""
    i = 0
    while i < k:
        name, score = ranked[i]
        if i > 0:
            output = output + "\n"
        output = output + name
        i = i + 1
    return output

assert solve("2 8 3\nKai Uma\nAva 50\nBen 80\nKai 100\nAva 90\nCora 85\nBen 70\nDax 40\nDax 88\n").split() == "Ava\nDax\nCora\n".split()
assert solve("1 5 2\nBad\nZoe 50\nBad 999\nAva 50\nZoe 40\nMia 60\n").split() == "Mia\nAva\n".split()
assert solve("1 4 2\nBad\nZoe 90\nZoe 10\nAva 50\nMia 40\n").split() == "Zoe\nAva\n".split()
assert solve("1 2 1\nBad\nSolo 7\nBad 99\n").split() == "Solo\n".split()
```
