# Unit 1 — Reading the Input

Contest-style problems hand you input as text and expect exact output text. Before we can solve the puzzle, we need to learn how to read that text. In Book 2, we will put that skill inside a function named `solve`: it takes the whole input as one string and returns the finished output string.

## Split the Input into Tokens

The new string method `.split()` breaks text wherever it finds whitespace, including spaces and newlines. It gives us a list of string tokens. Extra spaces and a final newline do not create extra tokens.

```python
data = "3\n4 9 2\n"
tokens = data.split()
print(tokens)
assert tokens == ["3", "4", "9", "2"]
```

## Turn Number Tokens into Integers

A token from `.split()` is still a string. Call `int(token)` before doing number arithmetic or comparisons.

```python
token = "42"
number = int(token)
print(number + 8)
assert number + 8 == 50
```

## Read N, Then N Values

A common input pattern puts `N` first and then supplies exactly `N` values. Read the first token as `N`, then use positions 1 through `N` to build the integer list.

```python
data = "4\n7 3 8 5"
tokens = data.split()
n = int(tokens[0])
values = []
for position in range(n):
    values.append(int(tokens[position + 1]))

print(values)
assert values == [7, 3, 8, 5]
```

## Running Example: Sum and Maximum

Our complete problem says: the first line is `N`; the second line has `N` integers; print their sum and maximum, separated by one space. All parsing and problem work happen inside `solve(data)`, and the function returns text instead of printing it.

```python
def solve(data):
    tokens = data.split()
    n = int(tokens[0])
    values = []
    for position in range(n):
        values.append(int(tokens[position + 1]))

    total = sum(values)
    biggest = max(values)
    return f"{total} {biggest}"


answer = solve("3\n4 9 2")
print(answer)
assert answer == "15 9"
```

## Read a Grid into a List of Lists

A grid begins with `R C`: its number of rows and columns. The next `R` rows each contain `C` integers. After splitting all the whitespace, a token-position counter helps us rebuild those rows as a list of lists.

```python
data = "2 3\n1 2 3\n4 5 6"
tokens = data.split()
rows = int(tokens[0])
columns = int(tokens[1])
token_position = 2
grid = []

for row in range(rows):
    current_row = []
    for column in range(columns):
        current_row.append(int(tokens[token_position]))
        token_position = token_position + 1
    grid.append(current_row)

print(grid)
assert grid == [[1, 2, 3], [4, 5, 6]]
```

## Build Exact Output Text

Judges compare output text exactly. An f-string is convenient for a fixed number of values. When a list can have many values, a loop can add one value at a time and put spaces only between values.

```python
values = [6, 10, 14]
output = ""
for position in range(len(values)):
    if position > 0:
        output = output + " "
    output = output + str(values[position])

print(output)
assert output == "6 10 14"
```
