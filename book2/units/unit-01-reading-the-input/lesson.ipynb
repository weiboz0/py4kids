# Unit 1 — Reading the Input

Contest problems hand you the input as **text** and expect exact output text. A real contest program reads ALL of that text from standard input, works out the answer, and prints it. This unit teaches that pipeline — **read stdin → parse → print** — one step at a time.

## The Real Program Reads stdin

A contest submission begins by reading everything from standard input into one string:

```python
import sys
data = sys.stdin.read()
```

The rest of the unit turns that one string into numbers and grids. In the notebook we practice the parsing on a **literal** `data = "..."` string (there is no contest input waiting here); the full programs at the end read real stdin and are run from a terminal.

## Lesson 1 — Read values

`.split()` breaks text on any whitespace (spaces and newlines) into a list of string tokens.

````python
data = "3\n4 9 2\n"
tokens = data.split()
print(tokens)
````

**Notice:** `.split()` ignores the newline and the trailing space — `"3\n4 9 2\n"` becomes `['3', '4', '9', '2']`, four tokens.

````python
token = "42"
number = int(token)
print(number + 8)
````

**Notice:** a token is still a STRING — call `int(token)` before doing arithmetic.

````python
data = "4\n7 3 8 5"
tokens = data.split()
n = int(tokens[0])
values = []
for position in range(n):
    values.append(int(tokens[position + 1]))
print(values)
````

**Notice:** the common pattern — `N` first, then `N` values: read `tokens[0]` as `n`, then positions `1..n`.

````python
values = [6, 10, 14]
output = ""
for position in range(len(values)):
    if position > 0:
        output = output + " "
    output = output + str(values[position])
print(output)
````

**Notice:** build exact output by adding a single space only BETWEEN values (not before the first).

**Put it together:** the full program reads the whole input from stdin, parses `N` and the values, and prints their sum and maximum separated by one space.

````python
import sys

data = sys.stdin.read()
tokens = data.split()
n = int(tokens[0])
values = []
for position in range(n):
    values.append(int(tokens[position + 1]))
total = sum(values)
biggest = max(values)
print(f"{total} {biggest}")
````

**Complexity:** building the list and its sum/max is `O(n)`.

## Lesson 2 — Read a grid

A grid begins with `R C`: the number of rows and columns, then `R` rows of `C` integers. Build it one step at a time.

````python
data = "2 3\n1 2 3\n4 5 6"
tokens = data.split()
rows = int(tokens[0])
columns = int(tokens[1])
print(rows, columns)
````

**Notice:** the header `R C` is just the first two tokens — read them before any grid values.

````python
position = 2
row = []
for column in range(columns):
    row.append(int(tokens[position]))
    position = position + 1
print(row)
````

**Notice:** one row is `C` values read with a moving `position` counter — `position` keeps its place in the token list across rows.

````python
tokens = "2 3\n1 2 3\n4 5 6".split()
rows = int(tokens[0])
columns = int(tokens[1])
position = 2
grid = []
for r in range(rows):
    current_row = []
    for c in range(columns):
        current_row.append(int(tokens[position]))
        position = position + 1
    grid.append(current_row)
print(grid)
print(grid[1][2])
````

**Notice:** nested loops build a list-of-lists; `grid[r][c]` reads one cell (here `grid[1][2]` is `6`).

**Put it together:** the full program reads an `R`×`C` grid from stdin and prints each row's sum, separated by spaces.

````python
import sys

data = sys.stdin.read()
tokens = data.split()
rows = int(tokens[0])
columns = int(tokens[1])
grid = []
position = 2
for r in range(rows):
    row = []
    for c in range(columns):
        row.append(int(tokens[position]))
        position = position + 1
    grid.append(row)
answer = ""
for r in range(rows):
    if r > 0:
        answer = answer + " "
    answer = answer + str(sum(grid[r]))
print(answer)
````

**Notice:** read `R C`, rebuild the grid, print one row-sum per row.

**Complexity:** reading every cell is `O(R*C)`.
