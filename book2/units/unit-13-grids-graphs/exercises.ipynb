# Unit 13 — Grids, Graphs & Traversal Exercises

Each exercise gives `solve(data)` the entire input string. Grid moves are up, down, left, and right only unless a statement explicitly says otherwise. Build every graph as a plain adjacency-list dictionary, store undirected edges at both endpoints, and mark nodes or cells visited before exploring onward.

## Exercise 1

### Friendship Degrees

The first line contains `N M A B`. The next `M` lines each contain an undirected friendship edge `u v`. Build a plain adjacency-list dictionary. Return two integers separated by one space: the degree of `A`, then the degree of `B`. The sample makes the final edge touch only `B`; it must be stored at both endpoints for `B`'s decisive degree to be correct.

### Constraints

`2 <= N <= 200000`, `1 <= M <= 200000`, and `1 <= A, B, u, v <= N`.
There are no self-edges or repeated edges, and both queried nodes occur in at least one edge.
Use a plain dictionary with `if u not in adj: adj[u] = []`, then append each undirected edge to both endpoint lists.

### Sample Input

```text
6 5 2 5
1 2
2 3
2 4
4 5
5 6
```

### Sample Output

```text
3 2
```

## Exercise 2

### Paint One Room

The first line contains `R C start_r start_c`, followed by `R` grid rows. An open cell is `.`, and a wall is `#`. The start is guaranteed open. Use recursive flood-fill with a `visited` set passed as an argument. Return the number of open cells in the start's four-neighbour region. Diagonal contact does not connect regions; the isolated final-row cell in the sample is not counted.

### Constraints

`1 <= R, C <= 20`, `R * C <= 400`, `0 <= start_r < R`, and `0 <= start_c < C`.
Each row contains exactly `C` characters from `.` and `#`.
Use recursive four-neighbour flood-fill and pass the mutable `visited` set to every call.

### Sample Input

```text
4 5 0 0
..###
.####
...##
####.
```

### Sample Output

```text
6
```

## Exercise 3

Use recursive flood-fill to return the number of islands. The sample has two side-connected islands plus one land cell that touches another island only diagonally, so that final cell is a separate third island.

### Constraints

`1 <= R, C <= 20` and `R * C <= 400`.
Each row contains exactly `C` characters from `#` and `.`.
Use recursive four-neighbour flood-fill with one `visited` set passed as an argument.

### Sample Input

```text
4 5
##...
#..#.
...##
..#..
```

### Sample Output

```text
3
```

## Exercise 4

### Fewest Maze Steps

The first line contains `R C`, followed by a maze. `S` is the start, `T` is the target, `.` is open, and `#` is a wall. Moving one cell up, down, left, or right costs one step. Use BFS with a FIFO `deque`, a `visited` set, and distances to return the fewest steps from `S` to `T`. Return `-1` if `T` is unreachable. The sample's target is reached only after the last downward move, so the decisive shortest distance is `8`.

### Constraints

`1 <= R, C <= 500` and `R * C <= 200000`.
The grid contains exactly one `S` and one `T`, with all other cells `.` or `#`.
Use `append` to enqueue, `popleft` to dequeue, and `while len(queue) > 0`.

### Sample Input

```text
5 6
S...##
.##...
...#..
##...#
....T#
```

### Sample Output

```text
8
```

## Exercise 5

### Emergency Exit

The first line contains `R C`, followed by a building map. `S` is the start, `E` is the exit, `.` is open floor, and `#` is a wall. Use BFS through four-neighbour open cells. Return `YES` if the exit is reachable and `NO` if it is impossible to reach. The sample keeps `E` visible but completely separated, so the required impossible sentinel is `NO`.

### Constraints

`1 <= R, C <= 500` and `R * C <= 200000`.
The grid contains exactly one `S` and one `E`, with all other cells `.` or `#`.
Use a FIFO `deque`, `visited`, and four-neighbour movement.

### Sample Input

```text
4 5
S...#
#####
...#.
.#..E
```

### Sample Output

```text
NO
```

## Exercise 6

### Can the Message Travel?

The first line contains `N M start target`. The next `M` lines contain undirected edges. Build a plain adjacency-list dictionary, then use recursive DFS with a `visited` set passed as an argument. Return `YES` if `target` is reachable from `start`, or `NO` otherwise. The sample contains the cycle `1-2-3-1`; `visited` must stop DFS from circling forever before it reaches the decisive final node `6`.

### Constraints

`2 <= N <= 400`, `1 <= M <= 2000`, and `1 <= start, target, u, v <= N`.
There are no self-edges or repeated edges.
Use recursive DFS; pass and mutate `visited`, and store every undirected edge at both endpoints.

### Sample Input

```text
6 6 1 6
1 2
2 3
3 1
3 4
4 5
5 6
```

### Sample Output

```text
YES
```

## Exercise 9
Challenge (stretch)

### Farthest Delivery

The first line contains `N M start`. The next `M` lines contain undirected roads. Every node is reachable from `start`. Build an adjacency-list dictionary and use BFS distances to return the greatest shortest-path distance from `start` to any node. The final road in the sample extends the only farthest route, making the decisive answer `5`. If the reachability guarantee were removed, an unreachable node's distance would use sentinel `-1`; this problem guarantees that sentinel is never needed.

### Constraints

`2 <= N <= 200000`, `1 <= M <= 200000`, and `1 <= start, u, v <= N`.
There are no self-edges or repeated edges, and every node is reachable from `start`.
Use a FIFO `deque`, `while len(queue) > 0`, a `visited` set, and a distance dictionary.

### Sample Input

```text
7 7 1
1 2
1 3
2 4
3 4
4 5
5 6
6 7
```

### Sample Output

```text
5
```
