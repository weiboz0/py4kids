# Unit 6 — Greedy Solutions

Reference solutions for all nine exercises.

## Exercise 1

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    events = []
    i = 0
    while i < n:
        start = int(parts[i * 2 + 1])
        end = int(parts[i * 2 + 2])
        events.append((start, end))
        i = i + 1

    def end_time(event):
        return event[1]

    events = sorted(events, key=end_time)
    attended = 0
    last_end = -1
    i = 0
    while i < n:
        start = events[i][0]
        end = events[i][1]
        if start >= last_end:
            attended = attended + 1
            last_end = end
        i = i + 1
    return str(attended)

assert solve("6\n1 4\n3 5\n0 7\n5 7\n5 9\n8 9\n").split() == "3".split()
assert solve("1\n0 100\n").split() == "1".split()
assert solve("4\n1 3\n0 3\n3 4\n4 6\n").split() == "3".split()
assert solve("3\n0 2\n2 4\n4 5\n").split() == "3".split()
```

## Exercise 2

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    jobs = []
    i = 0
    while i < n:
        duration = int(parts[i * 2 + 1])
        deadline = int(parts[i * 2 + 2])
        jobs.append((duration, deadline))
        i = i + 1

    def job_deadline(job):
        return job[1]

    jobs = sorted(jobs, key=job_deadline)
    finish_time = 0
    worst_lateness = 0
    i = 0
    while i < n:
        duration = jobs[i][0]
        deadline = jobs[i][1]
        finish_time = finish_time + duration
        lateness = finish_time - deadline
        if lateness > worst_lateness:
            worst_lateness = lateness
        i = i + 1
    return str(worst_lateness)

assert solve("3\n5 9\n1 1\n5 10\n").split() == "1".split()
assert solve("3\n1 10\n2 10\n3 10\n").split() == "0".split()
assert solve("2\n3 1\n2 2\n").split() == "3".split()
assert solve("3\n1 100\n10 10\n1 11\n").split() == "0".split()
```

## Exercise 3

```python
def solve(data: str) -> str:
    parts = data.split()
    amount = int(parts[0])
    coins = [25, 10, 5, 1]
    coins.sort(reverse=True)
    coin_count = 0
    i = 0
    while i < 4:
        coin_count = coin_count + amount // coins[i]
        amount = amount % coins[i]
        i = i + 1
    return str(coin_count)

assert solve("99\n").split() == "9".split()
assert solve("0\n").split() == "0".split()
assert solve("30\n").split() == "2".split()
assert solve("41\n").split() == "4".split()
```

## Exercise 4

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    robots = []
    stations = []
    i = 0
    while i < n:
        robots.append(int(parts[i + 1]))
        stations.append(int(parts[n + i + 1]))
        i = i + 1
    robots.sort()
    stations.sort()
    total_distance = 0
    i = 0
    while i < n:
        total_distance = total_distance + abs(robots[i] - stations[i])
        i = i + 1
    return str(total_distance)

assert solve("4\n10 1 4 30\n2 11 6 50\n").split() == "24".split()
assert solve("1\n-10\n15\n").split() == "25".split()
assert solve("3\n10 0 5\n6 11 1\n").split() == "3".split()
assert solve("3\n0 0 10\n0 1 12\n").split() == "3".split()
assert solve("2\n-5 5\n5 -5\n").split() == "0".split()
```

## Exercise 5

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    deadlines = []
    i = 0
    while i < n:
        deadlines.append(int(parts[i + 1]))
        i = i + 1
    deadlines.sort()
    completed = 0
    i = 0
    while i < n:
        if completed < deadlines[i]:
            completed = completed + 1
        i = i + 1
    return str(completed)

assert solve("5\n4 1 1 2 5\n").split() == "4".split()
assert solve("3\n1 1 1\n").split() == "1".split()
assert solve("3\n3 3 3\n").split() == "3".split()
assert solve("4\n2 1 2 1\n").split() == "2".split()
```

## Exercise 6

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    budget = int(parts[1])
    costs = []
    i = 0
    while i < n:
        costs.append(int(parts[i + 2]))
        i = i + 1
    costs.sort()
    spent = 0
    selected = 0
    i = 0
    while i < n and spent + costs[i] <= budget:
        spent = spent + costs[i]
        selected = selected + 1
        i = i + 1
    return str(selected)

assert solve("5 31\n9 2 4 6 10\n").split() == "5".split()
assert solve("3 1\n2 3 4\n").split() == "0".split()
assert solve("3 3\n3 1 2\n").split() == "2".split()
assert solve("4 10\n8 1 2 3\n").split() == "3".split()
assert solve("4 10\n3 1 3 3\n").split() == "4".split()
```

## Exercise 7

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    times = []
    i = 0
    while i < n:
        times.append(int(parts[i + 1]))
        i = i + 1
    times.sort()
    finish_time = 0
    total_completion_time = 0
    i = 0
    while i < n:
        finish_time = finish_time + times[i]
        total_completion_time = total_completion_time + finish_time
        i = i + 1
    return str(total_completion_time)

assert solve("4\n3 1 2 8\n").split() == "24".split()
assert solve("1\n9\n").split() == "9".split()
assert solve("3\n2 2 2\n").split() == "12".split()
assert solve("4\n4 1 1 5\n").split() == "20".split()
assert solve("4\n10 4 1 2\n").split() == "28".split()
```

## Exercise 8

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    segments = []
    i = 0
    while i < n:
        start = int(parts[i * 2 + 1])
        end = int(parts[i * 2 + 2])
        segments.append((start, end))
        i = i + 1

    def segment_end(segment):
        return segment[1]

    segments = sorted(segments, key=segment_end)
    checkpoint = -1
    checkpoint_count = 0
    i = 0
    while i < n:
        start = segments[i][0]
        end = segments[i][1]
        if checkpoint < start:
            checkpoint = end
            checkpoint_count = checkpoint_count + 1
        i = i + 1
    return str(checkpoint_count)

assert solve("6\n2 5\n1 3\n3 6\n8 10\n7 8\n12 12\n").split() == "3".split()
assert solve("2\n1 2\n2 3\n").split() == "1".split()
assert solve("3\n1 10\n2 9\n3 8\n").split() == "1".split()
assert solve("5\n0 10\n1 2\n2 2\n3 4\n10 10\n").split() == "3".split()
assert solve("3\n0 0\n2 2\n4 4\n").split() == "3".split()
```

## Exercise 9

```python
def solve(data: str) -> str:
    parts = data.split()
    n = int(parts[0])
    jobs = []
    i = 0
    while i < n:
        deadline = int(parts[i * 2 + 1])
        prize = int(parts[i * 2 + 2])
        jobs.append((deadline, prize))
        i = i + 1

    def prize_amount(job):
        return job[1]

    jobs = sorted(jobs, key=prize_amount, reverse=True)
    used = [False] * (n + 1)
    total_prize = 0
    i = 0
    while i < n:
        slot = jobs[i][0]
        prize = jobs[i][1]
        while slot > 0 and used[slot]:
            slot = slot - 1
        if slot > 0:
            used[slot] = True
            total_prize = total_prize + prize
        i = i + 1
    return str(total_prize)

assert solve("5\n2 60\n1 70\n1 80\n2 50\n3 40\n").split() == "180".split()
assert solve("1\n1 25\n").split() == "25".split()
assert solve("3\n2 5\n2 9\n2 7\n").split() == "16".split()
assert solve("4\n3 50\n1 40\n2 30\n3 20\n").split() == "120".split()
assert solve("3\n2 100\n1 99\n2 98\n").split() == "199".split()
assert solve("5\n1 100\n2 90\n2 90\n3 20\n4 10\n").split() == "220".split()
```
