## Milestone 1 — Warmups

### Problem 1 — Checkpoint Ledger  *(prefix sums)*

Line 1: `N`. Line 2: `N` integers. Line 3: `Q`. Then `Q` lines, each `l r` (1-indexed, inclusive). For each
query output the sum `a[l..r]`. Print the `Q` answers space-separated on one line.

Constraints: `1 <= N <= 100000`, `1 <= Q <= 100000`, `0 <= a_i <= 10^9`, `1 <= l <= r <= N`.

Sample Input
```text
5
2 4 1 3 5
3
1 3
2 5
4 4
```
Sample Output
```text
7 13 3
```

### Problem 2 — Warehouse Robot  *(simulation on a grid)*

Line 1: `R C`. Line 2: `sr sc` (0-indexed start). Line 3: a string of `U`/`D`/`L`/`R`. The robot ignores any
move that would leave the `R x C` grid. Output the final `r c`.

Constraints: `1 <= R, C <= 1000`, `0 <= sr < R`, `0 <= sc < C`, `0 <= len(moves) <= 100000`.

Sample Input
```text
3 3
1 1
UULLDDRR
```
Sample Output
```text
2 2
```

## Milestone 2 — Search & Pointers

### Problem 3 — Split the Load  *(binary search on the answer + greedy)*

Line 1: `N K`. Line 2: `N` positive integers (weights, in order). Split the sequence into at most `K`
contiguous groups so that the largest group sum is as small as possible. Output that minimum possible largest
sum.

Constraints: `1 <= K <= N <= 100000`, `1 <= w_i <= 10^9`.

Sample Input
```text
5 2
7 2 5 10 8
```
Sample Output
```text
18
```

### Problem 4 — Perfect Pair  *(two pointers)*

Line 1: `N T`. Then `N` lines each `id skill`. Is there a pair of two DIFFERENT players whose skills add up
to exactly `T`? Output `YES` or `NO`.

Constraints: `2 <= N <= 100000`, `0 <= id, skill <= 10^9`, `0 <= T <= 2*10^9`.

Sample Input
```text
4 10
1 8
2 3
3 7
4 1
```
Sample Output
```text
YES
```

## Milestone 3 — Graphs & Grids

### Problem 5 — City Network  *(graph BFS)*

Line 1: `N M S T` (nodes `1..N`, `M` edges, source `S`, target `T`). Then `M` lines each `u v`, an undirected
road. Output the fewest number of roads on a path from `S` to `T`, or `-1` if `T` is unreachable.

Constraints: `1 <= N <= 100000`, `0 <= M <= 200000`, `1 <= S, T <= N`.

Sample Input
```text
4 3 1 4
1 2
2 3
3 4
```
Sample Output
```text
3
```

### Problem 6 — Flood the Basin  *(flood fill)*

Line 1: `R C`. Then `R` rows of `#` (land) and `.` (water). Output the size (cell count) of the LARGEST
4-connected region of `#` (0 if there is none).

Constraints: `1 <= R, C`, `R*C <= 400`.

Sample Input
```text
3 4
#.#.
##..
...#
```
Sample Output
```text
3
```

## Milestone 4 — Recursion & Backtracking

### Problem 7 — Seating Plan  *(recursion + backtracking)*

A single line `N`. Count the arrangements of people `1..N` into seats `1..N` such that person `p` never sits
in seat `p`. Output the count. (Build it by backtracking: for each seat, try every unused person that is
allowed, mark them used, recurse, then unmark.)

Constraints: `1 <= N <= 9`.

Sample Input
```text
3
```
Sample Output
```text
2
```

## Milestone 5 — Trees

### Problem 8 — Team Roster  *(tree traversal)*

Line 1: `N` (nodes `1..N`, node `1` is the root). Then `N` lines each `label left right`, where `left`/`right`
are the child node ids (in `1..N`) or `-1` for none. Output the PRE-order sequence of labels (node, then left
subtree, then right subtree), space-separated.

Constraints: `1 <= N <= 300`, `0 <= label <= 10^9`; the input describes a valid rooted binary tree.

Sample Input
```text
3
1 2 3
2 -1 -1
3 -1 -1
```
Sample Output
```text
1 2 3
```
