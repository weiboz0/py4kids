# PET SIMULATOR

Adopt a virtual pet with its **OWN** name and mood. Feed it, play with it, and watch its stats change.

Each idea comes as a short **ladder** — simplest first, then one step up, then the real simulator — with a **Notice** naming the one new thing.

## Lesson 1: Meet Your Pet

A **class** is a set of instructions for making objects. The `__init__` method runs when we make a new pet, and `self` means **this particular pet**. The values it stores (`self.name`, `self.hunger`, ...) are called **attributes**.

````python
class Pet:
    def __init__(self, name):
        self.name = name

buddy = Pet("Buddy")
print(buddy.name)
````

**Notice:** `Pet("Buddy")` makes one object; `__init__` stored `"Buddy"` in `self.name`, read back with `buddy.name`.

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

buddy = Pet("Buddy")
print(buddy.name)
print(buddy.hunger)
````

**Notice:** `__init__` can set several attributes — every new pet starts with a name, hunger, and happiness.

A dot reads one of a pet's **attributes**. Each object carries its OWN attributes.

````python
buddy = Pet("Buddy")
print(buddy.name)
print(buddy.hunger)
````

**Notice:** `buddy.hunger` reads the hunger attribute of this particular pet.

````python
buddy = Pet("Buddy")
luna = Pet("Luna")
print(buddy.name)
print(luna.name)
````

**Notice:** two pets are two separate objects, each with its own attributes.

````python
buddy = Pet("Buddy")
luna = Pet("Luna")
luna.hunger = 7
print(buddy.hunger)
print(luna.hunger)
````

**Notice:** changing `luna.hunger` leaves `buddy.hunger` at 5 — each object's attributes are independent.

## Lesson 2: Teach Your Pet Actions

A **method** is a function inside a class. It receives `self`, so it reads or changes the attributes of this particular pet. (After adding methods to the class, make a NEW pet so it has them.)

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def play(self):
        self.happiness = self.happiness + 1

buddy = Pet("Buddy")
buddy.play()
print(buddy.happiness)
````

**Notice:** `play` is a method — calling `buddy.play()` changes this pet's `happiness`.

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def play(self):
        self.happiness = self.happiness + 1

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

buddy = Pet("Buddy")
new_hunger = buddy.feed(2)
print(new_hunger)
````

**Notice:** one step up — `feed` takes a parameter (`amount`) and RETURNS the new hunger.

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def play(self):
        self.happiness = self.happiness + 1

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

    def pass_time(self):
        self.hunger = self.hunger + 2

buddy = Pet("Buddy")
buddy.pass_time()
print(buddy.hunger)
````

**Notice:** another method — `pass_time` makes the pet hungrier as time passes.

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def play(self):
        self.happiness = self.happiness + 1

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

    def pass_time(self):
        self.hunger = self.hunger + 2

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "full"
        else:
            mood = "content"
        print(f"{self.name} is {mood}.")
        return mood

buddy = Pet("Buddy")
buddy.status()
````

**Notice:** a method can make a decision with `if`/`elif`/`else`, print a report, and RETURN a result for other code to use.

## Lesson 3: Run a Pet Day

**Put it together:** different foods fill a pet by different amounts. A dictionary maps each food to its filling power, and the number becomes `feed`'s `amount`.

````python
foods = {"apple": 2, "steak": 4}
buddy = Pet("Buddy")
buddy.feed(foods["steak"])
buddy.status()
````

**Put it together:** care for a whole LIST of pets with one loop — each gets a time change and reports its own status.

````python
pets = []
pets.append(Pet("Rex"))
pets.append(Pet("Mochi"))
pets.append(Pet("Ziggy"))

for pet in pets:
    pet.pass_time()
    pet.status()
````

**Put it together:** a `while` loop plays with Buddy until happiness reaches 10.

````python
buddy = Pet("Buddy")
while buddy.happiness < 10:
    buddy.play()
buddy.status()
````
