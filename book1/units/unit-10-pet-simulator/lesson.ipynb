## Lesson 1: Meet Your Pet

A **class** is a set of instructions for making objects.
Our `Pet` class describes what every new pet starts with.

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5
````

`__init__` runs when we make a new pet.
The word `self` means **this particular pet**.
So `self.name`, `self.hunger`, and `self.happiness` belong to the pet being made.
These pieces of information are called **attributes**.

````python
buddy = Pet("Buddy")

print(buddy.name)
print(buddy.hunger)
````

`Pet("Buddy")` makes one object from the class instructions.
A dot lets us read one of Buddy's attributes.
Now let's adopt two pets and change only one of them.

````python
buddy = Pet("Buddy")
luna = Pet("Luna")

luna.hunger = 7

print(buddy.name, buddy.hunger)
print(luna.name, luna.hunger)
````

Buddy's hunger stayed at `5` when Luna's hunger changed to `7`.
Each object carries its own attributes, so the two pets can have different stats.

## Lesson 2: Teach Your Pet Actions

A **method** is a function inside a class.
It tells one particular object what action to take.
We will add four methods to `Pet`.

````python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

    def play(self):
        self.happiness = self.happiness + 1

    def pass_time(self):
        self.hunger = self.hunger + 2

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}. Happiness: {self.happiness}.")
        return mood
````

Every method receives `self`, so it reads or changes the attributes of this particular pet.
`feed` also receives an `amount` and returns the new hunger.
`play` raises happiness by one.
`pass_time` raises hunger by two because time makes a pet hungrier.
`status` chooses a mood from the pet's hunger, prints a friendly report, and RETURNS the mood so other code can use it.

````python
buddy = Pet("Buddy")

buddy.feed(2)
buddy.play()
buddy.pass_time()
buddy.status()
````

Buddy starts with hunger `5`.
Food lowers it, passing time raises it, and playing raises happiness.

## Lesson 3: Run a Pet Day

Different foods can change hunger by different amounts.
This table connects each food name to its filling power.

````python
foods = {"apple": 2, "steak": 4}

buddy.feed(foods["steak"])
buddy.status()
````

Square brackets look up the number stored for `"steak"`.
That number becomes the `amount` sent to Buddy's `feed` method.
Now the simulator can care for a whole list of pets with one pass.

````python
pets = []
pets.append(Pet("Rex"))
pets.append(Pet("Mochi"))
pets.append(Pet("Ziggy"))

for pet in pets:
    pet.pass_time()
    pet.status()
````

The loop visits each pet once.
Each pet experiences one time change and then reports its own status.
