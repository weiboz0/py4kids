## Exercise 2: Two Pets, Two Stories

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

milo = Pet("Milo")
pepper = Pet("Pepper")
steps = 0
while milo.hunger > 2:
    milo.hunger = milo.hunger - 1
    steps = steps + 1
print(milo.name, milo.hunger)
print(pepper.name, pepper.hunger)
assert milo.hunger == 2
assert pepper.hunger == 5
assert steps == 3
```

## Exercise 4: Build the Mood Ladder

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}, happiness: {self.happiness}")
        return mood

buddy = Pet("Buddy")
buddy.hunger = 8
reported = buddy.status()
print(reported)
assert reported == "hungry"
```

## Exercise 5: Make a Pet List

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}, happiness: {self.happiness}")
        return mood

pets = []
pets.append(Pet("Ivy"))
pets.append(Pet("Rocket"))
for pet in pets:
    pet.status()
print(pets[0].name)
pets[1].status()
assert pets[0].name == "Ivy"
assert pets[1].status() == "content"
```

## Exercise 10: Three Snack Stops

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

snacks = {"biscuit": 1}
buddy = Pet("Buddy")
stops = 0
while stops < 3:
    buddy.feed(snacks["biscuit"])
    stops = stops + 1
print(buddy.hunger)
print(stops)
assert buddy.hunger == 2
assert stops == 3
```

## Exercise 11: Pet Roll Call

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}.")
        return mood

pets = []
pets.append(Pet("Coco"))
pets.append(Pet("Sunny"))
pets.append(Pet("Pip"))
pets[0].hunger = 8
pets[1].hunger = 1
moods = []
for pet in pets:
    moods.append(pet.status())
print(pets[2].name)
assert moods == ["hungry", "stuffed", "content"]
assert pets[2].name == "Pip"
```

## Exercise 12: Run a Pet Day

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

    def play(self):
        self.happiness = self.happiness + 1

    def pass_time(self):
        self.hunger = self.hunger + 2

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}, happiness: {self.happiness}")
        return mood

pet_day_name = "Buddy"
buddy = Pet(pet_day_name)
buddy.feed(2)
buddy.play()
buddy.pass_time()
buddy.status()
print(buddy.hunger)
print(buddy.happiness)
assert buddy.hunger == 5
assert buddy.happiness == 6
```

## Exercise 13: Happiest Pet

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

pets = []
pets.append(Pet("Coco"))
pets.append(Pet("Sunny"))
pets.append(Pet("Pip"))
pets[0].happiness = 7
pets[1].happiness = 9
pets[2].happiness = 6
best = 0
for pet in pets:
    if pet.happiness > best:
        best = pet.happiness
print(best)
assert best == 9
```

## Exercise 14: Play Until Happy

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def play(self):
        self.happiness = self.happiness + 1

buddy = Pet("Buddy")
plays = 0
while buddy.happiness < 10:
    buddy.play()
    plays = plays + 1
print(buddy.happiness)
assert buddy.happiness == 10
assert plays == 5
```
