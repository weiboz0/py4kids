## Exercise 1: Make a Pet

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}, happiness: {self.happiness}")

    def play(self):
        self.happiness = self.happiness + 1

buddy = Pet("Buddy")
print(buddy.name)
print(buddy.hunger)
assert buddy.name == "Buddy"
assert buddy.hunger == 5
```

## Exercise 2: Two Pets, Two Stories

```python
milo = Pet("Milo")
pepper = Pet("Pepper")
milo.hunger = 2
print(milo.name, milo.hunger)
print(pepper.name, pepper.hunger)
assert milo.hunger == 2
assert pepper.hunger == 5
```

## Exercise 3: Add a Feed Method

```python
buddy = Pet("Buddy")
new_hunger = buddy.feed(2)
print(new_hunger)
assert new_hunger == 3
assert buddy.hunger == 3
```

## Exercise 4: Build the Mood Ladder

```python
buddy = Pet("Buddy")
buddy.hunger = 8
buddy.status()
assert buddy.hunger == 8
```

## Exercise 5: Make a Pet List

```python
pets = []
pets.append(Pet("Ivy"))
pets.append(Pet("Rocket"))
for pet in pets:
    pet.status()
assert pets[0].name == "Ivy"
assert pets[1].name == "Rocket"
```

## Exercise 6: Feed from the Foods Table

```python
foods = {"apple": 2, "steak": 4}
buddy = Pet("Buddy")
buddy.feed(foods["steak"])
print(buddy.hunger)
assert buddy.hunger == 1
```

## Exercise 9: Happiest Pet

```python
pets = []
pets.append(Pet("Coco"))
pets.append(Pet("Sunny"))
pets.append(Pet("Pip"))
pets[0].happiness = 7
pets[1].happiness = 9
pets[2].happiness = 6
best = 0
for pet in pets:
    if pet.happiness > best:
        best = pet.happiness
print(best)
assert best == 9
```

## Exercise 10: Play Until Happy

```python
buddy = Pet("Buddy")
while buddy.happiness < 10:
    buddy.play()
print(buddy.happiness)
assert buddy.happiness == 10
```
