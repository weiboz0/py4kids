## Exercise 2: Two Pets, Two Stories

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

milo = Pet("Milo")
pepper = Pet("Pepper")
steps = 0
while milo.hunger > 2:
    milo.hunger = milo.hunger - 1
    steps = steps + 1
print(milo.name, milo.hunger)
print(pepper.name, pepper.hunger)
assert milo.hunger == 2
assert pepper.hunger == 5
assert steps == 3
```

## Exercise 4: Build the Mood Ladder

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}, happiness: {self.happiness}")
        return mood

buddy = Pet("Buddy")
buddy.hunger = 8
reported = buddy.status()
print(reported)
assert reported == "hungry"
```

## Exercise 12: Run a Pet Day

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def feed(self, amount):
        self.hunger = self.hunger - amount
        return self.hunger

    def play(self):
        self.happiness = self.happiness + 1

    def pass_time(self):
        self.hunger = self.hunger + 2

    def status(self):
        if self.hunger > 6:
            mood = "hungry"
        elif self.hunger < 3:
            mood = "stuffed"
        else:
            mood = "content"
        print(f"{self.name} is {mood}. Hunger: {self.hunger}, happiness: {self.happiness}")
        return mood

pet_day_name = "Buddy"
buddy = Pet(pet_day_name)
buddy.feed(2)
buddy.play()
buddy.pass_time()
buddy.status()
print(buddy.hunger)
print(buddy.happiness)
assert buddy.hunger == 5
assert buddy.happiness == 6
```

## Exercise 13: Happiest Pet

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

pets = []
pets.append(Pet("Coco"))
pets.append(Pet("Sunny"))
pets.append(Pet("Pip"))
pets[0].happiness = 7
pets[1].happiness = 9
pets[2].happiness = 6
best_pet = pets[0]
best = pets[0].happiness
for pet in pets:
    if pet.happiness > best:
        best = pet.happiness
        best_pet = pet
print(best_pet.name)
print(best)
assert best_pet is pets[1]
assert best_pet.name == "Sunny"
assert best == 9
```

## Exercise 14: Play Until Happy

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.happiness = 5

    def play(self):
        self.happiness = self.happiness + 1

buddy = Pet("Buddy")
plays = 0
while buddy.happiness < 10:
    buddy.play()
    plays = plays + 1
print(buddy.happiness)
assert buddy.happiness == 10
assert plays == 5
```

## Exercise 15: List the happy pets

```python
class Pet:
    def __init__(self, name, happiness):
        self.name = name
        self.happiness = happiness


pets = [Pet("Mochi", 6), Pet("Sunny", 7), Pet("Pip", 9), Pet("Coco", 4)]
happiness_threshold = 7
happy_pets = []

for pet in pets:
    if pet.happiness >= happiness_threshold:
        happy_pets.append(pet)

for pet in happy_pets:
    print(pet.name)

assert happy_pets == [pets[1], pets[2]]
assert (happy_pets[0].name, happy_pets[1].name) == ("Sunny", "Pip")
assert happy_pets[0].happiness == happiness_threshold
assert pets[0].happiness < happiness_threshold
assert pets[0] not in happy_pets
```

## Exercise 20: Hungriest and Least Hungry

```python
class Pet:
    def __init__(self, name, hunger):
        self.name = name
        self.hunger = hunger


pets = [Pet("Rex", 8), Pet("Mia", 3), Pet("Sam", 7)]
hungriest_name = pets[0].name
hungriest_hunger = pets[0].hunger
least_name = pets[0].name
least_hunger = pets[0].hunger

for pet in pets:
    if pet.hunger > hungriest_hunger:
        hungriest_name = pet.name
        hungriest_hunger = pet.hunger
    if pet.hunger < least_hunger:
        least_name = pet.name
        least_hunger = pet.hunger

print(f"Hungriest: {hungriest_name} with {hungriest_hunger}")
print(f"Least hungry: {least_name} with {least_hunger}")

assert hungriest_name == "Rex"
assert hungriest_hunger == 8
assert least_name == "Mia"
assert least_hunger == 3
```

## Exercise 21: Mood Tally

```python
class Pet:
    def __init__(self, name, mood):
        self.name = name
        self.mood = mood


pets = [
    Pet("Rex", "happy"),
    Pet("Mia", "sad"),
    Pet("Sam", "happy"),
    Pet("Buddy", "hungry"),
]
counts = {}

for pet in pets:
    mood = pet.mood
    counts[mood] = counts.get(mood, 0) + 1

print(f"Mood tally: {counts}")

assert counts == {"happy": 2, "sad": 1, "hungry": 1}
```

## Exercise 22: Feed Until the Snacks Would Overfeed

```python
class Pet:
    def __init__(self, name, hunger):
        self.name = name
        self.hunger = hunger


buddy = Pet("Buddy", 8)
snacks = [2, 4, 3, 5]
snacks_used = 0

for snack in snacks:
    if buddy.hunger - snack < 0:
        break
    buddy.hunger = buddy.hunger - snack
    snacks_used = snacks_used + 1

print(f"{buddy.name} has {buddy.hunger} hunger after {snacks_used} snacks.")
assert buddy.hunger == 2
assert snacks_used == 2
```

## Exercise 23: Feed Until Hunger Goes Negative

```python
class Pet:
    def __init__(self, name, hunger):
        self.name = name
        self.hunger = hunger


buddy = Pet("Buddy", 8)
snacks = [2, 4, 3, 5]
snacks_used = 0
tipping_snack = 0

for snack in snacks:
    buddy.hunger = buddy.hunger - snack
    snacks_used = snacks_used + 1
    if buddy.hunger < 0:
        tipping_snack = snack
        break

print(f"{buddy.name} has {buddy.hunger} hunger after {snacks_used} snacks.")
print(f"The tipping snack was {tipping_snack}.")
assert buddy.hunger == -1
assert snacks_used == 3
assert tipping_snack == 3
```
