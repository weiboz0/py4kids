## Exercise 3

```python
def show_double(number):
    print(number * 2)


def get_double(number):
    return number * 2


doubled_score = get_double(7)
print("get_double hands its result back, while show_double only displays it.")
assert doubled_score == 14
```

## Exercise 4

```python
def area(w, h):
    return w * h


card_area = area(8, 2.5)
area_message = f"The area is {card_area} square units."
print(area_message)
assert card_area == 20.0
# Prediction answered: the result is a FLOAT (20.0), because 2.5 is a float and
# any arithmetic touching a float produces a float.
```

## Exercise 5

```python
team = "Comets"


def player_label(name):
    label = f"{team}: {name}"
    print(label)


player_label("Jo")
print(team)
print("team is global. label and name are local to the function.")
print("Using label after the call would fail because that local name is out of scope.")
assert team == "Comets"
```

## Exercise 6

```python
stamp_count = 4
for stamp_number in range(stamp_count):
    size = 20 + stamp_number * 5
    distance = stamp_number * 40
    print(f"counter {stamp_number}: size {size}, distance {distance}")

print("The square turn is 90 degrees. One stamp function replaces four copies of its commands.")
assert stamp_number == 3
assert size == 35
assert distance == 120
```

### Challenge 2

```python
# The real flower that DRAWS petals lives in assets/solutions_l3.py (flower()).
# Here we can't run turtle in a notebook, so petal() just returns the size and we
# check the petal COUNT instead — the plan's structure, verified headlessly.
def petal(size):
    return size


def flower(size, petal_count):
    petals_drawn = 0
    for petal_number in range(petal_count):
        petal(size)
        petals_drawn = petals_drawn + 1
    return petals_drawn


flower_petals = flower(30, 6)
print(f"One flower call draws {flower_petals} petals.")
assert flower_petals == 6
```
