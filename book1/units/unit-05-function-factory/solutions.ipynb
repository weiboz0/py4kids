## Exercise 1

```python
def cheer(name):
    message = f"{name}, keep coding!"
    return message


maya_message = cheer("Maya")
leo_message = cheer("Leo")
print(maya_message)
print(leo_message)
print("The caller uses the returned values because message is local to cheer.")
assert maya_message == "Maya, keep coding!"
assert leo_message == "Leo, keep coding!"
```

## Exercise 2

```python
def greeting_card(name):
    greeting = f"Welcome, {name}!"
    return greeting


rin_greeting = greeting_card("Rin")
print(rin_greeting)
print("I added name as a parameter, built a local greeting, and returned it.")
assert rin_greeting == "Welcome, Rin!"
```

## Exercise 3

```python
def show_double(number):
    print(number * 2)


def get_double(number):
    doubled = number * 2
    return doubled


doubled_score = get_double(2.5)
double_message = f"Doubling 2.5 gives {doubled_score}."
print(double_message)
print("get_double hands its result back, while show_double only displays it.")
assert doubled_score == 5.0
assert double_message == "Doubling 2.5 gives 5.0."
```

## Exercise 4

```python
def area(w, h):
    card_area = w * h
    return card_area


calculated_area = area(8, 2.5)
area_message = f"The area is {calculated_area} square units."
print(area_message)
print("The returned result is a float value.")
assert calculated_area == 20.0
assert area_message == "The area is 20.0 square units."
```

## Exercise 5

```python
team = "Comets"


def player_label(name):
    label = f"{team}: {name}"
    print(label)
    return label


jo_label = player_label("Jo")
print(team)
print("team is global. label and name are local to the function.")
print("Using label after the call would fail because that local name is out of scope.")
assert team == "Comets"
assert jo_label == "Comets: Jo"
```

## Exercise 7

```python
def total_card_borders(number_of_cards):
    total = 0
    for card_number in range(number_of_cards):
        side = 10 + card_number * 5
        border_length = side * 4
        total = total + border_length
    return total


border_total = total_card_borders(4)
print(border_total)
print("The return line comes after the loop so it returns the completed total.")
assert border_total == 280
```

## Exercise 8

The broken raising line, `print(card_label)`, remains text only. The repaired program returns the local value instead.

```python
def make_label(name):
    card_label = f"Made for {name}"
    return card_label


saved_label = make_label("Nia")
print(saved_label)
print("The final print can see saved_label because the caller created it outside the function.")
assert saved_label == "Made for Nia"
```

## Exercise 9

```python
def turn_angle(n):
    return 360 / n


def angle_message(angle):
    return f"Turn {angle} degrees."


final_message = angle_message(turn_angle(7))
print(final_message)
print("n is local to turn_angle; angle is local to angle_message.")
assert turn_angle(7) == 51.42857142857143
assert final_message == "Turn 51.42857142857143 degrees."
```

## Exercise 11

```python
bands = 2
stamps_per_band = 5
stamps_drawn = 0
last_size = 0
for band_number in range(bands):
    size = 15 + band_number * 5
    last_size = size
    for stamp_number in range(stamps_per_band):
        stamps_drawn = stamps_drawn + 1
print(last_size)
print(stamps_drawn)
assert last_size == 20
assert stamps_drawn == 10
```

```python
# The real flower that DRAWS petals lives in assets/solutions_challenge2.py.
# Here we can't run turtle in a notebook, so petal() just returns the size and we
# check the petal COUNT instead — the plan's structure, verified headlessly.
def petal(size):
    return 1


def flower(size, petal_count):
    petals_drawn = 0
    for petal_number in range(petal_count):
        petals_drawn = petals_drawn + petal(size)
    return petals_drawn


flower_petals = flower(30, 6)
print(f"One flower call draws {flower_petals} petals.")
assert flower_petals == 6
```
