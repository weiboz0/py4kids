## Exercise 1

````python
def cheer(name):
    message = f"{name}, keep coding!"
    return message


maya_message = cheer("Maya")
leo_message = cheer("Leo")
print(maya_message)
print(leo_message)
print("The caller uses the returned values because message is local to cheer.")
assert maya_message == "Maya, keep coding!"
assert leo_message == "Leo, keep coding!"
````

**The real program**

```python
def cheer(name):
    message = f"{name}, keep coding!"
    return message


first_name = input("First name? ")
second_name = input("Second name? ")
first_message = cheer(first_name)
second_message = cheer(second_name)
print(first_message)
print(second_message)
```

## Exercise 3

````python
def show_double(number):
    print(number * 2)


def get_double(number):
    doubled = number * 2
    return doubled


doubled_score = get_double(2.5)
double_message = f"Doubling 2.5 gives {doubled_score}."
print(double_message)
print("get_double hands its result back, while show_double only displays it.")
assert doubled_score == 5.0
assert double_message == "Doubling 2.5 gives 5.0."
````

**The real program**

```python

    return doubled
```

## Exercise 11

````python
def total_card_borders(number_of_cards):
    total = 0
    for card_number in range(number_of_cards):
        side = 10 + card_number * 5
        border_length = side * 4
        total = total + border_length
    return total


border_total = total_card_borders(4)
print(border_total)
print("The return line comes after the loop so it returns the completed total.")
assert border_total == 280
````

## Exercise 16

### Check Before You Add

````python
def stamps_that_fit(limit):
    count = 0
    width = 0
    for i in range(5):
        size = 10 + 5 * i
        if width + size <= limit:
            width = width + size
            count = count + 1
        else:
            break
    return count


def width_used(limit):
    width = 0
    for i in range(5):
        size = 10 + 5 * i
        if width + size <= limit:
            width = width + size
        else:
            break
    return width


assert stamps_that_fit(60) == 3
assert width_used(60) == 45
print(stamps_that_fit(60))
print(width_used(60))
````

### Add First, Then Check

````python
def stamps_to_pass(limit):
    count = 0
    width = 0
    for i in range(5):
        size = 10 + 5 * i
        width = width + size
        count = count + 1
        if width > limit:
            break
    return count


def width_when_passed(limit):
    width = 0
    for i in range(5):
        size = 10 + 5 * i
        width = width + size
        if width > limit:
            break
    return width


assert stamps_to_pass(60) == 4
assert width_when_passed(60) == 70
print(stamps_to_pass(60))
print(width_when_passed(60))
````

````python
# The real flower that DRAWS petals lives in assets/solutions_challenge2.py.
# Here we can't run turtle in a notebook, so petal() just returns the size and we
# check the petal COUNT instead — the plan's structure, verified headlessly.
def petal(size):
    return 1


def flower(size, petal_count):
    petals_drawn = 0
    for petal_number in range(petal_count):
        petals_drawn = petals_drawn + petal(size)
    return petals_drawn


flower_petals = flower(30, 6)
print(f"One flower call draws {flower_petals} petals.")
assert flower_petals == 6
````
