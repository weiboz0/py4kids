# Function Factory

Imagine making a greeting card for everyone in your group. Copy-pasting one card over and over is slow, and one change means many repairs.

Today you build a **card factory**: define a machine once, then call it whenever you need it. Each idea comes as a short **ladder** — simplest first, then one step up — with a **Notice** naming the one new thing. At the end of each lesson you apply the same factory idea to turtle stamps.

## Lesson 1: Build a Card Factory

A **function** is a named block of code you define once with `def` and run later by calling its name. Defining it does not run it — a call does.

````python
def blank_card():
    print("+----------------------+")
    print("|                      |")
    print("+----------------------+")


blank_card()
````

**Notice:** `def blank_card():` defines the machine; the last line `blank_card()` **calls** it, and only the call prints a card.

````python
def blank_card():
    print("+----------------------+")
    print("|                      |")
    print("+----------------------+")


blank_card()
blank_card()
blank_card()
````

**Notice:** define once, **call many** — three calls reuse the same body with no copy-paste.

A **parameter** is a blank the function fills in when it is called. `greeting_card(name)` has one parameter, `name`, that can hold a different value on every call.

````python
def greeting_card(name):
    print("+----------------------+")
    print(f"Happy coding, {name}!")
    print("+----------------------+")


greeting_card("Maya")
greeting_card("Leo")
````

**Notice:** the parameter `name` is a blank — each call (`"Maya"`, `"Leo"`) fills it with a different value.

````python
def greeting_card(name, message):
    print("+----------------------+")
    print(f"{message}, {name}!")
    print("+----------------------+")


greeting_card("Ada", "Well done")
greeting_card("Sam", "Happy birthday")
````

**Notice:** one step up — **two parameters**, `name` and `message`, each a separate blank the call fills in order.

**Put it together:** the same factory idea draws turtle stamps. Read this excerpt from `assets/l1_cards.py` (turtle drawings run in their own window):

```python

```

## Lesson 2: Hand a Value Back with `return`

Some machines show their work; others hand a result back so the rest of the program can use it. `return` sends a value back to the line that called the function.

````python
def area(w, h):
    return w * h


card_area = area(7, 4.5)
message = f"The card area is {card_area} square units."
print(message)
````

**Notice:** `area` does not print — it **returns** `w * h`, so `card_area` can store the result and the f-string can use it.

````python
def polygon_points(n):
    return 360 / n


square_turn = polygon_points(4)
seven_side_turn = polygon_points(7)
print(f"A square turns {square_turn} degrees.")
print(f"A seven-sided shape turns {seven_side_turn} degrees.")
````

**Notice:** a returned value can be computed — `360 / n` is a **float**, and the returned number flows straight into the f-string.

Printing only DISPLAYS something; returning HANDS a value back for another job. That is why `polygon_points` can feed a turtle loop instead of repeating the division.

**Put it together:** the shape factory uses the returned angle. Read this excerpt from `assets/l2_shapes.py`:

```python
def polygon_points(n):
    return 360 / n

def polygon(n, side_length):
    turn_angle = polygon_points(n)
    for side in range(n):
        turtle.forward(side_length)
        turtle.right(turn_angle)
```

## Lesson 3: Keep Track of Scope

A name made INSIDE a function is **local** to that function. A name made outside functions is **global**. **Scope** is the part of a program where a name is available.

````python
factory_name = "Function Factory"


def pack_card(name):
    local_label = f"Card for {name}"
    print(factory_name)
    return local_label


first_card = pack_card("Ari")
print(f"Packed: {first_card}")
````

**Notice:** inside `pack_card`, `local_label` is **local** (made inside) while `factory_name` is **global** (made outside) — the function can read the global and return its own local.

Now meet a deliberate bug. Study the next cell, but do not run it — predict the last line of the traceback first.

````python
def make_label(name):
    card_label = f"Made for {name}"
    print(card_label)


make_label("Nia")
print(card_label)
````

```text
NameError: name 'card_label' is not defined
```

**Notice:** `card_label` lives only INSIDE `make_label`; the final `print` runs outside that local scope, so the name is gone. The repair is to `return` the label and store it outside, exactly as Lesson 2 returned an area.

**Put it together:** the final turtle factory nests loops, and each `stamp(size)` call keeps its own local `edge` counter. Read this excerpt from `assets/l3_stamps.py`:

```python
def stamp(size):
    for edge in range(4):
        turtle.forward(size)
        turtle.right(90)

def stamp_gallery(rings, stamps_per_ring):
    for ring_number in range(rings):
        size = 18 + ring_number * 5
        for stamp_number in range(stamps_per_ring):
            stamp(size)
            turtle.right(360 / stamps_per_ring)
```

Each `stamp` call finishes its own square before the next begins. Run the full gallery:

```text

```
