## L1: Build a Greeting-Card Factory

A **function** is a named block of code that you define once and call when you need it. Python starts a function definition with `def`.

A **parameter** is a blank the function fills when it is called. In `greeting_card(name)`, the parameter `name` can hold a different name on every call.

Predict before you run the next cell. How many cards will print? Which name will appear first? Does defining the function print a card, or does calling it print a card?

````python
def greeting_card(name):
    print("+----------------------+")
    print(f"Happy coding, {name}!")
    print("+----------------------+")


greeting_card("Maya")
greeting_card("Leo")
````

The indented lines are the function body. They wait until a call supplies a value for `name`. Two calls reuse the same body without copy-paste.

## L2: Hand a Value Back with `return`

Some factory machines show their work. Other machines hand a result back so the rest of the program can use it. `return` sends a value back to the line that called the function.

Predict the value of `card_area` and the complete message before running the cell. Notice that `4.5` is a **float**, a number with a decimal part.

````python
def area(w, h):
    return w * h


card_area = area(7, 4.5)
message = f"The card area is {card_area} square units."
print(message)
````

`area` does not print. It returns the product, so `card_area` can store it and the f-string can use it. Printing only displays something; returning hands a value back for another job.

A regular polygon must turn a total of 360 degrees. Python's `/` makes a float even when the result is a whole number. Predict the two printed values below. Which result displays `.0`, and which has more digits after the decimal point?

````python
def polygon_points(n):
    return 360 / n


square_turn = polygon_points(4)
seven_side_turn = polygon_points(7)
print(square_turn)
print(seven_side_turn)
````

## L3: Keep Track of Scope

A name made inside a function is **local** to that function. A name made outside functions is **global**. Scope is the part of a program where a name is available.

Predict all three printed lines. Which function call makes each local label?

````python
factory_name = "Function Factory"


def pack_card(name):
    local_label = f"Card for {name}"
    print(factory_name)
    return local_label


first_card = pack_card("Ari")
second_card = pack_card("Bo")
print(f"Packed: {first_card} and {second_card}")
````

Now meet a deliberate bug. Predict the last line of the traceback before running this cell. The error is useful evidence: it tells us where the name is unavailable.

````python
def make_label(name):
    card_label = f"Made for {name}"
    print(card_label)


make_label("Nia")
print(card_label)
````

Read a traceback from the bottom. `NameError` names the kind of problem, and `card_label` names the missing name. The call worked while Python was inside `make_label`; the final print is outside that local scope.

A repair is to return the label and store it outside, just as Lesson 2 returned an area.

The final turtle factory puts one loop inside another. Each call to `stamp(size)` draws one closed stamp and ends where it began. Its local `edge` counter belongs only to that call.

```python
def stamp(size):
    for edge in range(4):
        turtle.forward(size)
        turtle.right(90)

def stamp_gallery(rings, stamps_per_ring):
    for ring_number in range(rings):
        size = 18 + ring_number * 5
        for stamp_number in range(stamps_per_ring):
            stamp(size)
            turtle.right(360 / stamps_per_ring)
```
