## Exercise 2

`range(3)` takes 3 trips with counters 0, 1, 2. `range(5)` takes 5 trips with counters 0, 1, 2, 3, 4. `range(8)` takes 8 trips with counters 0, 1, 2, 3, 4, 5, 6, 7.

```python
n = 7
angle = 360 / n
for side_number in range(n):
    print(side_number + 1, angle)

# Division makes a decimal near 51.428571, and range(7) stops before 7, so its last counter is 6.
assert angle == 51.42857142857143
assert side_number == 6
assert side_number + 1 == 7
```

## Exercise 4

The outer loop takes 24 trips. During each one, the inner loop takes 6 trips, so the design draws 144 straight sides. The outer loop rotates between whole polygons. The inner loop must finish all six sides before the next polygon can begin.

```python
shape_count = 3
side_count = 4
for shape_number in range(shape_count):
    shape_label = shape_number + 1
    for side_number in range(side_count):
        side_label = side_number + 1
        print(shape_label, side_label)

assert 24 * 6 == 144
assert shape_count * side_count == 12
assert shape_label == 3
assert side_label == 4
```

## Exercise 10

The schedule has 3 shapes, 5 sides per shape, and 15 strokes in all.

```python
shape_count = 3
side_count = 5
# The counter formula numbers all five sides of each of three shapes, so the final stroke is 15.
for shape_number in range(shape_count):
    for side_number in range(side_count):
        stroke_number = shape_number * side_count + side_number + 1
        print(f"shape {shape_number + 1}, side {side_number + 1}, stroke {stroke_number}")

assert shape_number == 2
assert side_number == 4
assert stroke_number == 15
assert shape_count * side_count == 15
```
