## L1: Four Sides, Eight Commands

A turtle moves like a tiny robot. `forward(100)` moves it 100 steps. `right(90)` turns it one square corner. The turtle draws while its pen touches the page.


A drawing window will open. Close that window when you finish looking.

````python
import turtle

turtle.forward(100)
turtle.right(90)
turtle.forward(100)
turtle.right(90)
turtle.forward(100)
turtle.right(90)
turtle.forward(100)
turtle.right(90)

turtle.done()
````

The turtle finishes where it started and faces its first direction again.

Now notice the typing. The same two commands appear four times. A 20-sided shape would need 40 repeated commands. That repetition is our problem to solve.

## L2: Polygon Power with `for` and `range()`

A `for` loop repeats an indented block. `range(n)` supplies exactly `n` counter values. The counter starts at 0 and stops before `n`.

Cover the last column. Predict it first. Then uncover it to check.

| Loop | How many trips? | Counter values |
|---|---:|---|
| `range(4)` | 4 | 0, 1, 2, 3 |
| `range(6)` | 6 | 0, 1, 2, 3, 4, 5 |
| `range(1)` | 1 | 0 |

The counter name can describe the work. In our polygon, `side_number` tracks the current side.

A full turn is 360 degrees. Split that turn evenly across `n` corners:

`angle = 360 / n`

For some polygons, this division makes a number with a decimal part. Python calls that kind of number a **float**. The turtle can turn by that amount, so every side still fits around one full turn.

Predict the loop counters before the polygon appears.

| `n` | Counter starts at | Counter stops before | Number of sides |
|---:|---:|---:|---:|
| 3 | 0 | 3 | 3 |
| 5 | 0 | 5 | 5 |
| 7 | 0 | 7 | 7 |

````python
import turtle

n = 7
side_length = 70
angle = 360 / n

turtle.color("blue")
turtle.penup()
turtle.forward(20)
turtle.pendown()

for side_number in range(n):
    turtle.pensize(side_number + 1)
    turtle.forward(side_length)
    turtle.right(angle)

turtle.done()
````

The pen rises for a move that leaves no mark. It lowers before drawing. The color is the literal name `"blue"`. The loop counter also makes each side thicker than the last.

## L3: A Spirograph with Nested Loops

One loop can live inside another loop. These are **nested loops**. The inner loop draws one complete polygon. The outer loop turns the turtle and starts the next polygon.

Read the work from the outside inward.

| Counter | What one trip does |
|---|---|
| `shape_number` | Makes one whole polygon, then rotates |
| `side_number` | Makes one side of that polygon |

The inner loop finishes all six sides before the outer loop moves to its next counter value.

````python
import turtle

n = 6
side_length = 80
angle = 360 / n
shape_count = 24
turn_between_shapes = 15

turtle.color("purple")

for shape_number in range(shape_count):
    for side_number in range(n):
        turtle.forward(side_length)
        turtle.right(angle)
    turtle.right(turn_between_shapes)

turtle.done()
````
