# Word Wizard Exercises

Your WORD WIZARD needs a bigger phrasebook and a sharper spell-checker. Teach it to translate safely, tidy messy searches, and count the words in a growing log. Write your work in each blank code cell.

## Exercise 1: Build a Phrasebook

Build a dictionary named `translations` with these three English-to-Spanish pairs: `"sun"` maps to `"sol"`, `"moon"` maps to `"luna"`, and `"star"` maps to `"estrella"`. Use one dictionary literal, then print the dictionary.

## Exercise 2: Safe Lookup

## Exercise 5: Count the Words

Use the given log `words = ["owl", "fox", "owl", "cat", "owl"]` and start with `counts = {}`. Use a **single loop** over `words`. If a word is already in `counts`, add 1 to its count; otherwise, give its first sighting the count 1. Then use a second loop over `counts.items()` to print each word and count in a friendly f-string. Finally, compare the counts for `"owl"` and `"fox"` with `>` and print the resulting boolean.

## Exercise 7: Grow the Log

## Exercise 11: Fix the KeyError

A Word Wizard tried a missing key and Python actually raised this error:

```text
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
KeyError: 'fish'
```

Before moving on, explain aloud or in your notebook notes what `KeyError: 'fish'` tells you: name the missing key and why the lookup failed.

After reading and explaining the traceback, write `translations = {"cat": "gato"}` in your code cell. Use `.get("fish", "???")` instead of the crashing square-bracket lookup, store the safe result, and print it.

**Why one traceback rep is enough here:** `error-messages` is peripheral in this unit; dictionaries drive Word Wizard, while traceback reading is practiced again in other units.
