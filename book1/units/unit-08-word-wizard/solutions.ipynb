# Word Wizard Solutions

## Exercise 1: Build a Phrasebook

````python
translations = {"sun": "sol", "moon": "luna", "star": "estrella"}
print(translations)
assert translations == {"sun": "sol", "moon": "luna", "star": "estrella"}
````

## Exercise 2: Safe Lookup

````python
translations = {"cat": "gato", "dog": "perro"}
word = "fish"
translation = translations.get(word, "???")
print(translation)
assert translation == "???"
assert translations.get("cat", "???") == "gato"  # hit path
````

## Exercise 3: Is It in the Book?

````python
translations = {"cat": "gato", "dog": "perro"}
word = "cat"
known = word in translations
print(known)
assert known is True
if known:
    print(f"I know {word}!")
else:
    print(f"I do not know {word} yet.")

word = "fish"
known = word in translations
print(known)
assert known is False
if known:
    print(f"I know {word}!")
else:
    print(f"I do not know {word} yet.")
````

## Exercise 4: Add a Word

````python
translations = {"cat": "gato"}
translations["bird"] = "pajaro"
print(translations)
print(f"Bird means {translations['bird']}.")
new_words = 1
print("New words: " + str(new_words))
assert translations["bird"] == "pajaro"
assert "New words: " + str(new_words) == "New words: 1"
````

## Exercise 5: Count the Words

````python
words = ["owl", "fox", "owl", "cat", "owl"]
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
for word, count in counts.items():
    print(f"{word}: {count}")
owls_lead = counts["owl"] > counts["fox"]
print(owls_lead)
assert counts == {"owl": 3, "fox": 1, "cat": 1}
assert owls_lead is True
````

## Exercise 6: Most Common Word

````python
counts = {"owl": 3, "fox": 1, "cat": 1}
best_word = ""
best_count = 0
new_best_checks = []
for word, count in counts.items():
    is_new_best = count > best_count
    print(is_new_best)
    new_best_checks.append(is_new_best)
    if is_new_best:
        best_word = word
        best_count = count
print(best_word, best_count)
assert best_word == "owl"
assert best_count == 3
assert new_best_checks == [True, False, False]
````

## Exercise 7: Grow the Log

````python
words = ["cat", "dog", "cat"]
new_word = "cat"
words.append(new_word)
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
cats_lead = counts["cat"] > counts["dog"]
print(cats_lead)
assert counts == {"cat": 3, "dog": 1}
assert cats_lead is True
````

## Exercise 8: Tidy, Then Translate

````python
translations = {"cat": "gato", "dog": "perro"}
raw_word = "  CAT "
clean_word = raw_word.strip().lower()
translation = translations.get(clean_word, "???")
print(f"{clean_word} means {translation}.")
assert clean_word == "cat"
assert translations.get(raw_word.strip().lower(), "???") == "gato"
````

## Exercise 9: Print Every Pair

````python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
seen_words = []
seen_translations = []
for word, translation in translations.items():
    print(f"{word} means {translation}.")
    seen_words.append(word)
    seen_translations.append(translation)
assert seen_words == ["hello", "cat", "dog"]
assert seen_translations == ["hola", "gato", "perro"]
````

## Exercise 10: Translate a List

````python
def translate(word, dictionary):
    return dictionary.get(word, "???")

translations = {"hello": "hola", "cat": "gato"}
words = ["cat", "hello", "fish"]
translated_words = []
for word in words:
    translation = translate(word, translations)
    translated_words.append(translation)
print(translated_words)
assert translated_words == ["gato", "hola", "???"]
assert translate("fish", translations) == "???"
assert translate("cat", translations) == "gato"  # hit path
````

## Exercise 12: Scoreboard Lines

````python
counts = {"owl": 3, "fox": 1, "cat": 2}
scoreboard = []
champion_checks = []
for word, count in counts.items():
    line = word + " => " + str(count)
    print(line)
    scoreboard.append(line)
    is_champion = count >= 3
    print(is_champion)
    champion_checks.append(is_champion)
print(scoreboard)
assert scoreboard == ["owl => 3", "fox => 1", "cat => 2"]
assert champion_checks == [True, False, False]
assert "owl" + " => " + str(counts["owl"]) == "owl => 3"
````

## Exercise 13: Visit Labels

````python
visits = {"cave": 2, "tower": 4, "lake": 1}
labels = []
for place, visit_count in visits.items():
    label = place + " visits: " + str(visit_count)
    print(label)
    labels.append(label)
print(labels)
assert labels == ["cave visits: 2", "tower visits: 4", "lake visits: 1"]
assert "tower" + " visits: " + str(visits["tower"]) == "tower visits: 4"
````

## Exercise 14: Walk the Keys

````python
translations = {"sun": "sol", "moon": "luna", "star": "estrella"}
visited_words = []
visited_count = 0
for word in translations:
    print(word, translations[word])
    visited_words.append(word)
    visited_count = visited_count + 1
print(visited_words)
print(visited_count)
assert visited_words == ["sun", "moon", "star"]
assert visited_count == 3
assert translations[visited_words[1]] == "luna"
````

## Challenge 1: Reverse Lookup

````python
translations = {"cat": "gato", "dog": "perro", "bird": "pajaro"}
target = "perro"
english_word = "???"
for word, translation in translations.items():
    if translation == target:
        english_word = word
print(english_word)
assert english_word == "dog"
````

## Challenge 2: Merge Two Phrasebooks

````python
translations = {"cat": "gato", "dog": "perro"}
new_translations = {"bird": "pajaro", "fish": "pez"}
for word, translation in new_translations.items():
    translations[word] = translation
print(translations)
assert translations == {"cat": "gato", "dog": "perro", "bird": "pajaro", "fish": "pez"}
````
