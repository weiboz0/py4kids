# Word Wizard Solutions

## Exercise 1: Build a Phrasebook

```python
translations = {"sun": "sol", "moon": "luna", "star": "estrella"}
print(translations)
assert translations["moon"] == "luna"
```

## Exercise 2: Safe Lookup

```python
translations = {"cat": "gato", "dog": "perro"}
word = "fish"
translation = translations.get(word, "???")
print(translation)
assert translation == "???"
```

## Exercise 3: Is It in the Book?

```python
translations = {"cat": "gato", "dog": "perro"}
word = "cat"
if word in translations:
    print(f"I know {word}!")
else:
    print(f"I do not know {word} yet.")

word = "fish"
if word in translations:
    print(f"I know {word}!")
else:
    print(f"I do not know {word} yet.")
```

## Exercise 4: Add a Word

```python
translations = {"cat": "gato"}
translations["bird"] = "pajaro"
print(translations)
print(translations["bird"])
assert translations["bird"] == "pajaro"
```

## Exercise 5: Count the Words

```python
words = ["owl", "fox", "owl", "cat", "owl"]
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
assert counts["owl"] == 3 and counts["fox"] == 1
```

## Exercise 6: Most Common Word

```python
counts = {"owl": 3, "fox": 1, "cat": 1}
best_word = ""
best_count = 0
for word, count in counts.items():
    if count > best_count:
        best_word = word
        best_count = count
print(best_word, best_count)
assert best_word == "owl"
assert best_count == 3
```

## Exercise 7: Grow the Log

```python
words = ["cat", "dog", "cat"]
new_word = "cat"
words.append(new_word)
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
assert counts["cat"] == 3
```

## Exercise 8: Tidy, Then Translate

```python
translations = {"cat": "gato", "dog": "perro"}
raw_word = "  CAT "
clean_word = raw_word.strip().lower()
translation = translations.get(clean_word, "???")
print(translation)
assert clean_word == "cat"
assert translation == "gato"
```

## Exercise 9: Print Every Pair

```python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
for word, translation in translations.items():
    print(f"{word} means {translation}.")
```

## Exercise 10: Translate a List

```python
def translate(word, dictionary):
    return dictionary.get(word, "???")

translations = {"hello": "hola", "cat": "gato"}
words = ["cat", "hello", "fish"]
for word in words:
    print(translate(word, translations))
assert translate("fish", translations) == "???"
```

## Challenge 1: Reverse Lookup

```python
translations = {"cat": "gato", "dog": "perro", "bird": "pajaro"}
target = "perro"
english_word = "???"
for word, translation in translations.items():
    if translation == target:
        english_word = word
print(english_word)
assert english_word == "dog"
```

## Challenge 2: Merge Two Phrasebooks

```python
translations = {"cat": "gato", "dog": "perro"}
new_translations = {"bird": "pajaro", "fish": "pez"}
for word, translation in new_translations.items():
    translations[word] = translation
print(translations)
assert translations["fish"] == "pez"
```
