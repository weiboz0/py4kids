## Exercise 3: Is It in the Book?

````python
translations = {"cat": "gato", "dog": "perro"}
word = "cat"
known = word in translations
print(known)
assert known is True
if known:
    print(f"I know {word}!")
else:
    print(f"I do not know {word} yet.")

word = "fish"
known = word in translations
print(known)
assert known is False
if known:
    print(f"I know {word}!")
else:
    print(f"I do not know {word} yet.")
````

## Exercise 4: Add a Word

## Exercise 5: Grow the Log

````python
words = ["cat", "dog", "cat"]
new_word = "cat"
words.append(new_word)
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
cats_lead = counts["cat"] > counts["dog"]
print(cats_lead)
assert counts == {"cat": 3, "dog": 1}
assert cats_lead is True
````

**The real program**

```python

```

## Exercise 9: Scoreboard Lines

````python
counts = {"owl": 3, "fox": 1, "cat": 2}
scoreboard = []
champion_checks = []
for word, count in counts.items():
    line = word + " => " + str(count)
    print(line)
    scoreboard.append(line)
    is_champion = count >= 3
    print(is_champion)
    champion_checks.append(is_champion)
print(scoreboard)
assert scoreboard == ["owl => 3", "fox => 1", "cat => 2"]
assert champion_checks == [True, False, False]
assert "owl" + " => " + str(counts["owl"]) == "owl => 3"
````

## Exercise 10: Visit Labels

````python
visits = {"cave": 2, "tower": 4, "lake": 1}
labels = []
for place, visit_count in visits.items():
    label = place + " visits: " + str(visit_count)
    print(label)
    labels.append(label)
print(labels)
assert labels == ["cave visits: 2", "tower visits: 4", "lake visits: 1"]
assert "tower" + " visits: " + str(visits["tower"]) == "tower visits: 4"
````

## Exercise 11: Walk the Keys

````python
translations = {"sun": "sol", "moon": "luna", "star": "estrella"}
visited_words = []
visited_count = 0
for word in translations:
    print(word, translations[word])
    visited_words.append(word)
    visited_count = visited_count + 1
print(visited_words)
print(visited_count)
assert visited_words == ["sun", "moon", "star"]
assert visited_count == 3
assert translations[visited_words[1]] == "luna"
````

## Exercise 12: Count the Words

````python
words = ["owl", "fox", "owl", "cat", "owl"]
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
for word, count in counts.items():
    print(f"{word}: {count}")
owls_lead = counts["owl"] > counts["fox"]
print(owls_lead)
assert counts == {"owl": 3, "fox": 1, "cat": 1}
assert owls_lead is True
````

## Exercise 13: Most Common Word

````python
counts = {"owl": 3, "fox": 1, "cat": 1}
best_word = ""
best_count = 0
new_best_checks = []
for word, count in counts.items():
    is_new_best = count > best_count
    print(is_new_best)
    new_best_checks.append(is_new_best)
    if is_new_best:
        best_word = word
        best_count = count
print(best_word, best_count)
assert best_word == "owl"
assert best_count == 3
assert new_best_checks == [True, False, False]
````

## Exercise 14: Translate a List

````python
def translate(word, dictionary):
    return dictionary.get(word, "???")

translations = {"hello": "hola", "cat": "gato"}
words = ["cat", "hello", "fish"]
translated_words = []
for word in words:
    translation = translate(word, translations)
    translated_words.append(translation)
print(translated_words)
assert translated_words == ["gato", "hola", "???"]
assert translate("fish", translations) == "???"
assert translate("cat", translations) == "gato"  # hit path
````

**The real program**

```python
def translate(word, dictionary):
    return dictionary.get(word, "???")

translations = {"hello": "hola", "cat": "gato"}
words = [input("Word 1? "), input("Word 2? "), input("Word 3? ")]
translated_words = []
for word in words:
    translated_words.append(translate(word, translations))
print(translated_words)
```

## Exercise 16: Keep only the long words

````python
words = ["cat", "dragon", "owl", "wizard", "sun"]
minimum_length = 5
long_words = []

for word in words:
    if len(word) >= minimum_length:
        long_words.append(word)

print(long_words)
assert long_words == ["dragon", "wizard"]
assert long_words is not words
assert words == ["cat", "dragon", "owl", "wizard", "sun"]
````

**The real program**


## Exercise 18: Rarest Word

````python
counts = {"owl": 3, "fox": 1, "cat": 2}
first_pair = True

for word, count in counts.items():
    if first_pair:
        rarest_word = word
        rarest_count = count
        first_pair = False
    elif count < rarest_count:
        rarest_word = word
        rarest_count = count

print(f"Rarest word: {rarest_word} ({rarest_count})")
assert rarest_word == "fox"
assert rarest_count == 1
````

## Exercise 19: Known and Unknown

````python
words = ["cat", "zzz", "dog"]
translations = {"cat": "gato", "dog": "perro"}
known = 0
unknown = 0

for word in words:
    if word in translations:
        known = known + 1
    else:
        unknown = unknown + 1

print(f"Known: {known}; unknown: {unknown}")
assert known == 2
assert unknown == 1
````

## Exercise 20: Word Lengths That Fit

````python
words = ["owl", "dragon", "cat", "wizard", "sun"]
budget = 12
fit_count = 0
fit_total = 0

for word in words:
    word_length = len(word)
    if fit_total + word_length <= budget:
        fit_total = fit_total + word_length
        fit_count = fit_count + 1
    else:
        break

print(f"Words that fit: {fit_count}; total length: {fit_total}")
assert fit_count == 3
assert fit_total == 12
````

## Exercise 21: Word Lengths Until It Tips

````python
words = ["owl", "dragon", "cat", "wizard", "sun"]
budget = 12
tip_count = 0
tip_total = 0
tipping_word = ""

for word in words:
    tip_total = tip_total + len(word)
    tip_count = tip_count + 1
    if tip_total > budget:
        tipping_word = word
        break

print(f"Tipped at {tipping_word}: {tip_count} words, total length {tip_total}")
assert tip_count == 4
assert tip_total == 18
assert tipping_word == "wizard"
````

**The real program**


## Challenge 1: Merge Two Phrasebooks

````python
translations = {"cat": "gato", "dog": "perro"}
new_translations = {"bird": "pajaro", "fish": "pez"}
for word, translation in new_translations.items():
    translations[word] = translation
print(translations)
assert translations == {"cat": "gato", "dog": "perro", "bird": "pajaro", "fish": "pez"}
````

## Challenge 2: Flip the Phrasebook

````python
translations = {"cat": "gato", "dog": "perro", "bird": "pajaro"}
spanish_to_english = {}

for word, translation in translations.items():
    spanish_to_english[translation] = word

print(spanish_to_english)
assert spanish_to_english == {"gato": "cat", "perro": "dog", "pajaro": "bird"}
````
