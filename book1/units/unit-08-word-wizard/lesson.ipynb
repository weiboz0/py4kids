# Word Wizard

A **WORD WIZARD** translates a word in a blink and counts how often each word appears. Its magic is a **dictionary** — pairs of keys and values. Each idea comes as a short **ladder** (simplest → one step up → real use) with a **Notice** naming the one new thing.

## Lesson 1: Build the Wizard's Phrasebook

A **dictionary** (dict) pairs each **key** with a **value**. Curly braces hold the pairs; a colon joins each key to its value.

```python
translations = {"hello": "hola"}
print(translations)
```

**Notice:** one pair — the key `"hello"` maps to the value `"hola"`.

```python
translations = {"hello": "hola", "cat": "gato"}
print(translations)
```

**Notice:** separate more pairs with commas.

```python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
print(translations)
```

**Notice:** a dict can hold as many key→value pairs as you need — the wizard's phrasebook.

To look up a value, put its key in square brackets.

```python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
print(translations["hello"])
```

**Notice:** `translations["hello"]` returns that key's value, `"hola"`.

```python
translations["bird"] = "pajaro"
print(translations)
```

**Notice:** assign to a new key to ADD a pair (or to an existing key to change its value).

```python
print(translations.get("fish", "???"))
```

**Notice:** `.get(key, default)` returns the value, or the default (`"???"`) when the key is missing — no crash.

Why `.get`? A missing key in square brackets CRASHES. Here is the deliberate bug — leave it unrun and read the traceback from the bottom.

```python
translations = {"hello": "hola", "cat": "gato"}
translations["fish"]
```

`KeyError: 'fish'` means the key is not in the dictionary. `.get()` exists precisely to avoid this when a word might be missing. (Search words also arrive messy; `raw.strip().lower()` cleans them to match the tidy lowercase keys.)

## Lesson 2: Walk the Phrasebook

A `for` loop can walk a dictionary. Looping over the dict gives one KEY per trip.

```python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
for word in translations:
    print(word)
```

**Notice:** looping over a dict visits its keys, one per trip.

```python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
for word, translation in translations.items():
    print(f"{word} means {translation}")
```

**Notice:** `.items()` gives the key AND value together on each trip.

```python
def translate(word, dictionary):
    return dictionary.get(word, "???")


translations = {"hello": "hola", "cat": "gato", "dog": "perro", "bird": "pajaro"}
words_to_translate = ["hello", "bird", "fish"]
for word in words_to_translate:
    print(f"{word} means {translate(word, translations)}")
```

**Notice:** put `.get` in a helper and loop a whole list through it — a missing word gets the friendly default.

## Lesson 3: Count the Word Log

A spell-checker notices repeated words. Start with an empty dict `counts`, then one loop over the list: a seen word grows by 1; a first sighting begins at 1.

```python
words = ["cat", "dog", "cat", "bird", "cat"]
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
```

**Notice:** `if word in counts` chooses between growing a count and starting a new one at 1.

```python
counts = {"cat": 3, "dog": 1, "bird": 1}
for word, count in counts.items():
    label = word + " => " + str(count)
    print(label)
```

**Notice:** walk the pairs with `.items()` and build a label; `str(count)` turns the number into text so `+` can join it.

```python
counts = {"cat": 3, "dog": 1, "bird": 1}
best_word = ""
best_count = 0
for word, count in counts.items():
    if count > best_count:
        best_count = count
        best_word = word
print(f"Most common word: {best_word} ({best_count})")
```

**Notice:** track the best-so-far count while looping to find the most common word.
