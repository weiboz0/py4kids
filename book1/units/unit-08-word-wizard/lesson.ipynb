# Word Wizard

## Lesson 1: Build the Wizard's Phrasebook

A **WORD WIZARD** can translate a word in a blink. Like a tiny spell-checker, it will also learn to count how often each word appears. We will begin with its bilingual phrasebook: each English word is paired with one Spanish translation. How could our wizard find the right pair instantly?

````python
translations = {"hello": "hola", "cat": "gato", "dog": "perro"}
print(translations)
````

This is a **dictionary**, or **dict** for short. Each word before a colon is a **key**, and its partner after the colon is a **value**. To cast the wizard's first translation spell, put the key `"hello"` inside square brackets.

````python
print(translations["hello"])
````

A bird lands beside the wizard with a new word. We add the translation by putting the new key in square brackets and assigning its value. The dictionary now remembers four pairs.

````python
translations["bird"] = "pajaro"
print(translations)
````

Search words can arrive with extra spaces or capital letters. Our keys are tidy lowercase words, so we clean a messy search before looking it up. `.strip()` removes outside spaces, and `.lower()` changes the letters to lowercase. Now `"  Hello "` matches the key `"hello"`.

````python
raw_word = "  Hello "
clean = raw_word.strip().lower()
print(clean)
print(translations[clean])
````

Before a lookup, the wizard can check whether a key is in the phrasebook. The result of `"cat" in translations` is either `True` or `False`. A `True`/`False` value is called a **boolean**.

````python
print("cat" in translations)
````

What if the phrasebook does not know a word yet? The safe `.get()` lookup accepts a default value. For the missing key `"fish"`, the wizard returns `"???"` instead of crashing.

````python
print(translations.get("fish", "???"))
````

Here is a deliberate bug to inspect together. Square-bracket lookup expects its key to exist, but `"fish"` is missing. The cell is marked to stay unrun when the lesson is checked. Read the traceback below as a clue, starting from its final line.

````python
translations["fish"]
````


The final line names the problem: `KeyError` means that the requested key is not in the dictionary. `.get()` exists precisely to avoid this crash when a word might be missing.

## Lesson 2: Count and Translate the Word Log

Yesterday we looked words up one at a time; today we count them and translate a whole list. Our WORD WIZARD will walk through its phrasebook and a spell-checker's word log, one item at a time.

First, the wizard walks through the dictionary's keys. On each trip through the loop, `word` holds one English word.

````python
for word in translations:
    print(word)
````

Often we need each key and value together. `.items()` gives the loop one complete pair at a time: `word` receives the key, and `translation` receives its value.

````python
for word, translation in translations.items():
    print(f"{word} means {translation}")
````

Let's give the translation spell a name. The helper function receives a word and a dictionary as parameters. It returns the matching value, or `"???"` when the word is not there.

````python
def translate(word, dictionary):
    return dictionary.get(word, "???")


print(translate("cat", translations))
print(translate("fish", translations))
````

Now the wizard can translate a whole given list. The loop sends each word through the same helper spell. A missing word receives the friendly default.

````python
words_to_translate = ["hello", "bird", "fish"]
for word in words_to_translate:
    print(f"{word} means {translate(word, translations)}")
````

A spell-checker also needs to notice repeated words. Here is a given word log. We start with an empty dictionary named `counts`, then use one loop over the list. A word we have seen grows by 1; a first sighting begins at 1.

````python
words = ["cat", "dog", "cat", "bird", "cat"]
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
````

The counter's dictionary pairs each word with an integer count. We walk through those pairs and build the exact label `word + " => " + str(count)`. `str(count)` changes the number into text so the pieces can join.

````python
for word, count in counts.items():
    label = word + " => " + str(count)
    print(label)
````

Which word appears most often? Start with an empty best word and a best count of 0. One loop compares every pair with the best count so far. When a count is greater, that word becomes the new winner.

````python
best_word = ""
best_count = 0
for word, count in counts.items():
    if count > best_count:
        best_count = count
        best_word = word

print(f"Most common word: {best_word} ({best_count})")
````

A new `"cat"` arrives in the word log. After appending it, we **reset** `counts` to an empty dictionary and count the entire grown list from scratch. The reset matters: keeping the old counts and counting every word again would double-count the earlier words.

````python
words.append("cat")
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1

print(f"Cat count after the new word: {counts['cat']}")
````
