# High-Score Hall of Fame Exercises

The arcade's scrap-paper score board is overflowing. Build a Python list that can remember new scores, show every entry, and keep the hall ready for the next player. Write your work in each blank code cell.

## Exercise 1

### Build the Board from Scratch

Create a list named `scores` with the integer scores 700, 1250, and 980. Append the new integer score 1100. Print the whole score board in an f-string.

## Exercise 2

### Champion and Rookie

Start with `scores = [1200, 850, 990, 1500]`. Use `max` to find the champion score and `min` to find the rookie score. Store both results and print each one in an f-string.

## Exercise 3

### Number the Arrival-Order Score Sheet

Use the unsorted list `scores = [1200, 850, 990, 1500]`. Loop through its positions with `for position in range(len(scores)):`. Print each item exactly like `Entry #1: 1200`, using `position + 1` for the entry number and `scores[position]` for the score. Keep the scores in arrival order.

## Exercise 4

### Total and Average

Start with `scores = [600, 800, 1000]` and `total = 0`. Use one loop to add every score to the accumulator. Then calculate `average = total / len(scores)`. Print the total and the decimal average in f-strings.

## Exercise 5

### Award a Score Tier

Store an integer in `score`. If it is at least 1000, store `"gold"` in `tier`; elif it is at least 500, store `"silver"`; otherwise store `"bronze"`. Print the score and tier in an f-string. Test the branch with 1000, 999, and 499 one value at a time.

## Exercise 6

### Tidy a Sample Champion Name

Store the scalar string `"  grace hopper  "` in `winner`. Rebuild it without outside spaces and in uppercase by chaining `.strip().upper()`, then store the changed string back in `winner`. Print it in an f-string. This illustrative name is not paired with a score; true name-to-score pairing needs dictionaries in Unit 08.

## Exercise 7

### Add My Score and Keep the Hall Ranked

A game might ask for a score with `input("What score did you earn? " )`, but do **not** use that call in your code cell. Instead, define `add_my_score(scores, new_score)` so the score is a parameter. Inside the function, append `new_score`, immediately sort the same list top-first with `.sort(reverse=True)`, and return `scores`. Start with the already-ranked list `[1500, 1200, 990, 850]`, call the function with a fixed sample score of 1310, and print the result.

## Exercise 8

### Label an Already-Sorted Board

Define `board_line(place, score)` with two parameters. Return exactly `"Place " + str(place) + ": " + str(score)`. Use the already-sorted top-first list `scores = [1500, 1200, 990, 850]`. Loop with `for place in range(len(scores)):` and print the helper's result for each score, using `place + 1` for the displayed place.

## Exercise 9

### Guard Against a Repeated Score

Start with the already-ranked list `scores = [1500, 1200, 990, 850]` and a fixed `new_score`. Use `if new_score in scores` as a condition. If it is already present, print a helpful message without printing the truth value. Otherwise, append it and immediately sort the board top-first before printing the updated board. Test once with 990 and once with 1310.

## Challenge

### Challenge 1: Show the Top Three

Start with `scores = [980, 1500, 720, 1310, 1100]`. Sort the list top-first in place. Then print the top three scores by using indexes 0, 1, and 2.

### Challenge 2: Merge a Second Score Board

Start with the unranked, arrival-order first board `scores = [1200, 990, 1500]` and a second arcade's `second_scores = [1310, 700, 1050]`. Use **one loop** to append each score from `second_scores` to `scores`. Do not sort after each append. After the loop is completely finished, perform exactly one `scores.sort(reverse=True)`. Only then print the merged board; do not display any Place labels before that one final sort. Do not use a nested loop.
