# High-Score Hall of Fame Exercises

The arcade's scrap-paper score board is overflowing. Build a Python list that can remember new scores, show every entry, and keep the hall ready for the next player. Write your work in each blank code cell.

## Exercise 1

### Build the Board from Scratch

Create a list named `scores` with the integer scores 700, 1250, and 980. Store 1100 in `new_score`, append it, and print the whole score board in an f-string. Then print `new_score in scores` so the membership result appears directly as `True` or `False`.

## Exercise 2

### Champion and Rookie

Start with `scores = [1200, 850, 990, 1500]`. Use `max` to find the champion score and `min` to find the rookie score. Calculate the decimal midpoint between them with `(champion + rookie) / 2`. Sort the list top-first, then print `scores[0]`, `scores[-1]`, and the midpoint in f-strings. Finally, write an `if`/`else` that prints whether the champion is greater than the rookie.

## Exercise 3

### Read an IndexError, Then Number the Score Sheet

Use the unsorted list `scores = [1200, 850, 990, 1500]`. In the first blank cell, deliberately write `scores[len(scores)]`, predict the last line of the traceback, and run it only when your teacher asks. Expect `IndexError: list index out of range`. In the second blank cell, fix the bug by safely printing the last score with the negative index `scores[-1]`. Then loop through the positions with `for position in range(len(scores)):` and print each item exactly like `Entry #1: 1200`, using `position + 1` and `scores[position]`. Keep the scores in arrival order.

## Exercise 4

### Total and Average

Start with `scores = [600, 800, 1000]` and `total = 0`. Use one loop to add every score to the accumulator. Then calculate `average = total / len(scores)` and use `max(scores)` for the best score. Store `average >= 700` in `reached_goal`. Print the total, decimal average, best score, and the boolean value in f-strings.

## Exercise 5

### Award a Score Tier

Start with `scores = [1000, 999, 499]` and `gold_count = 0`. Loop over the scores. If a score is at least 1000, store `"gold"` in `tier` and add 1 to `gold_count`; elif it is at least 500, store `"silver"`; otherwise store `"bronze"`. Print each score and tier in an f-string, then print the accumulated gold count.

## Exercise 6

### Tidy the Champion Names

Start with `raw_winners = ["  grace hopper  ", "ADA", " ada "]` and an empty `clean_winners` list. Loop over the raw names and rebuild each one with `.strip().upper()`. If the cleaned name is already `in clean_winners`, print a repeated-name message; otherwise append it. At the end, print `"Clean winners: " + str(clean_winners)`. Keep this name list separate from the score list.

## Exercise 7

### Add My Score and Keep the Hall Ranked

Define `add_my_score(scores, new_score)` so the score is a parameter. Inside the function, append `new_score`, immediately sort the same list top-first with `.sort(reverse=True)`, and return `scores`. Start with `[1500, 1200, 990, 850]`, read a score with `int(input("What score did you earn? "))`, call the function, and print the ranked list and its champion at `scores[0]`. This interactive exercise cell is intentionally tagged `no-exec`; run it yourself and enter a score such as 1310.

## Exercise 8

### Label an Already-Sorted Board

Define `board_line(place, score)` with two parameters. Return exactly `"Place " + str(place) + ": " + str(score)`. Use the already-sorted top-first list `scores = [1500, 1200, 990, 850]`. Loop with `for place in range(len(scores)):` and print the helper's result for each score, using `place + 1` for the displayed place.
