# High-Score Hall of Fame

Every arcade game needs a leaderboard. A messy scrap of paper can hold a few scores, but our program needs to remember every score, grow when a new player finishes, and crown a champion. We build that hall with a Python **list**.

Each idea comes as a short **ladder** — simplest first, then one step up, then the real board — with a **Notice** naming the one new thing.

## L1: Build the Score Board

A **list** lets one variable hold many items in order. Square brackets mark its start and end; commas separate the items.

```python
scores = []
print(scores)
```

**Notice:** `[]` is an empty list — a board with no scores yet.

```python
scores = [1200, 850]
print(scores)
```

**Notice:** put items inside the brackets, separated by commas.

```python
scores = [1200, 850, 990, 1500]
print(f"Arcade scores: {scores}")
```

**Notice:** a list can hold as many items as you like — here, the four arcade scores.

List positions start at 0. Square brackets with a position select one item.

```python
scores = [1200, 850, 990, 1500]
print(scores[0])
```

**Notice:** `scores[0]` is the FIRST item (position 0).

```python
print(scores[2])
```

**Notice:** each position picks the item sitting there — `scores[2]` is the third.

```python
print(scores[-1])
```

**Notice:** `-1` counts back from the end — the LAST score.

The list method `.append()` adds a new item to the END of the same list.

```python
scores = [1200, 850]
scores.append(990)
print(scores)
```

**Notice:** `.append(990)` adds `990` to the end of `scores`.

```python
best_so_far = 1500
scores.append(best_so_far)
print(scores)
```

**Notice:** you can append a value held in a variable too, not only a literal.

```python
board = []
new_scores = [1200, 850, 990]
for one_score in new_scores:
    board.append(one_score)
print(board)
```

**Notice:** a `for` loop can append many items in a row to build a list.

A `for` loop can also VISIT each item in a list, one at a time.

```python
scores = [1200, 850, 990, 1500]
for one_score in scores:
    print(f"Score: {one_score}")
```

**Notice:** the loop runs once per item, with `one_score` holding each value in turn.

```python
scores = [1200, 850, 990, 1500]
total = 0
for one_score in scores:
    total = total + one_score
print(f"Total points: {total}")
```

**Notice:** add each item into an accumulator to build a running total over the whole list.

```python
scores = [1200, 850, 990, 1500]
for position in range(len(scores)):
    print(scores[position])
```

**Notice:** you can also loop by POSITION with `range(len(scores))`, reading each item as `scores[position]`.

```python
scores = [1200, 850, 990, 1500]
for position in range(len(scores)):
    print(f"Entry #{position + 1}: {scores[position]}")
```

**Notice:** one step more — add 1 to the position for a friendly entry number beside each score.

**Built-in functions** measure a list. `len` counts items, `max` finds the biggest, `min` the smallest.

```python
scores = [1200, 850, 990, 1500]
print(len(scores))
```

**Notice:** `len(scores)` is how many items the list holds.

```python
print(max(scores))
```

**Notice:** `max(scores)` finds the champion — the biggest score.

```python
print(min(scores))
```

**Notice:** `min(scores)` finds the smallest score.

Dividing the total by the count gives the average. Python writes that as a decimal, called a **float**.

```python
scores = [1200, 850, 990, 1500]
total = 0
for one_score in scores:
    total = total + one_score
average = total / len(scores)
print(f"Average score: {average}")
```

**Notice:** `/` gives a float (a decimal) — the average of the four scores.

## L2: Rank the Hall and Crown the Champion

The list method `.sort()` rearranges the SAME list in place (it changes the list, and returns `None`, so call it on its own line).

```python
scores = [1200, 850, 990, 1500]
scores.sort()
print(f"Low to high: {scores}")
```

**Notice:** `.sort()` puts the items in order, lowest first.

```python
scores.sort(reverse=True)
print(f"Top score first: {scores}")
```

**Notice:** `.sort(reverse=True)` orders them highest first — the leaderboard order.

```python
scores = [1200, 850, 990, 1500]
scores.sort(reverse=True)
print(f"Champion: {scores[0]}")
print(f"Runner-up: {scores[1]}")
```

**Notice:** once the list is sorted top-first, `scores[0]` is the champion and `scores[1]` the runner-up — sorting makes the ranks easy to read.

Unlike Unit 06's string methods (which RETURN a changed string), `.sort()` mutates the list in place and returns `None` — so write `scores.sort()`, never `best = scores.sort()`.

**Put it together:** a helper prints one ranked line, and we loop the sorted board to show every place.

```python
def board_line(place, score):
    return "Place " + str(place) + ": " + str(score)


scores = [1500, 1200, 990, 850]
for place in range(len(scores)):
    print(board_line(place + 1, scores[place]))
```

**Put it together:** a ranked hall must stay top-first when a new score arrives. This helper appends the score and immediately re-sorts the same list before returning it.

```python
def board_line(place, score):
    return "Place " + str(place) + ": " + str(score)


def add_score(scores, new_score):
    scores.append(new_score)
    scores.sort(reverse=True)
    return scores


scores = [1500, 1200, 990, 850]
scores = add_score(scores, 1310)
for place in range(len(scores)):
    print(board_line(place + 1, scores[place]))
```

**Notice:** append + re-sort keeps the hall ranked after every new score.

## L3: Polish the Hall

A tier rewards one score: at least 1000 earns gold, at least 500 silver, else bronze.

```python
score = 990
if score >= 1000:
    tier = "gold"
elif score >= 500:
    tier = "silver"
else:
    tier = "bronze"
print(f"A score of {score} earns the {tier} tier.")
```

Before adding a score, membership can guard against recording it twice. `in` belongs inside the `if` condition.

```python
scores = [1500, 1310, 1200, 990, 850]
new_score = 1310
if new_score in scores:
    print(f"The score {new_score} is already on the board.")
else:
    print(f"Adding {new_score}.")
```

Imagine a challenge where the qualifying score doubles each round. A `while` loop doubles a threshold until it reaches the champion score.

```python
champion = 1500
threshold = 10
while threshold < champion:
    threshold = threshold * 2
print(f"First doubled threshold at or above {champion}: {threshold}")
```
