# High-Score Hall of Fame

Every arcade game needs a leaderboard. A messy scrap of paper can hold a few scores, but our program needs to remember every score, grow when a new player finishes, and help us crown a champion. We will build that high-score hall with a Python list.

## L1: Build the Score Board

The arcade has four scores scribbled on its paper leaderboard. A **list** lets one variable hold all four integer scores in order. Square brackets mark the beginning and end of the list, and commas separate its items.

```python
scores = [1200, 850, 990, 1500]
print(f"Arcade scores: {scores}")
```

List positions start at 0. That makes `scores[0]` the first score. The position `-1` means the last score. Predict both values before you run the cell.

```python
print(f"First score: {scores[0]}")
print(f"Last score: {scores[-1]}")
```

A fifth player finishes the game with 1310 points. The list method `.append()` adds that new score to the end of our one score board.

```python
scores.append(1310)
print(f"Board after the new game: {scores}")
```

The score keeper should show every entry. A `for` loop can visit each score in the list, one at a time.

```python
for score in scores:
    print(f"Score: {score}")
```

The board is still in arrival order, so its numbers describe entries, not how well anybody scored. The loop counter `position` visits 0 through one less than the list length. We add 1 only for the friendly entry number that players see.

```python
for position in range(len(scores)):
    print(f"Entry #{position + 1}: {scores[position]}")
```

Built-in functions can measure the score board. `len(scores)` counts its entries, `max(scores)` finds the champion score, and `min(scores)` finds the rookie score.

```python
print(f"Number of scores: {len(scores)}")
print(f"Champion score: {max(scores)}")
print(f"Rookie score: {min(scores)}")
```

To find the arcade's average, an accumulator named `total` starts at 0 and grows by each score. Dividing the total by the number of scores gives `average`. Python writes this result as a decimal number, which is called a **float**.

```python
total = 0
for score in scores:
    total = total + score

average = total / len(scores)
print(f"Total points: {total}")
print(f"Average score: {average}")
```

## L2: Rank the Hall and Crown the Champion

Yesterday we listed scores; today we rank them and crown the champion. We will start with a fresh score board and save the champion score before we change the board's order.

```python
scores = [1200, 850, 990, 1500]
champion = max(scores)
print(f"Champion score: {champion}")
```

The list method `.sort()` rearranges the one list from low to high. Then `.sort(reverse=True)` rearranges that same list from high to low, putting the top score first.

```python
scores.sort()
print(f"Low to high: {scores}")

scores.sort(reverse=True)
print(f"Top score first: {scores}")
```

List sorting **mutates the one list in place**: the items inside `scores` move, and we keep using the same variable. In Unit 06, string methods rebuilt strings and returned changed strings because strings were not mutated. A list can change in place. Because it changes the list in place, `.sort()` returns `None` — so call it on its own line (`scores.sort()`), never `best = scores.sort()`.

Now that the board is sorted top-first, each item can receive a place label. The helper function has two parameters and returns one finished line of text.

```python
def board_line(place, score):
    return "Place " + str(place) + ": " + str(score)


for place in range(len(scores)):
    print(board_line(place + 1, scores[place]))
```

A ranked hall must stay top-first when a new score arrives. This helper appends the new score and **immediately** sorts the same list again before returning it. Our current board is already ranked, so we can safely demonstrate the update.

```python
def add_score(scores, new_score):
    scores.append(new_score)
    scores.sort(reverse=True)
    return scores


scores = add_score(scores, 1310)
for place in range(len(scores)):
    print(board_line(place + 1, scores[place]))
```

Here is an illustrative string-method demo, separate from the score data. The scalar variable `winner` is **not linked to a score**. True name-to-score pairing needs dictionaries in Unit 08; we will keep only integer scores here.

```python
winner = "  ada lovelace  ".strip().upper()
print(f"Sample champion name: {winner}")
```

The arcade also awards a tier for one score: at least 1000 points earns gold, at least 500 earns silver, and anything lower earns bronze.

```python
score = 990
if score >= 1000:
    tier = "gold"
elif score >= 500:
    tier = "silver"
else:
    tier = "bronze"

print(f"A score of {score} earns the {tier} tier.")
```

Before adding a score, membership can guard against recording the same score again. The `in` operator belongs inside the `if` condition; the program prints a useful message instead of printing a truth value.

```python
new_score = 1310
if new_score in scores:
    print(f"The score {new_score} is already on the board.")
else:
    scores = add_score(scores, new_score)
    print(f"Added {new_score} and kept the board top-first.")
```

Imagine an arcade challenge where the score needed to qualify doubles every round. A `while` loop can keep doubling the threshold until it reaches or passes the champion score.

```python
threshold = 10
while threshold < champion:
    threshold = threshold * 2

print(f"First doubled threshold at or above {champion}: {threshold}")
```
