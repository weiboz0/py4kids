## Exercise 2

```python
scores = [1200, 850, 990, 1500]
champion = max(scores)
rookie = min(scores)
midpoint = (champion + rookie) / 2
scores.sort(reverse=True)
print(f"Champion: {scores[0]}")
print(f"Rookie: {scores[-1]}")
print(f"Midpoint: {midpoint}")

if champion > rookie:
    print("The champion is greater than the rookie.")
else:
    print("The champion is not greater than the rookie.")

assert scores == [1500, 1200, 990, 850]
assert scores[0] == champion
assert scores[-1] == rookie
assert midpoint == 1175.0
```

## Exercise 4

```python
def add_my_score(scores, new_score):
    scores.append(new_score)
    scores.sort(reverse=True)
    return scores


scores = [1500, 1200, 990, 850]
new_score = 1310
scores = add_my_score(scores, new_score)
print(f"Ranked board: {scores}")
print(f"Champion: {scores[0]}")

assert scores == [1500, 1310, 1200, 990, 850]
assert scores[0] == 1500
```

## Exercise 9

```python
waiting_scores = [300, 450, 275, 600]
bonus_board = []
position = 0
total = 0

while total < 1000:
    score = waiting_scores[position]
    bonus_board.append(score)
    total = total + score
    position = position + 1

average = total / len(bonus_board)
print(f"Bonus board: {bonus_board}")
print(f"Last score: {bonus_board[-1]}")
print(f"Average: {average}")

assert bonus_board == [300, 450, 275]
assert bonus_board[-1] == 275
assert total == 1025
assert average == 341.6666666666667
```

## Exercise 18

### Best and Worst in One Pass

```python
scores = [720, 880, 1050, 940]
best = scores[0]
worst = scores[0]

for score in scores:
    if score > best:
        best = score
    if score < worst:
        worst = score

print(f"Best: {best}; worst: {worst}")
assert best == 1050
assert worst == 720
```

## Exercise 19

### Average of the Passers

```python
scores = [600, 800, 1000, 900, 700]
pass_mark = 800
passer_count = 0
passer_total = 0

for score in scores:
    if score >= pass_mark:
        passer_count = passer_count + 1
        passer_total = passer_total + score

if passer_count == 0:
    average = 0
else:
    average = passer_total / passer_count

print(f"Passers: {passer_count}; average: {average}")
assert passer_count == 3
assert average == 900.0
```

## Exercise 21

### Scores That Fit the Wait Budget

```python
waiting = [300, 450, 275, 600]
budget = 1000
fit_count = 0
fit_total = 0

for wait in waiting:
    if fit_total + wait <= budget:
        fit_total = fit_total + wait
        fit_count = fit_count + 1
    else:
        break

print(f"Fits: {fit_count}; total wait: {fit_total}")
assert fit_count == 2
assert fit_total == 750
```

## Exercise 22

### Waits Until the Budget Tips

```python
waiting = [300, 450, 275, 600]
budget = 1000
tip_count = 0
tip_total = 0
tipping_wait = 0

for wait in waiting:
    tip_total = tip_total + wait
    tip_count = tip_count + 1
    if tip_total > budget:
        tipping_wait = wait
        break

print(f"Tips after {tip_count} waits at {tip_total}; tipping wait: {tipping_wait}")
assert tip_count == 3
assert tip_total == 1025
assert tipping_wait == 275
```
