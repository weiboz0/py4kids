## Exercise 4

```python
scores = [600, 800, 1000]
total = 0

for score in scores:
    total = total + score

average = total / len(scores)
print(f"Total: {total}")
print(f"Average: {average}")

assert total == 2400
assert average == 800.0
```

## Exercise 7

```python
def add_my_score(scores, new_score):
    scores.append(new_score)
    scores.sort(reverse=True)
    return scores


scores = [1500, 1200, 990, 850]
new_score = 1310
starting_length = len(scores)
champion = max(scores)
scores = add_my_score(scores, new_score)
print(f"Ranked board: {scores}")

assert len(scores) == starting_length + 1
assert scores[0] == champion
```

## Exercise 8

```python
def board_line(place, score):
    return "Place " + str(place) + ": " + str(score)


scores = [1500, 1200, 990, 850]
for place in range(len(scores)):
    print(board_line(place + 1, scores[place]))

place_label = board_line(1, scores[0])
assert place_label == "Place 1: 1500"
```

## Exercise 9

```python
scores = [1500, 1200, 990, 850]
starting_length = len(scores)
new_score = 990
if new_score in scores:
    print(f"The score {new_score} is already on the board.")
else:
    scores.append(new_score)
    scores.sort(reverse=True)
    print(f"Updated board: {scores}")
assert len(scores) == starting_length

new_score = 1310
if new_score in scores:
    print(f"The score {new_score} is already on the board.")
else:
    scores.append(new_score)
    scores.sort(reverse=True)
    print(f"Updated board: {scores}")
assert len(scores) == starting_length + 1
assert scores[1] == 1310
```

## Challenge

### Challenge 1: Show the Top Three

```python
scores = [980, 1500, 720, 1310, 1100]
champion = max(scores)
scores.sort(reverse=True)
print(f"First: {scores[0]}")
print(f"Second: {scores[1]}")
print(f"Third: {scores[2]}")

assert scores[0] == champion
```

### Challenge 2: Merge a Second Score Board

```python
scores = [1500, 1200, 990]
second_scores = [1310, 700, 1050]
starting_length = len(scores)

for score in second_scores:
    scores.append(score)

scores.sort(reverse=True)
print(f"Merged board: {scores}")

assert len(scores) == starting_length + len(second_scores)
assert scores == [1500, 1310, 1200, 1050, 990, 700]
```
