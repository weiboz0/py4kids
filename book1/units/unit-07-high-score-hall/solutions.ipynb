## Exercise 2

```python
scores = [1200, 850, 990, 1500]
champion = max(scores)
rookie = min(scores)
midpoint = (champion + rookie) / 2
scores.sort(reverse=True)
print(f"Champion: {scores[0]}")
print(f"Rookie: {scores[-1]}")
print(f"Midpoint: {midpoint}")

if champion > rookie:
    print("The champion is greater than the rookie.")
else:
    print("The champion is not greater than the rookie.")

assert scores == [1500, 1200, 990, 850]
assert scores[0] == champion
assert scores[-1] == rookie
assert midpoint == 1175.0
```

## Exercise 4

```python
scores = [600, 800, 1000]
total = 0

for score in scores:
    total = total + score

average = total / len(scores)
best = max(scores)
reached_goal = average >= 700
print(f"Total: {total}")
print(f"Average: {average}")
print(f"Best: {best}")
print(f"Reached goal: {reached_goal}")

assert total == 2400
assert average == 800.0
assert best == 1000
assert reached_goal == True
```

## Exercise 5

```python
scores = [1000, 999, 499]
gold_count = 0
tiers = []

for score in scores:
    if score >= 1000:
        tier = "gold"
        gold_count = gold_count + 1
    elif score >= 500:
        tier = "silver"
    else:
        tier = "bronze"
    tiers.append(tier)
    print(f"{score}: {tier}")

print(f"Gold count: {gold_count}")

assert tiers == ["gold", "silver", "bronze"]
assert gold_count == 1
```

```python
raw_winners = ["  grace hopper  ", "ADA", " ada "]
clean_winners = []

for raw_name in raw_winners:
    clean_name = raw_name.strip().upper()
    if clean_name in clean_winners:
        print(clean_name + " is already on the winners list.")
    else:
        clean_winners.append(clean_name)

print("Clean winners: " + str(clean_winners))

assert clean_winners == ["GRACE HOPPER", "ADA"]
assert len(clean_winners) == 2
```

## Exercise 7

```python
def add_my_score(scores, new_score):
    scores.append(new_score)
    scores.sort(reverse=True)
    return scores


scores = [1500, 1200, 990, 850]
new_score = 1310
scores = add_my_score(scores, new_score)
print(f"Ranked board: {scores}")
print(f"Champion: {scores[0]}")

assert scores == [1500, 1310, 1200, 990, 850]
assert scores[0] == 1500
```

## Exercise 8

```python
def board_line(place, score):
    return "Place " + str(place) + ": " + str(score)


scores = [1500, 1200, 990, 850]
for place in range(len(scores)):
    print(board_line(place + 1, scores[place]))

assert board_line(1, scores[0]) == "Place 1: 1500"
assert board_line(4, scores[3]) == "Place 4: 850"
```

## Exercise 9

```python
scores = [1500, 1200, 990, 850]
new_score = 990
print(new_score in scores)

while new_score in scores:
    print("That score is already on the board. Try again.")
    new_score = 1310

scores.append(new_score)
scores.sort(reverse=True)
if new_score == scores[0]:
    print("The new score became the champion!")
else:
    print("The champion is still " + str(scores[0]) + ".")

assert new_score not in [1500, 1200, 990, 850]
assert scores == [1500, 1310, 1200, 990, 850]
assert scores[0] == 1500
```

## Exercise 12

```python
scores = [640, 1500, 980]
threshold = 10.0
doublings = 0
champion = max(scores)

while threshold <= champion:
    threshold = threshold * 2
    doublings = doublings + 1

print(f"Champion: {champion}")
print(f"First qualifying threshold: {threshold}")
print(f"Doublings: {doublings}")

assert champion == 1500
assert threshold == 2560.0
assert threshold / 2 <= champion
assert doublings == 8
```

## Exercise 13

```python
waiting_scores = [300, 450, 275, 600]
bonus_board = []
position = 0
total = 0

while total < 1000:
    score = waiting_scores[position]
    bonus_board.append(score)
    total = total + score
    position = position + 1

average = total / len(bonus_board)
print(f"Bonus board: {bonus_board}")
print(f"Last score: {bonus_board[-1]}")
print(f"Average: {average}")

assert bonus_board == [300, 450, 275]
assert bonus_board[-1] == 275
assert total == 1025
assert average == 341.6666666666667
```

### Challenge 2: Merge a Second Score Board

```python
scores = [1200, 990, 1500]
second_scores = [1310, 700, 1050]

for score in second_scores:
    scores.append(score)

scores.sort(reverse=True)
print(f"Merged board: {scores}")

assert len(scores) == 6
assert scores == [1500, 1310, 1200, 1050, 990, 700]
```
