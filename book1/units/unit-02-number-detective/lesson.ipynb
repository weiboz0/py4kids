# Number Detective

The computer picks a secret number. Can you find it in only a few guesses?

By the end of this unit, you will build the whole guessing game yourself.

## L1: Numbers + random

A whole number such as `7`, `0`, or `-12` is an **integer**. Python calls this the `int` type.

Integers can do arithmetic. Try adding, subtracting, and multiplying. `//` tells how many whole groups fit. `%` tells what is left over.

```python
marbles = 12
found = 5
lost = 3
print(marbles + found - lost)
print(4 * 3)
print(17 // 5)
print(17 % 5)
```

There is one catch. `input()` gives you text, even when someone types digits.

Use `int()` to turn digit text into an integer. Use `str()` to turn an integer back into text. This is **type conversion**.

```python
typed_guess = "27"
guess = int(typed_guess)
guess_text = str(guess)
print(f"The number is {guess_text}.")
```

Now try it with a number you type. Change the prompt to make it sound like a detective gadget.

```python
guess = int(input("Enter an integer: "))
print(f"You entered {guess}.")
```

Python can pick the secret for us.

`import random` brings in Python's random-number tools. Then `random.randint(1, 100)` picks an integer from 1 through 100. Both ends can be picked.

Run this teacher-led demo several times. Is the result always the same?

```python
import random
secret = random.randint(1, 100)
print(f"Teacher-only peek: the secret is {secret}.")
```

## L2: One-guess detective

The computer has a secret. Now it needs to judge your guess.

A comparison asks a yes-or-no question. Its answer is a **boolean**: `True` or `False`.

Try `==` for equal, `!=` for not equal, `<` for less than, and `>` for greater than.

```python
secret = 42
guess = 30
print(guess == secret)
print(guess != secret)
print(guess < secret)
print(guess > secret)
```

An `if` statement makes a choice.

Python checks the `if` comparison first. If that is `False`, it checks `elif`. If neither comparison is `True`, `else` handles the remaining case.

The indented lines belong to each choice. This first detective gives a verdict for one guess only.

```python
import random
secret = random.randint(1, 100)
guess = int(input("Your one guess: "))

if guess > secret:
    print("Too high!")
elif guess < secret:
    print("Too low!")
else:
    print("You found it!")
```

## L3: Full game

One guess is not much of a game. A `while` loop repeats while its comparison is `True`.

Our loop repeats while the guess is not the secret. Each trip through the loop gives a hint and asks again. When the guess matches, the loop stops.

Tally every guess **on paper** while you play. Do not add a guess-counting variable. The computer learns that trick in a later unit.

```python
import random
secret = random.randint(1, 100)
guess = int(input("Guess the secret number: "))

while guess != secret:
    if guess > secret:
        print("Too high!")
    else:
        print("Too low!")
    guess = int(input("Guess again: "))

print("Correct! Case closed.")
```

Detectives inspect clues. Programmers inspect error messages.

The next snippet is intentionally broken. Run it, type `20`, and read the traceback from the bottom. Find the line where Python got confused. What two types was it asked to compare?

```python
# Intentionally broken: read the traceback to find the clue.
secret = 50
guess = input("Guess a number: ")

if guess < secret:
    print("Too low!")
else:
    print("Too high or correct!")
```

The traceback points to the comparison. `input()` made text, but `secret` is an integer. Convert the typed guess before comparing it.

```python
# Fixed: the two compared values are integers.
secret = 50
guess = int(input("Guess a number: "))

if guess < secret:
    print("Too low!")
else:
    print("Too high or correct!")
```
