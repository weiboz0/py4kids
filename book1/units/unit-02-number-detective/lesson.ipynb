# Number Detective

The computer picks a secret number. Can you find it in only a few guesses? By the end of this unit you will build the whole guessing game yourself.

Each new idea comes as a short **ladder**: the simplest version, then one small twist, then a real use. The **Notice** line after each rung points at the single new thing.

## L1: Numbers

A whole number such as `7`, `0`, or `-12` is an **integer** — Python calls this the `int` type. No quotes are needed.

```python
age = 11
print(age)
```

**Notice:** `11` is an integer — a whole number stored without quotes.

```python
temperature = -4
score = 0
print(temperature)
print(score)
```

**Notice:** integers include negatives and zero, not just counting numbers.

```python
clue = 7
label = "7"
print(clue)
print(label)
```

**Notice:** `7` is an integer; `"7"` is text that only looks like one — Python treats them differently (you will see how in a moment).

Integers can do **arithmetic**: add `+`, subtract `-`, multiply `*`. `//` tells how many whole groups fit, and `%` tells what is left over.

```python
print(5 + 3)
```

**Notice:** Python works out the math and prints the result.

```python
print(10 - 4)
print(6 * 7)
```

**Notice:** `-` subtracts and `*` multiplies.

```python
marbles = 12
found = 5
print(marbles + found)
```

**Notice:** arithmetic works on saved numbers, not only typed-in ones.

```python
print(17 // 5)
```

**Notice:** `//` gives how many whole groups fit — `17 // 5` is `3`.

```python
print(17 % 5)
```

**Notice:** `%` gives what is left over after those groups — `17 % 5` is `2`.

One catch: `input()` always gives **text**, even when someone types digits. Use `int()` to turn digit-text into an integer and `str()` to turn an integer back into text. This is **type conversion**.

```python
typed = "27"
number = int(typed)
print(number)
```

**Notice:** `int("27")` becomes the number `27`, ready for arithmetic.

```python
count = 5
count_text = str(count)
print(f"As text, the count is {count_text}.")
```

**Notice:** `str()` turns a number into text — the reverse of `int()`.

```python
guess = int(input("Enter an integer: "))
print(f"You entered {guess}.")
```

**Notice:** wrap `input()` in `int()` whenever you need a number from the keyboard.

## L2: The Computer Picks and Judges

`import random` brings in Python's random-number tools. `random.randint(1, 100)` then picks an integer from 1 through 100 — both ends can be picked.

```python
import random
print(random.randint(1, 6))
```

**Notice:** `import random` loads the tools; `randint(1, 6)` rolls a number 1 through 6.

```python
import random
print(random.randint(1, 100))
```

**Notice:** change the two numbers to change the range.

```python
import random
secret = random.randint(1, 10)
print(f"(teacher peek) the secret is {secret}")
```

**Notice:** save the pick in a variable to use later — this is how the game hides its secret.

Now the computer must judge a guess. A **comparison** asks a yes-or-no question; its answer is a **boolean**, `True` or `False`. Use `==` equal, `!=` not equal, `<` less than, `>` greater than.

```python
print(7 == 7)
```

**Notice:** a comparison reports `True` or `False` — a boolean.

```python
print(3 < 10)
print(10 < 3)
```

**Notice:** `<` is True only when the left side is smaller.

```python
print(5 == 5)
print(5 != 4)
print(5 > 2)
print(5 < 2)
```

**Notice:** the four tests are `==`, `!=`, `>`, `<`.

```python
secret = 42
guess = 30
print(guess == secret)
print(guess < secret)
```

**Notice:** you can compare saved numbers — exactly what the game needs to judge a guess.

```python
secret = 42
guess = 30
is_too_low = guess < secret
print(is_too_low)
```

**Notice:** the `True`/`False` answer can be saved in a variable and reused later.

## L3: Make a Choice

An **`if` statement** runs its indented lines only when a comparison is `True`.

```python
guess = 30
secret = 42
if guess < secret:
    print("Too low!")
```

**Notice:** the indented `print` runs only because `30 < 42` is True.

```python
guess = 50
secret = 42
if guess < secret:
    print("Too low!")
print("(done checking)")
```

**Notice:** when the `if` is False, Python skips the indented line and carries on.

```python
guess = 30
secret = 42
if guess == secret:
    print("Found it!")
else:
    print("Keep trying.")
```

**Notice:** `else` runs when the `if` is False — exactly one of the two always runs.

**`elif`** adds more choices between `if` and `else`. Python checks them top to bottom and takes the first comparison that is `True`.

```python
guess = 30
secret = 42
if guess > secret:
    print("Too high!")
elif guess < secret:
    print("Too low!")
else:
    print("Just right!")
```

**Notice:** `elif` is checked only when the tests above it were False.

```python
temperature = 15
if temperature > 25:
    print("Hot")
elif temperature > 10:
    print("Warm")
else:
    print("Cold")
```

**Notice:** Python takes the FIRST branch whose test is True, then stops checking the rest.

```python
import random
secret = random.randint(1, 100)
guess = int(input("Your one guess: "))

if guess > secret:
    print("Too high!")
elif guess < secret:
    print("Too low!")
else:
    print("You found it!")
```

**Notice:** put it together and the detective gives a verdict for one real guess.

## L4: The Full Game

One guess is not much of a game. A **`while` loop** repeats its indented lines while a comparison stays `True`. Tally guesses **on paper** — do not add a guess-counting variable (the computer learns that trick in a later unit).

```python
answer = input("Type the word stop to finish: ")
while answer != "stop":
    print("Not yet...")
    answer = input("Type the word stop to finish: ")
print("Stopped!")
```

**Notice:** the loop repeats while the test is True; typing `stop` makes it False and the loop ends.

```python
secret = 7
guess = int(input("Guess the number: "))
while guess != secret:
    print("Nope, guess again.")
    guess = int(input("Guess again: "))
print("Correct!")
```

**Notice:** each trip asks again; when the guess matches the secret, the loop stops.

```python
secret = 7
guess = int(input("Guess the number: "))
while guess != secret:
    if guess > secret:
        print("Too high!")
    else:
        print("Too low!")
    guess = int(input("Guess again: "))
print("Got it!")
```

**Notice:** put an `if` inside the loop to give a smarter hint on every guess.

```python
import random
secret = random.randint(1, 100)
guess = int(input("Guess the secret number: "))

while guess != secret:
    if guess > secret:
        print("Too high!")
    else:
        print("Too low!")
    guess = int(input("Guess again: "))

print("Correct! Case closed.")
```

**Notice:** the full detective game — `random` picks, `while` repeats, `if/else` hints.
