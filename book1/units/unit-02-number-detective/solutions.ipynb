## Exercise 1

### Six-sided dice roller

## Exercise 2

### Higher-or-lower verdict

```python
secret = 37
# This is a sample value standing in for the player's one guess.
guess = int("52")

if guess > secret:
    verdict = "Too high!"
elif guess < secret:
    verdict = "Too low!"
else:
    verdict = "Correct!"

print(verdict)
assert verdict == "Too high!"
```

## Exercise 3

### Number Detective from 1 through 1,000

```python
import random
secret = random.randint(1, 1000)
# This sample first guess replaces a typed guess.
guess = 500
print("Guess a secret integer from 1 through 1,000.")

while guess != secret:
    if guess > secret:
        print("Too high!")
    else:
        print("Too low!")
    # This sample follow-up guess lets the headless game finish.
    guess = secret

print("Correct! The 1-through-1,000 mystery is solved.")
# The larger range felt wide; 500 was a useful first guess near the middle.
assert guess == secret
```

## Exercise 4

### Two-player take-turns game

```python
# These are sample values standing in for Player A's secret and Player B's guess.
player_a_secret = int("63")
player_b_guess = int("50")

while player_b_guess != player_a_secret:
    if player_b_guess > player_a_secret:
        print("Too high, Player B!")
    else:
        print("Too low, Player B!")
    # This sample follow-up guess ends the demonstration.
    player_b_guess = player_a_secret

print("Correct! Now swap roles.")
assert player_b_guess == player_a_secret
```

## Exercise 5

### Repair the type problem

```python
secret = 25
# This sample text stands in for the typed reply.
typed_guess = "12"
# Text and an integer cannot be compared with >, so convert the text first.
guess = int(typed_guess)

if guess > secret:
    verdict = "Too high!"
else:
    verdict = "Too low or correct!"

print(verdict)
assert verdict == "Too low or correct!"
```

## Exercise 6

### Buried-treasure search

```python
import random
secret_depth = random.randint(1, 50)
# This sample first guess replaces a typed depth.
guessed_depth = 25

while guessed_depth != secret_depth:
    if guessed_depth > secret_depth:
        print("Too deep!")
    else:
        print("Too shallow!")
    # This sample follow-up guess uncovers the treasure.
    guessed_depth = secret_depth

print("Treasure found! Raise the pirate flag!")
assert guessed_depth == secret_depth
```

## Exercise 7

### Range-width report

```python
low = 12
high = 47
span = high - low
midpoint = (low + high) // 2

if span % 2 == 0:
    even_or_odd = "even"
else:
    even_or_odd = "odd"

if span < 10:
    span_size = "narrow"
elif span < 30:
    span_size = "medium"
else:
    span_size = "wide"

signal = low + high*2
span_message = "The span is " + str(span) + "."
print(span_message)
print(f"The midpoint is {midpoint}. The span is {even_or_odd} and {span_size}.")
print(f"The signal is {signal}.")

assert span == 35
assert midpoint == 29
assert even_or_odd == "odd"
assert span_size == "wide"
assert signal == 106
assert span_message == "The span is 35."
```

## Challenge

### Challenge 1

### Hot and cold hints

```python
import random
random.seed(4)
secret = random.randint(1, 100)
# Sample values stand in for the player's typed guesses (no typing needed here).
guess = 50  # first guess — wrong, so the game prints a hot/cold hint
while guess != secret:
    distance = secret - guess
    if distance < 0:
        distance = guess - secret
    if distance < 6:
        print("HOT! You are within 5.")
    else:
        print("cold... more than 5 away.")
    guess = secret  # second sample guess — correct, so the loop ends
print(f"Got it! The secret was {secret}.")
assert guess == secret
```

### Challenge 2

### The computer guesses

```python
# The computer hunts the number by halving. Your replies are numbers:
# 1 = too high, 2 = too low, 3 = correct. The computer NEVER sees your number —
# it only sees your replies. Scripted sample replies stand in for the player's typing:
# (secretly, the player is thinking of 68)
low = 1
high = 100
guess = (low + high) // 2   # 50
reply = 2                    # "too low"
while reply != 3:
    if reply == 1:
        high = guess - 1
    if reply == 2:
        low = guess + 1
    guess = (low + high) // 2
    # scripted replies: 75 -> 1, 62 -> 2, 68 -> 3
    if guess == 75:
        reply = 1
    if guess == 62:
        reply = 2
    if guess == 68:
        reply = 3
print(f"Found it: {guess}!")
assert guess == 68
```
