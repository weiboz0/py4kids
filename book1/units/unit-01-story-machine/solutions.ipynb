## Exercise 6

### Story Machine remix

```python
# These sample values are the ingredients for a space-school remix.
character = "Dr. Waffles"
place = "the cafeteria on Mars"
object_name = "a musical lunchbox"
action = "dance with the potatoes"
title = "The Strange Day of " + character
story_line_one = f"At {place}, {character} discovered {object_name}."
story_line_two = f"Suddenly, {character} began to {action}!"
print(title)
print(story_line_one)
print(story_line_two)
assert title == "The Strange Day of Dr. Waffles"
```

### Challenge 1

A multi-paragraph story reuses the same hero and place.

```python
hero = "Zee the Zebra"
place = "the cloud castle"
paragraph_one = f"{hero} climbed the silver stairs to {place}."
paragraph_two = f"Inside {place}, {hero} found a choir of tiny frogs."
print(paragraph_one)
print("")
print(paragraph_two)
assert paragraph_one == "Zee the Zebra climbed the silver stairs to the cloud castle."
```
