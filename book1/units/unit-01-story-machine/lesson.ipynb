# Story Machine

Your teacher runs a ridiculous Mad-Libs story. A heroic pickle may zoom through a library on a singing skateboard. Now you will make the story yours.

## Lesson One: Wake up the Story Machine

A program is a set of instructions for the computer. Running a program tells the computer to follow those instructions.

Start with `print()`. It puts a message where you can see it. Run the cell below.

```python
print("Today, a sleepy dragon borrowed my backpack.")
print("It sneezed glitter all over the moon!")
```

The text inside matching quotes is a **string literal**. The quotes show Python where the text begins and ends. Change the words inside the quotes, then run the program again.

```python
print("A nervous llama opened the secret door.")
print("Behind it waited a mountain of jellybeans.")
```

A **comment** is a note for people who read the program. Python does not print it. A comment begins with `#`.

```python
# This note tells us what the next instruction does.
print("The Story Machine is warming up!")
```

Programs sometimes break. That is normal. Run this broken line only when your teacher asks. Read the error message from the bottom.

```python
print("The penguin wore a purple wig.)
```

The last line of the traceback names a `SyntaxError`. It says the string literal never ended. Python found the opening quote but could not find its matching closing quote. Look near the marked line, add the missing quote, and try again.

```python
print("The penguin wore a purple wig.")
```

## Lesson Two: Make the Story Yours

The Story Machine needs words it can save and reuse. A **variable** is a name that saves something for later.

```python
hero = "Captain Pickle"
place = "the moon library"
print(hero)
print(place)
```

A clear variable name helps you remember what it saves. Use letters and underscores. Do not put spaces in a name. Use the exact same spelling every time.

Names such as `hero_name` and `silly_object` give useful clues. A name such as `stuff` hides the clue.

```python
hero_name = "Professor Pancake"
silly_object = "a squeaky crown"
print(hero_name)
print(silly_object)
```

`input()` pauses the program and shows a prompt. You type a reply. The variable saves the words you typed. This cell waits for you, so run it with your teacher.

```python
hero = input("Name a hero: ")
print(hero)
```

**String concatenation** joins string literals and saved strings with `+`. Each piece becomes one longer message. Add spaces inside the string literals where you want spaces to appear.

```python
hero = "Captain Pickle"
print("Meet " + hero + "!")
```

An **f-string** is another way to mix saved strings with story text. Put `f` before the opening quote. Put each variable name inside curly braces.

```python
hero = "Captain Pickle"
place = "the moon library"
print(f"{hero} raced to {place}.")
```

Now build the full Mad-Libs Story Machine. It asks for your ideas, saves them in clearly named variables, and reuses them in the story. Run it and make a wonderfully ridiculous tale.

```python
# Collect the story ingredients.
hero = input("Name a hero: ")
place = input("Name a strange place: ")
object_name = input("Name a silly object: ")
action = input("Name an action: ")

# Print the title and the story.
print("The Amazing Adventure of " + hero)
print(f"One day, {hero} arrived at {place}.")
print(f"A mysterious voice offered {hero} {object_name}.")
print(f"Without warning, {hero} began to {action}!")
print(f"Everyone at {place} cheered for {hero}.")
```

Remix the string literals, prompts, and variable names. Keep matching quotes and exact spellings. If the program stops, read the final traceback line and use its clue. Your Story Machine is now yours.
