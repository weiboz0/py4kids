## Lesson One: Wake up the Story Machine

A program is a set of instructions for the computer. Running a cell tells the computer to follow them.

`print()` puts a message where you can see it. Here it is three ways, each a small step up.

```python
print("The Story Machine is awake!")
```

**Notice:** `print()` shows whatever sits inside the quotes.

```python
print("First, the dragon yawned.")
print("Then it sneezed glitter.")
```

**Notice:** each `print()` starts its own new line, top to bottom.

```python
print("Chapter One")
print()
print("A sleepy dragon borrowed my backpack.")
```

**Notice:** an empty `print()` leaves a blank line — handy for spacing a story out.

The text inside matching quotes is a **string literal**. The quotes show Python where the text begins and ends.

```python
print("A nervous llama opened the secret door.")
```

**Notice:** everything between the quotes is the text that gets shown.

```python
print('Behind it waited a mountain of jellybeans.')
```

**Notice:** you may use `'single'` or `"double"` quotes — just start and end with the same kind.

```python
print("Careful — it's a trap!")
```

**Notice:** punctuation like `—`, `!`, and `'` is part of the text too; Python shows it exactly.

A **comment** is a note for people who read the program. Python does not print it. A comment begins with `#`.

```python
# This note explains the next line.
print("The Story Machine is warming up!")
```

**Notice:** Python skips the `#` line and runs only the `print()`.

```python
print("Loading the story...")  # a friendly reminder for whoever reads the code
```

**Notice:** a comment can also sit at the *end* of a line, after the code.

Programs sometimes break. That is normal. Run this broken line only when your teacher asks. Read the error message from the bottom.

```python
print("The penguin wore a purple wig.)
```

The last line of the traceback names a `SyntaxError`. It says the string literal never ended: Python found the opening quote but not its matching closing quote. Add the missing quote and try again.

```python
print("The penguin wore a purple wig.")
```

## Lesson Two: Save Words to Reuse

The Story Machine needs words it can save and reuse. A **variable** is a name that saves something for later.

```python
hero = "Captain Pickle"
print(hero)
```

**Notice:** the name `hero` now stands for the saved words, so `print(hero)` shows them.

```python
hero = "Captain Pickle"
place = "the moon library"
print(hero)
print(place)
```

**Notice:** you can save as many things as you like, each in its own name.

```python
mood = "sleepy"
print(mood)
mood = "thrilled"
print(mood)
```

**Notice:** saving a new value *replaces* the old one — the most recent value wins.

A clear variable name helps you remember what it saves. Use letters and underscores, no spaces, and the exact same spelling every time.

```python
hero_name = "Professor Pancake"
print(hero_name)
```

**Notice:** `hero_name` tells you what it holds — far better than a vague name like `stuff`.

```python
silly_object = "a squeaky crown"
print(silly_object)
```

**Notice:** join words in a name with `_` — spaces are not allowed inside a name.

```python
favorite_snack = "pickle pizza"
print(favorite_snack)
print(favorite_snack)
```

**Notice:** spell the name the exact same way each time, or Python will not find it.

`input()` pauses the program, shows a prompt, and saves what you type into a variable. These cells wait for you, so run them with your teacher.

```python
name = input("Your name? ")
print(name)
```

**Notice:** `input()` waits, then saves your reply in the variable.

```python
color = input("Name a color: ")
print("You chose this color:")
print(color)
```

**Notice:** the saved reply is a string like any other — reuse it with `print()`.

```python
hero = input("Name a hero: ")
sidekick = input("Name a sidekick: ")
print(hero)
print(sidekick)
```

**Notice:** ask as many questions as your story needs; each reply gets its own variable.

## Lesson Three: Combine Words into a Story

**String concatenation** joins string literals and saved strings with `+`, making one longer message.

```python
hero = "Captain Pickle"
print("Meet " + hero + "!")
```

**Notice:** `+` glues the pieces into a single message.

```python
hero = "Captain Pickle"
place = "the moon"
print(hero + " lives on " + place)
```

**Notice:** you can join several saved strings and literals in a row.

```python
word = "jelly"
other = "bean"
print("stuck: " + word + other)
print("spaced: " + word + " " + other)
```

**Notice:** `+` adds no spaces on its own — put a `" "` between the pieces where you want a space.

An **f-string** is another way to mix saved strings into story text. Put `f` before the opening quote and each variable name inside curly braces.

```python
hero = "Captain Pickle"
print(f"{hero} saved the day.")
```

**Notice:** the `f` lets `{hero}` become the saved words inside the sentence.

```python
hero = "Captain Pickle"
place = "the moon library"
print(f"{hero} raced to {place}.")
```

**Notice:** drop in as many `{names}` as you like.

```python
hero = "Captain Pickle"
print(f"{hero} tripped, but {hero} jumped right back up!")
```

**Notice:** the same `{name}` can appear more than once.

Now build the full Mad-Libs Story Machine. It asks for your ideas, saves them in clearly named variables, and reuses them in the story. Run it and make a wonderfully ridiculous tale.

```python
# Collect the story ingredients.
hero = input("Name a hero: ")
place = input("Name a strange place: ")
object_name = input("Name a silly object: ")
action = input("Name an action: ")

# Print the title and the story.
print("The Amazing Adventure of " + hero)
print(f"One day, {hero} arrived at {place}.")
print(f"A mysterious voice offered {hero} {object_name}.")
print(f"Without warning, {hero} began to {action}!")
print(f"Everyone at {place} cheered for {hero}.")
```

Remix the string literals, prompts, and variable names. Keep matching quotes and exact spellings. If the program stops, read the final traceback line and use its clue. Your Story Machine is now yours.
