## Exercise 4

### Sudden-death remix

````python
score = 0
asked = 0
shown = False

while asked < 3:
    if asked == 0:
        print("How many legs does a cat have?")
        answer = 4  # Sample stand-in answer for question 1.
        correct = answer == 4
    elif asked == 1:
        print("What is 5 times 3?")
        answer = 15  # Sample stand-in answer for question 2.
        correct = answer == 15
    elif asked == 2:
        print("How many hours are in a day?")
        answer = 24  # Sample stand-in answer for question 3.
        correct = answer == 24

    asked = asked + 1
    if correct:
        score = score + 1
        if asked == 3:
            print("Perfect sudden-death round!")
            shown = True
    else:
        print("Wrong answer — the lightning round is over!")
        break

print(f"Score: {score}")
assert asked == 3 and score == 3
assert shown
````

## Exercise 7

### Escape the penalty loop

`not on_streak` flips the streak flag; combine it with the wrong answer using `and`.

````python
score = 3
on_streak = False
correct = False

while True:
    penalty = not on_streak and not correct
    if penalty:
        if score > 0:
            score = score - 1
        print(f"Penalty! Score: {score}")
        break
    else:
        print("Safe from the penalty!")
        break

assert penalty is True and score == 2
# With an on-streak player, `and` keeps the penalty false; `or` would make it true.
on_streak = True
penalty = not on_streak and not correct
assert penalty is False
````

**The real program** (reads both booleans):

```python
score = 3
on_streak = input("On a streak? (yes/no) ") == "yes"
correct = input("Answer correct? (yes/no) ") == "yes"

while True:
    penalty = not on_streak and not correct
    if penalty:
        if score > 0:
            score = score - 1
        print(f"Penalty! Score: {score}")
        break
    else:
        print("Safe from the penalty!")
        break
```

## Exercise 9

### Two strikes

````python
c1 = False
c2 = False
c3 = True
q = 0
score = 0
strikes = 0

while q < 3:
    if q == 0:
        correct = c1
    elif q == 1:
        correct = c2
    else:
        correct = c3

    q = q + 1
    if not correct:
        strikes = strikes + 1
        if strikes == 2:
            print("Two strikes — game over!")
            break
    else:
        score = score + 1

print(f"Score: {score}; strikes: {strikes}")
# The second strike stops the loop before the third, correct result is used.
assert q == 2 and strikes == 2 and score == 0
````

**The real program** (reads each answer as the loop runs):

## Exercise 10

### Last-chance qualifier

````python
score = 5  # Fixed stand-in for the typed starting score.
attempts = 0

while attempts < 2:
    print("What is 4 + 4?")
    if attempts == 0:
        answer = 7  # Fixed first attempt: wrong.
    else:
        answer = 8  # Fixed second attempt: correct.

    attempts = attempts + 1
    correct = answer == 8
    eligible = correct and score >= 4

    if eligible:
        if attempts == 1:
            score = score + 2
        else:
            score = score + 1
        print("You qualify!")
        break
    else:
        if not correct:
            score = score - 1
        if attempts == 2:
            print("Qualifier over.")
            break

print(f"Final score: {score}")
# `or` would qualify the wrong first answer; correct `and` reaches attempt 2.
assert eligible is True and attempts == 2 and score == 5
````

**The real program** (reads the starting score and each answer):


## Exercise 19
