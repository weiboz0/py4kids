```python
score = 4
category = "science"
answer_correct = False

if answer_correct:
    score = score + 1
    if category == "science":
        score = score + 2

print(f"Score after the category question: {score}")
# A science answer earns no points unless the outer condition is true.
assert score == 4
```

## Exercise 4

### Sudden-death remix

```python
score = 0
questions_asked = 0
third_answer_message_shown = False

while questions_asked < 3:
    if questions_asked == 0:
        print("How many legs does a cat have?")
        answer = 4  # Sample stand-in answer for question 1.
        answer_correct = answer == 4
    elif questions_asked == 1:
        print("What is 5 times 3?")
        answer = 15  # Sample stand-in answer for question 2.
        answer_correct = answer == 15
    elif questions_asked == 2:
        print("How many hours are in a day?")
        answer = 24  # Sample stand-in answer for question 3.
        answer_correct = answer == 24

    questions_asked = questions_asked + 1
    if answer_correct:
        score = score + 1
        if questions_asked == 3:
            print("Perfect sudden-death round!")
            third_answer_message_shown = True
    else:
        print("Wrong answer — the lightning round is over!")
        break

print(f"Score: {score}")
assert questions_asked == 3 and score == 3
assert third_answer_message_shown
```

## Exercise 6

### Predict the score

The final score is 5, and the final value of `questions_asked` is 2.

```python
score = 0
streak_alive = True
first_correct = True
second_correct = True

if first_correct:
    score = score + 1
else:
    streak_alive = False

if second_correct:
    score = score + 1
else:
    streak_alive = False

if second_correct and streak_alive:
    score = score + 2

questions_asked = 0
while questions_asked < 3:
    if questions_asked == 0:
        correct = True
    elif questions_asked == 1:
        correct = False
    elif questions_asked == 2:
        correct = True

    questions_asked = questions_asked + 1
    if correct:
        score = score + 1
    else:
        break

print(f"Score: {score}; questions asked: {questions_asked}")
assert score == 5 and questions_asked == 2
```

## Exercise 7

### Escape the penalty loop

`not on_streak` flips the streak flag; combine it with the wrong answer using `and`.

```python
score = 3
on_streak = False
answer_correct = False

while True:
    penalty = not on_streak and not answer_correct
    if penalty:
        if score > 0:
            score = score - 1
        print(f"Penalty! Score: {score}")
        break
    else:
        print("Safe from the penalty!")
        break

assert penalty is True and score == 2
# Using `or` would wrongly penalize this on-streak player.
assert (not True and not answer_correct) is False
```

## Exercise 8

### Accept either spelling

```python
score = 0
attempts = 0

while attempts < 3:
    print("Type true or True:")
    answer = "true"  # Fixed stand-in for the typed answer.
    attempts = attempts + 1
    accepted = answer == "true" or answer == "True"

    if accepted:
        score = score + 1
        if attempts == 1:
            score = score + 1
        print(f"Accepted! Score: {score}")
        break
    elif attempts == 3:
        print("No accepted answer — round over!")
        break

# The lowercase spelling distinguishes `or` from the incorrect `and` rule.
assert accepted is True and attempts == 1 and score == 2
```

## Exercise 9

### Two strikes

```python
answer_1_correct = False
answer_2_correct = False
answer_3_correct = True
question_number = 0
score = 0
strikes = 0

while question_number < 3:
    if question_number == 0:
        answer_correct = answer_1_correct
    elif question_number == 1:
        answer_correct = answer_2_correct
    else:
        answer_correct = answer_3_correct

    question_number = question_number + 1
    if not answer_correct:
        strikes = strikes + 1
        if strikes == 2:
            print("Two strikes — game over!")
            break
    else:
        score = score + 1

print(f"Score: {score}; strikes: {strikes}")
# The second strike stops the loop before the third, correct result is used.
assert question_number == 2 and strikes == 2 and score == 0
```

## Exercise 10

### Last-chance qualifier

```python
score = 5  # Fixed stand-in for the typed starting score.
attempts = 0

while attempts < 2:
    print("What is 4 + 4?")
    if attempts == 0:
        answer = 7  # Fixed first attempt: wrong.
    else:
        answer = 8  # Fixed second attempt: correct.

    attempts = attempts + 1
    answer_correct = answer == 8
    eligible = answer_correct and score >= 4

    if eligible:
        if attempts == 1:
            score = score + 2
        else:
            score = score + 1
        print("You qualify!")
        break
    else:
        if not answer_correct:
            score = score - 1
        if attempts == 2:
            print("Qualifier over.")
            break

print(f"Final score: {score}")
# `or` would qualify the wrong first answer; correct `and` reaches attempt 2.
assert eligible is True and attempts == 2 and score == 5
```

```python
seconds_left = 5
last_shown = 0
while seconds_left > 0:
    print(seconds_left)
    last_shown = seconds_left
    seconds_left = seconds_left - 1
print("Time!")
# Self-check: the loop counted down and the final number shown was 1.
assert last_shown == 1
assert seconds_left == 0
```
