## Exercise 3

### Add a category bonus

```python
score = 4
category = "science"
answer_correct = True

if answer_correct:
    score = score + 1
    if category == "science":
        score = score + 2

print(f"Score after the category question: {score}")
assert score == 7
```

## Exercise 4

### Sudden-death remix

```python
questions_asked = 0
score = 0

while questions_asked < 3:
    if questions_asked == 0:
        print("How many legs does a cat have?")
        answer = 4  # Sample stand-in answer for question 1.
        correct = answer == 4
    elif questions_asked == 1:
        print("What is 5 times 3?")
        answer = 15  # Sample stand-in answer for question 2.
        correct = answer == 15
    elif questions_asked == 2:
        print("How many hours are in a day?")
        answer = 24  # Sample stand-in answer for question 3.
        correct = answer == 24

    questions_asked = questions_asked + 1
    if correct:
        score = score + 1
        print("Still alive!")
    else:
        print("Wrong answer — the lightning round is over!")
        break

print(f"Score: {score}")
assert questions_asked == 3 and score == 3
```

## Exercise 5

### Error-message debugging

The broken program tried to add an integer (`score`) and a string (`bonus`). Converting the stand-in response to an integer repairs the calculation.

The repaired line is `bonus = int(input("Bonus points: "))` — wrapping the input in `int(...)` turns the typed text into a number so the math works.

```python
score = 2
bonus = int("3")  # Sample stand-in value for the typed bonus points.
score = score + bonus
print(f"New score: {score}")
assert score == 5
```

## Exercise 6

### Predict the score

The final score is 5, and the final value of `questions_asked` is 2.

```python
score = 0
streak_alive = True
first_correct = True
second_correct = True

if first_correct:
    score = score + 1
else:
    streak_alive = False

if second_correct:
    score = score + 1
else:
    streak_alive = False

if second_correct and streak_alive:
    score = score + 2

questions_asked = 0
while questions_asked < 3:
    if questions_asked == 0:
        correct = True
    elif questions_asked == 1:
        correct = False
    elif questions_asked == 2:
        correct = True

    questions_asked = questions_asked + 1
    if correct:
        score = score + 1
    else:
        break

print(f"Score: {score}; questions asked: {questions_asked}")
assert score == 5 and questions_asked == 2
```

## Challenge

### Challenge 1 — Double or Nothing

```python
score = 4
risk_choice = "yes"  # Sample stand-in value for the player's risk choice.

if risk_choice == "yes":
    print("What is 6 + 6?")
    answer = 12  # Sample stand-in answer for the hard-coded question.
    if answer == 12:
        score = score * 2
    else:
        score = 0

print(f"Final score: {score}")
assert score == 8
```

### Challenge 2 — Lightning-Round Countdown

```python
seconds_left = 5
last_shown = 0
while seconds_left > 0:
    print(seconds_left)
    last_shown = seconds_left
    seconds_left = seconds_left - 1
print("Time!")
# Self-check: the loop counted down and the final number shown was 1.
assert last_shown == 1
assert seconds_left == 0
```
