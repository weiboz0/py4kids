## Exercise 6

### Predict the score

Cover the last `print` line. Trace the code on paper and predict the final score and the final value of `questions_asked`. Pay special attention to the streak bonus and the point where `break` runs. Then uncover the line and run the cell to check.

*This is a trace-and-predict exercise: it keeps fixed values on purpose so you can predict the output, so it has no `input()` version.*

````python
score = 0
streak_alive = True
first_correct = True
second_correct = True

if first_correct:
    score = score + 1
else:
    streak_alive = False

if second_correct:
    score = score + 1
else:
    streak_alive = False

if second_correct and streak_alive:
    score = score + 2

questions_asked = 0
while questions_asked < 3:
    if questions_asked == 0:
        correct = True
    elif questions_asked == 1:
        correct = False
    elif questions_asked == 2:
        correct = True

    questions_asked = questions_asked + 1
    if correct:
        score = score + 1
    else:
        break

print(f"Score: {score}; questions asked: {questions_asked}")
````
