# Quiz Show

The lights flash. The audience cheers. You are about to host a quiz show with scores, streak bonuses, secret follow-ups, and a SUDDEN DEATH round. Can your code keep the whole game fair?

## L1: Score and Streaks

A quiz needs a score that can grow. Start it at zero. Each time an answer is right, use `score = score + 1`. Python finds the old value on the right, adds one, and stores the new value back in `score`.

A variable updated this way is called an **accumulator** because it collects a total as the game continues.

```python
score = 0
score = score + 1
score = score + 1
print(f"Practice score: {score}")
```

Unit 03 introduced counters while the turtle drew its sides and shapes. We are not learning counting again here. We are bringing that idea into a game: `questions_asked` records how many questions the host has asked.

A score and a question counter both update an old value, but they answer different questions: points earned versus questions asked.

Here is the three-question opening round. It is straight-line code: question 1, then question 2, then question 3. Run it and answer using the spelling shown in each prompt.

Question 2 accepts either `Pacific` or `pacific`. The word `or` makes the condition true when either comparison is true. At the end, the bonus needs two things at once: the last answer is right **and** the streak is still alive. The word `and` makes that rule exact.

There is a third logical operator: `not`, which flips True and False. `not on_streak` is True exactly when the player is **not** on a streak — handy for a rule that punishes coasting.

```python
score = 0
questions_asked = 0
streak_alive = True

answer_1 = int(input("What is 5 + 3? "))
questions_asked = questions_asked + 1
if answer_1 == 8:
    score = score + 1
else:
    streak_alive = False

answer_2 = input("Which ocean is west of California: Pacific or Atlantic? ")
questions_asked = questions_asked + 1
if answer_2 == "Pacific" or answer_2 == "pacific":
    score = score + 1
else:
    streak_alive = False

answer_3 = input("What color do blue and yellow paint make? Type green: ")
questions_asked = questions_asked + 1
third_correct = answer_3 == "green"
if third_correct:
    score = score + 1
else:
    streak_alive = False

if third_correct and streak_alive:
    print("Three-answer streak! Bonus: 2 points.")
    score = score + 2

print(f"Questions asked: {questions_asked}")
print(f"Final score: {score}")
```

Logical operators join boolean conditions.

- `A and B` is true only when both A and B are true.
- `A or B` is true when at least one of them is true.

Use the operator that matches the rule your host says aloud.

```python
answer_correct = True
streak_alive = True
print(answer_correct and streak_alive)

accepted_first_spelling = False
accepted_second_spelling = True
print(accepted_first_spelling or accepted_second_spelling)
```

## L2: Secret Follow-ups and SUDDEN DEATH

The final round raises the drama. First, a contestant must unlock a follow-up question. Then one wrong answer can end a whole round.

A condition can live inside another condition. This is **conditional nesting**. The inner `if` is reached only when the outer `if` is true. Here, part two stays hidden unless part one is correct.

```python
final_round_score = 0
part_one = int(input("How many months are in one year? "))

if part_one == 12:
    final_round_score = final_round_score + 1
    part_two = input("Follow-up unlocked! Which planet has rings? Type Saturn: ")
    if part_two == "Saturn" or part_two == "saturn":
        final_round_score = final_round_score + 2
        print("Follow-up bonus won!")
    else:
        print("Part one scored, but the follow-up missed.")
else:
    print("The follow-up stays locked.")

print(f"Final-round score: {final_round_score}")
```

Now for **SUDDEN DEATH**. The counter controls which of three hard-coded questions is asked. A `while` loop keeps the round moving. After a correct answer, the counter grows and the next trip begins. After a wrong answer, `break` ends the loop immediately.

Notice the dispatch chain: `if` handles counter value 0, and the two `elif` branches handle 1 and 2. No collection of questions is hiding behind the curtain; all three are written directly in the round.

```python
questions_asked = 0
sudden_death_score = 0

while questions_asked < 3:
    if questions_asked == 0:
        answer = int(input("SUDDEN DEATH 1 — How many sides does a triangle have? "))
        correct = answer == 3
    elif questions_asked == 1:
        answer = int(input("SUDDEN DEATH 2 — What is 6 times 7? "))
        correct = answer == 42
    elif questions_asked == 2:
        answer = int(input("SUDDEN DEATH 3 — How many days are in a week? "))
        correct = answer == 7

    questions_asked = questions_asked + 1
    if correct:
        sudden_death_score = sudden_death_score + 1
        print("Still alive!")
    else:
        print("Wrong answer — the round ends now!")
        break

print(f"Questions faced: {questions_asked}")
print(f"Sudden-death score: {sudden_death_score}")
```

`break` belongs inside a loop. It does not merely skip one line; it jumps to the first line after the loop. That is exactly what gives sudden death its drama.
