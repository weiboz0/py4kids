# Quiz Show

The lights flash. The audience cheers. You are about to host a quiz show with scores, streak bonuses, secret follow-ups, and a SUDDEN DEATH round. Can your code keep the whole game fair?

Each idea comes as a short **ladder** — the simplest version, then one small step up, then the real game use — with a **Notice** naming the one new thing each time.

## L1: Score and Streaks

A quiz needs a score that can grow. An **accumulator** is a variable that collects a total: `score = score + 1` takes the old value, adds one, and stores it back.

```python
score = 0
score = score + 1
print(f"Score: {score}")
```

**Notice:** `score = score + 1` turns the old `0` into `1` and saves it back in `score`.

```python
score = 0
score = score + 1
score = score + 1
print(f"Score: {score}")
```

**Notice:** each update builds on the last, so the total climbs to `2`.

```python
score = 0
round_number = 0
while round_number < 3:
    score = score + 1
    round_number = round_number + 1
print(f"Score after the round: {score}")
```

**Notice:** inside a `while` loop the accumulator grows once per trip — three trips, three points.

```python
score = 0
questions_asked = 0
round_number = 0
while round_number < 3:
    questions_asked = questions_asked + 1
    score = score + 1
    round_number = round_number + 1
print(f"Asked {questions_asked}, scored {score}")
```

**Notice:** two accumulators can run side by side — points earned versus questions asked — each answering its own question.

A quiz also needs rules that combine yes/no facts. **Logical operators** join booleans: `and`, `or`, `not`.

```python
print(True and True)
print(True and False)
```

**Notice:** `and` is True only when BOTH sides are True.

```python
print(True or False)
print(False or False)
```

**Notice:** `or` is True when AT LEAST ONE side is True.

```python
print(not True)
print(not False)
```

**Notice:** `not` flips True and False.

```python
answer_correct = True
streak_alive = True
print(answer_correct and streak_alive)
```

**Notice:** combine them for a real rule — the streak bonus needs the answer correct **and** the streak alive.

Here is the opening round put together. Run it and answer using the spelling shown in each prompt; `or` accepts either spelling, and the bonus rule uses `and`.

```python
score = 0
questions_asked = 0
streak_alive = True

answer_1 = int(input("What is 5 + 3? "))
questions_asked = questions_asked + 1
if answer_1 == 8:
    score = score + 1
else:
    streak_alive = False

answer_2 = input("Which ocean is west of California? Type Pacific: ")
questions_asked = questions_asked + 1
if answer_2 == "Pacific" or answer_2 == "pacific":
    score = score + 1
else:
    streak_alive = False

if answer_1 == 8 and streak_alive:
    score = score + 1
    print("Streak bonus!")

print(f"Score {score} after {questions_asked} questions.")
```

**Notice:** the opening round in full — accumulators keep the tallies, `or` accepts either spelling, and `and` guards the bonus.

## L2: Secret Follow-ups

A condition can live inside another condition. This is **conditional nesting**: the inner `if` is reached only when the outer `if` is True.

```python
unlocked = True
correct = True
if unlocked:
    if correct:
        print("Bonus won!")
```

**Notice:** the inner `if` runs only because the outer `if` (`unlocked`) was True.

```python
unlocked = True
correct = False
if unlocked:
    if correct:
        print("Bonus won!")
    else:
        print("Unlocked, but the answer was wrong.")
```

**Notice:** the inner `if`/`else` lives entirely inside the outer branch.

```python
part_one = 12
final_score = 0
if part_one == 12:
    final_score = final_score + 1
    part_two = "Saturn"
    if part_two == "Saturn":
        final_score = final_score + 2
        print("Follow-up bonus!")
    else:
        print("Follow-up missed.")
else:
    print("The follow-up stays locked.")
print(f"Final score: {final_score}")
```

**Notice:** the follow-up stays locked unless part one is correct — a gate inside a gate.

## L3: SUDDEN DEATH

In the final round, one wrong answer ends everything. A **`break`** statement leaves a loop immediately, jumping to the first line after it.

```python
count = 0
while count < 5:
    print(count)
    if count == 2:
        break
    count = count + 1
```

**Notice:** `break` stops the loop the instant `count` reaches 2 — it never prints 3 or 4.

```python
round_number = 0
while round_number < 3:
    correct = round_number != 1
    if not correct:
        print("Wrong — round over!")
        break
    print("Correct! Keep going.")
    round_number = round_number + 1
```

**Notice:** a wrong answer triggers `break`, ending the whole round early.

```python
questions_asked = 0
sudden_death_score = 0

while questions_asked < 3:
    if questions_asked == 0:
        answer = int(input("SUDDEN DEATH 1 — sides on a triangle? "))
        correct = answer == 3
    elif questions_asked == 1:
        answer = int(input("SUDDEN DEATH 2 — 6 times 7? "))
        correct = answer == 42
    else:
        answer = int(input("SUDDEN DEATH 3 — days in a week? "))
        correct = answer == 7

    if not correct:
        print("Wrong! Sudden death ends.")
        break
    sudden_death_score = sudden_death_score + 1
    questions_asked = questions_asked + 1

print(f"Survived {sudden_death_score} sudden-death questions.")
```

**Notice:** the full SUDDEN DEATH round — a `while` loop, an `if`/`elif`/`else` dispatch, and `break` the moment an answer is wrong. Tally your questions on paper.
