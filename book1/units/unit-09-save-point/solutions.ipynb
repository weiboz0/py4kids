## Exercise 6: Save Then Load — `savegame.txt`

````python
def save_scores(scores, filename):
    with open(filename, "w") as f:
        for score in scores:
            f.write(str(score) + "\n")
    return filename


def load_scores(filename):
    loaded_scores = []
    with open(filename) as f:
        for line in f:
            loaded_scores.append(int(line.strip()))
    return loaded_scores


saved_filename = save_scores([300, 450, 725, 1200, 900], "savegame.txt")
loaded_result = load_scores(saved_filename)
print(loaded_result)
assert (saved_filename, loaded_result) == ("savegame.txt", [300, 450, 725, 1200, 900])
````

**The real program**


## Exercise 7: Save My Score — `savegame.txt`

````python
sample_score = "1350"
new_score = int(sample_score)
saved_scores = load_scores("savegame.txt")

if new_score in saved_scores:
    save_message = f"{new_score} is already saved."
else:
    saved_scores.append(new_score)
    save_scores(saved_scores, "savegame.txt")
    save_message = f"Saved {new_score}; {len(saved_scores)} scores total."

print(save_message)
assert (save_message, load_scores("savegame.txt")) == (
    "Saved 1350; 6 scores total.",
    [300, 450, 725, 1200, 900, 1350],
)
````

## Exercise 9: Prove That `"w"` Replaces — `overwrite_scores_tmp.txt`

````python
practice_scores = [42, 77, 15, 103, 64]

with open("overwrite_scores_tmp.txt", "w") as f:
    for score in practice_scores:
        f.write(str(score) + "\n")

with open("overwrite_scores_tmp.txt") as f:
    first_save = f.read()

with open("overwrite_scores_tmp.txt", "w") as f:
    for score in practice_scores:
        f.write(str(score) + "\n")

with open("overwrite_scores_tmp.txt") as f:
    second_save = f.read()

print(first_save)
print(second_save)
print(len(first_save))
print(len(second_save))
assert (first_save, second_save, len(first_save), len(second_save)) == (
    "42\n77\n15\n103\n64\n",
    "42\n77\n15\n103\n64\n",
    16,
    16,
)
````

## Exercise 10: Save and Load a Player Profile — `settings.txt`

````python
profile = {"player": "Mina", "level": "forest"}


def save_profile(profile, filename):
    with open(filename, "w") as f:
        f.write(profile["player"] + "\n")
        f.write(profile["level"] + "\n")
    return filename


def load_profile(filename):
    loaded_profile = []
    with open(filename) as f:
        for line in f:
            loaded_profile.append(line.strip())
    return loaded_profile


profile_filename = save_profile(profile, "settings.txt")
loaded_profile = load_profile(profile_filename)

if profile["level"] in loaded_profile:
    profile_message = f"{profile['player']} can resume {profile['level']}."
else:
    profile_message = f"No {profile['level']} save found for {profile['player']}."

print(profile_message)
assert (loaded_profile, profile_message) == (
    ["Mina", "forest"],
    "Mina can resume forest.",
)
````

**The real program**

## Exercise 11: Load a Labeled Scoreboard — `savegame.txt`

````python
def load_scoreboard(filename, game_name):
    scores = []
    with open(filename) as f:
        for line in f:
            scores.append(int(line.strip()))
    print(f"{game_name} loaded {len(scores)} scores.")
    return scores


scoreboard = load_scoreboard("savegame.txt", "Space Race")
print(scoreboard)
assert scoreboard == [300, 450, 725, 1200, 900, 1350]
````

## Exercise 21: Saves That Fit the Budget

````python
scores = [300, 450, 725, 1350]
budget = 1000
fit_count = 0
fit_total = 0

for score in scores:
    if fit_total + score <= budget:
        fit_total = fit_total + score
        fit_count = fit_count + 1
    else:
        break

print(f"{fit_count} saves fit for a total of {fit_total}")
assert fit_count == 2
assert fit_total == 750
````

## Exercise 22: Saves Until the Budget Tips

````python
scores = [300, 450, 725, 1350]
budget = 1000
tip_count = 0
tip_total = 0
tipping_score = 0

for score in scores:
    tip_total = tip_total + score
    tip_count = tip_count + 1
    if tip_total > budget:
        tipping_score = score
        break

print(f"Save {tipping_score} tipped the total to {tip_total} after {tip_count} saves")
assert tip_count == 3
assert tip_total == 1475
assert tipping_score == 725
````
