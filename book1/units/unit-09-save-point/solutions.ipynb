```python
loaded = []

with open("savegame.txt") as f:
    for line in f:
        loaded.append(int(line.strip()))

assert loaded == [1200, 850, 990]
```

## Exercise 7: Save Then Load — `savegame.txt`

```python
def save_scores(scores, filename):
    with open(filename, "w") as f:
        for score in scores:
            f.write(str(score) + "\n")
    return filename


def load_scores(filename):
    loaded_scores = []
    with open(filename) as f:
        for line in f:
            loaded_scores.append(int(line.strip()))
    return loaded_scores


assert load_scores(save_scores([100, 200], "savegame.txt")) == [100, 200]
round_trip_scores = [1200, 850, 990]
save_scores(round_trip_scores, "savegame.txt")
loaded_round_trip = load_scores("savegame.txt")
print(loaded_round_trip)
assert loaded_round_trip == round_trip_scores
```

## Exercise 8: Save My Score — `savegame.txt`

```python
sample_score = "1350"
new_score = int(sample_score)
saved_scores = load_scores("savegame.txt")
saved_scores.append(new_score)
save_scores(saved_scores, "savegame.txt")
assert load_scores("savegame.txt") == [1200, 850, 990, 1350]
```

## Exercise 9: Highest Saved Score — `savegame.txt`

```python
high_scores = load_scores("savegame.txt")
highest_score = max(high_scores)
print("Highest saved score:", highest_score)
assert highest_score == 1350
```

## Exercise 10: Add a New High — `savegame.txt`

```python
scores_with_new_high = load_scores("savegame.txt")
scores_with_new_high.append(1600)
save_scores(scores_with_new_high, "savegame.txt")
assert load_scores("savegame.txt") == [1200, 850, 990, 1350, 1600]
```
