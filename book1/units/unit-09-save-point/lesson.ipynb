# SAVE POINT

## Lesson 1: Write a Save File

A game that forgets everything when you close it is no fun.
Let's teach our program to **WRITE** the player's progress to a file and **LOAD** it back next time.

We will begin with three scores.
The file `savegame.txt` is only for whole-number scores, one score on each line.

````python
scores = [1200, 850, 990]

with open("savegame.txt", "w") as f:
    for score in scores:
        f.write(str(score) + "\n")
````

The line `for score in scores:` loops over a **list**.
Each time around the loop, `score` is one number from the list.
`str(score)` turns that number into text so `f.write()` can put it in the file.

### New syntax: `\n`

`\n` is the invisible **newline character** that ends a line.
Adding it after each score starts the next value on a fresh line.
Without `\n`, the scores would run together.

The `with` block closes the file automatically when the indented work is finished.

Let's open `savegame.txt` for reading right away and display exactly what we saved.

````python
with open("savegame.txt") as f:
    saved_scores = f.read()

print(saved_scores)
````

Now we need a different file for words and settings.
`settings.txt` will hold the player's name, volume, and difficulty as text lines.
Keeping settings separate means every line in `savegame.txt` stays safe to turn back into an integer.

````python
settings = {"volume": 8, "difficulty": "easy"}

with open("settings.txt", "w") as f:
    f.write("Ada\n")
    f.write(str(settings["volume"]) + "\n")
    f.write(settings["difficulty"] + "\n")
````

Read `settings.txt` straight back too.
This file is text, so we will keep its contents as text.

````python
with open("settings.txt") as f:
    saved_settings = f.read()

print(saved_settings)
````

## Lesson 2: Load the Save Point

We saved our game in Lesson 1.
Now it is time to load it.
First, `f.read()` brings the whole scores file into one text string.

````python
with open("savegame.txt") as f:
    content = f.read()

print(content)
````

A text string is useful for looking, but our game needs a list of integer scores again.

This time, `for line in f:` loops over the **file object line-by-line**.
That is different from `for score in scores:`, which loops over a **list**.
For each file line, `.strip()` drops the trailing newline, `int()` changes the remaining text into an integer, and `.append()` adds the integer to the new list.

````python
loaded = []

with open("savegame.txt") as f:
    for line in f:
        loaded.append(int(line.strip()))

print(loaded)
````

We can package that line-by-line loading pattern into a helper function.
The parameter `filename` tells the helper which score file to read, and `return` sends the rebuilt list back.

````python
def load_scores(filename):
    result = []
    with open(filename) as f:
        for line in f:
            result.append(int(line.strip()))
    return result

loaded = load_scores("savegame.txt")
print(f"Loaded {len(loaded)} scores")
````

The settings file has words as well as a number, so we do **not** turn its lines into integers.
Instead, read `settings.txt` as text and use `in` to search it for the player's name.

````python
with open("settings.txt") as f:
    info = f.read()

if "Ada" in info:
    print("Welcome back, Ada!")
````

What happens if a game tries to load a file that was never saved?
The next example is here to study, not run.

````python
with open("missing.txt") as f:
    f.read()
````

```text
Traceback (most recent call last):
  FileNotFoundError: [Errno 2] No such file or directory: 'missing.txt'
```

`FileNotFoundError` tells us the program tried to load a file that does not exist yet.
For our save-point game, the safe plan is simple: save first, then load.
