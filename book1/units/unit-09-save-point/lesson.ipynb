# SAVE POINT

A game that forgets everything when you close it is no fun. Let's teach our program to **WRITE** the player's progress to a file and **LOAD** it back next time.

Each idea comes as a short **ladder** — simplest first, then one step up, then the real save system — with a **Notice** naming the one new thing.

## Lesson 1: Write a Save File

`with open("savegame.txt", "w") as f:` opens a file for **writing** and, when the block ends, closes it for you. `f.write(...)` puts text in the file. `\n` is the invisible **newline** that ends a line.

````python
with open("savegame.txt", "w") as f:
    f.write("1200\n")
print("Saved one score.")
````

**Notice:** the `with` block opens `savegame.txt` for writing and closes it automatically when the block ends.

````python
scores = [1200, 850, 990]
with open("savegame.txt", "w") as f:
    for score in scores:
        f.write(str(score) + "\n")
print("Saved three scores.")
````

**Notice:** a loop writes several lines — `str(score)` turns each number into text, and `\n` starts the next score on a fresh line.

````python
settings = {"volume": 8, "difficulty": "easy"}
with open("settings.txt", "w") as f:
    f.write("Ada\n")
    f.write(str(settings["volume"]) + "\n")
    f.write(settings["difficulty"] + "\n")
print("Saved the settings.")
````

**Notice:** a different file can hold mixed lines — a name, a number, and a word — for the player's settings.

## Lesson 2: Read a Save File

`with open("savegame.txt") as f:` (no `"w"`) opens a file for **reading**. `f.read()` brings the whole file in as one text string.

````python
with open("savegame.txt") as f:
    content = f.read()
print(content)
````

**Notice:** `f.read()` returns the entire file as one string — exactly what was saved.

````python
with open("savegame.txt") as f:
    for line in f:
        print(line)
````

**Notice:** one step up — `for line in f:` walks the file one LINE at a time (each line still carries its `\n`, so the output looks double-spaced).

````python
with open("savegame.txt") as f:
    for line in f:
        print(line.strip())
````

**Notice:** `.strip()` removes the trailing newline, so each line prints cleanly.

````python
loaded = []
with open("savegame.txt") as f:
    for line in f:
        loaded.append(int(line.strip()))
print(loaded)
````

**Notice:** now turn each cleaned line back into an integer with `int(...)` and `.append()` it — the scores are a list of numbers again.

## Lesson 3: Load the Save Point

**Put it together:** package the line-by-line loader into a helper. `filename` says which file to read; `return` sends the rebuilt list back.

````python
def load_scores(filename):
    result = []
    with open(filename) as f:
        for line in f:
            result.append(int(line.strip()))
    return result


loaded = load_scores("savegame.txt")
print(f"Loaded {len(loaded)} scores")
````

**Put it together:** the settings file has words as well as a number, so read it as text and use `in` to search it for the player's name.

````python
with open("settings.txt") as f:
    info = f.read()

if "Ada" in info:
    print("Welcome back, Ada!")
````
