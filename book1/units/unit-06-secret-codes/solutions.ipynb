## Exercise 6

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


def decode(message, shift):
    return encode(message, 26 - shift)


sample_message = "meet me at 4!"
shift = 3
coded_message = encode(sample_message, shift)
print(f"Send this code: {coded_message}")

msg = "meet me at 4!"
assert encode("z", 3) == "c"
assert encode("meet me at 4!", 3) == "phhw ph dw 4!"
assert decode(encode(msg, 3), 3) == msg
```

## Exercise 7

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


fixed_message = encode("zoo", 3)
print(fixed_message)
assert fixed_message == "crr"
```

The `z` starts at position 25. Adding 3 gives 28, and `28 % 26` is 2, so the shifted character at position 2 is `c`.

### Challenge 1: Case-Insensitive Keyword Check

```python
message = "Meet Me At The Library"
keyword = "LIBRARY"
quiet_message = message.lower()
quiet_keyword = keyword.lower()
if quiet_keyword in quiet_message:
    print("The keyword appears in the message.")
else:
    print("The keyword does not appear in the message.")
assert quiet_keyword in quiet_message
```

### Challenge 2: Two-Step Cipher

```python
def two_step(message, shift):
    reversed_message = message[::-1]
    return encode(reversed_message, shift)


two_step_message = two_step("abc!", 2)
print(two_step_message)
assert two_step_message == "!edc"
```

A friend first decodes the Caesar shift, then reverses the decoded string to recover the original message.
