## Exercise 6

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        elif letter == " ":
            result = result + letter
        else:
            result = result + letter
    return result


sample_message = "meet me at 4!"
shift = 3
coded_message = encode(sample_message, shift)
print(f"Send this code: {coded_message}")
assert coded_message == "phhw ph dw 4!"
assert encode("z!", 3) == "c!"
```

## Exercise 7

```python
# IndexError: string index out of range
# The failing operation is indexing letters with letters[new_position].
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


fixed_message = encode("zoo", 3)
print(fixed_message)
# The z shifts from position 25 to position 2, so the first output character is c.
assert fixed_message == "crr"

# IndexError: string index out of range
# The failing operation is indexing letters with letters[last_position].
letters = "abcdefghijklmnopqrstuvwxyz"
last_position = 25
last_letter = letters[last_position]
print(last_letter)
assert last_position == 25 and last_letter == "z"
```

## Exercise 9

```python
def classify_code_character(sample_text):
    cleaned_text = sample_text.strip().lower()
    character = cleaned_text[0]
    first_two = cleaned_text[:2]
    letters = "abcdefghijklmnopqrstuvwxyz"
    vowels = "aeiou"
    is_letter = character in letters
    print(is_letter)
    if character in vowels:
        category = "vowel"
        print(f"{first_two}: {category}")
    elif character in letters:
        category = "consonant letter"
        print(f"{first_two}: {category}")
    else:
        category = "mark"
        print(f"{first_two}: {category}")
    return first_two, is_letter, category


vowel_report = classify_code_character("  Echo  ")
consonant_report = classify_code_character("  Zebra  ")
mark_report = classify_code_character("  !go  ")
assert vowel_report == ("ec", True, "vowel")
assert consonant_report == ("ze", True, "consonant letter")
assert mark_report == ("!g", False, "mark")
```

## Exercise 10

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        elif letter == " ":
            result = result + letter
        else:
            result = result + letter
    return result


shift_three_output = encode("Secret zoo!", 3)
shift_five_output = encode("Secret zoo!", 5)
same_output = shift_three_output == shift_five_output
print(f"Shift 3: {shift_three_output}")
print(f"Shift 5: {shift_five_output}")
print(f"Same output: {same_output}")
print(f"Shift 3 first: {shift_three_output[0]}, slice: {shift_three_output[:3]}")
print(f"Shift 5 first: {shift_five_output[0]}, slice: {shift_five_output[:3]}")
# Shift 5 moves the first letter farther through the alphabet than shift 3.
assert shift_three_output == "vhfuhw crr!"
assert shift_five_output == "xjhwjy ett!"
assert not same_output
assert (shift_three_output[0], shift_three_output[:3]) == ("v", "vhf")
assert (shift_five_output[0], shift_five_output[:3]) == ("x", "xjh")
```

## Exercise 11

```python
def decode_atbash(message):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    decoded_text = ""
    for character in message:
        if character in letters:
            for position in range(26):
                if letters[position] == character:
                    decoded_text = decoded_text + letters[25 - position]
        elif character == " ":
            decoded_text = decoded_text + character
        else:
            decoded_text = decoded_text + character
    return decoded_text


decoded_message = decode_atbash("hvxivg!")
print(f"Decoded: {decoded_message}")
print(decoded_message[0])
print(decoded_message[:3])
assert decoded_message == "secret!"
assert (decoded_message[0], decoded_message[:3]) == ("s", "sec")
```

### Challenge 2: Two-Step Cipher

```python
def two_step(message, shift):
    reversed_message = message[::-1]
    return encode(reversed_message, shift)


two_step_message = two_step("abc!", 2)
print(two_step_message)
assert two_step_message == "!edc"
```

A friend first decodes the Caesar shift, then reverses the decoded string to recover the original message.
