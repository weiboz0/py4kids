## Exercise 7

### Fix the Caesar Wrap

This version fails when a shifted position steps past the end of the alphabet string. Copy it into the blank cell, repair the `new_position` calculation with `% 26`, and test `encode("zoo", 3)`. Explain why the repaired first output character is `c`.

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = position + shift
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result
```
