## L1: Read and Decode a Secret Note

A string is text stored in order. Each character has a numbered position, starting at 0. We will store the note's first coded word in `word`. Square brackets let us select one character. A negative position counts backward from the end.

Predict the three printed characters before you run the cell.

```python
word = "nvvg"
print(word[0])
print(word[1])
print(word[-1])
```

A **slice** selects a run of characters. In `text[start:stop]`, Python includes the start position and stops just before the stop position. Leaving one side blank means continue to that end of the string.

Predict the three pieces of `word`. Why does `word[1:4]` include positions 1, 2, and 3?

```python
print(word[1:4])
print(word[:2])
print(word[2:])
```

Messages often arrive with extra spaces, mixed case, or separator marks. String methods make a changed string that we can store in a new variable:

- `.strip()` removes extra space from both ends.
- `.lower()` changes letters to lowercase.
- `.upper()` changes letters to uppercase.
- `.replace(old, new)` swaps one piece of text for another.

Predict every printed line. Notice that `messy_note` itself stays the same.

```python
messy_note = "  MEET_ME_AT_NOON  "
trimmed_note = messy_note.strip()
quiet_note = trimmed_note.lower()
spaced_note = quiet_note.replace("_", " ")
print(trimmed_note)
print(quiet_note)
print(spaced_note)
print(spaced_note.upper())
```

A slice can also move backward. The slice `[::-1]` starts at the end and steps backward one character at a time. That makes a quick **reverse cipher**.

```python
word = "noon ta teem"
forward_note = word[::-1]
print(forward_note)
```

Our folded note uses a different trick. The first alphabet character swaps with the last, the second swaps with the second-to-last, and so on.

The loop below checks all 26 positions to locate each coded character. If a character is at `position`, its partner is at `25 - position`. Each character begins unchanged, so a space or punctuation mark is copied when the scan does not find it in the lowercase alphabet.

```python
letters = "abcdefghijklmnopqrstuvwxyz"
coded_note = "nvvg nv zg gsv oryizib"
coded_note = coded_note.lower()
decoded_note = ""

for letter in coded_note:
    new_letter = letter
    for position in range(26):
        if letters[position] == letter:
            new_letter = letters[25 - position]
    decoded_note = decoded_note + new_letter

print(decoded_note)
```

## L2: Write a Code Your Friend Can Crack

Yesterday we read codes. Today we write one only your friend can crack. We will shift each letter forward while leaving spaces and punctuation untouched.

The `in` operator asks whether one value appears inside another. That lets the cipher choose between shifting a letter and copying a mark unchanged. Predict the three printed lines.

```python
letters = "abcdefghijklmnopqrstuvwxyz"
print("m" in letters)
print(" " in letters)
print("!" in letters)
```

Here is a deliberate bug. The code shifts letters forward, but a late alphabet letter can step past the end of the alphabet string. Predict which source line the traceback will point to when the message begins with `z`.

```python
def encode_without_wrap(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = position + shift
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


print(encode_without_wrap("zoo", 3))
```

Read a traceback from the bottom: `IndexError` tells us that a numbered position fell outside the string. The remainder operator repairs the wrap. `(position + shift) % 26` always produces a position from 0 through 25.

The function also changes the message to lowercase first. Spaces, punctuation, and other characters are copied in the `else` branch.

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


coded_message = encode("Meet me at 4!", 3)
print(coded_message)
```

Trace the first `m`: the scan finds position 12, adds the shift 3, and selects position 15, which holds `p`. For `z`, the sum 28 leaves a remainder of 2, which wraps to `c`.

The `result` variable is a string accumulator. Each pass adds one transformed or unchanged character, building the answer from left to right.

To undo a shift, move forward by the rest of one full 26-character trip. A shift of 3 is undone by a shift of `26 - 3`. The decoder can reuse the encoder instead of copying its whole body.

```python
def decode(message, shift):
    return encode(message, 26 - shift)


original_message = "meet me at 4!"
coded_message = encode(original_message, 3)
decoded_message = decode(coded_message, 3)
print(f"Coded: {coded_message}")
print(f"Decoded: {decoded_message}")
```

Encoding and then decoding makes a **round trip**. Because the encoder changes text to lowercase, the decoded result matches the lowercase form of the original message. The spaces, number, and exclamation mark survive unchanged.

Now let a person type the message. The shift stays a fixed integer, so both friends know the key. This interactive cell waits for typing, so it is not part of headless execution.

```python
message = input("Type a secret message: ")
shift = 3
coded_message = encode(message, shift)
print(f"Send this code: {coded_message}")
```
