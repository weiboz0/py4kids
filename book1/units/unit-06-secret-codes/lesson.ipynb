# Secret Codes

A folded note lands on your desk:

`nvvg nv zg gsv oryizib`

It looks like nonsense, but your friend promises it hides a real message. Today we crack it by taking text apart, flipping it, and rebuilding it one character at a time.

Each tool comes as a short **ladder** — simplest first, then one step up, then the real cipher — with a **Notice** naming the one new thing each time.

## L1: Take the Note Apart

A string is text stored in order. Each character has a numbered position, starting at 0. Square brackets select one character by its position.

```python
word = "nvvg"
print(word[0])
```

**Notice:** positions start at 0, so `word[0]` is the FIRST character.

```python
word = "nvvg"
print(word[1])
print(word[2])
```

**Notice:** each number picks the character sitting at that position.

```python
word = "nvvg"
print(word[-1])
```

**Notice:** a negative position counts back from the end — `[-1]` is the last character.

```python
word = "code"
for position in range(4):
    print(word[position])
```

**Notice:** a loop can walk every position in order, one character at a time.

A **slice** selects a RUN of characters. In `text[start:stop]` Python includes `start` and stops just before `stop`.

```python
word = "nvvg"
print(word[1:4])
```

**Notice:** `[1:4]` takes positions 1, 2, 3 — up to but NOT including 4.

```python
word = "nvvg"
print(word[:2])
```

**Notice:** leave the start blank to begin at position 0.

```python
word = "nvvg"
print(word[2:])
```

**Notice:** leave the stop blank to run all the way to the end.

```python
word = "noon ta teem"
print(word[::-1])
```

**Notice:** `[::-1]` steps backward, reversing the whole string — a quick reverse cipher.

Messages often arrive with extra spaces, mixed case, or separator marks. **String methods** make a CHANGED copy (the original stays the same). This unit uses four: `.lower()`, `.upper()`, `.strip()`, `.replace(old, new)`.

```python
note = "MEET ME"
print(note.lower())
print(note)
```

**Notice:** `.lower()` returns a lowercase copy; `note` itself is unchanged.

```python
quiet = "meet me"
print(quiet.upper())
```

**Notice:** `.upper()` returns an uppercase copy.

```python
padded = "   meet me   "
print(padded.strip())
```

**Notice:** `.strip()` removes the spaces from both ends.

```python
coded = "meet_me_at_noon"
print(coded.replace("_", " "))
```

**Notice:** `.replace(old, new)` swaps every `old` piece for `new`.

```python
messy = "   MEET_ME_AT_NOON   "
clean = messy.strip().lower().replace("_", " ")
print(clean)
```

**Notice:** methods can be CHAINED — each one works on the result of the one before it.

## L2: Decode the Folded Note

Our note uses a swap: the first alphabet letter trades with the last, the second with the second-to-last, and so on. The loop checks all 26 positions to find each coded letter; a letter at `position` has its partner at `25 - position`. Characters not found (a space or mark) are copied unchanged.

```python
letters = "abcdefghijklmnopqrstuvwxyz"
coded_note = "nvvg nv zg gsv oryizib"
coded_note = coded_note.lower()
decoded_note = ""

for letter in coded_note:
    new_letter = letter
    for position in range(26):
        if letters[position] == letter:
            new_letter = letters[25 - position]
    decoded_note = decoded_note + new_letter

print(decoded_note)
```

The note says **meet me at the library**. Trace one coded character through the inner loop: which position is found, and which partner is used? Why does the accumulator begin as an empty string?

To WRITE a cipher we must tell letters from marks. The **`in` operator** asks whether one value appears inside another.

```python
letters = "abcdefghijklmnopqrstuvwxyz"
print("m" in letters)
```

**Notice:** `in` reports True/False — is `"m"` somewhere in `letters`? Yes.

```python
letters = "abcdefghijklmnopqrstuvwxyz"
print(" " in letters)
print("!" in letters)
```

**Notice:** a space and a `!` are not letters, so `in` is False for them.

```python
letters = "abcdefghijklmnopqrstuvwxyz"
message = "hi!"
for character in message:
    if character in letters:
        print(character + " is a letter")
    else:
        print(character + " is a mark")
```

**Notice:** inside a loop, `in` lets the cipher decide per character — shift a letter, copy a mark.

## L3: Write a Code Your Friend Can Crack

Now we shift each letter forward, leaving spaces and punctuation untouched. Here is a deliberate bug: a late letter can step past the end of the alphabet. Predict which line the traceback points to when the message starts with `z`.

```python
def encode_without_wrap(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = position + shift
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


print(encode_without_wrap("zoo", 3))
```

Read a traceback from the bottom: `IndexError` means a position fell outside the string. The remainder operator repairs the wrap — `(position + shift) % 26` is always 0 through 25.

```python
def encode(message, shift):
    letters = "abcdefghijklmnopqrstuvwxyz"
    message = message.lower()
    result = ""
    for letter in message:
        if letter in letters:
            for position in range(26):
                if letters[position] == letter:
                    new_position = (position + shift) % 26
                    result = result + letters[new_position]
        else:
            result = result + letter
    return result


coded_message = encode("Meet me at 4!", 3)
print(coded_message)
```

Trace the first `m`: the scan finds position 12, adds shift 3, selects position 15 — `p`. For `z`, 25 + 3 = 28, and `28 % 26` wraps to 2 — `c`. The `result` variable is a string accumulator, built left to right.

```python
def decode(message, shift):
    return encode(message, 26 - shift)


original_message = "meet me at 4!"
coded_message = encode(original_message, 3)
decoded_message = decode(coded_message, 3)
print(f"Coded: {coded_message}")
print(f"Decoded: {decoded_message}")
```

Encoding then decoding is a **round trip**. Because the encoder lowercases first, the decoded text matches the lowercase original; spaces, the number, and `!` survive unchanged.

```python
message = input("Type a secret message: ")
shift = 3
coded_message = encode(message, shift)
print(f"Send this code: {coded_message}")
```
