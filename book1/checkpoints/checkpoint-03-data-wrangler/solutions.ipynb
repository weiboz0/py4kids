## Question 4

```python
scores = [88, 92, 75, 100]
for score in scores:
    print(score)
print(len(scores))
print(max(scores))
print(min(scores))
total = 0
for score in scores:
    total = total + score
print(total)
assert total == 355
```

## Question 6

```python
prices = {"apple": 3, "pear": 2, "plum": 4}
print(prices["pear"])
print(prices.get("fig", 0))
print("apple" in prices)
if "plum" in prices:
    print(prices["plum"])
elif "pear" in prices:
    print(prices["pear"])
else:
    print("sold out")
assert prices.get("fig", 0) == 0
```

## Question 7

```python
words = ["cat", "dog", "cat", "bird", "cat"]
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
for word, count in counts.items():
    print(word, count)
assert counts["cat"] == 3
```
