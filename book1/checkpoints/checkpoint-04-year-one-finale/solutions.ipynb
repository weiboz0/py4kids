## Question 1

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.health = 10

hero = Hero("Ada")
print(hero.name)
print(hero.health)
assert hero.name == "Ada" and hero.health == 10
```

## Question 2

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.health = 10

    def heal(self, amount):
        self.health = self.health + amount
        return self.health

hero = Hero("Ada")
new_health = hero.heal(3)
print(new_health)
assert new_health == 13 and hero.health == 13
```

## Question 3

```python
scores = [40, 90, 20]
with open("finale.txt", "w") as f:
    for score in scores:
        f.write(str(score) + "\n")
```

## Question 4

```python
loaded = []
with open("finale.txt") as f:
    for line in f:
        loaded.append(int(line.strip()))
print(loaded)
assert loaded == [40, 90, 20]
```

## Question 5

```python
loaded.sort(reverse=True)
print(loaded[0])
assert loaded == [90, 40, 20] and loaded[0] == 90
```

## Question 6

```python
inventory = {"sword": 1, "potion": 3}
lines = []
for item, count in inventory.items():
    line = f"{item}: {count}"
    print(line)
    lines.append(line)
assert lines == ["sword: 1", "potion: 3"]
```
