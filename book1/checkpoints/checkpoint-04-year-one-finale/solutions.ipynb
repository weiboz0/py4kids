## Question 1

````python
class Hero:
    def __init__(self, name):
        self.name = name
        self.health = 10

hero = Hero("Ada")
print(hero.name)
print(hero.health)
assert hero.name == "Ada" and hero.health == 10
````

```python
        self.health = 10

print(hero.health)
```

## Question 2

````python
class Hero:
    def __init__(self, name):
        self.name = name
        self.health = 10

    def heal(self, amount):
        self.health = self.health + amount
        return self.health

hero = Hero("Ada")
new_health = hero.heal(3)
print(new_health)
assert new_health == 13 and hero.health == 13
````

**The real program** (reads the hero's name and a whole-number heal amount)

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.health = 10

    def heal(self, amount):
        self.health = self.health + amount
        return self.health

name = input("Hero name? ")
hero = Hero(name)
amount = int(input("Heal by how much? "))
new_health = hero.heal(amount)
print(new_health)
```

## Question 4

````python
loaded = []
with open("finale.txt") as f:
    for line in f:
        loaded.append(int(line.strip()))
print(loaded)
assert loaded == [40, 90, 20]
````

## Question 6
