## Question 2

```python
count = 1
total = 0

while count <= 4:
    total = total + count
    count = count + 1

print(total)
print("The comparison becomes false when count reaches 5.")
assert total == 10
assert count == 5
```

## Question 3

```python
def greeting(name):
    return f"Hello, {name}!"


message = greeting("Maya")
print(message)
assert message == "Hello, Maya!"
```

## Question 4

```python
def show_bonus(points):
    print(points + 2)


def get_bonus(points):
    return points + 2


show_bonus(5)
returned_bonus = get_bonus(5)
print("get_bonus hands a value back. show_bonus only displays a value.")
assert returned_bonus == 7
```

## Question 5

```python
team = "Owls"


def make_tag(player):
    tag = f"{team}: {player}"
    print(tag)


make_tag("Dev")
print(team)
print("team is global. tag is local and is unavailable after the call.")
assert team == "Owls"
```

## Question 6

```python
energy = 7
has_key = True
knows_code = False

if energy >= 8:
    door_message = "Power door opens"
elif energy >= 5:
    if has_key or knows_code:
        door_message = "Side door opens"
    else:
        door_message = "Find a key"
else:
    door_message = "Recharge first"

print(door_message)
print("7 >= 8 is false; 7 >= 5 is true; has_key or knows_code is true.")
assert door_message == "Side door opens"
```

## Question 8

```python
def score_round(first_correct, second_correct, third_correct):
    score = 0
    if first_correct:
        score = score + 1
    if second_correct:
        score = score + 1
    if third_correct:
        score = score + 1
    return score


round_score = score_round(True, False, True)
print(round_score)
assert round_score == 2
```
