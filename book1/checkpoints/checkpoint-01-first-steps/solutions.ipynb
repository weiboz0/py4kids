## Question 4

```python
secret = 20
guess = 27
if guess < secret:
    result = "Too low"
elif guess > secret:
    result = "Too high"
else:
    result = "Correct"
print(result)
# Why 'Too high' prints: Python checks top to bottom. guess < secret (27 < 20) is
# False, so that branch is skipped. guess > secret (27 > 20) is True, so 'Too high'
# runs; once one branch runs, the remaining elif/else are not checked at all.
assert result == "Too high"
```
