## Question 4

`Too high` prints because 27 is greater than 20. After that true `elif` branch runs, Python skips the earlier false branch and the final `else`.

```python
secret = 20
guess = 27

if guess < secret:
    message = "Too low"
elif guess > secret:
    message = "Too high"
else:
    message = "Correct"

print(message)
```
