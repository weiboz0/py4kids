Build the year's biggest program: a text adventure with a hero, a world to explore, and progress you can save. Your hero will travel through rooms, collect useful items, and return for another adventure later.

## Milestone 1

### Meet your hero

Every great adventure begins with a hero. Build one `Hero` class whose `__init__` method sets `self.name`, `self.health`, and `self.inventory`.

Run this starter, enter a hero name, and print the hero's starting stats. Then personalize the starting health or welcome message.

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.health = 100
        self.inventory = []


hero_name = input(f"Name your hero: ")
hero = Hero(hero_name)

print(f"Hero: {hero.name}")
print(f"Health: {hero.health}")
print(f"Inventory: {hero.inventory}")
```

```python
descriptions = {"cave": "A damp cave.", "forest": "A sunny forest.", "river": "A rushing river."}
exits = {"cave east": "forest", "forest west": "cave", "forest east": "river"}  # key is "room direction"
room_items = {"forest": "sword", "river": "shield"}

current = "cave"
print(descriptions[current])

# TODO: Extend all three dictionaries with another room.
```

## Milestone 3

### Explore the adventure

Bring the world to life with an exploration loop. Each turn, ask for a direction, join the current room and direction into `move_key`, move when that key is in `exits`, and collect an item when the new room is in `room_items`. Enter `q` to end the loop.

After the basic journey works, you could add a random `trap` event with `random.randint`. Keep that idea for later—the scaffold does not implement it for you.

```python
while True:
    print(f"You are in the {current}.")
    print(descriptions[current])
    direction = input(f"Choose a direction, or q to quit: ").strip()

    if direction == "q":
        print(f"Your hero rests for now.")
        break

    move_key = current + " " + direction

    if move_key in exits:
        current = exits[move_key]
        print(f"You travel to the {current}.")
    else:
        print("You can't go that way.")

    if current in room_items:
        item = room_items[current]
        if item not in hero.inventory:
            hero.inventory.append(item)
            print(f"You pick up the {item}.")

    # TODO: Add your own room messages or choices.
```

## Milestone 4

### Save and load the journey

Let the adventure continue another day. Save ONE value per line in `adventure_save.txt`: line 0 is the hero's name, line 1 is health, and lines 2 and beyond are inventory items.

Use one `with` block to write the file. Then use a separate `with` block with `readlines()` to load it, use `.strip()` on each saved line, convert health back to an integer, and confirm that the loaded values match the hero you saved.

```python
with open("adventure_save.txt", "w") as save_file:
    save_file.write(hero.name + "\n")
    save_file.write(str(hero.health) + "\n")
    for item in hero.inventory:
        save_file.write(item + "\n")

loaded_name = ""
loaded_health = 0
loaded_inventory = []
line_number = 0

with open("adventure_save.txt", "r") as save_file:
    for saved_line in save_file:
        saved_value = saved_line.strip()
        if line_number == 0:
            loaded_name = saved_value
        elif line_number == 1:
            loaded_health = int(saved_value)
        else:
            loaded_inventory.append(saved_value)
        line_number = line_number + 1

print(f"Loaded name matches: {loaded_name == hero.name}")
print(f"Loaded health matches: {loaded_health == hero.health}")

# TODO: Connect saving and loading to your adventure's choices.
```
