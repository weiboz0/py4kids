## Milestone 2

### Build game one: Lucky Guess

Make a function for a quick guessing game. The computer picks an integer from 1 through 5 with `random.randint`, and the player tries to match it.

The starter always returns 0 points. Complete the TODO so a correct guess returns a real reward. Keep `return points` inside the function.

```python
import random


def lucky_guess():
    secret_number = random.randint(1, 5)
    guess = input(f"Pick an integer from 1 through 5: ")
    points = 0

    if guess == f"{secret_number}":
        print(f"Jackpot! You matched {secret_number}.")
        points = 0  # TODO: Change 0 to the points for a match.
    else:
        print(f"The lucky number was {secret_number}.")

    return points
```

## Milestone 3

### Build game two: Quick Quiz

Make another function that returns points. This game asks two quick questions, then uses `and`, `or`, and an `if`/`elif`/`else` chain to choose a score.

Complete both scoring TODOs. Award more points for two correct answers than for one correct answer.

```python
def quick_quiz():
    answer_one = input(f"Which is larger: 1) 8 or 2) 12? ")
    answer_two = input(f"Which is even: 1) 6 or 2) 7? ")
    points = 0

    if answer_one == f"2" and answer_two == f"1":
        print(f"Two correct answers!")
        points = 0  # TODO: Award points for two correct answers.
    elif answer_one == f"2" or answer_two == f"1":
        print(f"One correct answer!")
        points = 0  # TODO: Award fewer points for one correct answer.
    else:
        print(f"No points this round. Try again!")

    return points
```

## Milestone 4

### Connect the games and keep score

Now turn the pieces into one arcade. `score` accumulates the points returned by each game. `rounds_played` counts a round only when the player chooses a game. The f-string scoreboard shows both totals.

This scaffold calls your two functions. Finish the TODOs in those functions first, then personalize the menu messages here.

```python
score = 0
rounds_played = 0
choice = f""

while choice != f"q":
    print(f"--- ARCADE NIGHT ---")
    print(f"1: Lucky Guess")
    print(f"2: Quick Quiz")
    print(f"q: Quit")
    choice = input(f"Choose a game: ")

    if choice == f"1":
        score = score + lucky_guess()
        rounds_played = rounds_played + 1
    elif choice == f"2":
        score = score + quick_quiz()
        rounds_played = rounds_played + 1
    elif choice == f"q":
        print(f"The arcade is closing.")
        break
    else:
        print(f"Choose 1, 2, or q.")

    if choice == f"1" or choice == f"2":
        print(f"Score: {score} points after {rounds_played} rounds.")

print(f"Final score: {score} points in {rounds_played} rounds.")
```
