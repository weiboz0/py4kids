# Arcade Night Reference

This reference builds two games that return points, then runs two fixed sample rounds and quits.

## Lucky Guess

The player's fixed guess is passed in as a parameter. A match earns 5 points.

```python
def lucky_guess(guess):
    secret_number = random.randint(1, 5)
    points = 0

    if guess == f"{secret_number}":
        print(f"Jackpot! You matched {secret_number}.")
        points = 5
    else:
        print(f"The lucky number was {secret_number}.")

    return points
```

## Quick Quiz

The two fixed answers are parameters. Two correct answers earn 4 points, and one correct answer earns 2 points.

```python
def quick_quiz(answer_one, answer_two):
    points = 0

    if answer_one == f"2" and answer_two == f"1":
        print(f"Two correct answers!")
        points = 4
    elif answer_one == f"2" or answer_two == f"1":
        print(f"One correct answer!")
        points = 2
    else:
        print(f"No points this round. Try again!")

    return points
```

## Fixed arcade driver

The fixed choices play each game once. After two rounds, the choice becomes `q` and the arcade closes.

```python
score = 0
rounds_played = 0
choice = f"1"
lucky_points = 0
quiz_points = 0

# A real arcade loops until the player picks quit. Here the choices are scripted
# (1, then 2, then q) so the quit branch is genuinely exercised, no typed menu needed.
while True:
    print(f"--- ARCADE NIGHT ---")
    if choice == f"1":
        lucky_points = lucky_guess(f"2")
        score = score + lucky_points
        rounds_played = rounds_played + 1
        choice = f"2"
    elif choice == f"2":
        quiz_points = quick_quiz(f"2", f"1")
        score = score + quiz_points
        rounds_played = rounds_played + 1
        choice = f"q"
    else:
        print(f"The arcade is closing.")
        break
    print(f"Score: {score} points after {rounds_played} rounds.")

print(f"Final score: {score} points in {rounds_played} rounds.")
```

```python
assert lucky_points == 5
```

```python
assert quiz_points == 4
```

```python
assert score == 9
assert rounds_played == 2
assert choice == f"q"
# Pin all three quiz branches so the points can't be hard-coded:
assert quick_quiz(f"2", f"1") == 4   # both correct
assert quick_quiz(f"2", f"2") == 2   # one correct
assert quick_quiz(f"1", f"2") == 0   # neither correct
```
