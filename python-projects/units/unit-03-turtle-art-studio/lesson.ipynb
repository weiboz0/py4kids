## Lesson 1: Move the Pen

`import turtle` brings in the drawing tools. `forward(100)` moves the turtle 100 steps; `right(90)` turns it one square corner. The turtle draws wherever its pen touches the page.

````python
import turtle

turtle.forward(100)
turtle.right(90)

turtle.done()
````

**Notice:** `forward` moves the turtle and `right` turns it — one move, one turn leaves an open corner.

## Lesson 2: Repeat with `for` and `range()`

Writing the same two commands four times is a lot of typing. A `for` loop repeats an indented block, and `range(n)` runs it exactly `n` times. The counter starts at 0 and stops before `n`.

````python
import turtle

for side in range(4):
    turtle.forward(100)
    turtle.right(90)

turtle.done()
````

**Notice:** `for side in range(4):` repeats the block four times — the loop replaces the written-out square.

Run it from this unit folder — a drawing window opens; close it when done:

```text
python assets/l2_square_loop.py
```

````python
import turtle

angle = 120
for side in range(3):
    turtle.forward(100)
    turtle.right(angle)

turtle.done()
````

**Notice:** change `range(3)` and the corner turn to draw a **triangle** — the number of sides is now yours to choose.

Run it from this unit folder — a drawing window opens; close it when done:

```text
python assets/l2_triangle.py
```

````python
import turtle

n = 7
side_length = 70
angle = 360 / n
for side in range(n):
    turtle.forward(side_length)
    turtle.right(angle)

turtle.done()
````

**Notice:** let Python compute `angle = 360 / n`. The `/` always makes a **float** — a decimal like `360 / 7 = 51.428…` — and the turtle turns by it just the same.

Run it from this unit folder — a drawing window opens; close it when done:

```text
python assets/l2_polygon.py
```

````python
import turtle

n = 7
side_length = 70
angle = 360 / n

turtle.color("blue")
turtle.pensize(2)
turtle.penup()
turtle.forward(20)
turtle.pendown()

for side_number in range(n):
    turtle.pensize(side_number + 1)
    turtle.forward(side_length)
    turtle.right(angle)

turtle.done()
````

**Notice:** the loop **counter** `side_number` can do work too — here `pensize(side_number + 1)` makes each side a little thicker than the last.

## Lesson 3: Loops Inside Loops

One loop can live inside another — **nested loops**. The inner loop draws one whole shape; the outer loop turns the turtle and starts the next.

````python
import turtle

for shape_number in range(2):
    for side_number in range(4):
        turtle.forward(90)
        turtle.right(90)
    turtle.right(180)

turtle.done()
````

**Notice:** the inner loop finishes a whole square before the outer loop turns 180° and starts the next — a loop inside a loop.

Run it from this unit folder — a drawing window opens; close it when done:

```text
python assets/l3_two_squares.py
```

````python
import turtle

for shape_number in range(6):
    for side_number in range(4):
        turtle.forward(60)
        turtle.right(90)
    turtle.right(60)

turtle.done()
````

**Notice:** more repeats with a smaller turn between them — six squares fan out into a ring.

Run it from this unit folder — a drawing window opens; close it when done:


**Put it together:** the gallery spirograph nests the same two loops, using `angle = 360 / n` for the shape and a small `turn_between_shapes` for the outer sweep. Compare the result with your teacher's opening demo — you now know what both loops are doing.

````python
import turtle

n = 6
side_length = 80
angle = 360 / n
shape_count = 24
turn_between_shapes = 15

turtle.color("purple")
turtle.pensize(2)

for shape_number in range(shape_count):
    for side_number in range(n):
        turtle.forward(side_length)
        turtle.right(angle)
    turtle.right(turn_between_shapes)

turtle.done()
````
