## Lesson 2: Repeat with `for` and `range()`

Writing the same two commands four times is a lot of typing. A `for` loop repeats an indented block, and `range(n)` runs it exactly `n` times. The counter starts at 0 and stops before `n`.

````python
import turtle

for side in range(4):
    turtle.forward(100)
    turtle.right(90)

turtle.done()
````

**Notice:** `for side in range(4):` repeats the block four times — the loop replaces the written-out square.

Run `assets/l2_square_loop.py` the same way — a drawing window opens; close it when done.

````python
import turtle

angle = 120
for side in range(3):
    turtle.forward(100)
    turtle.right(angle)

turtle.done()
````

**Notice:** change `range(3)` and the corner turn to draw a **triangle** — the number of sides is now yours to choose.

Run `assets/l2_triangle.py` the same way — a drawing window opens; close it when done.

````python
import turtle

n = 7
side_length = 70
angle = 360 / n
for side in range(n):
    turtle.forward(side_length)
    turtle.right(angle)

turtle.done()
````

**Notice:** let Python compute `angle = 360 / n`. The `/` always makes a **float** — a decimal like `360 / 7 = 51.428…` — and the turtle turns by it just the same.

Run `assets/l2_polygon.py` the same way — a drawing window opens; close it when done.

````python
import turtle

n = 7
side_length = 70
angle = 360 / n

turtle.color("blue")
turtle.pensize(2)
turtle.penup()
turtle.forward(20)
turtle.pendown()

for side_number in range(n):
    turtle.pensize(side_number + 1)
    turtle.forward(side_length)
    turtle.right(angle)

turtle.done()
````

**Notice:** the loop **counter** `side_number` can do work too — here `pensize(side_number + 1)` makes each side a little thicker than the last.

Run `assets/l2_polygons.py` the same way — a drawing window opens; close it when done.

## Lesson 3: Loops Inside Loops

One loop can live inside another — **nested loops**. The inner loop draws one whole shape; the outer loop turns the turtle and starts the next.

````python
import turtle

for shape_number in range(2):
    for side_number in range(4):
        turtle.forward(90)
        turtle.right(90)
    turtle.right(180)

turtle.done()
````

**Notice:** the inner loop finishes a whole square before the outer loop turns 180° and starts the next — a loop inside a loop.

Run `assets/l3_two_squares.py` the same way — a drawing window opens; close it when done.

````python
import turtle

for shape_number in range(6):
    for side_number in range(4):
        turtle.forward(60)
        turtle.right(90)
    turtle.right(60)

turtle.done()
````

**Notice:** more repeats with a smaller turn between them — six squares fan out into a ring.

Run `assets/l3_rings.py` the same way — a drawing window opens; close it when done.

**Put it together:** the gallery spirograph nests the same two loops, using `angle = 360 / n` for the shape and a small `turn_between_shapes` for the outer sweep. Compare the result with the spirograph promised at the start of the unit — you now know what both loops are doing.

````python
import turtle

n = 6
side_length = 80
angle = 360 / n
shape_count = 24
turn_between_shapes = 15

turtle.color("purple")
turtle.pensize(2)

for shape_number in range(shape_count):
    for side_number in range(n):
        turtle.forward(side_length)
        turtle.right(angle)
    turtle.right(turn_between_shapes)

turtle.done()
````

Run `assets/l3_spirograph.py` the same way — a drawing window opens; close it when done.

### Recap

- `forward`, `right`, `penup`, `pendown`, `color` and `pensize` control the turtle's pen.
- `for side in range(n):` repeats a block exactly `n` times, counting from 0.
- `angle = 360 / n` gives the turn for an `n`-sided shape, and `/` always makes a float.
- The loop counter can do work too, such as making each side thicker.
- In nested loops the inner loop draws a whole shape before the outer loop turns and starts the next.

You can now design and run a spirograph of your own.
