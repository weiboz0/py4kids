## Task

A teammate ran self-training on a three-class shape task with 30 labelled images, a 600-image unlabelled pool, and 30 labelled validation images. Their recipe: threshold $\tau=0.6$, **no** per-class cap, the student **continues** training from the teacher's weights each round (no re-initialization), no student noise. Each round re-selects pseudo-labels from the full unlabelled pool, so the admitted counts are per round, not cumulative. This is the log:

| round | pseudo-labels admitted (class 0, 1, 2) | mean confidence of admitted | validation accuracy | validation recall (class 0, 1, 2) | training loss at end of round |
|---:|---|---:|---:|---|---:|
| 0 (labelled only) | — | — | 0.80 | 0.80, 0.80, 0.80 | 0.004 |
| 1 | 180, 150, 90 | 0.91 | 0.83 | 0.90, 0.80, 0.80 | 0.003 |
| 2 | 260, 210, 40 | 0.95 | 0.80 | 1.00, 0.80, 0.60 | 0.002 |
| 3 | 330, 240, 12 | 0.98 | 0.73 | 1.00, 0.90, 0.30 | 0.001 |
| 4 | 352, 246, 2 | 0.99 | 0.70 | 1.00, 0.90, 0.20 | 0.001 |

The pool is known to be roughly balanced across the three classes.

(a) **Diagnose.** Name the failure modes and support each with evidence from at least three different columns. Explain why the falling training loss and rising confidence are not evidence of improvement.

(b) **Mechanism.** Explain in at most four sentences how continuing from the teacher's weights without noise lets the student's mistakes reinforce themselves (confirmation bias), and why class 2 is the one that disappears.

(c) **Stopping.** If you must use this run's models, which round's model do you submit, and on what evidence? Why must that choice not use the test set?

(d) **Remedy.** Choose a revised recipe from these options and justify each choice, explaining the risk of each option you reject: (i) raise $\tau$ to 0.99 and keep everything else; (ii) per-class caps that grow across rounds; (iii) re-initialize the student each round; (iv) add Gaussian input noise to the student's training inputs; (v) add the 30 validation images to the labelled set to get more labels; (vi) choose the number of rounds on validation accuracy.

(e) **Verification without pool labels.** State three quantities you will log in the revised run, with thresholds, that would tell you the problem is fixed, using only labelled validation data and pseudo-label counts.
