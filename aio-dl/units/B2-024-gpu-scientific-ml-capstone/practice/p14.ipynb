## Task

A three-bump mixture is $f(x;\theta)=\sum_{k=1}^{3}w_k\exp\!\big(-(x-c_k)^2/(2s_k^2)\big)$ with $\theta=(w_1,c_1,s_1,w_2,c_2,s_2,w_3,c_3,s_3)$, $w_k>0$, $s_k>0$.
For a permutation $\pi$ of $\{1,2,3\}$ let $\pi\theta$ be the vector whose $k$-th triple is $(w_{\pi(k)},c_{\pi(k)},s_{\pi(k)})$.

(a) Prove that $f(x;\pi\theta)=f(x;\theta)$ for every $x$ and every $\pi$.

(b) Suppose the centers are strictly distinct. Define $\operatorname{can}(\theta)$ as the vector whose triples are ordered by increasing center. Prove that **exactly one** of the six vectors $\pi\theta$ has strictly increasing centers, so $\operatorname{can}(\theta)$ is well defined.

(c) Prove that $\operatorname{can}(\pi\theta)=\operatorname{can}(\theta)$ for every $\pi$, and that if $\theta$ and $\theta'$ have strictly distinct centers and $\operatorname{can}(\theta)=\operatorname{can}(\theta')$, then $\theta'=\pi\theta$ for some $\pi$.

(d) Give an explicit two-component example (you may set one triple's weight equal to another's) showing that sorting by weight does **not** give a well-defined canonical form, and explain in one sentence why this unit's generator guarantees the hypothesis of (b).

(e) Conclude in two sentences why a regressor should be trained on $\operatorname{can}(\theta)$ rather than on $\theta$ in the generator's draw order.
