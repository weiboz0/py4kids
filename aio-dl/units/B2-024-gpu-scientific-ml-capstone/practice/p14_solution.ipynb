## Solution

Notation: $\theta$ has triples $\theta_k=(w_k,c_k,s_k)$, $k=1,2,3$, and $(\pi\theta)_k=\theta_{\pi(k)}$. Write $g(x;w,c,s)=w\exp(-(x-c)^2/(2s^2))$, so $f(x;\theta)=\sum_{k=1}^3 g(x;\theta_k)$.

**Composition rule (used below).** For permutations $\pi,\pi'$: $(\pi'(\pi\theta))_k=(\pi\theta)_{\pi'(k)}=\theta_{\pi(\pi'(k))}$, so $\pi'(\pi\theta)=(\pi\circ\pi')\theta$.

**(a) Invariance.** $f(x;\pi\theta)=\sum_{k=1}^3 g(x;\theta_{\pi(k)})$. Since $\pi$ is a bijection of $\{1,2,3\}$, substituting $j=\pi(k)$ runs $j$ over $\{1,2,3\}$ exactly once each, and a finite sum does not depend on the order of its terms (commutativity and associativity of addition). Hence $f(x;\pi\theta)=\sum_{j=1}^3 g(x;\theta_j)=f(x;\theta)$ for every $x$.

**(b) Exactly one ordering has strictly increasing centers.** The centers of $\pi\theta$ are $(c_{\pi(1)},c_{\pi(2)},c_{\pi(3)})$.
*Existence:* the three centers are distinct real numbers, so they have a smallest, a middle, and a largest value, attained at unique indices $i_1,i_2,i_3$ (uniqueness because the values are distinct). The map $\sigma(1)=i_1,\sigma(2)=i_2,\sigma(3)=i_3$ is a permutation and $\sigma\theta$ has $c_{i_1}<c_{i_2}<c_{i_3}$.
*Uniqueness:* suppose $\pi\theta$ has strictly increasing centers. Then $c_{\pi(1)}$ is smaller than the other two, so it is the minimum and $\pi(1)=i_1$ (the minimum is attained at only one index); similarly $c_{\pi(3)}$ is the maximum so $\pi(3)=i_3$, and then $\pi(2)=i_2$. Hence $\pi=\sigma$. Among the six vectors $\pi\theta$ exactly one, $\sigma\theta$, has strictly increasing centers, so $\operatorname{can}(\theta)=\sigma\theta$ is well defined.

**(c) Canonical form is a complete invariant of relabelling.**
*Invariance:* by the composition rule, the six relabellings of $\pi\theta$ are $\{\pi'(\pi\theta)\}=\{(\pi\circ\pi')\theta:\pi'\in S_3\}=\{\rho\theta:\rho\in S_3\}$, because $\pi'\mapsto\pi\circ\pi'$ is a bijection of $S_3$. So $\theta$ and $\pi\theta$ have the same set of six relabellings, and (by (b), which applies since $\pi\theta$ has the same, distinct, centers) the unique member with strictly increasing centers is the same: $\operatorname{can}(\pi\theta)=\operatorname{can}(\theta)$.
*Completeness:* let $\operatorname{can}(\theta)=\sigma\theta$ and $\operatorname{can}(\theta')=\sigma'\theta'$ and suppose they are equal. Apply $\sigma'^{-1}$ to both sides: by the composition rule $\sigma'^{-1}(\sigma'\theta')=(\sigma'\circ\sigma'^{-1})\theta'=\theta'$ and $\sigma'^{-1}(\sigma\theta)=(\sigma\circ\sigma'^{-1})\theta$. Hence $\theta'=\pi\theta$ with $\pi=\sigma\circ\sigma'^{-1}$.

**(d) Sorting by weight is not well defined.** Take two components with equal weights, $\theta=\big((1,0.2,0.05),(1,0.7,0.05)\big)$, and its swap $\pi\theta=\big((1,0.7,0.05),(1,0.2,0.05)\big)$, which by (a) is the same function. Both vectors already have non-decreasing weights, so "sort by weight" (with any stable sort) returns each unchanged: the two representations of one function get two different "canonical" vectors, and none of them is distinguished (there are two weight-sorted orderings, not exactly one).
This unit's generator guarantees (b)'s hypothesis because it redraws until the minimum gap between sorted centers is at least $3\times$ the largest width, and widths are at least $0.04$, so every gap is $\ge0.12>0$ (and the generator re-checks the rule after rounding).

**(e) Why train on $\operatorname{can}(\theta)$.** By (a) the observed function, hence the input, is the same for all six relabellings, so in draw order the same input appears with up to six different targets and a squared-error regressor is pulled toward their average (e.g. all three predicted centers drift toward the mean center), which is not a valid answer at all. By (b)–(c), $\operatorname{can}(\theta)$ is a single well-defined function of the mixture itself, so the regression target becomes a genuine function of the input and the loss can actually reach zero.

## Numeric sanity check

```python
import importlib.util
import itertools
from pathlib import Path

import torch

SEED = 20261022
# Boilerplate: find and load this unit's data module
LOADER_CANDIDATES = (
    Path("../data/capstone_data.py"),
    Path("units/B2-024-gpu-scientific-ml-capstone/data/capstone_data.py"),
    Path("aio-dl/units/B2-024-gpu-scientific-ml-capstone/data/capstone_data.py"),
)
loader_path = next(path for path in LOADER_CANDIDATES if path.is_file())
loader_spec = importlib.util.spec_from_file_location("capstone_data", loader_path)
capstone_data = importlib.util.module_from_spec(loader_spec)
loader_spec.loader.exec_module(capstone_data)
torch.set_num_threads(1)

x = capstone_data.MIXTURE_X.double()
theta = capstone_data.train_rows("mixture")[1].double()                  # (800, 9), canonical
p = theta.reshape(-1, 3, 3)
perms = list(itertools.permutations(range(3)))

# (a) invariance of f, (c) can(pi theta) = can(theta) for all six pi
f0 = capstone_data.mixture_values(x, theta)
max_df, can_ok = 0.0, True
for perm in perms:
    tp = p[:, list(perm)].reshape(-1, 9)
    max_df = max(max_df, (capstone_data.mixture_values(x, tp) - f0).abs().max().item())
    can_ok &= torch.equal(capstone_data.canonical_sort(tp), theta)

# (b) exactly one of the six orderings is strictly increasing in the centers
counts = torch.zeros(p.shape[0], dtype=torch.long)
for perm in perms:
    c = p[:, list(perm), 1]
    # .long() converts to int64 (True -> 1, False -> 0)
    counts += ((c[:, 0] < c[:, 1]) & (c[:, 1] < c[:, 2])).long()

# (d) equal weights: sorting by weight does not identify the two representations
t_d = torch.tensor([[1.0, 0.2, 0.05], [1.0, 0.7, 0.05]])
sort_by_weight = lambda t: t[torch.argsort(t[:, 0], stable=True)]
weight_forms_differ = not torch.equal(sort_by_weight(t_d), sort_by_weight(t_d[[1, 0]]))
center_forms_equal = torch.equal(t_d[torch.argsort(t_d[:, 1])], t_d[[1, 0]][torch.argsort(t_d[[1, 0]][:, 1])])
# generator separation rule on the stored training targets
gaps = (p[:, 1:, 1] - p[:, :-1, 1]).min(dim=1).values
rule = (gaps >= 3 * p[..., 2].max(dim=1).values - 1e-6)
# .unique: distinct values
print("max |f(pi theta) - f(theta)|:", max_df, "| can invariant:", bool(can_ok), "| increasing orderings per row:", counts.unique().tolist())
print("weight-sort forms differ:", weight_forms_differ, "| min center gap in train:", gaps.min().item(), "| rule holds:", bool(rule.all()))
```

### Answer check

```python
# Checks: the mixture-regression answers against pinned values
assert max_df < 1e-12
assert can_ok
assert counts.unique().tolist() == [1]
assert weight_forms_differ and center_forms_equal
assert bool(rule.all()) and gaps.min().item() >= 0.12 - 1e-6
```
