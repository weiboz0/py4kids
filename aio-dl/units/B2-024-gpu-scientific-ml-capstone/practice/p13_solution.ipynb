## Solution

Throughout, $A\in\mathbb R^{m\times n}$, $y\in\mathbb R^m$, $\lambda>0$, and $M=A^\top A+\lambda I\in\mathbb R^{n\times n}$. Note $M^\top=A^\top A+\lambda I=M$, so $M$ is symmetric.

**(a) Quadratic form and gradient.**
Using $\lVert z\rVert_2^2=z^\top z$ and expanding the product,
$$\lVert Au-y\rVert^2=(Au-y)^\top(Au-y)=u^\top A^\top Au-2y^\top Au+y^\top y,$$
because the two cross terms $u^\top A^\top y$ and $y^\top Au$ are equal scalars (a scalar equals its transpose). Adding $\lambda\lVert u\rVert^2=\lambda u^\top u=u^\top(\lambda I)u$,
$$J(u)=u^\top Mu-2(A^\top y)^\top u+y^\top y .$$
By the sum-of-squares gradient rule, $\nabla_u\lVert Au-y\rVert^2=2A^\top(Au-y)$, and $\nabla_u\lambda\lVert u\rVert^2=2\lambda u$ (the same rule with $A=I$, $y=0$, scaled by $\lambda$). The gradient is linear, so
$$\nabla J(u)=2A^\top(Au-y)+2\lambda u=2\big(Mu-A^\top y\big).$$

**(b) $M$ is invertible.** Let $v\in\mathbb R^n$, $v\neq0$. Then
$$v^\top Mv=v^\top A^\top Av+\lambda v^\top v=\lVert Av\rVert^2+\lambda\lVert v\rVert^2\ \ge\ \lambda\lVert v\rVert^2>0,$$
using $\lVert Av\rVert^2\ge0$, $\lambda>0$, and $\lVert v\rVert>0$ for $v\ne0$. If $Mv=0$ we would have $v^\top Mv=0$, a contradiction; so the null space of $M$ is $\{0\}$. A square $n\times n$ matrix with trivial null space has rank $n$ (rank–nullity), and a square matrix of full rank is invertible. No assumption on the rank of $A$ was used.

**(c) Unique stationary point.** By (a), $u$ is stationary iff $\nabla J(u)=0$ iff $Mu=A^\top y$. By (b), $M$ is invertible, so this linear system has exactly one solution, $u_\lambda=M^{-1}A^\top y$.

**(d) Unique global minimizer.** Write $b=A^\top y=Mu_\lambda$ (definition of $u_\lambda$). By (a), $J(u)=u^\top Mu-2b^\top u+y^\top y$. Then, for any $u$,
$$J(u)-J(u_\lambda)=u^\top Mu-2u_\lambda^\top Mu-u_\lambda^\top Mu_\lambda+2u_\lambda^\top Mu_\lambda
=u^\top Mu-2u_\lambda^\top Mu+u_\lambda^\top Mu_\lambda,$$
where $b^\top u=u_\lambda^\top M^\top u=u_\lambda^\top Mu$ uses the symmetry of $M$. Expanding $(u-u_\lambda)^\top M(u-u_\lambda)=u^\top Mu-u^\top Mu_\lambda-u_\lambda^\top Mu+u_\lambda^\top Mu_\lambda$ and using $u^\top Mu_\lambda=u_\lambda^\top Mu$ (symmetry again) gives exactly the same expression, so
$$J(u)-J(u_\lambda)=(u-u_\lambda)^\top M(u-u_\lambda).$$
By (b), the right side is $>0$ whenever $u-u_\lambda\ne0$ and $=0$ when $u=u_\lambda$. Hence $J(u)>J(u_\lambda)$ for every $u\ne u_\lambda$: $u_\lambda$ is the unique global minimizer.

**(e) Why it fails for $\lambda=0$ on the $8\times81$ grid operator.**
$A^\top A$ is $81\times81$ but $\operatorname{rank}(A^\top A)=\operatorname{rank}(A)\le 8$ (it is $8$ here, checked below), so $A^\top A$ is singular and step (b)'s bound $v^\top Mv\ge\lambda\lVert v\rVert^2$ collapses to $\lVert Av\rVert^2$, which is $0$ for every $v$ in the 73-dimensional null space of $A$.
The minimizers of $\lVert Au-y\rVert^2$ are then the solutions of the normal equations $A^\top Au=A^\top y$, which always exist but form an affine set $u^\ast+\operatorname{null}(A)$ of dimension $81-8=73$: infinitely many density maps fit the eight readings equally well.
So there is no unique minimizer for $\lambda=0$; the penalty $\lambda\lVert u\rVert^2$ is what selects one of them (for small $\lambda$, close to the minimum-norm one).

## Numeric sanity check

```python
import importlib.util
from pathlib import Path

import torch

SEED = 20261022
# Boilerplate: find and load this unit's data module
LOADER_CANDIDATES = (
    Path("../data/capstone_data.py"),
    Path("units/B2-024-gpu-scientific-ml-capstone/data/capstone_data.py"),
    Path("aio-dl/units/B2-024-gpu-scientific-ml-capstone/data/capstone_data.py"),
)
loader_path = next(path for path in LOADER_CANDIDATES if path.is_file())
loader_spec = importlib.util.spec_from_file_location("capstone_data", loader_path)
capstone_data = importlib.util.module_from_spec(loader_spec)
loader_spec.loader.exec_module(capstone_data)
torch.set_num_threads(1)

gen = torch.Generator().manual_seed(SEED)
m, n, lam = 3, 6, 0.3                       # wide A: rank(A) <= 3 < n, yet M is invertible
A = torch.randn(m, n, generator=gen, dtype=torch.float64)
y = torch.randn(m, generator=gen, dtype=torch.float64)
M = A.T @ A + lam * torch.eye(n, dtype=torch.float64)
u_lam = torch.linalg.solve(M, A.T @ y)

J = lambda u: ((A @ u - y) ** 2).sum() + lam * (u ** 2).sum()

# (a) gradient formula vs autograd
u = torch.randn(n, generator=gen, dtype=torch.float64, requires_grad=True)
J(u).backward()
grad_formula = 2 * (M @ u.detach() - A.T @ y)
# (b) smallest eigenvalue of M is at least lam
# torch.linalg.eigvalsh: symmetric eigenvalues
min_eig = torch.linalg.eigvalsh(M).min().item()
# (d) identity for several random u
gaps = []
for _ in range(5):
    v = torch.randn(n, generator=gen, dtype=torch.float64)
    gaps.append(((J(v) - J(u_lam)) - (v - u_lam) @ M @ (v - u_lam)).abs().item())
# (e) the grid operator
d2 = ((capstone_data.SENSORS_XY[:, None, :] - capstone_data.GRID_XY[None, :, :]) ** 2).sum(-1)
A_grid = 1.0 / (d2 + capstone_data.SENSOR_HEIGHT ** 2)
rank_grid = int(torch.linalg.matrix_rank(A_grid.T @ A_grid))
# a null-space direction leaves the data misfit unchanged
# torch.linalg.svd: rows of Vh
null_dir = torch.linalg.svd(A_grid).Vh[-1]
y_grid = capstone_data.val_rows("inverse")[0][0].double()
u_star = A_grid.T @ torch.linalg.solve(A_grid @ A_grid.T, y_grid)        # an exact fit (minimum-norm): A u* = y
misfit = lambda u: ((A_grid @ u - y_grid) ** 2).sum().item()
print("grad error", (u.grad - grad_formula).abs().max().item(), "| min eig", min_eig, "| identity gaps", max(gaps))
print("rank(A^T A) for the grid:", rank_grid, "| misfit at u*", misfit(u_star), "and at u* + 5 n:", misfit(u_star + 5 * null_dir))
```

### Answer check

```python
# Checks: the inverse-problem answers against pinned values
assert torch.allclose(u.grad, grad_formula, atol=1e-10, rtol=1e-05)
assert min_eig >= lam - 1e-12
assert max(gaps) < 1e-10
assert int(torch.linalg.matrix_rank(A)) == m < n          # rank-deficient A, invertible M
assert rank_grid == 8
assert abs(misfit(u_star) - misfit(u_star + 5 * null_dir)) < 1e-8
```
