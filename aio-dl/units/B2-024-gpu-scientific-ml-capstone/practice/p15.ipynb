## Task

Let $X_1,\dots,X_n$ be independent random variables with $E[X_i]=\mu$ and $\operatorname{Var}(X_i)=\sigma^2$, and $\bar X=\frac1n\sum_iX_i$.

(a) Prove $E[\bar X]=\mu$ and $\operatorname{Var}(\bar X)=\sigma^2/n$, citing linearity of expectation, the variance of a sum, independence, and $\operatorname{Var}(aY)=a^2\operatorname{Var}(Y)$ at the steps where each is used.

(b) Two training configurations $A$ and $B$ are each run with seeds $1,\dots,r$; run $s$ of $A$ and run $s$ of $B$ share seed $s$. Their scores are $(A_s,B_s)$, independent across $s$, with $\operatorname{Var}(A_s)=\operatorname{Var}(B_s)=\sigma^2$ and $\operatorname{Cov}(A_s,B_s)=\rho\sigma^2$ for some $\rho\in[-1,1]$. Let $D_s=A_s-B_s$ and $\bar D=\frac1r\sum_sD_s$. Prove $\operatorname{Var}(\bar D)=\frac{2\sigma^2(1-\rho)}{r}$.

(c) Compare with *unpaired* runs (different seeds, so $\rho=0$). Prove that pairing lowers the variance exactly when $\rho>0$, and compute the ratio of the two variances for $\rho=0.75$.

(d) With $\sigma=0.002$, $r=3$, and $\rho=0.75$, compute the standard deviation of $\bar D$ and decide whether an observed $\bar D=0.0015$ is larger than two standard deviations.

(e) In at most three sentences, explain which assumption of (a) fails when the same 200 validation examples are reused across seeds, and why that does not invalidate (b).
