## Task

Derive the optimal discriminator for a fixed generator. Reasoning is required.

Let $p_{\text{data}}$ and $p_g$ be densities on $\mathbb R^d$, where $p_g$ is the density of $G(z)$ for $z\sim N(0,I)$, and let the discriminator range over all functions $D:\mathbb R^d\to(0,1)$.

(a) Starting from $V(D,G)=E_{x\sim p_{\text{data}}}[\log D(x)]+E_{z}[\log(1-D(G(z)))]$, write $V$ as a single integral over $x$, and state exactly which fact justifies replacing the expectation over $z$ by an expectation over $x\sim p_g$.

(b) Fix a point $x$ with $a=p_{\text{data}}(x)>0$ and $b=p_g(x)>0$. Prove that $f(y)=a\log y+b\log(1-y)$ has the unique maximizer $y^*=\frac{a}{a+b}$ on $(0,1)$. Use single-variable calculus and state how you rule out a larger value near $y=0$ or $y=1$.

(c) Explain why choosing $D(x)=y^*(x)$ separately at every point maximizes the integral over the whole function class. State which property of the function class this pointwise argument needs.

(d) Treat the boundary cases $a>0=b$, $a=0<b$, and $a=b=0$: for each, give the supremum of $f$ on $(0,1)$, say whether it is attained, and state the convention for $D^*(x)$.

(e) For $p_{\text{data}}(x)=2x$ and $p_g(x)=1$ on $[0,1]$, give $D^*(x)$ in closed form, evaluate it at $x=\tfrac14$ and $x=1$, and say which case of (b) or (d) applies at $x=0$.
