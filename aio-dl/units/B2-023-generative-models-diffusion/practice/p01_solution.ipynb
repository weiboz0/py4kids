## Solution

**$p_g$ is a density.** $p_g(x)=x/8\ge0$ on $[0,4]$ and
$$\int_0^4\frac{x}{8}\,dx=\Big[\frac{x^2}{16}\Big]_0^4=\frac{16}{16}=1 .$$

**Both densities at $x=3$.** $p_{\text{data}}(3)=\tfrac14>0$ and $p_g(3)=\tfrac38>0$, so both are positive and the interior formula for $D^*$ applies (no boundary convention is needed).

**Substitute.**
$$D^*(3)=\frac{\tfrac14}{\tfrac14+\tfrac38}=\frac{\tfrac28}{\tfrac58}=\frac25 .$$

**Normal form and sign convention.** $\gcd(2,5)=1$, the denominator $b=5>0$, and the value is positive, so there is no minus sign to place: $a=2$, $b=5$.
Hence $a+b=7$.

**Answer: C.**

The distractors match the standard slips: A $=3$ is $\tfrac12$ (the confused discriminator, as if the densities were equal); B $=5$ is the ratio $p_{\text{data}}/p_g=\tfrac23$ (a ratio instead of a share); D $=8$ is the swapped-role value $p_g/(p_{\text{data}}+p_g)=\tfrac35$; E $=28$ is the unreduced $\tfrac{8}{20}$.

```python
from fractions import Fraction
from math import gcd

# Fraction: exact rational number
p_data = lambda x: Fraction(1, 4)
p_g = lambda x: Fraction(x, 8)

# p_g integrates to 1: the antiderivative of x/8 is x^2/16
integral = Fraction(4 ** 2, 16) - Fraction(0, 16)

a_dens, b_dens = p_data(3), p_g(3)
d_star = a_dens / (a_dens + b_dens)            # Fraction keeps lowest terms with b > 0
a, b = d_star.numerator, d_star.denominator
ANSWER = "C"

# what each distractor computes
confused = Fraction(1, 2)
ratio = a_dens / b_dens
swapped = b_dens / (a_dens + b_dens)
unreduced = (8, 20)
print(integral, a_dens, b_dens, d_star, a + b)
print(confused.numerator + confused.denominator, ratio.numerator + ratio.denominator,
      swapped.numerator + swapped.denominator, sum(unreduced))
```
