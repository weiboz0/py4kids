# Unit 5 solutions: matrix factorization -- latent factors, the logistic loss, and gradient descent

These reference solutions build the latent-factor path from the ground up, then reveal the shipped `bookrec.MatrixFactorizationPath` and score it honestly on the Unit 1 validation scoreboard. First the model itself: every reader and every book is a short vector of hidden **latent factors**, and the score for a pair is their plain **dot product**. Then the **logistic implicit-feedback loss** over the train positives plus complement-sampled negatives, **full-batch, per-entity-averaged gradient-descent** steps taken by hand on a tiny two-cluster toy log, and the loss falling as it trains. Only then do we fit `MatrixFactorizationPath` on the real log at the pinned config and score it against the random floor, popularity, the Unit-3 lexical path, and Unit 4's item-item CF -- where MF lands **on par with CF** (within noise over 500 readers), not beating it. The two Challenge exercises explore how many factors the model needs and prove that MF's score is *literally* an embedding dot product -- the bridge to Unit 8's neural two-tower. Every toy experiment is seeded with `SEED = 0` and the only random retrieval path is the seeded `RandomRetrievalPath` floor, so the notebook runs top-to-bottom deterministically, and every exercise ends with an `assert` on a computed result so a wrong number fails loudly instead of printing quietly. A reader's `seen` set is always their train-positive history; `val` is the sealed holdout we grade against.

```python
import json

import bookrec
import numpy as np
import pandas as pd

SEED = 0  # seed every toy experiment so the loss curve and the learned factors are reproducible

# --- the real interaction log (the same substrate as Units 1-4) ----------------------------
data_dir = bookrec.generated_dir()
catalog = bookrec.load_catalog(data_dir / "catalog.csv.gz")
catalog_ids = sorted(catalog)
interactions_path = data_dir / "interactions.csv.gz"
interactions = pd.read_csv(interactions_path)
rows = interactions.to_dict("records")
keywords = bookrec.load_keywords(data_dir / "keywords.csv.gz")

cold = json.loads((data_dir / "cold_partitions.json").read_text())
cold_readers = set(cold["cold_readers"])

# Each reader's train-positive history is their `seen` set (the exclusion set at retrieval).
train = interactions[interactions["split"] == "train"]
train_positive = train[train["label"] == 1]
seen_by = (
    train_positive.groupby("reader_id")["item_id"]
    .apply(lambda items: {int(i) for i in items})
    .to_dict()
)
col_of = {item_id: col for col, item_id in enumerate(catalog_ids)}

# The pinned matrix-factorization config, measured on this data. Do NOT shrink the epochs:
# below ~300 the model silently drops under the lexical path.
PINNED = {
    "n_factors": 32,
    "n_epochs": 300,
    "learning_rate": 0.5,
    "reg": 0.05,
    "n_negatives": 10,
    "seed": SEED,
}

# --- a tiny, hand-checkable toy log: two taste clusters ------------------------------------
# Readers 0,1,2 each read books 10,11,12; readers 3,4,5 each read books 20,21,22. Six readers,
# six books -- small enough to trace the factors and the loss by hand.
TOY_CATALOG = [10, 11, 12, 20, 21, 22]
TOY_PAIRS = [(r, b) for r in (0, 1, 2) for b in (10, 11, 12)]
TOY_PAIRS += [(r, b) for r in (3, 4, 5) for b in (20, 21, 22)]
TOY_ROWS = [
    {"reader_id": r, "item_id": b, "split": "train", "label": 1} for r, b in TOY_PAIRS
]
toy_book_col = {b: j for j, b in enumerate(TOY_CATALOG)}
n_readers_toy = 6
n_books_toy = len(TOY_CATALOG)


def sigmoid(z):
    """The logistic function 1 / (1 + e^-z): maps a raw score to a (0, 1) like-probability."""
    return 1.0 / (1.0 + np.exp(-z))


def accumulate(index, weighted, size):
    """Per-entity gradient sum (GIVEN helper): row e of the result is the sum of `weighted`
    over every training pair whose entity index is e -- a per-column np.bincount, the fast
    per-entity gradient path the shipped path uses."""
    out = np.zeros((size, weighted.shape[1]))
    for d in range(weighted.shape[1]):
        out[:, d] = np.bincount(index, weights=weighted[:, d], minlength=size)
    return out
```

## Exercise 1 - The latent-factor model: a score is a dot product of hidden taste factors

Matrix factorization describes every reader and every book by a short vector of hidden **latent factors** -- a handful of "taste dials". Reader $u$ is a row $P_u$ of the reader-factor matrix $P$ (readers x factors); book $i$ is a row $Q_i$ of the item-factor matrix $Q$ (books x factors). The model's raw score for the pair is their **dot product** $P_u \cdot Q_i$, and the predicted like-probability is $\sigma(P_u \cdot Q_i)$ -- a large positive dot product means the reader's taste dials line up with the book's, so a likely read.

Below are tiny hand-made factors for 2 readers and 3 books, with 2 latent factors each. Form the full score matrix $P Q^{\top}$ (readers x books) with a single matrix product, squash it to like-probabilities with `sigmoid`, and read back reader 0's highest-scoring book. In two or three sentences, say what a factor "means" here (it is learned, not labelled) and why the score being a plain dot product is the fact this whole unit -- and Unit 8's two-tower -- rests on.

```python
P_demo = np.array([[1.5, -0.4], [-1.1, 1.3]])             # 2 readers x 2 latent factors
Q_demo = np.array([[1.4, -0.3], [-0.9, 1.2], [0.2, 0.1]])  # 3 books x 2 latent factors

score_matrix = P_demo @ Q_demo.T       # (readers x books); entry [u, i] = P_u . Q_i
prob_matrix = sigmoid(score_matrix)    # predicted like-probability in (0, 1)
reader0_scores = score_matrix[0]       # reader 0's raw score for each of the 3 books
reader0_top_book = int(np.argsort(reader0_scores)[-1])  # index of reader 0's best book
factors_explanation = (
    "A latent factor is a LEARNED, unlabelled coordinate of taste: nobody tells the model what "
    "factor 0 or factor 1 stands for -- training just arranges each reader and each book in the "
    "same low-dimensional space so that readers sit near the books they engage. The score for a "
    "(reader, book) pair is nothing more than the dot product of their two factor vectors, which is "
    "large exactly when the vectors point the same way (aligned taste). That plain dot product is "
    "the whole bet of this unit and of Unit 8's two-tower: learn a vector per reader and per book, "
    "and ranking becomes a dot product between embeddings."
)

print("score matrix (readers x books):")
print(score_matrix)
print("like-probabilities:")
print(np.round(prob_matrix, 4))
print(f"reader 0 scores: {np.round(reader0_scores, 4)} -> top book index {reader0_top_book}")

# entry [u, i] must equal the dot product of row u of P and row i of Q
assert np.allclose(score_matrix, P_demo @ Q_demo.T)
assert np.allclose(prob_matrix, sigmoid(score_matrix))
assert ((prob_matrix > 0.0) & (prob_matrix < 1.0)).all(), "like-probabilities live strictly in (0, 1)"
assert reader0_top_book == 0, "reader 0's own-taste book (index 0) wins the dot product"
assert np.isclose(score_matrix[0, 0], 1.5 * 1.4 + (-0.4) * (-0.3)), "the score is literally P_0 . Q_0"
```

## Exercise 2 - Implicit feedback: positives, sampled negatives, and the logistic loss

Our feedback is **implicit**: a train positive (`label == 1`) says a reader engaged a book, but there are no negative *ratings*. To learn, we pair each observed positive with `n_negatives` **sampled negatives** drawn from that reader's **unobserved complement** -- random books the reader has *not* read (resampling any draw that happens to hit one of their own positives). These complement negatives, **not** the log's `label == 0` exposure rows (which carry taste signal and would wreck training -- the lesson measures both), are the negatives that work.

With labels $y \in \{0, 1\}$ and model score $s = P_u \cdot Q_i$, the **logistic implicit-feedback loss** of one pair is $-[\,y \log \sigma(s) + (1 - y)\log(1 - \sigma(s))\,]$, averaged over all pairs; driving it down pushes $\sigma$ toward $1$ on positives and toward $0$ on sampled negatives. On the toy log, build the positive index arrays, sample `n_negatives = 10` complement negatives **per positive**, stack the positive-then-negative `u_idx` / `i_idx` / `y` arrays, and write `mean_logistic_loss(P, Q, u_idx, i_idx, y)`. Report the per-positive negative count and the loss of a fresh random $P, Q$ (expect roughly $\log 2 \approx 0.69$ at tiny random init, since $\sigma(0) = 0.5$).

```python
rng = np.random.default_rng(SEED)
n_negatives = 10

# Positive pairs as entity-index arrays (a toy reader id already is its index; map books to cols).
pos_u = np.array([r for r, b in TOY_PAIRS])
pos_i = np.array([toy_book_col[b] for r, b in TOY_PAIRS])
n_pos = len(TOY_PAIRS)

# Sample n_negatives complement negatives PER positive: repeat each positive reader n_negatives
# times, draw random books, and resample any (reader, book) that collides with an observed pair.
observed = {(r, toy_book_col[b]) for r, b in TOY_PAIRS}   # observed (reader_index, book_column) pairs
neg_u = np.repeat(pos_u, n_negatives)
neg_i = rng.integers(0, n_books_toy, size=n_pos * n_negatives)
pending = [j for j in range(neg_i.shape[0]) if (int(neg_u[j]), int(neg_i[j])) in observed]
while pending:
    neg_i[pending] = rng.integers(0, n_books_toy, size=len(pending))
    pending = [j for j in pending if (int(neg_u[j]), int(neg_i[j])) in observed]

u_idx = np.concatenate([pos_u, neg_u])
i_idx = np.concatenate([pos_i, neg_i])
y = np.concatenate([np.ones(n_pos), np.zeros(n_pos * n_negatives)])


def mean_logistic_loss(P, Q, u_idx, i_idx, y):
    scores = (P[u_idx] * Q[i_idx]).sum(axis=1)
    p = sigmoid(scores)
    return float(np.mean(-(y * np.log(p) + (1 - y) * np.log(1 - p))))


P_init = rng.normal(0.0, 0.1, size=(n_readers_toy, 2))   # 2 latent factors for the toy run
Q_init = rng.normal(0.0, 0.1, size=(n_books_toy, 2))
neg_per_positive = n_negatives
loss_at_init = mean_logistic_loss(P_init, Q_init, u_idx, i_idx, y)

print(f"positives: {n_pos}   negatives per positive: {neg_per_positive}   total pairs: {len(u_idx)}")
print(f"loss at tiny random init: {loss_at_init:.4f}   (log 2 = {np.log(2):.4f}, since sigmoid(0)=0.5)")

assert len(u_idx) == n_pos * (1 + n_negatives), "positives + n_negatives negatives per positive"
assert int(y.sum()) == n_pos, "exactly the positives carry label 1"
assert all((int(neg_u[j]), int(neg_i[j])) not in observed for j in range(len(neg_i))), (
    "every sampled negative is from the reader's unobserved complement"
)
assert abs(loss_at_init - np.log(2)) < 0.05, "tiny random init scores ~0, so the loss is near log 2"
```

## Exercise 3 - One full-batch, per-entity-averaged gradient-descent step

To *fit* the factors we descend the logistic loss. For a pair with score $s = P_u \cdot Q_i$ the gradient of its loss with respect to $s$ is $\sigma(s) - y$; by the chain rule the pair's contribution to $\nabla_{P_u}$ is $(\sigma(s) - y)\,Q_i$ and to $\nabla_{Q_i}$ is $(\sigma(s) - y)\,P_u$. We take a **full-batch** step: sum every pair's contribution per entity, divide each entity's sum by its **own interaction count** (so a reader with many pairs is not pushed harder than a quiet one -- the *per-entity averaging*), add the L2 penalty $\mathrm{reg}\cdot P$, and step **downhill**. Writing $e = y - \sigma(s)$ (the negative of the score gradient), the update reads $P \mathrel{+}= \eta\,(\mathrm{grad}_P / \mathrm{count}_u - \mathrm{reg}\cdot P)$.

Re-initialize $P, Q$ at the same seed, compute the per-entity gradients with the given `accumulate` helper, divide by each entity's interaction count (`np.bincount`, floored at 1), take one step, and confirm the loss **after** the step is below the loss before it. Record both losses and the bool that the step reduced the loss.

```python
learning_rate = PINNED["learning_rate"]
reg = PINNED["reg"]

rng_step = np.random.default_rng(SEED)
P = rng_step.normal(0.0, 0.1, size=(n_readers_toy, 2))
Q = rng_step.normal(0.0, 0.1, size=(n_books_toy, 2))

loss_before = mean_logistic_loss(P, Q, u_idx, i_idx, y)

# Per-entity interaction counts (floored at 1 so an unseen entity never divides by zero).
count_u = np.maximum(np.bincount(u_idx, minlength=n_readers_toy), 1)[:, None]
count_i = np.maximum(np.bincount(i_idx, minlength=n_books_toy), 1)[:, None]

scores = (P[u_idx] * Q[i_idx]).sum(axis=1)
err = y - sigmoid(scores)                       # the negative score-gradient per pair (e = y - sigma)
grad_p = accumulate(u_idx, err[:, None] * Q[i_idx], n_readers_toy)
grad_q = accumulate(i_idx, err[:, None] * P[u_idx], n_books_toy)
P += learning_rate * (grad_p / count_u - reg * P)   # one downhill step, L2-regularized, in place
Q += learning_rate * (grad_q / count_i - reg * Q)

loss_after = mean_logistic_loss(P, Q, u_idx, i_idx, y)
step_reduced_loss = bool(loss_after < loss_before)

print(f"loss before step: {loss_before:.6f}")
print(f"loss after  step: {loss_after:.6f}")
print(f"one GD step reduced the loss: {step_reduced_loss}")

assert step_reduced_loss, "one full-batch per-entity-averaged step must reduce the logistic loss"
assert loss_after < loss_before
```

## Exercise 4 - Train the toy factors and watch the loss fall

One step helps; **gradient descent** repeats it. Re-initialize $P, Q$ and run the exact step from Exercise 3 for `n_epochs` epochs (use `PINNED["n_epochs"]`), recording the loss every 30 epochs into a list. The toy log has two clean taste clusters, so the loss should fall steeply and then flatten as $\sigma(P_u \cdot Q_i)$ saturates toward $1$ on the positives. Record the loss curve, confirm the final loss is well below the first recorded loss, and -- as the honest read -- confirm the curve is (weakly) **monotone non-increasing** from one recorded point to the next. In one or two sentences, relate `reg` to this curve: why a nonzero L2 penalty keeps the factors from growing without bound even as the loss falls.

```python
n_epochs = PINNED["n_epochs"]

rng_fit = np.random.default_rng(SEED)
P = rng_fit.normal(0.0, 0.1, size=(n_readers_toy, 2))
Q = rng_fit.normal(0.0, 0.1, size=(n_books_toy, 2))

count_u = np.maximum(np.bincount(u_idx, minlength=n_readers_toy), 1)[:, None]
count_i = np.maximum(np.bincount(i_idx, minlength=n_books_toy), 1)[:, None]

loss_curve = []
for epoch in range(n_epochs):
    scores = (P[u_idx] * Q[i_idx]).sum(axis=1)
    err = y - sigmoid(scores)
    grad_p = accumulate(u_idx, err[:, None] * Q[i_idx], n_readers_toy)
    grad_q = accumulate(i_idx, err[:, None] * P[u_idx], n_books_toy)
    P += learning_rate * (grad_p / count_u - reg * P)
    Q += learning_rate * (grad_q / count_i - reg * Q)
    if epoch % 30 == 0:
        loss_curve.append(mean_logistic_loss(P, Q, u_idx, i_idx, y))

first_recorded_loss = loss_curve[0]
final_loss = loss_curve[-1]
loss_fell = bool(final_loss < first_recorded_loss)
is_monotone = bool(all(loss_curve[t] <= loss_curve[t - 1] + 1e-12 for t in range(1, len(loss_curve))))
reg_explanation = (
    "The L2 penalty subtracts reg * P (and reg * Q) on every step, a constant pull back toward the "
    "origin that grows with the size of the factors; so as the data term keeps pushing the positive "
    "scores up, the penalty balances it and the factor norms settle at a finite size instead of "
    "running away to make sigmoid exactly 1. That bounded-norm solution is what keeps the model from "
    "memorizing the toy log and lets the same training generalize on the real, sparse log."
)

print("loss every 30 epochs:")
print([round(c, 4) for c in loss_curve])
print(f"first recorded loss: {first_recorded_loss:.4f}   final loss: {final_loss:.4f}")
print(f"loss fell: {loss_fell}   monotone non-increasing: {is_monotone}")

assert loss_fell, "the final recorded loss must be well below the first"
assert final_loss < 0.5 * first_recorded_loss, "the clean two-cluster toy trains to a far lower loss"
assert is_monotone, "full-batch GD drives the recorded loss down monotonically on this toy"
```

## Exercise 5 - Reveal `MatrixFactorizationPath`: fit the real log, and the retrieve contract

The package ships exactly this training as a standard retrieval path. `bookrec.MatrixFactorizationPath(**PINNED)` learns $P, Q$ from the **train positives** of the real log by full-batch, per-entity-averaged gradient descent -- `fit(interactions, catalog=None)` duck-types the row-mapping log (no pandas inside the package) and takes `catalog` as the item universe, so **every** catalog book gets a $Q$ row (MF can *score* the whole catalog). `retrieve(query, context, k)` scores a fitted reader's whole catalog by $P[\text{reader}] \cdot Q^{\top}$, excludes `context["seen"]`, and returns the top $k$.

Fit it on `rows` with `catalog=catalog_ids` at the pinned config. Pick a known reader (the first of `mf.artifact()["reader_ids"]`), retrieve its top 5 given its real `seen` history, and retrieve again with an **empty** `seen`. Confirm the known reader still gets 5 recs with empty `seen` -- MF builds its query from the learned factor, not from `seen` (which only *excludes*). Then retrieve for an **unknown** reader id and confirm the answer is `[]`. In two sentences, contrast the two cold cases: a cold *reader* (no learned factor) versus a cold *book* (gets a $Q$ row but only negative-shaped signal).

```python
mf = bookrec.MatrixFactorizationPath(**PINNED).fit(rows, catalog=catalog_ids)
reader_id = int(mf.artifact()["reader_ids"][0])   # a reader with a learned factor
seen = seen_by[reader_id]                          # that reader's train-positive history

top5 = mf.retrieve(reader_id, {"seen": seen}, k=5)
top5_ids = [c.item_id for c in top5]

empty_seen = mf.retrieve(reader_id, {"seen": set()}, k=5)
known_reader_still_recommends = bool(len(empty_seen) == 5)   # empty seen still returns k recs

unknown = mf.retrieve(10_000_000, {"seen": set()}, k=5)
unknown_is_empty = bool(unknown == [])
cold_cases_explanation = (
    "A cold READER has zero train positives, so MF never learns a P row for them and retrieve returns "
    "[] -- there is simply no taste vector to dot against the catalog. A cold BOOK is different: it "
    "still gets an initialized Q row (the item universe is the whole catalog), but with no positive "
    "signal it is only ever touched as a sampled negative, so the gradient pushes its score down and "
    "it ranks near the bottom -- MF can score it but cannot meaningfully recommend it, which is the "
    "feature-based cold-start work of Unit 9."
)

print(f"known reader {reader_id}: top-5 item ids with real seen = {top5_ids}")
print(f"same reader, empty seen -> {len(empty_seen)} recs (query comes from the factor, not seen)")
print(f"unknown reader -> {unknown}  (no learned factor)")

assert len(top5) == 5, "a fitted reader gets k recs"
assert all(i not in seen for i in top5_ids), "retrieve excludes the reader's seen books"
assert known_reader_still_recommends, "empty seen excludes nothing, so a known reader still gets k"
assert unknown_is_empty, "an unknown / cold reader has no factor -> []"
```

## Exercise 6 - Register the paths and read the validation scoreboard honestly

Line MF up against the earlier units under the Unit 1 rules unchanged. Build a `bookrec.RandomRetrievalPath(catalog_ids, seed=SEED)` floor, fit a `bookrec.PopularityRetrievalPath()`, a `bookrec.LexicalRetrievalPath(keywords)`, and an item-item `bookrec.ItemItemRetrievalPath()` on `rows` with `catalog=catalog_ids`, and register all of them plus the `mf` from Exercise 5 in a fresh `bookrec.PathRegistry`. Score each with `bookrec.run_validation_scoreboard(path, interactions_path, catalog_ids=catalog_ids, k=10, cold_readers=cold_readers)` and record the `hit_rate_at_k`, plus MF's ratios over popularity and the random floor.

```python
random_floor = bookrec.RandomRetrievalPath(catalog_ids, seed=SEED)
popularity = bookrec.PopularityRetrievalPath().fit(rows, catalog=catalog_ids)
lexical = bookrec.LexicalRetrievalPath(keywords).fit(rows, catalog=catalog_ids)
cf = bookrec.ItemItemRetrievalPath().fit(rows, catalog=catalog_ids)

registry = bookrec.PathRegistry()
for path in (random_floor, popularity, lexical, cf, mf):
    registry.register(path)


def score(path):
    return bookrec.run_validation_scoreboard(
        path, interactions_path, catalog_ids=catalog_ids, k=10, cold_readers=cold_readers
    ).hit_rate_at_k


mf_hit = score(mf)
cf_hit = score(cf)
lexical_hit = score(lexical)
popularity_hit = score(popularity)
floor_hit = score(random_floor)
mf_vs_popularity = mf_hit / popularity_hit         # expect ~2.5
mf_vs_floor = mf_hit / floor_hit                   # expect ~23
scoreboard_reading = (
    "MF lands ON PAR with item-item CF on this data -- about 0.25-0.28 hit@10 across seeds (0.276 here) against CF's ~0.25, a "
    "gap well inside the sampling noise over 500 readers, so this is NOT evidence that MF beats CF; a "
    "0.24 draw on another seed would sit comfortably in the same band. What the ratios DO establish "
    "is that MF reads real collaborative signal: it clears popularity by ~2.5x and the random floor by "
    "~23x, and edges past the Unit-3 lexical content path -- the latent factors are the generalization "
    "of co-occurrence, the same who-read-what signal CF memorizes, compressed into compact embeddings. "
    "They do NOT make MF a strong recommender in absolute terms (0.25 hit@10 still misses most readers) "
    "and say nothing about the cold catalog MF can score but cannot meaningfully rank."
)

print(f"{'path':>14} {'hit@10':>8} {'vs floor':>9}")
for label, hit in (("random", floor_hit), ("popularity", popularity_hit),
                   ("lexical", lexical_hit), ("item-item CF", cf_hit), ("MF (this unit)", mf_hit)):
    print(f"{label:>14} {hit:>8.4f} {hit / floor_hit:>8.1f}x")
print(f"\nMF vs popularity {mf_vs_popularity:.2f}x   vs floor {mf_vs_floor:.1f}x")

assert mf_hit >= cf_hit - 0.03, "MF is on par with CF (within noise), not reliably below it"
assert mf_hit > lexical_hit, "MF beats the lexical content path"
assert mf_hit >= 1.2 * popularity_hit, "MF clears popularity by a clear margin"
assert mf_hit >= popularity_hit + 0.03, "MF clears popularity by an absolute margin too"
assert floor_hit < 0.05, "the random floor is far below every learned path"
```

## Exercise 7 - Challenge: how many latent factors? (capacity and the exec budget)

`n_factors` is the model's **capacity** -- the number of taste dials. Too few and the factors cannot separate the tastes (underfitting); the shipped default is 32. Sweep a **small grid** of `n_factors` values and watch the val hit@10. To respect the per-cell exec budget (each fit at the pinned 300 epochs is ~16-21 s, and six fits in one cell would brush the 120 s cap), keep the sweep to **at most 4 fits in this cell** -- here three values. Fit `MatrixFactorizationPath` at each `n_factors` (every other setting pinned, `seed=SEED`), score each on `val` with your Exercise-6 `score` helper, and record the `{n_factors: hit}` map. Expect very small factor counts to underfit and 32 to be at or near the best. In one or two sentences, connect factor count to the under/overfitting trade-off (and why L2 `reg` matters more as capacity grows).

```python
# At most 4 fits in this cell (budget): three factor counts, everything else pinned.
factor_grid = [2, 8, 32]
hit_by_factors = {}
for n in factor_grid:
    path = bookrec.MatrixFactorizationPath(
        n_factors=n,
        n_epochs=PINNED["n_epochs"],
        learning_rate=PINNED["learning_rate"],
        reg=PINNED["reg"],
        n_negatives=PINNED["n_negatives"],
        seed=SEED,
    ).fit(rows, catalog=catalog_ids)
    hit_by_factors[n] = score(path)

smallest_underfits = bool(hit_by_factors[2] < hit_by_factors[32])
capacity_explanation = (
    "With only 2 factors the model has too little capacity to lay the many distinct tastes out in "
    "separate directions, so it underfits and scores below 32 factors. More factors add capacity to "
    "fit finer structure, but past a point the extra dimensions start memorizing the sparse log rather "
    "than generalizing -- which is why the L2 penalty matters MORE as n_factors grows: it is the brake "
    "that keeps the larger factor space from overfitting."
)

print(f"{'n_factors':>10} {'val hit@10':>12}")
for n in factor_grid:
    print(f"{n:>10} {hit_by_factors[n]:>12.4f}")
print(f"smallest (2) underfits vs 32: {smallest_underfits}")

assert smallest_underfits, "2 latent factors underfit relative to the shipped 32"
assert np.isclose(hit_by_factors[32], mf_hit), "n_factors=32 at seed 0 reproduces the Exercise-5 fit"
```

## Exercise 8 - Challenge: the score IS an embedding dot product (the bridge to Unit 8)

The hinge of Part 1. MF's learned factors are **embeddings**: a reader embedding $P_u$, a book embedding $Q_i$, and the recommendation score is nothing but their **dot product** -- the exact shape Unit 8's neural two-tower will learn (with a network in place of gradient descent on a table). Prove it. Take the fitted `mf` from Exercise 5, pull its `reader_factors` ($P$) and `item_factors` ($Q$), and for the Exercise-5 `reader_id` compute the by-hand score vector $P[\text{row}] \cdot Q^{\top}$ over the whole catalog yourself. Confirm the score `mf.retrieve` reports for each returned book equals your by-hand dot product to floating-point precision -- the retriever is *literally* ranking embedding dot products.

Then compare recommenders: retrieve the same reader's top 5 from the Exercise-6 `cf` (item-item CF) and count how many of MF's top-5 books also appear in CF's top-5.
