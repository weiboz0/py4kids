## Question 1

**Notice:** parse the grid, sum each column with nested loops, track the leftmost column with the greatest sum, and print its 1-based index and that sum.

```python
import sys

data = sys.stdin.read()
parts = data.split()
rows = int(parts[0])
columns = int(parts[1])
grid = []
position = 2
row_number = 0
while row_number < rows:
    row = []
    column_number = 0
    while column_number < columns:
        row.append(int(parts[position]))
        position = position + 1
        column_number = column_number + 1
    grid.append(row)
    row_number = row_number + 1

column_sums = []
column_number = 0
while column_number < columns:
    total = 0
    row_number = 0
    while row_number < rows:
        total = total + grid[row_number][column_number]
        row_number = row_number + 1
    column_sums.append(total)
    column_number = column_number + 1

best_column = 0
column_number = 1
while column_number < columns:
    if column_sums[column_number] > column_sums[best_column]:
        best_column = column_number
    column_number = column_number + 1
print(str(best_column + 1) + " " + str(column_sums[best_column]))
```

## Question 2

**Notice:** read the five 0/1 flags and print `OPEN` when not blocked and (registered or guest) and (badge or coach), otherwise `CLOSED`.

```python
import sys

data = sys.stdin.read()
parts = data.split()
registered = parts[0] == "1"
guest = parts[1] == "1"
badge = parts[2] == "1"
blocked = parts[3] == "1"
coach = parts[4] == "1"
is_open = not blocked and (registered or guest) and (badge or coach)
if is_open:
    print("OPEN")
else:
    print("CLOSED")
```

## Question 3

**Notice:** one pass keeping the smallest price seen so far and the best `price - smallest` gain; print the best gain.

```python
import sys

data = sys.stdin.read()
parts = data.split()
n = int(parts[0])
prices = []
i = 0
while i < n:
    prices.append(int(parts[i + 1]))
    i = i + 1

smallest = prices[0]
best_gain = 0
i = 1
while i < n:
    gain = prices[i] - smallest
    if gain > best_gain:
        best_gain = gain
    if prices[i] < smallest:
        smallest = prices[i]
    i = i + 1
print(str(best_gain))
```

## Question 4

**Notice:** store each pick once as a `(picker, picked)` tuple in a set, which also drops repeated picks. A pick is half of a mutual pair exactly when its reversed tuple `(picked, picker)` is in the set too; keep only the half with `picker < picked`, so each pair is recorded once with the alphabetically earlier name first. Sort the pairs with the named key `pair_key` (first name, then second name) and print the count followed by one pair per line; with no pairs, only `0` is printed. Each set lookup takes about the same time however many picks there are, so `100000` picks need one pass plus one sort instead of comparing every pick with every other pick.

```python
import sys


def pair_key(pair):
    return (pair[0], pair[1])


data = sys.stdin.read()
tokens = data.split()
n = int(tokens[0])
picks = set()
order = []
i = 0
while i < n:
    picker = tokens[1 + 2 * i]
    picked = tokens[2 + 2 * i]
    if (picker, picked) not in picks:
        picks.add((picker, picked))
        order.append((picker, picked))
    i = i + 1

pairs = []
i = 0
while i < len(order):
    picker = order[i][0]
    picked = order[i][1]
    if picker < picked and (picked, picker) in picks:
        pairs.append((picker, picked))
    i = i + 1

ranked = sorted(pairs, key=pair_key)
print(str(len(ranked)))
i = 0
while i < len(ranked):
    print(ranked[i][0] + " " + ranked[i][1])
    i = i + 1
```

## Question 5

**Notice:** sort the stations once. For each house, binary-search for `lo`, the first station at or after the house (the first position whose station is not smaller than the house). The nearest station is either `stations[lo]` (when `lo < n`) or the one just before it, `stations[lo - 1]` (when `lo > 0`); every other station is farther away than one of these two. Compare the two distances that exist and print the smaller. Sorting costs about `N log N` steps and each house needs only about `17` halvings, so `100000` houses stay fast where checking every station for every house would take ten billion steps.

```python
import sys

data = sys.stdin.read()
tokens = data.split()
n = int(tokens[0])
q = int(tokens[1])
stations = []
i = 0
while i < n:
    stations.append(int(tokens[2 + i]))
    i = i + 1
stations.sort()

i = 0
while i < q:
    house = int(tokens[2 + n + i])
    lo = 0
    hi = n
    while lo < hi:
        mid = (lo + hi) // 2
        if stations[mid] < house:
            lo = mid + 1
        else:
            hi = mid
    best = -1
    if lo < n:
        best = stations[lo] - house
    if lo > 0:
        gap = house - stations[lo - 1]
        if best == -1 or gap < best:
            best = gap
    print(str(best))
    i = i + 1
```

## Question 6

**Notice:** three nested loops choose positions `i < j < k` in increasing order, so every pick is examined exactly once and the values stay in list order. A pick counts when `values[j] - values[i] == values[k] - values[j]`; a negative step passes this test too, so decreasing picks count. With `N` at most `150` there are fewer than `600000` triples, which is quick without any cleverness.

```python
import sys

data = sys.stdin.read()
tokens = data.split()
n = int(tokens[0])
values = []
i = 0
while i < n:
    values.append(int(tokens[1 + i]))
    i = i + 1

count = 0
i = 0
while i < n:
    j = i + 1
    while j < n:
        k = j + 1
        while k < n:
            if values[j] - values[i] == values[k] - values[j]:
                count = count + 1
            k = k + 1
        j = j + 1
    i = i + 1
print(str(count))
```
