## Question 1

### Snack Sizes

The first line contains `N M`, the second line contains the smallest snack size each of `N` kids will accept, and the third line contains the sizes of `M` snacks. Print the maximum number of happy kids. Sort both lists, then sweep the snacks from smallest to largest, giving each snack to the least demanding kid still waiting when it is big enough and skipping it otherwise. In the sample, the snack of size `2` is too small for every kid still waiting and is skipped, and the final snack `8` is the only one large enough for the kid who needs `7`, raising the answer from `3` to `4`.

### Constraints

`1 <= N, M <= 100000`.

```text
4 5
3 1 7 5
2 4 1 6 8
```

```text
4
```
