## Question 4

### Biggest Rain Trough

A row of `N` fence posts stands at positions `0` through `N - 1`, and post `i` has height `h[i]`. Any two
different posts `i < j` form a trough that holds `min(h[i], h[j]) * (j - i)` units of rainwater. Print the
most water any trough can hold.

Use converging two pointers: start `lo` at the first post and `hi` at the last, record the current trough,
then move the pointer at the SHORTER post one step inward (move `hi` when the two heights are equal). Stop
when the pointers meet. In the sample, the final post is one wall of the best trough; without it the answer
would be only `18`.

First line: `N`. Then `N` non-negative integer heights (they may span several lines). Print the largest
amount of water as an integer string.

### Constraints

`2 <= N <= 200000`, and every height is between `0` and `1000000000`. Checking every pair of posts will not
finish in time.

### Sample Input

```text
7
2 8 3 1 6 4 7
```

### Sample Output

```text
35
```

## Question 5

### Longest Sunny Stretch

A weather log lists `N` days in order, each `1` (sunny) or `0` (rainy). A stretch of consecutive days is
allowed to contain at most `K` rainy days. Find the length of the LONGEST allowed stretch.

Use a sliding window: grow `right` one day at a time while counting the rainy days inside the window;
whenever that count is larger than `K`, shrink from `left` (lowering the count when a rainy day leaves)
until the window is allowed again. In the sample, the window must shrink past the first rainy day, and the
unique longest stretch ends on the final day.

First line: `N K`. Then `N` values, each `0` or `1`. Print the longest length (0 if no single day is
allowed) as an integer string.

### Constraints

`1 <= N <= 200000`, `0 <= K <= N`. Rechecking every stretch will not finish in time.

### Sample Input

```text
10 2
1 0 1 1 0 0 1 1 1 1
```

### Sample Output

```text
8
```

## Question 6

### Giant Power Odometer

A `D`-digit odometer shows only the last `D` digits of a number, with leading zeros. Given a base `A`, an
exponent `E`, and a digit count `D`, print what the odometer shows for `A` to the power `E` (any base to the
power `0` is `1`): the power modulo `10^D`, written with exactly `D` digits.

Use repeated squaring with modulus `M = 10^D` and reduce modulo `M` after every multiplication and every
squaring; computing the unreduced power first is infeasible for the enormous exponent. **Do not use a
built-in modular-power shortcut** (no three-argument `pow`). Your algorithm must use O(log E) modular
multiplications. Finally, put `0` characters in front of the result until it has exactly `D` digits. In the
sample, the power modulo `100000` is `3289`, so one leading zero fills the fifth digit.

Input: `A E D` on one line. Print the `D`-character result.

### Constraints

`0 <= A <= 10^18`, `0 <= E <= 10^18`, `1 <= D <= 18`.

### Sample Input

```text
3 123456789012345678 5
```

### Sample Output

```text
03289
```

### Trace the Traversal

Trace this exact breadth-first routine and predict what it prints. Neighbours are visited in the order the
edges are given.

```text
seed visited with the single start node          # visited = {start}
put start on the back of a FIFO queue
while the queue is not empty:
    take a node u from the FRONT of the queue
    print u
    for each out-neighbour v of u, in input order:
        if v is not visited:
            mark v visited and put it on the BACK of the queue
```

Input first line: `N M start` (nodes `1..N`, `M` directed edges, and the BFS start node). Then `M` lines,
each `u v`, a directed edge `u -> v`; a node's out-neighbours are in input order. Print the printed node
ids in the order the routine prints them, space-separated.

### Constraints

`1 <= N <= 30`, `0 <= M <= 200`.

### Sample Input

```text
3 2 1
1 2
1 3
```

### Sample Output

```text
1 2 3
```
