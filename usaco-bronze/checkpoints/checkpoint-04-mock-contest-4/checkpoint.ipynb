### Pasture Lookup

## Question 6

### Chain Letter Total

On day `1`, one letter is sent, even when `A` is `0`. On every later day, each letter sent the day before causes `A` new letters, so day `k` sends `A` to the power `k - 1` letters. Given `A`, a number of days `E`, and a modulus `M`, print the total number of letters sent during the first `E` days modulo `M`: the remainder of `1 + A + A^2 + ... + A^(E - 1)` after dividing by `M` (the total is `0` when `E` is `0`).

Input: `A E M` on one line. Print the total modulo `M`.

`E` can be as large as `10^18`, so adding one day at a time is infeasible, and because `M` need not be prime you cannot divide by `A - 1`. Instead, use repeated squaring on a pair of values reduced modulo `M`: `power` for `A^e` and `total` for the letters of the first `e` days. Doubling the days turns the pair into `total + power * total` and `power * power`; adding one more day turns it into `total + power` and `power * A`. Write the bits of `E` from the most significant bit down, start from `e = 0` (`power = 1`, `total = 0`), double for every bit, and then add one day when that bit is `1`, reducing modulo `M` after every multiplication. **Do not use a built-in modular-power shortcut** (no three-argument `pow`). Your algorithm must use O(log E) modular multiplications. In the sample, the six days send `1 + 3 + 9 + 27 + 81 + 243 = 364` letters, which is `64` modulo `100`.

### Constraints

`0 <= A <= 10^18`, `0 <= E <= 10^18`, `1 <= M <= 10^18`.

### Sample Input

```text
3 6 100
```

### Sample Output

```text
64
```
