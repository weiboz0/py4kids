## Exercise 1

**Notice:** gate access via `(member or guest) and not banned`.

```python
import sys

data = sys.stdin.read()
values = data.split()
member = int(values[0]) == 1
guest = int(values[1]) == 1
banned = int(values[2]) == 1
if (member or guest) and not banned:
    print("GRANTED")
else:
    print("DENIED")
```

## Exercise 2

**Notice:** `not (a and not b)` — an asymmetric rule (swapping operands changes the answer).

```python
import sys

data = sys.stdin.read()
values = data.split()
a = int(values[0]) == 1
b = int(values[1]) == 1
result = not (a and not b)
if result:
    print("TRUE")
else:
    print("FALSE")
```

## Exercise 3

**Notice:** any-true over N flags (`or` across the list).

```python
import sys

data = sys.stdin.read()
values = data.split()
n = int(values[0])
found_one = False
for i in range(n):
    if int(values[i + 1]) == 1:
        found_one = True
if found_one:
    print("YES")
else:
    print("NO")
```

## Exercise 4

**Notice:** all-true over N flags (`and` across the list).

```python
import sys

data = sys.stdin.read()
values = data.split()
n = int(values[0])
every_check_passed = True
for i in range(n):
    if int(values[i + 1]) == 0:
        every_check_passed = False
if every_check_passed:
    print("YES")
else:
    print("NO")
```

## Exercise 5

**Notice:** strict majority of `X` with a dissenter.

```python
import sys

data = sys.stdin.read()
values = data.split()
n = int(values[0])
x_count = 0
has_dissenter = False
for i in range(n):
    if values[i + 1] == "X":
        x_count = x_count + 1
    else:
        has_dissenter = True
if x_count * 2 > n and has_dissenter:
    print("YES")
else:
    print("NO")
```

## Exercise 6

**Notice:** quorum AND strict-majority YES.

```python
import sys

data = sys.stdin.read()
values = data.split()
n = int(values[0])
quorum = int(values[1])
yes_count = 0
for i in range(n):
    if values[i + 2] == "YES":
        yes_count = yes_count + 1
if n >= quorum and yes_count * 2 > n:
    print("PASS")
else:
    print("FAIL")
```

## Exercise 7

**Notice:** a positive non-SKIP value exists (short-circuit `!= "SKIP"` before `int`).

```python
import sys

data = sys.stdin.read()
values = data.split()
n = int(values[0])
found_positive = False
for i in range(n):
    token = values[i + 1]
    if token != "SKIP" and int(token) > 0:
        found_positive = True
if found_positive:
    print("YES")
else:
    print("NO")
```

## Exercise 8

**Notice:** multi-rule access with De Morgan-style grouping.

```python
import sys

data = sys.stdin.read()
values = data.split()
student = int(values[0]) == 1
mentor = int(values[1]) == 1
badge = int(values[2]) == 1
banned = int(values[3]) == 1
emergency = int(values[4]) == 1
if ((not banned) and ((student or mentor) and badge)) or (emergency and mentor):
    print("OPEN")
else:
    print("CLOSED")
```

## Exercise 9

**Notice:** evaluate one of several named boolean rules.

```python
import sys

data = sys.stdin.read()
values = data.split()
rule = values[0]
a = int(values[1]) == 1
b = int(values[2]) == 1
if rule == "A_AND_NOT_B":
    result = a and not b
elif rule == "NOT_A_OR_B":
    result = (not a) or b
elif rule == "A_OR_B":
    result = a or b
else:
    result = a and b
if result:
    print("TRUE")
else:
    print("FALSE")
```
