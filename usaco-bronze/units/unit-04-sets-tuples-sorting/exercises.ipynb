# Unit 4 — Sets, Tuples & Sorting Exercises

Each problem is a program that **reads the whole input from stdin** and **prints the exact output** (real contest style — no `solve()` wrapper): `import sys; data = sys.stdin.read()`, parse it, then print. Write each program in a `.py` file and run it with piped input, e.g. `python my_solution.py < case.txt`.

## Exercise 1

### Distinct Badge Count

The first line is `N`.
The second line contains `N` badge colors.
Print the number of distinct colors.
Use a set so repeated colors count only once.
In the sample, the last color is a new distinct color and must be counted.

### Constraints

`1 <= N <= 100000`.
Each color is one word.
A repeated scan for every color will not finish in time.

### Sample Input

```text
7
red blue red green blue yellow purple
```

### Sample Output

```text
5
```

## Exercise 2

### Students in Both Clubs

The first line contains `N M`.
The second line contains the `N` distinct names in the robotics club.
The third line contains the `M` distinct names in the art club.
Print how many students are in both clubs.
Build the intersection with membership checks and `.add()`.
In the sample, the final art-club name creates the third match.

### Constraints

`1 <= N, M <= 100000`.
Each name is one word, and names do not repeat within one club.
Comparing every robotics name with every art name will not finish in time.

### Sample Input

```text
4 4
Ava Ben Cy Dia
Zoe Ava Cy Dia
```

### Sample Output

```text
3
```

## Exercise 5

### Final Leaderboard

The first line is `N`.
The next `N` lines each contain a one-word name and an integer score.
Print all contestants, one per line as `name score`, sorted by score from highest to lowest and then by name alphabetically when scores tie.
Store each contestant as a tuple and call `sorted` with a named key function that returns `(-score, name)`.
The sample's lowest-ranked contestant is both the final input record and the final record after sorting, so do not drop it.

### Constraints

`1 <= N <= 100000`.
Names are distinct one-word strings, and `0 <= score <= 1000000`.
Repeatedly searching for the next contestant will not finish in time.

### Sample Input

```text
5
Mia 90
Zoe 100
Ava 100
Noah 80
Liam 70
```

### Sample Output

```text
Ava 100
Zoe 100
Mia 90
Noah 80
Liam 70
```

## Exercise 6

### Last Player in the Top K

The first line contains `N K`.
The next `N` lines each contain a distinct one-word name and an integer score.
Rank players by score from highest to lowest and then by name alphabetically when scores tie.
Print the `name score` record at rank `K`, where the best player has rank 1.
Use tuples and `sorted` with a named compound key function.
In the sample, the final input player changes the top K, and the answer is the last record in the retained top K after sorting.

### Constraints

`1 <= K <= N <= 100000`.
`0 <= score <= 1000000`.
Repeatedly searching for the next-highest score will not finish in time.

### Sample Input

```text
6 4
Ana 95
Bo 88
Cy 91
Di 88
Eli 70
Fay 90
```

### Sample Output

```text
Bo 88
```

## Exercise 9
Challenge (stretch)

### Allowed Finalists

The first line contains `B N K`.
The second line contains `B` distinct banned names.
The next `N` lines each contain a one-word name and an integer score, and a name may appear more than once.
Ignore every record with a banned name.
For each other name, keep only that contestant's highest score.
Rank the remaining distinct contestants by score from highest to lowest and then by name alphabetically when scores tie.
Print the first `K` names, one per line.
Use set membership, tuple records, and `sorted` with a named compound key function.
In the sample, the final record changes the finalists, and the Kth finalist is the last retained record after sorting.

### Constraints

`1 <= B <= 100000`.
`1 <= K <= N <= 100000`.
After banned names and repeats are handled, at least `K` distinct allowed contestants remain.
`0 <= score <= 1000000`.
Repeatedly rescanning records to find each finalist will not finish in time.

### Sample Input

```text
2 8 3
Kai Uma
Ava 50
Ben 80
Kai 100
Ava 90
Cora 85
Ben 70
Dax 40
Dax 88
```

### Sample Output

```text
Ava
Dax
Cora
```
