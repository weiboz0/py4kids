## Question 1

### Prime Test

````python
def is_prime(n):
    if n < 2:
        return False
    divisor = 2
    while divisor < n:
        if n % divisor == 0:
            return False
        divisor = divisor + 1
    return True

assert is_prime(0) == False
assert is_prime(1) == False
assert is_prime(9) == False
assert is_prime(19) == True
````

**The real program** (reads a number):


## Question 2

### Parameterized Travel Cost

````python
def travel_cost(km_text, rate_text):
    km = float(km_text)
    rate = float(rate_text)
    base = 3.5
    return round(base + km * rate, 2)

cost = travel_cost("8", "1.25")
print(f"Trip cost: ${cost}")
assert travel_cost("8", "1.25") == 13.5
assert travel_cost("15", "0.8") == 15.5
assert travel_cost("0", "2.5") == 3.5
````

**The real program** (reads distance and rate text):

```python

    km = float(km_text)
    rate = float(rate_text)
    base = 3.5
    return round(base + km * rate, 2)

km_text = input()
rate_text = input()
print(f"Trip cost: ${travel_cost(km_text, rate_text)}")
```

Sample input:

```text
8
1.25
```

Expected output:

```text

```

## Question 3

### Running-Total Function

````python
def sum_to_n(n):
    total = 0
    for number in range(1, n + 1):
        total = total + number
    return total

assert sum_to_n(3) == 6
assert sum_to_n(7) == 28
assert sum_to_n(20) == 210
````

## Question 4

### Count Primes

````python
def count_primes(n):
    total = 0
    for candidate in range(2, n + 1):
        if is_prime(candidate):
            total = total + 1
    return total

assert count_primes(5) == 3
assert count_primes(10) == 4
assert count_primes(20) == 8
````

## Question 5

### Award Ladder

````python
def award_level(points):
    if points >= 90:
        return "Gold"
    elif points >= 70:
        return "Silver"
    elif points >= 50:
        return "Bronze"
    else:
        return "Keep practicing"

assert award_level(93) == "Gold"
assert award_level(74) == "Silver"
assert award_level(58) == "Bronze"
assert award_level(30) == "Keep practicing"
````

## Question 6

### Number-Tool Score

````python
def balanced_score(a, b, c):
    high = max(max(a, b), c)
    low = min(a, b, c)
    spread = abs(high - low)
    bonus = sum(range(1, 4))
    return round((a + b + c + bonus - spread) / 3, 1)

assert balanced_score(8, 12, 10) == 10.7
assert balanced_score(20, 5, 11) == 9.0
assert balanced_score(7, 7, 7) == 9.0
````

**The real program** (reads three numbers):

## Question 7

### Seeded High-Roll Count

````python
def count_high_rolls(trials):
    random.seed(4)
    total = 0
    for trial in range(trials):
        roll = random.randint(1, 6)
        if roll >= 5:
            total = total + 1
    return total

assert count_high_rolls(1) == 0
assert count_high_rolls(5) == 1
assert count_high_rolls(12) == 2
assert count_high_rolls(30) == 5
````

**The real program** (reads a trial count):

