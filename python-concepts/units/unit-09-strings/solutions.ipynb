## Exercise 1

### Badge Line


````python
# Return two initials, a code slice, and the code length.
def badge_line(first, last, code):
    initials = first[0] + last[0]
    piece = code[2:5]
    return initials + " | " + piece + " | " + str(len(code))

assert badge_line("Mira", "Stone", "AX742Q") == "MS | 742 | 6"
assert badge_line("Leo", "Park", "ZZcat9") == "LP | cat | 6"
assert badge_line("Nia", "Young", "12SUN") == "NY | SUN | 5"
print(badge_line("Mira", "Stone", "AX742Q"))
````

## Exercise 3

### Palindrome Gate


````python
# Report whether the cleaned text reads the same from both ends.
def is_palindrome(text):
    cleaned = text.lower().replace(" ", "")
    for i in range(len(cleaned) // 2):
        if cleaned[i] != cleaned[len(cleaned) - 1 - i]:
            return False
    return True

assert is_palindrome("Taco cat") is True
assert is_palindrome("python") is False
assert is_palindrome("Level") is True
assert is_palindrome("a") is True
assert is_palindrome("") is True
print(is_palindrome("Taco cat"))
````

````python
# Return the matching character category.
def classify_char(ch):
    lowered = ch.lower()
    if lowered in "aeiou":
        return "vowel"
    elif lowered in "abcdefghijklmnopqrstuvwxyz":
        return "consonant"
    else:
        return "other"

assert classify_char("U") == "vowel"
assert classify_char("m") == "consonant"
assert classify_char("7") == "other"
assert classify_char("?") == "other"
print(classify_char("U"))
````

## Exercise 5

### Mask the Vowels


````python
# Return a new string with each vowel replaced by an asterisk.
def mask_vowels(text):
    masked = ""
    for ch in text:
        if ch.lower() in "aeiou":
            masked = masked + "*"
        else:
            masked = masked + ch
    return masked

assert mask_vowels("Rocket") == "R*ck*t"
assert mask_vowels("SKY") == "SKY"
assert mask_vowels("A-OK!") == "*-*K!"
assert mask_vowels("") == ""
print(mask_vowels("Rocket"))
````

### Alphabet Scan Report


````python
# Return how many times ch appears in text.
def count_char(text, ch):
    total = 0
    for current in text:
        if current == ch:
            total = total + 1
    return total

# Print one count row for every lowercase alphabet letter.
def print_alphabet_scan(text):
    cleaned = text.lower()
    print("letter | total")
    for letter in "abcdefghijklmnopqrstuvwxyz":
        total = count_char(cleaned, letter)
        print(f"{letter} | {total}")

assert count_char("pepper", "p") == 3
assert count_char("pepper", "z") == 0
assert count_char("Mississippi", "s") == 4
assert count_char("", "a") == 0
print_alphabet_scan("Bee")
````

## Exercise 13

### More Practice: Run-Length Encoding


````python
def rle(text):
    if text == "":
        return ""
    result = ""
    current = text[0]
    count = 0
    for ch in text:
        if ch == current:
            count = count + 1
        else:
            result = result + current + str(count)
            current = ch
            count = 1
    return result + current + str(count)

assert rle("aaabcc") == "a3b1c2"
assert rle("") == ""
assert rle("zz") == "z2"
print(rle("aaabcc"))
````

### More Practice: Atbash Cipher


````python
def atbash(text):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    result = ""
    for ch in text:
        if ch in alphabet:
            result = result + alphabet[25 - alphabet.find(ch)]
        else:
            result = result + ch
    return result

assert atbash("abc xyz!") == "zyx cba!"
assert atbash("Abc!") == "Ayx!"
assert atbash("z") == "a"
print(atbash("abc xyz!"))
````

````python
def center_text(text, width):
    left = (width - len(text)) // 2
    right = width - len(text) - left
    return " " * left + text + " " * right

assert center_text("gnat", 10) == "   gnat   "
assert center_text("gnat", 9) == "  gnat   "
assert center_text("A", 1) == "A"
print(center_text("gnat", 10))
````

````python
# Return text with case chosen by each character's position.
def alternating_case(text):
    result = ""
    for i in range(len(text)):
        if i % 2 == 0:
            result = result + text[i].upper()
        else:
            result = result + text[i].lower()
    return result

assert alternating_case("python") == "PyThOn"
assert alternating_case("moon base") == "MoOn bAsE"
assert alternating_case("A1b2!") == "A1B2!"
assert alternating_case("") == ""
print(alternating_case("python"))
````

## Exercise 27 — Challenge

### Challenge 2: Caesar Shift


````python
# Return the first matching alphabet position, or -1.
def position(alphabet, ch):
    for i in range(len(alphabet)):
        if alphabet[i] == ch:
            return i
    return -1

# Return a lowercase message shifted forward by k positions.
def shift(message, k):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    cleaned = message.lower()
    shifted = ""
    for ch in cleaned:
        pos = position(alphabet, ch)
        if pos == -1:
            shifted = shifted + ch
        else:
            new_pos = (pos + k) % 26
            shifted = shifted + alphabet[new_pos]
    return shifted

assert position("abcdefghijklmnopqrstuvwxyz", "q") == 16
assert position("abcdefghijklmnopqrstuvwxyz", "a") == 0
assert position("abcdefghijklmnopqrstuvwxyz", "!") == -1
assert shift("Code 9!", 2) == "eqfg 9!"
assert shift("XYZ", 3) == "abc"
assert shift("stay", 0) == "stay"
assert shift("zebra", 1) == "afcsb"
assert shift("Hi, Z!", 27) == "ij, a!"
print(shift("Code 9!", 2))
````

## Exercise 28 — Challenge

### Challenge 3: Word Wrap


````python
def wrap(line, width):
    result = ""
    current = ""
    for word in line.split():
        if current == "":
            current = word
        elif len(current) + 1 + len(word) <= width:
            current = current + " " + word
        else:
            if result != "":
                result = result + "\n"
            result = result + current
            current = word
    if current != "":
        if result != "":
            result = result + "\n"
        result = result + current
    return result

assert wrap("the quick brown fox jumps", 10) == "the quick\nbrown fox\njumps"
assert wrap("elephant cat", 3) == "elephant\ncat"
assert wrap("", 10) == ""
print(wrap("the quick brown fox jumps", 10))
````
