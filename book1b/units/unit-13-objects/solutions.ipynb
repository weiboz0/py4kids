## Exercise 1

### Two Separate Points


```python
# Store a map position in one object.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

start = Point(2, 6)
finish = Point(9, 4)
assert start.x == 2
assert start.y == 6
assert finish.x == 9
assert finish.y == 4
start.x = 5
assert start.x == 5
assert finish.x == 9
```

## Exercise 2

### Rectangle Record


```python
# Keep a sign order's code and dimensions together.
class Rectangle:
    def __init__(self, code, width, height):
        self.code = code
        self.width = width
        self.height = height

small = Rectangle("N-12", 6, 4)
large = Rectangle("S-30", 12, 7)
assert small.code == "N-12"
assert small.width == 6
assert small.height == 4
assert large.code == "S-30"
assert large.width == 12
assert large.height == 7
```

## Exercise 3

### Rectangle Measurements


```python
# Calculate measurements from this rectangle's dimensions.
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * self.width + 2 * self.height

tile = Rectangle(9, 4)
assert tile.width == 9
assert tile.height == 4
assert tile.area() == 36
assert tile.perimeter() == 26
assert Rectangle(2.5, 6).area() == 15.0
assert Rectangle(1, 8).perimeter() == 18
```

## Exercise 4

### Distance Between Points


```python
# Measure the straight-line distance to another point.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx * dx + dy * dy) ** 0.5

origin = Point(0, 0)
three_four = Point(3, 4)
assert origin.x == 0
assert three_four.y == 4
assert origin.distance(three_four) == 5.0
assert Point(2, 1).distance(Point(2, 6)) == 5.0
assert Point(1, 1).distance(Point(1, 1)) == 0.0
```

## Exercise 5

### Short Code and Description


```python
# Report a rectangle's region and full description.
class Rectangle:
    def __init__(self, code, width, height):
        self.code = code
        self.width = width
        self.height = height

    def short_code(self):
        return self.code[0:2]

    def describe(self):
        return f"{self.code}: {self.width} by {self.height}"

crate = Rectangle("NE-48", 7, 3)
assert crate.code == "NE-48"
assert crate.width == 7
assert crate.short_code() == "NE"
assert crate.describe() == "NE-48: 7 by 3"
assert Rectangle("SW-9", 2, 2).short_code() == "SW"
assert Rectangle("C-1", 5, 1).describe() == "C-1: 5 by 1"
```

## Exercise 6

### Score Counter


```python
# Track one labeled score that can change.
class Counter:
    def __init__(self, label):
        self.label = label
        self.count = 0

    def increment(self, amount):
        self.count = self.count + amount
        return self.count

blue = Counter("blue")
gold = Counter("gold")
assert blue.label == "blue"
assert gold.count == 0
assert blue.increment(3) == 3
assert blue.count == 3
assert blue.increment(2) == 5
assert blue.count == 5
assert gold.count == 0
```

## Exercise 7

### Save and Load a Point


```python
# Save this point and rebuild a new point from its file.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def save(self, path):
        with open(path, "w") as f:
            f.write(f"{self.x}\n")
            f.write(f"{self.y}\n")
        return path

def load_point(path):
    with open(path, "r") as f:
        x = int(f.readline().strip())
        y = int(f.readline().strip())
    return Point(x, y)

original = Point(14, 9)
assert original.x == 14
assert original.save("ex7_point.txt") == "ex7_point.txt"
loaded = load_point("ex7_point.txt")
assert loaded.x == 14
assert loaded.y == 9
second = Point(-5, 12)
assert second.save("ex7_point.txt") == "ex7_point.txt"
reloaded = load_point("ex7_point.txt")
assert reloaded.x == -5
assert reloaded.y == 12
```

## Exercise 8

### Challenge 1: Rectangle Size Band


```python
# Label a rectangle from the area computed by its method.
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def size_band(self):
        area = self.area()
        if area >= 60:
            return "large"
        elif area >= 25:
            return "medium"
        else:
            return "small"

large = Rectangle(10, 7)
assert large.width == 10
assert large.area() == 70
assert large.size_band() == "large"
assert Rectangle(7, 5).size_band() == "medium"
assert Rectangle(4, 3).size_band() == "small"
assert Rectangle(5, 5).size_band() == "medium"
```

## Exercise 9

### Challenge 2: Counter Snapshot


```python
# Save a counter's current state and rebuild it later.
class Counter:
    def __init__(self, label, count):
        self.label = label
        self.count = count

    def increment(self, amount):
        self.count = self.count + amount
        return self.count

    def save(self, path):
        with open(path, "w") as f:
            f.write(f"{self.label}\n")
            f.write(f"{self.count}\n")
        return path

def load_counter(path):
    with open(path, "r") as f:
        label = f.readline().strip()
        count = int(f.readline().strip())
    return Counter(label, count)

visitors = Counter("north gate", 12)
assert visitors.label == "north gate"
assert visitors.increment(4) == 16
assert visitors.count == 16
assert visitors.save("ex9_counter.txt") == "ex9_counter.txt"
restored = load_counter("ex9_counter.txt")
assert restored.label == "north gate"
assert restored.count == 16
quiet = Counter("east door", 0)
assert quiet.save("ex9_counter.txt") == "ex9_counter.txt"
restored_quiet = load_counter("ex9_counter.txt")
assert restored_quiet.label == "east door"
assert restored_quiet.count == 0
```
