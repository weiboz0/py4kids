````python
# Calculate measurements from this rectangle's dimensions.
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * self.width + 2 * self.height

tile = Rectangle(9, 4)
assert tile.width == 9
assert tile.height == 4
assert tile.area() == 36
assert tile.perimeter() == 26
assert Rectangle(2.5, 6).area() == 15.0
assert Rectangle(1, 8).perimeter() == 18

print(tile.area())
print(tile.perimeter())
print(Rectangle(2.5, 6).area())
````

## Exercise 4

### Distance Between Points


````python
# Measure the straight-line distance to another point.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx * dx + dy * dy) ** 0.5

here = Point(1, 2)
there = Point(4, 6)
assert here.x == 1
assert there.y == 6
assert here.distance(there) == 5.0
assert Point(2, 1).distance(Point(2, 6)) == 5.0
assert Point(1, 1).distance(Point(1, 1)) == 0.0

print(here.distance(there))
print(Point(2, 1).distance(Point(2, 6)))
print(Point(1, 1).distance(Point(1, 1)))
````

## Exercise 7

### Save and Load a Point


````python
# Save this point and rebuild a new point from its file.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def save(self, path):
        with open(path, "w") as f:
            f.write(f"{self.x}\n")
            f.write(f"{self.y}\n")
        return path

def load_point(path):
    with open(path, "r") as f:
        x = int(f.readline().strip())
        y = int(f.readline().strip())
    return Point(x, y)

original = Point(14, 9)
assert original.x == 14
assert original.save("ex7_point.txt") == "ex7_point.txt"
loaded = load_point("ex7_point.txt")
assert loaded.x == 14
assert loaded.y == 9
second = Point(-5, 12)
assert second.save("ex7_point.txt") == "ex7_point.txt"
reloaded = load_point("ex7_point.txt")
assert reloaded.x == -5
assert reloaded.y == 12

sample_point = Point(14, 9)
print(sample_point.save("ex7_point.txt"))
sample_loaded = load_point("ex7_point.txt")
print(sample_loaded.x)
print(sample_loaded.y)
````

### Draw a Rectangle


````python
# Draw a sign as one string made from rows.
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def draw(self):
        rows = []
        for row in range(self.height):
            rows.append("#" * self.width)
        return "\n".join(rows)

assert Rectangle(4, 2).draw() == "####\n####"
assert Rectangle(2, 3).draw() == "##\n##\n##"
assert Rectangle(1, 1).draw() == "#"
print(Rectangle(4, 2).draw())
````

### Bank Account


````python
# Keep an account owner's money in one changing object.
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds"
        self.balance = self.balance - amount
        return self.balance

account = BankAccount("Idris", 50)
assert account.owner == "Idris"
assert account.deposit(30) == 80
assert account.withdraw(100) == "Insufficient funds"
assert account.balance == 80
assert account.withdraw(25) == 55
other = BankAccount("Mara", 0)
assert other.deposit(7) == 7
assert other.withdraw(7) == 0
sample = BankAccount("Idris", 50)
print(sample.deposit(30))
print(sample.withdraw(100))
print(sample.balance)
print(sample.withdraw(25))
````

### Traffic Light


````python
# Cycle one light through its three states.
class TrafficLight:
    def __init__(self):
        self.state = "green"

    def next(self):
        if self.state == "green":
            self.state = "yellow"
        elif self.state == "yellow":
            self.state = "red"
        else:
            self.state = "green"
        return self.state

light = TrafficLight()
assert light.state == "green"
assert light.next() == "yellow"
assert light.next() == "red"
assert light.next() == "green"
assert light.next() == "yellow"
other = TrafficLight()
assert other.next() == "yellow"
sample = TrafficLight()
print(sample.state)
print(sample.next())
print(sample.next())
print(sample.next())
````

````python
class VendingMachine:
    def __init__(self, price):
        self.price = price
        self.credit = 0

    def insert(self, amount):
        self.credit = self.credit + amount
        if self.credit < self.price:
            return f"Insert {self.price - self.credit} more"
        else:
            change = self.credit - self.price
            self.credit = 0
            return f"Vend! Change: {change}"

machine = VendingMachine(3)
assert machine.insert(1) == "Insert 2 more"
assert machine.insert(1) == "Insert 1 more"
assert machine.insert(2) == "Vend! Change: 1"
assert machine.credit == 0
assert machine.insert(3) == "Vend! Change: 0"
other = VendingMachine(5)
assert other.insert(2) == "Insert 3 more"
sample = VendingMachine(3)
print(sample.insert(1))
print(sample.insert(1))
print(sample.insert(2))
print(sample.credit)
````

````python
class Stockroom:
    def __init__(self):
        self.stock = {}

    def add(self, name, amount):
        if name in self.stock:
            self.stock[name] = self.stock[name] + amount
        else:
            self.stock[name] = amount

    def remove(self, name, amount):
        self.stock[name] = self.stock[name] - amount

room = Stockroom()
room.add("rivet", 10)
room.add("washer", 4)
room.remove("rivet", 3)
assert room.stock == {"rivet": 7, "washer": 4}
room.add("rivet", 2)
assert room.stock["rivet"] == 9
other = Stockroom()
other.add("bolt", 1)
assert other.stock == {"bolt": 1}
sample = Stockroom()
sample.add("rivet", 10)
sample.add("washer", 4)
sample.remove("rivet", 3)
print(sample.stock)
````

````python
class GameBoard:
    def __init__(self, size):
        self.cells = []
        for row_number in range(size):
            row = []
            for col_number in range(size):
                row.append(".")
            self.cells.append(row)

    def place(self, row, col, mark):
        self.cells[row][col] = mark

    def render(self):
        rows = []
        for row in self.cells:
            rows.append("".join(row))
        return "\n".join(rows)

board = GameBoard(3)
assert board.render() == "...\n...\n..."
board.place(0, 0, "X")
board.place(2, 2, "O")
assert board.render() == "X..\n...\n..O"
small = GameBoard(1)
small.place(0, 0, "Z")
assert small.render() == "Z"
print(board.render())
````

### More Practice: Total Area


````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

def total_area(rectangles):
    total = 0
    for rectangle in rectangles:
        total = total + rectangle.area()
    return total

assert total_area([Rectangle(2, 3), Rectangle(4, 5)]) == 26
assert total_area([Rectangle(1, 7)]) == 7
assert total_area([]) == 0
print(total_area([Rectangle(2, 3), Rectangle(4, 5)]))
````

## Exercise 17

### More Practice: Closest Point


````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx * dx + dy * dy) ** 0.5

def closest(origin, points):
    best = points[0]
    best_distance = origin.distance(best)
    for point in points:
        distance = origin.distance(point)
        if distance < best_distance:
            best = point
            best_distance = distance
    return best

points = [Point(3, 4), Point(1, 1), Point(6, 8)]
best = closest(Point(0, 0), points)
assert best is points[1]
assert Point(0, 0).distance(Point(3, 4)) == 5.0
other = [Point(2, 0), Point(-2, 0)]
assert closest(Point(0, 0), other) is other[0]
print(f"{best.x} {best.y}")
````

## Exercise 21

### Challenge 1: Rectangle Size Band


````python
# Label a rectangle from the area computed by its method.
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def size_band(self):
        area = self.area()
        if area >= 60:
            return "large"
        elif area >= 25:
            return "medium"
        else:
            return "small"

large = Rectangle(10, 7)
assert large.width == 10
assert large.area() == 70
assert large.size_band() == "large"
assert Rectangle(7, 5).size_band() == "medium"
assert Rectangle(4, 3).size_band() == "small"
assert Rectangle(5, 5).size_band() == "medium"

print(Rectangle(10, 7).size_band())
print(Rectangle(7, 5).size_band())
print(Rectangle(4, 3).size_band())
````

## Exercise 22

### Challenge 2: Counter Snapshot


````python
# Save a counter's current state and rebuild it later.
class Counter:
    def __init__(self, label, count):
        self.label = label
        self.count = count

    def increment(self, amount):
        self.count = self.count + amount
        return self.count

    def save(self, path):
        with open(path, "w") as f:
            f.write(f"{self.label}\n")
            f.write(f"{self.count}\n")
        return path

def load_counter(path):
    with open(path, "r") as f:
        label = f.readline().strip()
        count = int(f.readline().strip())
    return Counter(label, count)

visitors = Counter("north gate", 12)
assert visitors.label == "north gate"
assert visitors.increment(4) == 16
assert visitors.count == 16
assert visitors.save("ex22_counter.txt") == "ex22_counter.txt"
restored = load_counter("ex22_counter.txt")
assert restored.label == "north gate"
assert restored.count == 16
quiet = Counter("east door", 0)
assert quiet.save("ex22_counter.txt") == "ex22_counter.txt"
restored_quiet = load_counter("ex22_counter.txt")
assert restored_quiet.label == "east door"
assert restored_quiet.count == 0

sample_counter = Counter("north gate", 12)
print(sample_counter.increment(4))
print(sample_counter.save("ex22_counter.txt"))
sample_restored = load_counter("ex22_counter.txt")
print(sample_restored.label)
print(sample_restored.count)
````

## Exercise 23

### Challenge 3: Rover Commands


````python
class Rover:
    def __init__(self):
        self.x = 0
        self.y = 0
        self.facing = "N"

    def run(self, commands):
        for command in commands:
            if command == "R":
                self.facing = "NESW"[("NESW".find(self.facing) + 1) % 4]
            elif command == "L":
                self.facing = "NESW"[("NESW".find(self.facing) - 1) % 4]
            else:
                if self.facing == "N":
                    self.y = self.y + 1
                elif self.facing == "E":
                    self.x = self.x + 1
                elif self.facing == "S":
                    self.y = self.y - 1
                else:
                    self.x = self.x - 1

    def position(self):
        return f"({self.x}, {self.y}) facing {self.facing}"

rover = Rover()
rover.run("FFRFF")
assert rover.position() == "(2, 2) facing E"
other = Rover()
other.run("LFLF")
assert other.position() == "(-1, -1) facing S"
wrap = Rover()
wrap.run("LFR")
assert wrap.position() == "(-1, 0) facing N"
print(rover.position())
````

## Exercise 24

### Challenge 4: Save the Stockroom


````python
class Stockroom:
    def __init__(self):
        self.stock = {}

    def add(self, name, amount):
        if name in self.stock:
            self.stock[name] = self.stock[name] + amount
        else:
            self.stock[name] = amount

    def save(self, path):
        with open(path, "w") as f:
            for name in self.stock:
                f.write(f"{name},{self.stock[name]}\n")
        return path

def load_stockroom(path):
    room = Stockroom()
    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            room.add(parts[0], int(parts[1]))
    return room

room = Stockroom()
room.add("rivet", 10)
room.add("washer", 4)
room.stock["rivet"] = 7
assert room.save("ex24_stockroom.txt") == "ex24_stockroom.txt"
loaded = load_stockroom("ex24_stockroom.txt")
assert loaded.stock == {"rivet": 7, "washer": 4}
second = Stockroom()
second.add("bolt", 2)
assert second.save("ex24_stockroom.txt") == "ex24_stockroom.txt"
assert load_stockroom("ex24_stockroom.txt").stock == {"bolt": 2}
sample = Stockroom()
sample.add("rivet", 10)
sample.add("washer", 4)
sample.stock["rivet"] = 7
print(sample.save("ex24_stockroom.txt"))
print(load_stockroom("ex24_stockroom.txt").stock)
````
