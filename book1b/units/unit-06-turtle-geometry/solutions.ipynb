## Exercise 1

**The real program (`assets/solutions_ex1.py`)**

```python
# Bring in Python's turtle drawing tools.
import turtle

# Set up the four-sided courtyard outline.
n = 4
side = 62
angle = 360 / n
turtle.pencolor("royalblue")

# Each trip traces one courtyard edge and aims the turtle toward the next corner.
for corner in range(n):
    turtle.forward(side)
    turtle.left(angle)

turtle.done()
```

````python
n = 4
side = 62
angle = 360 / n
total_turn = n * angle
pen_down_moves = n
ends_where_started = abs(total_turn - 360) < 1e-6

assert isinstance(n, int)
assert isinstance(side, int)
assert isinstance(pen_down_moves, int)
assert isinstance(ends_where_started, bool)
assert pen_down_moves == 4
assert angle == 90.0
assert abs(total_turn - 360) < 1e-6
assert ends_where_started is True
````

## Exercise 6

**The real program (`assets/solutions_ex6.py`)**

```python
# Bring in Python's turtle drawing tools.
import turtle

# Set up one hexagon and the nine-shape wheel.
n = 6
side = 41
angle = 360 / n
shape_count = 9
turn_between_shapes = 360 / shape_count
turtle.pencolor("turquoise")

# The outer loop repeats whole hexagons around the wheel.
for shape in range(shape_count):
    # The inner loop finishes all six sides of one hexagon.
    for corner in range(n):
        turtle.forward(side)
        turtle.left(angle)
    turtle.left(turn_between_shapes)

turtle.done()
```

````python
n = 6
side = 41
angle = 360 / n
total_turn = n * angle
shape_count = 9
turn_between_shapes = 360 / shape_count
ring_turn = shape_count * turn_between_shapes
pen_down_moves = shape_count * n
ends_where_started = (
    abs(total_turn - 360) < 1e-6 and abs(ring_turn - 360) < 1e-6
)

assert isinstance(n, int)
assert isinstance(side, int)
assert isinstance(shape_count, int)
assert isinstance(pen_down_moves, int)
assert isinstance(ends_where_started, bool)
assert pen_down_moves == 54
assert turn_between_shapes == 40.0
assert abs(total_turn - 360) < 1e-6
assert abs(ring_turn - 360) < 1e-6
assert ends_where_started is True
````

## Exercise 9: The Eight-Degree Gap

**The real program (`assets/solutions_ex9.py`)**

```python
# Bring in Python's turtle drawing tools.
import turtle

# Use the honest decimal angle for a closed eleven-sided comparison shape.
n = 11
side = 34
angle = 360 / n
turtle.pencolor("magenta")

# Eleven equal decimal turns complete the regular hendecagon.
for corner in range(n):
    turtle.forward(side)
    turtle.left(angle)

turtle.done()
```

The correct program uses true division, so its angle is `360 / 11`, about 32.727 degrees, and the shape closes. The shortcut gives `360 // 11 == 32`, and `11 * 32 == 352`, so it leaves an 8-degree gap. On screen the gap is small, but the turtle's direction is off and it does not return to its starting point. The real tell is that the shape does not close.

````python
n = 11
side = 34
angle = 360 / n
total_turn = n * angle
pen_down_moves = n
ends_where_started = abs(total_turn - 360) < 1e-6
whole_number_angle = 360 // 11
whole_number_total_turn = 11 * (360 // 11)
whole_number_gap = 360 - whole_number_total_turn
whole_number_ends_where_started = False

assert isinstance(n, int)
assert isinstance(side, int)
assert isinstance(pen_down_moves, int)
assert isinstance(ends_where_started, bool)
assert pen_down_moves == 11
assert abs(total_turn - 360) < 1e-6
assert ends_where_started is True
assert 360 // 11 == 32
assert 11 * (360 // 11) == 352
assert whole_number_angle == 32
assert whole_number_total_turn == 352
assert whole_number_gap == 8
assert whole_number_ends_where_started is False
````
