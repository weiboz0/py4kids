# For & Range

A game club needs a 5-by-5 multiplication board. Each row must be assembled as one string and printed on its own line:

```text
1 2 3 4 5
2 4 6 8 10
3 6 9 12 15
4 8 12 16 20
5 10 15 20 25
```

How can a program visit every row and every column without a separate instruction for each product?

We will build that payoff in three lessons. First, `for` visits a known sequence of integers. Then familiar total-and-count patterns use that sequence. Finally, one loop inside another builds tables and triangles.

## Lesson One: Visit a Range

### Reproduce the 5050 sum

Unit 02 showed this total as a boxed peek. Unit 04 rebuilt it with `while`. Here is the same running total with `for` and `range`.

````python
total = 0

for n in range(1, 101):
    total = total + n

print(f"Sum: {total}")
````

**Notice:** this reproduces `Sum: 5050`. `range(1, 101)` supplies the integers 1 through 100, and `n` receives one value per pass. Do not modify the `for` variable inside its loop body.

### Two forms of `range`

`range(n)` starts at 0. It supplies 0 through `n - 1`.

````python
for number in range(5):
    print(number)
````

**Notice:** the output is 0, 1, 2, 3, 4. The stop value 5 is **not included**.

`range(a, b)` starts at `a` and stops before `b`.

````python
for number in range(3, 8):
    print(number)
````

**Notice:** `range(3, 8)` includes 3 but does not include 8, so the final printed value is 7. When an expected output includes the last value, set the stop one beyond it.

Python also has `range(a, b, step)`. Keep that third form as a Notice or Challenge for now; core work finds evens by testing `number % 2 == 0`.

### Comment the job, not every pass

A short comment can explain why a loop exists.

````python
# Print the five badge numbers.
for badge_number in range(1, 6):
    print(f"Badge: {badge_number}")
````

**Notice:** the name `badge_number` explains the changing value, and the comment explains the loop's purpose.

## Lesson Two: Calculate across a Range

### Running totals and conditional counts

The accumulator patterns from Unit 04 work unchanged. Only the source of each number changes.

````python
square_total = 0

for number in range(1, 6):
    square_total = square_total + number * number

print(f"Sum of squares: {square_total}")
````

**Notice:** the range supplies 1 through 5, and the accumulator produces `Sum of squares: 55`.

````python
count = 0

for number in range(1, 31):
    if number % 4 == 0:
        count = count + 1

print(f"Count: {count}")
````

**Notice:** every integer from 1 through 30 is checked, but only seven values pass the divisibility test.

### FizzBuzz with an `elif` ladder

FizzBuzz classifies each integer. Multiples of both 3 and 5 must be checked first because they also match the separate tests.

````python
for number in range(1, 16):
    if number % 15 == 0:
        line = "FizzBuzz"
    elif number % 3 == 0:
        line = "Fizz"
    elif number % 5 == 0:
        line = "Buzz"
    else:
        line = f"{number}"
    print(line)
````

**Notice:** each pass assigns exactly one string to `line`, then one unindented-within-the-ladder `print(line)` shows it. The order lets 15 become `FizzBuzz`.

### Primality: flag, test, and early exit

A prime number has no divisor from 2 through `n - 1`. For this rung, the given integer always satisfies **`n >= 2`**. A boolean **flag** starts as `True`, changes to `False` when a divisor is found, and `break` ends the search early.

````python
n = 29
is_prime = True

for divisor in range(2, n):
    if n % divisor == 0:
        is_prime = False
        break

print(f"{n} is prime: {is_prime}")
````

**Notice:** `break` stops this one loop as soon as evidence proves the flag false. For `n == 2`, `range(2, 2)` is empty, so the body runs zero times and `is_prime` correctly stays `True`. This is the final Lesson Two rung; if time is short, return to it after the earlier total, count, and FizzBuzz patterns.

## Lesson Three: Put One Loop inside Another

### Nested loops build rows

A **nested loop** places one loop inside another. The outer loop chooses a row. The inner loop performs every column action in that row.

````python
for row in range(1, 4):
    row_text = ""
    for col in range(1, 5):
        row_text = row_text + "*"
    print(row_text)
````

**Notice:** the inner body has 8 spaces; it adds four stars. `print(row_text)` has 4 spaces, so it runs once per completed row. If it were indented 8 spaces, it would print unfinished rows. After seeing the nested build, you may notice that Python can repeat text with `"*" * 4`, but the loop version exposes how each row is assembled.

A `break` inside the inner loop exits only that inner loop; the outer loop continues with its next row.

### A multiplication table

Build each table row as a string. A small condition inserts spaces between products without leaving a hidden trailing space.

````python
table_size = 3

for row in range(1, table_size + 1):
    row_text = ""
    for col in range(1, table_size + 1):
        product = row * col
        if col == 1:
            row_text = f"{product}"
        else:
            row_text = row_text + f" {product}"
    print(row_text)
````

**Notice:** the first row is `1 2 3` and the last is `3 6 9`. The inner loop builds one complete `row_text`; the outer loop prints three rows. No special print options are needed.

### A growing triangle

The row number can control how many columns the inner loop visits.

````python
height = 5

for row in range(1, height + 1):
    row_text = ""
    for col in range(1, row + 1):
        row_text = row_text + "*"
    print(row_text)
````

**Notice:** row 1 gets one inner pass, while row 5 gets five. The first accumulated row is `*`; the last is `*****`.

## Final build

Return to the 5-by-5 multiplication board. The outer loop selects rows 1 through 5, the inner loop selects columns 1 through 5, and string accumulation makes one assertable row at a time.

````python
table_size = 5

for row in range(1, table_size + 1):
    row_text = ""
    for col in range(1, table_size + 1):
        product = row * col
        if col == 1:
            row_text = f"{product}"
        else:
            row_text = row_text + f" {product}"
    print(row_text)
````

**Notice:** the finished board matches the opening exactly. Known ranges choose the positions, nested loops visit every row-column pair, and each row is printed only after its string is complete.
