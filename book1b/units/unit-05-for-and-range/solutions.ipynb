## Exercise 1

### The 5050 Total

```python
total = 0
for n in range(1, 101):
    total = total + n
line = f"Sum: {total}"
print(line)
assert total == 5050
assert line == "Sum: 5050"
```

## Exercise 2

### Stop-Bound Inspector

```python
last_seen = 0
for number in range(3, 9):
    last_seen = number
line = f"Last: {last_seen}"
print(line)
assert last_seen == 8
assert line == "Last: 8"
```

## Exercise 3

### Divisible-by-Four Census

```python
count = 0
for number in range(1, 31):
    if number % 4 == 0:
        count = count + 1
line = f"Count: {count}"
print(line)
assert count == 7
assert line == "Count: 7"
```

## Exercise 4

### FizzBuzz Board

```python
iterations = 0
for number in range(1, 16):
    if number % 15 == 0:
        line = "FizzBuzz"
    elif number % 3 == 0:
        line = "Fizz"
    elif number % 5 == 0:
        line = "Buzz"
    else:
        line = f"{number}"
    print(line)
    if number == 1:
        assert line == "1"
    if number == 15:
        assert line == "FizzBuzz"
    iterations = iterations + 1
assert iterations == 15
```

## Exercise 5

### Prime-Flag Test

```python
n = 29
is_prime = True
for divisor in range(2, n):
    if n % divisor == 0:
        is_prime = False
        break
line = f"{n} is prime: {is_prime}"
print(line)
assert is_prime == True
assert line == "29 is prime: True"
```

## Exercise 6

### Three-by-Three Times Table

```python
table_size = 3
row_count = 0
for row in range(1, table_size + 1):
    row_text = ""
    for col in range(1, table_size + 1):
        if col == 1:
            row_text = f"{row * col}"
        else:
            row_text = row_text + f" {row * col}"
    print(row_text)
    if row == 1:
        assert row_text == "1 2 3"
    if row == table_size:
        assert row_text == "3 6 9"
    row_count = row_count + 1
assert row_count == 3
```

## Exercise 7

### Nested Classification Count

```python
count = 0
for number in range(1, 13):
    if number % 2 == 0:
        if number % 3 == 0:
            count = count + 1
line = f"Count: {count}"
print(line)
assert count == 2
assert line == "Count: 2"
```

## Exercise 8

### Growing Signal Triangle

```python
height = 5
row_count = 0
for row in range(1, height + 1):
    row_text = ""
    for col in range(1, row + 1):
        row_text = row_text + "*"
    print(row_text)
    if row == 1:
        assert row_text == "*"
    if row == height:
        assert row_text == "*****"
    row_count = row_count + 1
assert row_count == 5
```

## Exercise 9

### Factorial over a Range

```python
n = 7
product = 1
for number in range(1, n + 1):
    product = product * number
line = f"Product: {product}"
print(line)
assert product == 5040
assert line == "Product: 5040"
```

## Exercise 11

### Challenge 2: Count the Primes

```python
prime_count = 0
for candidate in range(2, 21):
    is_prime = True
    for divisor in range(2, candidate):
        if candidate % divisor == 0:
            is_prime = False
            break
    if is_prime:
        prime_count = prime_count + 1
line = f"Count: {prime_count}"
print(line)
assert prime_count == 8
assert line == "Count: 8"
```
