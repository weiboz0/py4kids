## Exercise 1

### Target Message

```python
score = 12
if score >= 10:
    line_1 = "Target reached!"
    print(line_1)
assert line_1 == "Target reached!"
```

## Exercise 2

### Even or Odd

```python
number = 17
if number % 2 == 0:
    line_1 = "Even"
else:
    line_1 = "Odd"
print(line_1)
assert line_1 == "Odd"
```

## Exercise 3

### Inclusive Range

```python
x = 10
if x >= 0 and x <= 10:
    line_1 = f"{x} is in range."
else:
    line_1 = f"{x} is out of range."
print(line_1)
assert line_1 == "10 is in range."
```

## Exercise 4

### Weekend Label

```python
day = "Sunday"
if day == "Saturday" or day == "Sunday":
    line_1 = "Weekend"
else:
    line_1 = "Weekday"
print(line_1)
assert line_1 == "Weekend"
```

## Exercise 5

### Medal Ladder

```python
score = 95
if score >= 90:
    line_1 = "Gold"
elif score >= 80:
    line_1 = "Silver"
elif score >= 70:
    line_1 = "Bronze"
else:
    line_1 = "Practice round"
print(line_1)
assert line_1 == "Gold"
```

## Exercise 6

### Repair the Exact-Score Check

```python
score = 90
if score == 90:
    print("Exact score!")
```

```python
score = 90
if score == 90:
    line_1 = "Exact score!"
assert line_1 == "Exact score!"
```

## Exercise 7

### Permission and Score

```python
has_permission = True
score = 5
if has_permission:
    if score >= 7:
        line_1 = "Activity unlocked"
    else:
        line_1 = "Raise the score"
else:
    line_1 = "Permission needed"
print(line_1)
assert line_1 == "Raise the score"
```

## Exercise 8

### Fixed-Value Temperature

```python
temperature_text = "30"
temperature = int(temperature_text)
if temperature <= 32:
    line_1 = "Freezing"
else:
    line_1 = "Above freezing"
print(line_1)
assert line_1 == "Freezing"
```

## Exercise 9

### Challenge 1: Triangle Logic Puzzle

```python
side_a = 6
side_b = 6
side_c = 6
if side_a == side_b and side_b == side_c:
    line_1 = "Equilateral"
elif side_a == side_b or side_a == side_c or side_b == side_c:
    line_1 = "Isosceles"
else:
    line_1 = "Scalene"
print(line_1)
assert line_1 == "Equilateral"
```

## Exercise 10

### Challenge 2: Leap-Year Logic Puzzle

```python
year = 1900
if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    line_1 = f"{year} is a leap year."
else:
    line_1 = f"{year} is not a leap year."
print(line_1)
assert line_1 == "1900 is not a leap year."
```

## Exercise 11

### Challenge 3: Closed Sign

```python
is_open = False
if not is_open:
    line_1 = "Closed"
else:
    line_1 = "Open"
print(line_1)
assert line_1 == "Closed"
```
