## Exercise 12

        if player2 == "paper":

## Exercise 13

### More Practice: Quadrant Finder


````python
x = -3
y = 5
if x == 0:
    if y == 0:
        location = "the Origin"
    else:
        location = "on an axis"
elif y == 0:
    location = "on an axis"
elif x > 0:
    if y > 0:
        location = "in Quadrant I"
    else:
        location = "in Quadrant IV"
else:
    if y > 0:
        location = "in Quadrant II"
    else:
        location = "in Quadrant III"
line_1 = f"({x}, {y}) is {location}"
print(line_1)
assert line_1 == "(-3, 5) is in Quadrant II"
````
