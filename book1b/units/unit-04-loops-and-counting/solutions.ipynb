## Exercise 1

### Training-Lap Counter

```python
lap_goal = 6
laps = 0
while laps < lap_goal:
    laps = laps + 1
line = f"Count: {laps}"
print(line)
assert laps == 6
assert line == "Count: 6"
```

## Exercise 2

### Repair the Last Number

```python
i = 0
printed = ""
while i <= 5:
    print(i)
    printed = printed + str(i) + "\n"
    i = i + 1
```

```python
assert printed == "0\n1\n2\n3\n4\n5\n"
assert i == 6
```

## Exercise 3

### Fundraiser Running Total

```python
last_amount = 12
amount = 1
total = 0
while amount <= last_amount:
    total = total + amount
    amount = amount + 1
line = f"Sum: {total}"
print(line)
assert total == 78
assert line == "Sum: 78"
```

## Exercise 4

### Multiples-of-Four Count

```python
last_turn = 20
turn = 1
count = 0
while turn <= last_turn:
    if turn % 4 == 0:
        count = count + 1
    turn = turn + 1
line = f"Count: {count}"
print(line)
assert count == 5
assert line == "Count: 5"
```

## Exercise 5

### Factorial Machine

```python
n = 6
number = 1
product = 1
while number <= n:
    product = product * number
    number = number + 1
line = f"Product: {product}"
print(line)
assert product == 720
assert line == "Product: 720"
```

## Exercise 6

### Digit-Sum Scanner

```python
n = 7314
digit_total = 0
while n != 0:
    digit = n % 10
    digit_total = digit_total + digit
    n = n // 10
line = f"Digit sum: {digit_total}"
print(line)
assert n == 0
assert digit_total == 15
assert line == "Digit sum: 15"
```

## Exercise 7

### Computed Stop: Collatz Steps

```python
n = 7
steps = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = n * 3 + 1
    steps = steps + 1
line = f"Steps: {steps}"
print(line)
assert n == 1
assert steps == 16
assert line == "Steps: 16"
```

## Exercise 8

### Shared-Tile Size

```python
a = 84
b = 36
while a != b:
    if a > b:
        a = a - b
    else:
        b = b - a
line = f"GCD: {a}"
print(line)
assert a == b
assert a == 12
assert line == "GCD: 12"
```

## Exercise 9

### Score-Tier Census

```python
score = 1
low_count = 0
middle_count = 0
high_count = 0
while score <= 12:
    if score >= 9:
        high_count = high_count + 1
    elif score >= 5:
        middle_count = middle_count + 1
    else:
        low_count = low_count + 1
    score = score + 1
printed = ""
line_1 = f"Low: {low_count}"
print(line_1)
printed = printed + line_1 + "\n"
line_2 = f"Middle: {middle_count}"
print(line_2)
printed = printed + line_2 + "\n"
line_3 = f"High: {high_count}"
print(line_3)
printed = printed + line_3 + "\n"
assert printed == "Low: 4\nMiddle: 4\nHigh: 4\n"
assert low_count == 4
assert middle_count == 4
assert high_count == 4
```

## Exercise 10

### Challenge 1: Remainder GCD

```python
a = 252
b = 105
while b != 0:
    remainder = a % b
    a = b
    b = remainder
line = f"GCD: {a}"
print(line)
assert b == 0
assert a == 21
assert line == "GCD: 21"
```

### Challenge 2: Probe Mission 27

```python
n = 19
steps = 0
while True:
    if n == 1:
        break
    if n % 2 == 0:
        n = n // 2
    else:
        n = n * 3 + 1
    steps = steps + 1
line = f"Steps: {steps}"
print(line)
assert n == 1
assert steps == 20
assert line == "Steps: 20"
```
