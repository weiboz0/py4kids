## Exercise 9

### Score-Tier Census

````python
score = 1
low_count = 0
middle_count = 0
high_count = 0
while score <= 12:
    if score >= 9:
        high_count = high_count + 1
    elif score >= 5:
        middle_count = middle_count + 1
    else:
        low_count = low_count + 1
    score = score + 1
printed = ""
line_1 = f"Low: {low_count}"
print(line_1)
printed = printed + line_1 + "\n"
line_2 = f"Middle: {middle_count}"
print(line_2)
printed = printed + line_2 + "\n"
line_3 = f"High: {high_count}"
print(line_3)
printed = printed + line_3 + "\n"
assert printed == "Low: 4\nMiddle: 4\nHigh: 4\n"
assert low_count == 4
assert middle_count == 4
assert high_count == 4
````

## Exercise 28

### More Practice: Guessing Robot

````python
secret = 42
low = 1
high = 100
count = 0
transcript = ""
while True:
    guess = (low + high) // 2
    count = count + 1
    if transcript == "":
        transcript = str(guess)
    else:
        transcript = transcript + " " + str(guess)
    if guess == secret:
        break
    if guess < secret:
        low = guess + 1
    else:
        high = guess - 1
line_1 = "Guesses: " + transcript
line_2 = f"Found {secret} in {count} guesses"
print(line_1)
print(line_2)
assert line_1 == "Guesses: 50 25 37 43 40 41 42"
assert line_2 == "Found 42 in 7 guesses"
````

## Exercise 34

### Challenge 5: Count the Steps

````python
a = 270
b = 192
sub_a = a
sub_b = b
sub_steps = 0
while sub_a != sub_b:
    if sub_a > sub_b:
        sub_a = sub_a - sub_b
    else:
        sub_b = sub_b - sub_a
    sub_steps = sub_steps + 1
rem_a = a
rem_b = b
rem_steps = 0
while rem_b != 0:
    remainder = rem_a % rem_b
    rem_a = rem_b
    rem_b = remainder
    rem_steps = rem_steps + 1
line_1 = f"Subtraction steps: {sub_steps}"
line_2 = f"Remainder steps: {rem_steps}"
line_3 = f"GCD: {rem_a}"
print(line_1)
print(line_2)
print(line_3)
assert line_1 == "Subtraction steps: 10"
assert line_2 == "Remainder steps: 4"
assert line_3 == "GCD: 6"
````

**The real program** (reads two positive numbers, one per line):

```python

b = int(input())
sub_a = a
sub_b = b
sub_steps = 0

        sub_a = sub_a - sub_b
    else:
        sub_b = sub_b - sub_a
    sub_steps = sub_steps + 1
rem_a = a
rem_b = b
rem_steps = 0
while rem_b != 0:
    remainder = rem_a % rem_b
    rem_a = rem_b
    rem_b = remainder
    rem_steps = rem_steps + 1
line_1 = f"Subtraction steps: {sub_steps}"
line_2 = f"Remainder steps: {rem_steps}"
line_3 = f"GCD: {rem_a}"
print(line_1)
print(line_2)
print(line_3)
```
