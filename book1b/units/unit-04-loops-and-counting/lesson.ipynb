# Loops & Counting

A space probe starts with the number 27. If the number is even, the probe divides it by 2. Otherwise it triples the number and adds 1. It repeats until the number reaches 1.

The mission board needs one summary line:

```text
Steps: 111
```

How can one small block of code repeat without being copied 111 times?

We will build that payoff in three lessons. First we repeat with `while`, then collect totals and counts, and finally stop when a calculated value reaches a goal.

## Lesson One: Repeat and Keep Count

### Reassignment comes first

A loop counter depends on **reassignment**: calculate a new value from the saved value, then save the result back under the same name.

Run this rung before writing a loop.

````python
count = 0
count = count + 1
print(count)
````

**Notice:** in `count = count + 1`, Python computes the right side with the **OLD value** of `count`, then saves the new value. Python also allows `count += 1`, but this course uses the longer form so every step stays visible.

### A `while` loop

A `while` loop repeats its indented body while its condition is `True`. A **loop counter** records how many passes have happened.

**Minimal:** count three passes, then print one summary.

````python
count = 0

while count < 3:
    count = count + 1

print(f"Count: {count}")
````

**Notice:** the counter starts at 0, changes once each pass, and eventually makes `count < 3` false. That change is what guarantees the loop stops.

**One twist:** output can happen during each pass.

````python
number = 1

while number <= 5:
    line = f"Number: {number}"
    print(line)
    number = number + 1
````

**Notice:** all four spaces of indentation belong to the loop. The update happens after printing, so the visible numbers are 1 through 5.

### When the counter never changes

This broken example forgets `i = i + 1`. Read it, but do not run it.

````python
i = 0

while i < 5:
    print(i)
````

**Notice:** `i` stays 0, so the condition stays true forever. If you accidentally run an infinite loop, choose **Kernel → Interrupt** in Jupyter or press **Ctrl-C** in a terminal. Then add the missing update before running again.

````python
i = 0

while i < 5:
    print(i)
    i = i + 1
````

**Notice:** now `i` advances toward 5, so the loop prints 0 through 4 and stops.

### Repair an off-by-one boundary

A loop can stop safely but still include one too few or one too many values. Compare the two boundaries.

````python
i = 0

while i < 5:
    print(i)
    i = i + 1
````

**Notice:** `i < 5` prints 0, 1, 2, 3, 4. It stops before 5. Use `<=` when 5 itself must be included.

````python
i = 0

while i <= 5:
    print(i)
    i = i + 1
````

**Notice:** `i <= 5` prints 0 through 5. Reading the expected first and last values helps repair an off-by-one mistake.

### Read a real `NameError`

A counter or total must exist before the program can update it. The broken cell uses `total` before `total = 0`. Read the final traceback line, then run the fixed partner.

````python
number = 7
total = total + number
print(f"Total: {total}")
````

The traceback ends with `NameError: name 'total' is not defined`. Python cannot add to a value that has never been saved. Add the initializer before the update.

````python
number = 7
total = 0
total = total + number
print(f"Total: {total}")
````

**Notice:** the initializer gives `total` a starting value. This traceback beat is different from an infinite loop: a `NameError` stops with a message, while a logic mistake can keep running.

## Lesson Two: Accumulate Totals and Counts

### A running total

An **accumulator** saves a result that grows one pass at a time. A **running total** starts at 0, adds the current number, and saves the result back into `total`.

**Minimal:** add the integers from 1 through 10.

````python
number = 1
total = 0

while number <= 10:
    total = total + number
    number = number + 1

print(f"Sum: {total}")
````

**Notice:** `total` must begin at 0 because 0 changes no sum. The loop prints one summary after all ten additions.

You saw the 1-through-100 total in Unit 02 as a boxed peek. Now you can write its repeated work yourself.

````python
number = 1
total = 0

while number <= 100:
    total = total + number
    number = number + 1

print(f"Sum: {total}")
````

**Notice:** the result is `Sum: 5050`. The counter chooses each number once; the accumulator remembers the growing sum.

### Count by a condition

Sometimes we do not add the values themselves. Instead, we add 1 only when a value passes a test. That pattern is **count by condition**.

````python
number = 1
count = 0

while number <= 20:
    if number % 3 == 0:
        count = count + 1
    number = number + 1

print(f"Count: {count}")
````

**Notice:** every number is checked, but the counter changes only for 3, 6, 9, 12, 15, and 18. The summary is `Count: 6`.

**One twist:** an `elif` ladder can count several tiers in the same walk.

````python
score = 1
low_count = 0
middle_count = 0
high_count = 0

while score <= 10:
    if score >= 8:
        high_count = high_count + 1
    elif score >= 4:
        middle_count = middle_count + 1
    else:
        low_count = low_count + 1
    score = score + 1

print(f"Low: {low_count}")
print(f"Middle: {middle_count}")
print(f"High: {high_count}")
````

**Notice:** exactly one branch counts each score. Checking the highest tier first produces 3 low, 4 middle, and 3 high scores.

### Products and digit totals

A factorial multiplies the integers from 1 through `n`. Its accumulator must start at 1, not 0, because multiplying by 0 would erase the whole product. Keep `n` at 10 or below in this unit.

````python
n = 5
number = 1
product = 1

while number <= n:
    product = product * number
    number = number + 1

print(f"Product: {product}")
````

**Notice:** `product = 1` preserves the first multiplication. The summary is `Product: 120`.

The digit tools from Unit 02 can also drive a loop. `% 10` peels off the last digit, and `// 10` removes it.

````python
n = 4827
digit_total = 0

while n != 0:
    digit = n % 10
    digit_total = digit_total + digit
    n = n // 10

print(f"Digit sum: {digit_total}")
````

**Notice:** `n` becomes 482, then 48, then 4, then 0. Reaching 0 is a calculated stop value, and the digit sum is 21.

## Lesson Three: Stop at a Computed Goal

### A plain-condition sentinel

A **sentinel loop** repeats until a computed stop value appears. Begin with the plain condition `while n != 1`. For the Collatz rule below, every shown starting value is a positive integer chosen to reach 1.

````python
n = 6
steps = 0

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = n * 3 + 1
    steps = steps + 1

print(f"Steps: {steps}")
````

**Notice:** `steps` is the one source of truth: it counts each update and supplies the printed summary `Steps: 8`. The condition states the stop goal directly.

### Exit from the middle with `break`

Sometimes the stop test belongs after a calculation. `while True:` begins an intentionally open loop; `break` exits it as soon as a guaranteed-reachable condition is true.

````python
value = 3
steps = 0

while True:
    value = value * 2
    steps = steps + 1
    if value > 1000:
        break

print(f"Steps: {steps}")
````

**Notice:** positive doubling must eventually pass 1000, so the `break` is reachable. The summary is `Steps: 9`. Never write `while True:` unless you can explain why its `break` will be reached.

### Greatest common divisor by subtraction

For positive integers, any common divisor of `a` and `b` also divides `a - b`. That lets us replace the larger value with the difference until both values match.

A trace for `(48, 18)` is `(30, 18)`, `(12, 18)`, `(12, 6)`, `(6, 6)`. The matching value is the greatest common divisor.

````python
a = 48
b = 18

while a != b:
    if a > b:
        a = a - b
    else:
        b = b - a

print(f"GCD: {a}")
````

**Notice:** each subtraction keeps the common divisors but shrinks the larger positive value. The process stops at `GCD: 6`. A faster remainder-based version is saved as a later Challenge.

## Final build

Return to the probe mission. The fixed start value is 27, and the loop stops only when the computed value reaches 1. Run the build and check the single summary line.

````python
# Count the updates from 27 until the value reaches 1.
n = 27
steps = 0

while True:
    if n == 1:
        break
    if n % 2 == 0:
        n = n // 2
    else:
        n = n * 3 + 1
    steps = steps + 1

print(f"Steps: {steps}")
````

**Notice:** the build prints `Steps: 111`. A changing value reaches the sentinel, `break` exits, and the same `steps` counter both records the work and creates the final summary.
