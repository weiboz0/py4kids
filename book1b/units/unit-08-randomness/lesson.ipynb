## Lesson 1: Chance

A board game needs a six-sided die. We could type one fixed roll, but then the game would never surprise anyone. Python's `random` module can choose an integer for us.

### A die we can test

`import random` brings in the module. `random.seed(4)` fixes the starting point of its number recipe. That makes the same program repeat the same choices, which is useful while debugging and essential when a test must have a reproducible answer.

`random.randint(1, 6)` chooses an integer from 1 through 6. Both endpoints can be chosen. Predict the total before running this cell.

````python
import random

def roll_total(rolls):
    random.seed(4)
    total = 0
    for roll_number in range(rolls):
        roll = random.randint(1, 6)
        total = total + roll
    return total

print(roll_total(6))
````

**Notice:** The seed comes before the first random choice. Run the cell again: the total repeats because seed 4 restarts the same tested sequence. Change only the seed and the choices can change. During a real game we might not restart the seed inside the function, but seeded functions let this unit's tests compare exact results.

### Choose from a range

`random.choice(range(2))` chooses either 0 or 1. We can let 0 mean tails and 1 mean heads. `random.randint(0, 1)` is the everyday tool for choosing between those two integers, while `random.choice(range(...))` can pick from any supplied run of values. In this course, `random.choice` always receives a `range(...)`.

````python
import random

def count_heads(flips):
    random.seed(4)
    heads = 0
    for flip_number in range(flips):
        coin = random.choice(range(2))
        if coin == 1:
            heads = heads + 1
    return heads

print(count_heads(12))
````

**Notice:** Use module style: write `import random`, then names such as `random.randint` and `random.choice`. We use only `randint` and `choice(range(...))` for random choices in this unit.

### Dice-and-coin score

Now the game has rounds. Each round rolls a die and flips a coin. Heads adds the roll; tails subtracts the roll. The score must survive from one round to the next, so it is a running total.

````python
import random

def chance_score(rounds):
    random.seed(4)
    score = 0
    for round_number in range(rounds):
        roll = random.randint(1, 6)
        coin = random.randint(0, 1)
        if coin == 1:
            score = score + roll
        else:
            score = score - roll
    return score

print(chance_score(8))
````

**Trace it:** `score` starts once, before the loop. Each round reads the old score and stores the new score. If `score = 0` were inside the loop, every round would erase the earlier rounds instead of keeping a cumulative score.

## Lesson 2: Simulate & Estimate

The booth owner says, “Sixes happen about one-sixth of the time.” Instead of arguing from six rolls, we can simulate many trials and count every six.

### Count an outcome

A trial is one repeat of an experiment. Here, `total` is an accumulator: it changes only when the condition is true.

````python
import random

def count_sixes(trials):
    random.seed(4)
    total = 0
    for trial in range(trials):
        roll = random.randint(1, 6)
        if roll == 6:
            total = total + 1
    return total

print(count_sixes(60))
print(count_sixes(600))
````

**Notice:** This is count by condition. The loop performs every trial, but `total = total + 1` runs only for a six. The larger experiment gives more evidence, although its count is not required to be exactly one-sixth of the trials.

### Report a percentage

A raw count is hard to compare across experiment sizes. This function turns it into a percentage. `max(trials, 1)` keeps the divisor at least 1, and `round(..., 1)` reports one decimal place.

````python
import random

def six_percent(trials):
    random.seed(4)
    total = 0
    for trial in range(trials):
        roll = random.randint(1, 6)
        if roll == 6:
            total = total + 1
    safe_trials = max(trials, 1)
    return round(100 * total / safe_trials, 1)

print(six_percent(60))
print(six_percent(600))
````

### A quarter-circle picture for estimating π

Imagine a quarter-circle of radius 1000 inside a 1000-by-1000 square. The curved boundary runs from `(0, 1000)` to `(1000, 0)`.

```text
y
1000  *------. + outside corner
      |       `.
      |        )  quarter-circle edge
      |      .'
   0  +-------* x
      0      1000
```

The arc bulges outward, so the quarter-circle covers about 78.5% of the square — that fraction is `π / 4`, so multiply by 4. The square's area is `1000 * 1000`, and the quarter-circle's area is `π * 1000 * 1000 / 4`. Count the inside fraction and multiply by 4 to estimate π.

For each point, choose `x = random.randint(0, 1000)` and `y = random.randint(0, 1000)`. Each axis has **1001 possible integer values**, including both 0 and 1000, so do not call this a “1000×1000 grid.” A point is inside when `x*x + y*y <= 1000*1000`.

````python
import random

def estimate_pi(trials):
    random.seed(4)
    inside = 0
    for trial in range(trials):
        x = random.randint(0, 1000)
        y = random.randint(0, 1000)
        if x * x + y * y <= 1000 * 1000:
            inside = inside + 1
    safe_trials = max(trials, 1)
    return round(4 * inside / safe_trials, 2)

print(estimate_pi(50))
print(estimate_pi(500))
print(estimate_pi(5000))
````

**Notice:** The result is an estimate **close to π, not exactly π**. Expect a value around 3.1, not exactly 3.14159. More trials usually make the estimate steadier, but randomness does not promise that every larger experiment is closer. Seed 4 makes our classroom result reproducible.

## Lesson 3: Random Turtle Walk

A rescue robot can move a fixed distance, but debris forces it to turn left or right after every step. A regular polygon knows its next turn. This robot must choose. Can we make the route surprising and still make the program testable?

### Build the walk as a real program

The walk runs from `assets/l3_random_walk.py`, just like the turtle programs in Unit 06. Its step length stays fixed. `random.randint(0, 1)` chooses a left or right turn. The exact `# turtle-check: open-path` comment records that a random walk is not expected to return to its start.

**The real program (`assets/l3_random_walk.py`)**

```python
# Bring in the drawing and random-choice tools.
import turtle
import random

random.seed(4)

# turtle-check: open-path
# Take a fixed step, then randomly choose a left or right turn.
step = 28
move_count = 24
turtle.pencolor("darkviolet")
turtle.pensize(3)

for move_number in range(move_count):
    turtle.forward(step)
    turn = random.randint(0, 1)
    if turn == 0:
        turtle.left(90)
    else:
        turtle.right(90)

turtle.done()
```


Close the turtle window when you finish observing the route. Change only `move_count`, save, and rerun to see how a longer walk extends the same seeded route.

**Notice:** `step`, `move_count`, and `turn` have clear jobs. The turtle always moves with its pen down, so 24 loop trips make 24 drawn segments. Seed 4 makes the left/right choices repeat for testing. The route is intentionally open, unlike Unit 06's closed polygons.

### Put the ideas together

Random programs still need structure. A function gives an experiment inputs and returns its result. A loop runs the trials. An accumulator remembers a score or count. A condition decides which outcomes count. A fixed seed makes the experiment reproducible while you test it.
