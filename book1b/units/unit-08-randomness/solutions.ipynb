## Exercise 1

### Dice Total

````python
def roll_total(rolls):
    random.seed(4)
    total = 0
    for roll_number in range(rolls):
        roll = random.randint(1, 6)
        total = total + roll
    return total

assert roll_total(1) == 2
assert roll_total(4) == 12
assert roll_total(7) == 22
assert roll_total(10) == 25
````

## Exercise 2

### Heads Counter

````python
def count_heads(flips):
    random.seed(4)
    total = 0
    for flip_number in range(flips):
        coin = random.choice(range(2))
        if coin == 1:
            total = total + 1
    return total

assert count_heads(1) == 0
assert count_heads(6) == 3
assert count_heads(14) == 6
assert count_heads(20) == 8
````

## Exercise 3

### Bonus-Round Score

````python
def bonus_score(rounds):
    random.seed(4)
    score = 0
    for round_number in range(rounds):
        roll = random.randint(1, 6)
        if roll >= 4:
            score = score + roll
        else:
            score = score - 1
    return score

assert bonus_score(1) == -1
assert bonus_score(5) == 7
assert bonus_score(8) == 9
assert bonus_score(12) == 16
````

## Exercise 4

### Target Roll Counter

````python
def count_targets(trials, target):
    random.seed(4)
    total = 0
    for trial in range(trials):
        roll = random.randint(1, 6)
        if roll == target:
            total = total + 1
    return total

assert count_targets(1, 2) == 1
assert count_targets(12, 3) == 1
assert count_targets(30, 6) == 2
assert count_targets(50, 1) == 10
````

## Exercise 5

### Even-Roll Percentage

````python
def even_percent(trials):
    random.seed(4)
    total = 0
    for trial in range(trials):
        roll = random.randint(1, 6)
        if roll % 2 == 0:
            total = total + 1
    return round(100 * total / max(trials, 1), 1)

assert even_percent(0) == 0.0
assert even_percent(10) == 50.0
assert even_percent(25) == 40.0
assert even_percent(100) == 47.0
````

## Exercise 6

### Quarter-Circle Estimate

````python
def estimate_pi(trials):
    random.seed(4)
    inside = 0
    for trial in range(trials):
        x = random.randint(0, 1000)
        y = random.randint(0, 1000)
        if x * x + y * y <= 1000 * 1000:
            inside = inside + 1
    return round(4 * inside / max(trials, 1), 2)

assert abs(estimate_pi(0) - 0.0) < 1e-9
assert abs(estimate_pi(20) - 3.4) < 1e-9
assert abs(estimate_pi(200) - 3.14) < 1e-9
assert abs(estimate_pi(2000) - 3.08) < 1e-9
````

## Exercise 7

### Rescue-Robot Random Walk

**The real program (`assets/solutions_ex7_random_walk.py`)**

```python
# Bring in the drawing and random-choice tools.
import turtle
import random

random.seed(4)

# turtle-check: open-path
# Take a fixed step, then randomly choose a left or right turn.
step = 31
move_count = 20
turtle.pencolor("seagreen")
turtle.pensize(3)

for move_number in range(move_count):
    turtle.forward(step)
    turn = random.randint(0, 1)
    if turn == 0:
        turtle.left(90)
    else:
        turtle.right(90)

turtle.done()
```

The code cell below checks the same seeded turn choices without opening a turtle window.

````python
def random_walk_left_turns(move_count):
    random.seed(4)
    left_turns = 0
    for move_number in range(move_count):
        turn = random.randint(0, 1)
        if turn == 0:
            left_turns = left_turns + 1
    return left_turns

step = 31
move_count = 20
left_turns = random_walk_left_turns(move_count)
right_turns = move_count - left_turns

assert step == 31
assert move_count == 20
assert left_turns == 12
assert right_turns == 8
assert random_walk_left_turns(10) == 6
assert random_walk_left_turns(20) == left_turns
````

## Exercise 8

### Challenge 1: Highest Twenty-Sided Roll

````python
def highest_roll(trials):
    random.seed(4)
    high = 0
    for trial in range(trials):
        roll = random.randint(1, 20)
        high = max(high, roll)
    return high

assert highest_roll(1) == 8
assert highest_roll(3) == 10
assert highest_roll(10) == 16
assert highest_roll(30) == 18
````

## Exercise 9

### Challenge 2: Multiples in a Random Range

````python
def count_multiples(trials, n):
    random.seed(4)
    total = 0
    for trial in range(trials):
        value = random.choice(range(1, 21))
        if value % n == 0:
            total = total + 1
    return total

assert count_multiples(1, 4) == 1
assert count_multiples(10, 4) == 3
assert count_multiples(25, 5) == 3
assert count_multiples(40, 3) == 15
````
