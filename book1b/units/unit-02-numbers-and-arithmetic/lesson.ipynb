# Numbers & Arithmetic

A school trip lasts 135 minutes. The schedule needs that time written as hours and leftover minutes:

```text
2 hours 15 minutes
```

How can Python split one number into those two useful parts?

We will build that payoff in three lessons. Each idea climbs the same ladder: **minimal**, **one twist**, then **realistic**. Read every **Notice** line before moving up.

## Lesson One: Calculate with Whole Numbers

### Arithmetic operators

Python can calculate with whole numbers called **integers**. The symbols `+`, `-`, and `*` mean add, subtract, and multiply.

**Minimal:** add two integers.

````python
print(8 + 4)
````

**Notice:** numbers need no quotation marks. Python calculates `8 + 4` before printing `12`.

**One twist:** saved numbers can take part in calculations.

````python
tickets = 6
price_each = 4
print(tickets * price_each)
````

**Notice:** `tickets * price_each` uses the values saved in both variables.

**One twist:** multiplication happens before addition.

````python
print(2 + 3 * 4)
````

**Notice:** Python multiplies `3 * 4` first, then adds 2, giving 14.

**One twist:** parentheses choose a different order.

````python
print((2 + 3) * 4)
````

**Notice:** parentheses make `2 + 3` happen first, giving 20 after multiplication.

**Realistic:** combine operations to find how many seats remain.

````python
rows = 5
seats_per_row = 8
students = 34
open_seats = rows * seats_per_row - students
print(f"Open seats: {open_seats}")
````

**Notice:** multiplication happens before subtraction, so 40 seats are counted before 34 students are removed. Parentheses can make a chosen order clear.

### Powers

**Minimal:** `**` raises a number to a power.

````python
print(2 ** 5)
````

**Notice:** `2 ** 5` means 2 multiplied by itself five times, giving 32.

### Quotients and remainders

The operator `//` gives the whole-number quotient. The operator `%` gives the remainder. We will use both only with non-negative numbers.

**Minimal:** divide 17 items into groups of 5.

````python
print(17 // 5)
print(17 % 5)
````

**Notice:** 17 makes 3 complete groups of 5 with 2 left over.

**One twist:** an f-string can calculate inside its braces.

````python
print(f"{3 + 4}")
````

**Notice:** Python calculates `3 + 4` inside the braces and prints `7`.

**One twist:** label both quotient and remainder results.

````python
stickers = 29
team_size = 6
print(f"Each team gets: {stickers // team_size}")
print(f"Stickers left: {stickers % team_size}")
````

**Notice:** the same two given values answer two related questions.

**Realistic:** return to the trip schedule.

````python
total_minutes = 135
hours = total_minutes // 60
minutes_left = total_minutes % 60
print(f"{hours} hours {minutes_left} minutes")
````

**Notice:** `//` finds the 2 complete hours, and `%` finds the 15 minutes left over.

### Repeated text

**Minimal:** multiply a string by an integer to repeat it.

````python
print("=" * 12)
````

**Notice:** multiplying `"="` by 12 prints twelve equal signs in one line.

**One twist:** use the repeated line above and below a title.

````python
divider = "=" * 12
print(divider)
print("MATH CLUB")
print(divider)
````

**Notice:** saving the divider lets the same twelve-character line frame both sides of the title.

## Lesson Two: Work with Number Types

### Division and floats

A number with a decimal point is a **float**. The `/` operator performs regular division.

**Minimal:** divide 10 by 2.

````python
print(10 / 2)
````

**Notice:** `/` gives a float, `//` gives an int. The output above is exactly `5.0`.

**One twist:** a division can produce a half.

````python
print(7 / 2)
````

**Notice:** Python shows the computed float exactly as `3.5`.

**Realistic:** calculate the average of three given scores.

````python
score_one = 3
score_two = 4
score_three = 5
average = (score_one + score_two + score_three) / 3
print(f"Average: {average}")
````

**Notice:** the parentheses add the three scores first. The exact printed average is `4.0`.

**One twist:** show a price with two digits after the decimal point.

````python
price = 3.5
print(f"{price:.2f}")
````

**Notice:** `:.2f` displays `3.5` as `3.50`; it changes the printed format, not the saved number.

### Convert between text and numbers

`int()` turns suitable text into an integer. `float()` turns suitable text into a float. `str()` turns a value into text.

**Minimal:** convert quoted digits before calculating.

````python
age_text = "12"
age = int(age_text)
print(age + 1)
````

**Notice:** the sample value is `"12"` with quotes, so it begins as text. After `int(age_text)`, Python can calculate `13`.

**One twist:** joining text and adding numbers use the same symbol but do different jobs.

````python
print("3" + "4")
print(3 + 4)
````

**Notice:** `"3" + "4"` joins text to make `34`, while `3 + 4` calculates `7`.

A mixed text-and-number join stops with a `TypeError`. Read the final traceback line in this broken example.

````python
print("Age: " + 12)
````

The final line says `TypeError: can only concatenate str (not "int") to str`. Convert the number to text, or let an f-string place it in the sentence.

````python
age = 12
print("Age: " + str(age))
print(f"Age: {age}")
````

**Notice:** both repaired lines print `Age: 12`; each line keeps text and number types compatible.

**Try a broken conversion:** read the final traceback line to find out why these letters cannot become an integer.

````python
int("abc")
````

**Notice:** `int("abc")` raises a `ValueError` because the text does not contain digits for an integer.

**Try real input:** run the next cell with your teacher. It waits for an age, then converts it to an integer.

````python
age = int(input("Age: "))
print(f"Next year: {age + 1}")
````

**Notice:** `input()` supplies text, and `int()` converts it before the addition.

**Realistic:** use a fixed value to make the same calculation repeatable in the next cell.

````python
age_text = "12"
age = int(age_text)
print(f"Next year: {age + 1}")
````

**Notice:** the fixed value keeps the lesson repeatable; the real-input cell above lets you try your own age.

## Lesson Three: Ask Number Questions

### Comparisons and booleans

A comparison asks a yes-or-no question. Its result is a **boolean** value: `True` or `False`.

**Minimal:** compare two numbers.

````python
print(4 < 9)
print(9 > 4)
````

**Notice:** `<` asks whether the left value is less, while `>` asks whether it is greater. Both comparisons print the boolean `True`.

**One twist:** include the boundary with less-than-or-equal and greater-than-or-equal.

````python
score = 8
print(score <= 8)
print(score >= 8)
````

**Notice:** `<=` and `>=` include equality, so both comparisons print `True` when the two values match.

**Minimal:** `==` asks whether two values are equal.

````python
print(8 == 8)
````

**Notice:** `8 == 8` is `True`; use two equal signs to compare, rather than one equal sign to save a value.

**One twist:** `!=` asks whether two values differ.

````python
print(8 != 9)
````

**Notice:** `8 != 9` is `True` because the numbers are different.

**Realistic:** test equality and inequality directly.

````python
score = 8
print(score == 8)
print(score != 10)
````

**Notice:** `==` asks whether two values are equal, while `!=` asks whether they are different. Each comparison prints a boolean. A single `=` saves a value instead.

### Split digits

Quotient and remainder can separate the digits of a non-negative integer.

**Minimal:** split a two-digit number from 10 through 99.

````python
number = 47
tens = number // 10
ones = number % 10
print(tens)
print(ones)
````

**Notice:** complete tens come from `// 10`; the leftover ones come from `% 10`.

**One twist:** rebuild the original value.

````python
number = 63
tens = number // 10
ones = number % 10
rebuilt = tens * 10 + ones
print(rebuilt)
````

**Notice:** place value reverses the split: 6 tens plus 3 ones makes 63.

**Realistic:** print a labeled digit report for a two-digit number from 10 through 99.

````python
number = 82
tens = number // 10
ones = number % 10
print(f"Number: {number}")
print(f"Tens: {tens}")
print(f"Ones: {ones}")
````

**Notice:** the stated 10-through-99 range guarantees exactly one tens digit and one ones digit.

**Try real input:** type a two-digit number from 10 through 99 to see its place values.

````python
number = int(input("Two-digit number: "))
tens = number // 10
ones = number % 10
print(f"Tens: {tens}")
print(f"Ones: {ones}")
````

**Notice:** `int()` turns the typed text into a number before `//` and `%` split its digits.
