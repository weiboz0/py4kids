## Exercise 1

### Supply Total

```python
packs = 6
pencils_per_pack = 8
total = packs * pencils_per_pack
line_1 = f"Total pencils: {total}"
print(line_1)
assert line_1 == "Total pencils: 48"
```

## Exercise 2

### Trip Time

```python
total_minutes = 185
hours = total_minutes // 60
minutes = total_minutes % 60
line_1 = f"{hours} hours {minutes} minutes"
print(line_1)
assert line_1 == "3 hours 5 minutes"
```

## Exercise 3

### Equal Share

```python
ribbon_meters = 14
teammates = 4
share = ribbon_meters / teammates
line_1 = f"Meters each: {share}"
print(line_1)
assert line_1 == "Meters each: 3.5"
```

## Exercise 4

### Next-Year Age

```python
age_text = "12"
age = int(age_text) + 1
line_1 = f"Next year: {age}"
print(line_1)
assert line_1 == "Next year: 13"
```

## Exercise 5

### Compare Two Scores

```python
first_score = 18
second_score = 15
line_1 = first_score > second_score
print(line_1)
assert line_1 == True
```

## Exercise 6

### Two-Digit Report

```python
number = 73
tens = number // 10
ones = number % 10
line_1 = f"Tens: {tens}"
print(line_1)
assert line_1 == "Tens: 7"
line_2 = f"Ones: {ones}"
print(line_2)
assert line_2 == "Ones: 3"
```

## Exercise 7

### Repair the Age Label

```python
age = 12
print("Age: " + str(age))
```

```python
age = 12
line_1 = "Age: " + str(age)
assert line_1 == "Age: 12"
```

## Exercise 8

### Fixed-Value Price

```python
price_text = "3.5"
price = float(price_text)
total = price * 2
line_1 = f"Two cost: {total}"
print(line_1)
assert line_1 == "Two cost: 7.0"
```

## Exercise 9

### Challenge 1: Reverse Two Digits

```python
number = 42
tens = number // 10
ones = number % 10
reversed_number = ones * 10 + tens
line_1 = f"Reversed: {reversed_number}"
print(line_1)
assert line_1 == "Reversed: 24"
```

## Exercise 10

### Challenge 2: Three-Digit Places

```python
number = 582
hundreds = number // 100
tens = number // 10 % 10
ones = number % 10
line_1 = f"Hundreds: {hundreds}"
print(line_1)
assert line_1 == "Hundreds: 5"
line_2 = f"Tens: {tens}"
print(line_2)
assert line_2 == "Tens: 8"
line_3 = f"Ones: {ones}"
print(line_3)
assert line_3 == "Ones: 2"
```
