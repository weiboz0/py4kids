````python
# Prediction 1: 14
# Prediction 2: 20
# Prediction 3: 2
# Prediction 4: 16
# Prediction 5: False
line_1 = 6 + 4 * 2
print(line_1)
assert line_1 == 14
line_2 = (6 + 4) * 2
print(line_2)
assert line_2 == 20
line_3 = 20 // 3 % 4
print(line_3)
assert line_3 == 2
line_4 = 2 ** 3 * 2
print(line_4)
assert line_4 == 16
line_5 = 7 + 1 > 2 * 4
print(line_5)
assert line_5 == False
````

## Exercise 21

### Challenge 3: The 1089 Trick

````python
number = 841
hundreds = number // 100
tens = number // 10 % 10
ones = number % 10
reversed_number = ones * 100 + tens * 10 + hundreds
difference = number - reversed_number
difference_hundreds = difference // 100
difference_tens = difference // 10 % 10
difference_ones = difference % 10
reversed_difference = difference_ones * 100 + difference_tens * 10 + difference_hundreds
line_1 = difference + reversed_difference
print(line_1)
assert line_1 == 1089
````

**The real program** (reads three-digit number):

```python
number = int(input())
hundreds = number // 100
tens = number // 10 % 10
ones = number % 10
reversed_number = ones * 100 + tens * 10 + hundreds
difference = number - reversed_number
difference_hundreds = difference // 100
difference_tens = difference // 10 % 10
difference_ones = difference % 10
reversed_difference = difference_ones * 100 + difference_tens * 10 + difference_hundreds
line_1 = difference + reversed_difference
print(line_1)
```

Sample input:

```text
841
```

Expected output:

```text
1089
```
