````python
# Find the station with the greatest rider count.
def busiest_station(riders):
    best_station = ""
    best_count = -1
    for station, count in riders.items():
        if count > best_count:
            best_station = station
            best_count = count
    return best_station

assert busiest_station({"Oak": 12, "Pine": 19, "Elm": 8}) == "Pine"
assert busiest_station({"North": 4}) == "North"
assert busiest_station({"Red": 7, "Blue": 7, "Gold": 3}) == "Red"
assert busiest_station({"Red": 7, "Blue": 2, "Gold": 9}) == "Gold"

print(busiest_station({"Oak": 12, "Pine": 19, "Elm": 8}))
````

````python
# Count lowercase versions of the words.
def word_counts(words):
    counts = {}
    for word in words:
        key = word.lower()
        if key in counts:
            counts[key] = counts[key] + 1
        else:
            counts[key] = 1
    return counts

assert word_counts(["Cloud", "rain", "cloud"]) == {"cloud": 2, "rain": 1}
assert word_counts(["GO", "go", "Go", "stop"]) == {"go": 3, "stop": 1}
assert word_counts([]) == {}

print(word_counts(["Cloud", "rain", "cloud"]))
````

````python
# Group words by first letter while keeping input order.
def shelve_by_first(words):
    shelves = {}
    for word in words:
        letter = word[0]
        if letter in shelves:
            shelves[letter].append(word)
        else:
            shelves[letter] = [word]
    return shelves

assert shelve_by_first(["cat", "crow", "dog", "camel"]) == {"c": ["cat", "crow", "camel"], "d": ["dog"]}
assert shelve_by_first(["sun", "moon", "star"]) == {"s": ["sun", "star"], "m": ["moon"]}
assert shelve_by_first([]) == {}

print(shelve_by_first(["cat", "crow", "dog", "camel"]))
````

````python
# Turn each uppercase letter into a Morse code, then join the codes.
def to_morse(message):
    table = {"A": ".-", "E": ".", "N": "-.", "O": "---", "S": "...", "T": "-"}
    codes = []
    for letter in message:
        codes.append(table[letter])
    return " ".join(codes)

assert to_morse("NOTE") == "-. --- - ."
assert to_morse("SOS") == "... --- ..."
assert to_morse("TEN") == "- . -."

print(to_morse("NOTE"))
````

````python
# Add Roman symbol values, subtracting a symbol before a larger one.
def roman_to_int(s):
    values = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100}
    total = 0
    for i in range(len(s)):
        value = values[s[i]]
        if i == len(s) - 1:
            total = total + value
        elif value < values[s[i + 1]]:
            total = total - value
        else:
            total = total + value
    return total

assert roman_to_int("XLII") == 42
assert roman_to_int("LXIX") == 69
assert roman_to_int("IV") == 4

print(roman_to_int("XLII"))
````

````python
# Tally letters in first-appearance order and draw one bar per letter.
def tally_chart(word):
    counts = {}
    for ch in word:
        if ch in counts:
            counts[ch] = counts[ch] + 1
        else:
            counts[ch] = 1
    rows = []
    for ch, count in counts.items():
        rows.append(f"{ch} " + "#" * count)
    return "\n".join(rows)

assert tally_chart("papaya") == "p ##\na ###\ny #"
assert tally_chart("moo") == "m #\no ##"

print(tally_chart("papaya"))
````

````python
def is_anagram(first, second):
    first_counts = {}
    for ch in first:
        first_counts[ch] = first_counts.get(ch, 0) + 1
    second_counts = {}
    for ch in second:
        second_counts[ch] = second_counts.get(ch, 0) + 1
    return first_counts == second_counts

assert is_anagram("listen", "silent") == True
assert is_anagram("cider", "crumb") == False
assert is_anagram("aab", "aba") == True

print(is_anagram("listen", "silent"))
````

````python
def luhn_valid(s):
    total = 0
    for i in range(len(s)):
        digit = int(s[len(s) - 1 - i])
        if i % 2 == 1:
            digit = digit * 2
            if digit > 9:
                digit = digit - 9
        total = total + digit
    return total % 10 == 0

assert luhn_valid("79927398713") == True
assert luhn_valid("79927398710") == False
assert luhn_valid("0") == True

print(luhn_valid("79927398713"))
````

````python
def parse_records(records):
    scores = {}
    for record in records:
        parts = record.split()
        name = parts[0]
        score = int(parts[1])
        if name in scores:
            scores[name] = scores[name] + score
        else:
            scores[name] = score
    return scores

assert parse_records(["Rin 12", "Oto 7", "Rin 5"]) == {"Rin": 17, "Oto": 7}
assert parse_records(["Pax 3"]) == {"Pax": 3}

print(parse_records(["Rin 12", "Oto 7", "Rin 5"]))
````

````python
def group_by_length(words):
    groups = {}
    for word in words:
        size = len(word)
        if size in groups:
            groups[size].append(word)
        else:
            groups[size] = [word]
    return groups

assert group_by_length(["elk", "ibex", "yak", "hare", "gnu"]) == {3: ["elk", "yak", "gnu"], 4: ["ibex", "hare"]}
assert group_by_length(["a", "to"]) == {1: ["a"], 2: ["to"]}

print(group_by_length(["elk", "ibex", "yak", "hare", "gnu"]))
````

````python
def translate(line, dictionary):
    result = []
    for word in line.split():
        result.append(dictionary.get(word, word))
    return " ".join(result)

assert translate("a red hen", {"red": "rojo", "hen": "gallina"}) == "a rojo gallina"
assert translate("blue sky", {"blue": "azul"}) == "azul sky"

print(translate("a red hen", {"red": "rojo", "hen": "gallina"}))
````

````python
def from_morse(message):
    table = {"A": ".-", "E": ".", "N": "-.", "O": "---", "S": "...", "T": "-"}
    backwards = {}
    for letter, code in table.items():
        backwards[code] = letter
    letters = []
    for code in message.split():
        letters.append(backwards[code])
    return "".join(letters)

assert from_morse("- . -.") == "TEN"
assert from_morse("... --- ...") == "SOS"

print(from_morse("- . -."))
````
