## Exercise 1

### Planet Day Lookup


```python
# Return the day length stored for a planet.
def day_hours(planet):
    hours = {"Mercury": 4223, "Earth": 24, "Mars": 25}
    return hours[planet]

assert day_hours("Earth") == 24
assert day_hours("Mars") == 25
assert day_hours("Mercury") == 4223
```

## Exercise 2

### Safe Snack Count


```python
# Return a snack count, using zero for a missing snack.
def snack_count(snacks, name):
    return snacks.get(name, 0)

assert snack_count({"apple": 5, "cracker": 8}, "apple") == 5
assert snack_count({"apple": 5, "cracker": 8}, "banana") == 0
assert snack_count({}, "cracker") == 0
```

## Exercise 3

### Supply Total


```python
# Add every labelled supply amount.
def total_supplies(supplies):
    total = 0
    for item, amount in supplies.items():
        total = total + amount
    return total

assert total_supplies({"pens": 9, "paper": 20, "tape": 2}) == 31
assert total_supplies({"clips": 7}) == 7
assert total_supplies({}) == 0
```

## Exercise 4

### Busiest Station


```python
# Find the station with the greatest rider count.
def busiest_station(riders):
    best_station = ""
    best_count = -1
    for station, count in riders.items():
        if count > best_count:
            best_station = station
            best_count = count
    return best_station

assert busiest_station({"Oak": 12, "Pine": 19, "Elm": 8}) == "Pine"
assert busiest_station({"North": 4}) == "North"
assert busiest_station({"Red": 7, "Blue": 7, "Gold": 3}) == "Red"
assert busiest_station({"Red": 7, "Blue": 2, "Gold": 9}) == "Gold"
```

## Exercise 5

### Case-Folding Word Tally


```python
# Count lowercase versions of the words.
def word_counts(words):
    counts = {}
    for word in words:
        key = word.lower()
        if key in counts:
            counts[key] = counts[key] + 1
        else:
            counts[key] = 1
    return counts

assert word_counts(["Cloud", "rain", "cloud"]) == {"cloud": 2, "rain": 1}
assert word_counts(["GO", "go", "Go", "stop"]) == {"go": 3, "stop": 1}
assert word_counts([]) == {}
```

## Exercise 6

### First-Letter Shelves


```python
# Group words by first letter while keeping input order.
def shelve_by_first(words):
    shelves = {}
    for word in words:
        letter = word[0]
        if letter in shelves:
            shelves[letter].append(word)
        else:
            shelves[letter] = [word]
    return shelves

assert shelve_by_first(["cat", "crow", "dog", "camel"]) == {"c": ["cat", "crow", "camel"], "d": ["dog"]}
assert shelve_by_first(["sun", "moon", "star"]) == {"s": ["sun", "star"], "m": ["moon"]}
assert shelve_by_first([]) == {}
```

## Exercise 7

### Sorted Score Board


```python
# Build score/name rows from lowest score to highest.
def score_board(scores):
    board = []
    for name, score in scores.items():
        board.append([score, name])
    board.sort()
    return board

assert score_board({"Mia": 14, "Leo": 9, "Zoe": 18}) == [[9, "Leo"], [14, "Mia"], [18, "Zoe"]]
assert score_board({"Ivy": 3, "Kai": 11}) == [[3, "Ivy"], [11, "Kai"]]
assert score_board({}) == []
```

## Exercise 8

### Challenge 1: Qualifying Players


```python
# Keep names whose scores meet the minimum.
def qualifiers(scores, minimum):
    passing = []
    for name, score in scores.items():
        if score >= minimum:
            passing.append(name)
    return passing

assert qualifiers({"Ana": 16, "Bo": 7, "Cy": 12}, 10) == ["Ana", "Cy"]
assert qualifiers({"Dee": 5, "Eli": 6}, 6) == ["Eli"]
assert qualifiers({"Fox": 4}, 9) == []
```

## Exercise 9

### Challenge 2: Word Winner Report


```python
# Tally lowercase words, then report the unique winner.
def word_winner(words):
    counts = {}
    for word in words:
        key = word.lower()
        counts[key] = counts.get(key, 0) + 1

    best_word = ""
    best_count = -1
    for word, count in counts.items():
        if count > best_count:
            best_word = word
            best_count = count
    return f"{best_word}: {best_count}"

assert word_winner(["Red", "blue", "red"]) == "red: 2"
assert word_winner(["owl", "fox", "FOX", "fox"]) == "fox: 3"
assert word_winner(["solo"]) == "solo: 1"
```
