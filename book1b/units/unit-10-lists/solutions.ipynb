## Exercise 1

### Score Snapshot


```python
# Return the first score, score count, and total in a new list.
def score_snapshot(scores):
    return [scores[0], len(scores), sum(scores)]

assert score_snapshot([6, 9, 4]) == [6, 3, 19]
assert score_snapshot([12]) == [12, 1, 12]
assert score_snapshot([-2, 5, 1, 3]) == [-2, 4, 7]
assert score_snapshot([0, -5, 5]) == [0, 3, 0]
```

## Exercise 2

### Scale Every Score


```python
# Return a new list with every score multiplied by factor.
def scaled_scores(scores, factor):
    scaled = []
    for score in scores:
        scaled.append(score * factor)
    return scaled

assert scaled_scores([3, 7, 2], 4) == [12, 28, 8]
assert scaled_scores([-2, 5, 0], 3) == [-6, 15, 0]
assert scaled_scores([], 9) == []
assert scaled_scores([1, -3, 4], -2) == [-2, 6, -8]
```

## Exercise 3

### Round-to-Round Changes


```python
# Return the change from each score to the score after it.
def round_changes(scores):
    changes = []
    for position in range(1, len(scores)):
        change = scores[position] - scores[position - 1]
        changes.append(change)
    return changes

assert round_changes([5, 9, 8, 14]) == [4, -1, 6]
assert round_changes([20, 15, 10]) == [-5, -5]
assert round_changes([7]) == []
assert round_changes([2, 2, 5, 1]) == [0, 3, -4]
```

## Exercise 4

### Repair the In-Place Sorter


```python
# Sort the received list in place and return the changed list.
def sort_in_place(scores):
    scores.sort()
    return scores

assert sort_in_place([8, 3, 6]) == [3, 6, 8]
assert sort_in_place([-1, -7, -4]) == [-7, -4, -1]
assert sort_in_place([5, 5, 2]) == [2, 5, 5]
scores_to_sort = [4, 1, 3]
returned_scores = sort_in_place(scores_to_sort)
assert scores_to_sort == [1, 3, 4]
assert returned_scores == [1, 3, 4]
```

## Exercise 5

### Winning Position


```python
# Return the earliest position containing the largest score.
def position_of_largest(scores):
    best_position = 0
    for position in range(1, len(scores)):
        if scores[position] > scores[best_position]:
            best_position = position
    return best_position

assert position_of_largest([14, 22, 17]) == 1
assert position_of_largest([-8, -3, -10]) == 1
assert position_of_largest([9, 12, 12, 4]) == 1
assert position_of_largest([5, 5, 1]) == 0
```

## Exercise 6

### Progress Totals


```python
# Return the running page total after every session.
def progress_totals(pages):
    totals = []
    running_total = 0
    for page_count in pages:
        running_total = running_total + page_count
        totals.append(running_total)
    return totals

assert progress_totals([4, 6, 3]) == [4, 10, 13]
assert progress_totals([10, 0, 5, 2]) == [10, 10, 15, 17]
assert progress_totals([]) == []
assert progress_totals([-2, 5, -1]) == [-2, 3, 2]
```

## Exercise 7

### Above the Team Average


```python
# Return the scores that are strictly above the team average.
def above_average(scores):
    average = sum(scores) / len(scores)
    kept = []
    for score in scores:
        if score > average:
            kept.append(score)
    return kept

assert above_average([9, 15, 12, 20]) == [15, 20]
assert above_average([4, 4, 4]) == []
assert above_average([-6, -2, -4]) == [-2]
assert above_average([0, 10]) == [10]
```

## Exercise 8

### Challenge 1: Keep Approved Values


```python
# Return approved values in their original order, including repeats.
def keep_approved(values, approved):
    kept = []
    for value in values:
        if value in approved:
            kept.append(value)
    return kept

assert keep_approved([7, 2, 7, 5], [2, 5]) == [2, 5]
assert keep_approved([3, 3, 8], [3, 9]) == [3, 3]
assert keep_approved([4, 6], [1, 2]) == []
assert keep_approved([0, -1, 0, 2], [0, 2]) == [0, 0, 2]
```

## Exercise 9

### Challenge 2: Longest Word Position


```python
# Return the earliest position containing a longest word.
def position_of_longest(words):
    best_position = 0
    for position in range(1, len(words)):
        if len(words[position]) > len(words[best_position]):
            best_position = position
    return best_position

assert position_of_longest(["owl", "panther", "fox"]) == 1
assert position_of_longest(["kiwi", "apricot", "fig"]) == 1
assert position_of_longest(["sun", "map", "key"]) == 0
assert position_of_longest(["tiny", "enormous", "medium", "enormous"]) == 1
```
