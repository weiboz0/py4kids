## Exercise 1

### Score Snapshot


````python
# Return the first score, score count, and total in a new list.
def score_snapshot(scores):
    return [scores[0], len(scores), sum(scores)]

assert score_snapshot([6, 9, 4]) == [6, 3, 19]
assert score_snapshot([12]) == [12, 1, 12]
assert score_snapshot([-2, 5, 1, 3]) == [-2, 4, 7]
assert score_snapshot([0, -5, 5]) == [0, 3, 0]

print(score_snapshot([6, 9, 4]))
print(score_snapshot([12]))
print(score_snapshot([-2, 5, 1, 3]))
````

## Exercise 3

### Round-to-Round Changes


````python
# Return the change from each score to the score after it.
def round_changes(scores):
    changes = []
    for position in range(1, len(scores)):
        change = scores[position] - scores[position - 1]
        changes.append(change)
    return changes

assert round_changes([5, 9, 8, 14]) == [4, -1, 6]
assert round_changes([20, 15, 10]) == [-5, -5]
assert round_changes([7]) == []
assert round_changes([2, 2, 5, 1]) == [0, 3, -4]

print(round_changes([5, 9, 8, 14]))
print(round_changes([20, 15, 10]))
print(round_changes([7]))
````

## Exercise 7

### Above the Team Average


````python
# Return the scores that are strictly above the team average.
def above_average(scores):
    average = sum(scores) / len(scores)
    kept = []
    for score in scores:
        if score > average:
            kept.append(score)
    return kept

assert above_average([9, 15, 12, 20]) == [15, 20]
assert above_average([4, 4, 4]) == []
assert above_average([-6, -2, -4]) == [-2]
assert above_average([0, 10]) == [10]

print(above_average([9, 15, 12, 20]))
print(above_average([4, 4, 4]))
print(above_average([-6, -2, -4]))
````

## Exercise 9

### Second Largest


````python
# Find the greatest number strictly below the maximum.
def second_largest(numbers):
    top = max(numbers)
    best = min(numbers)
    for number in numbers:
        if number < top and number > best:
            best = number
    return best

assert second_largest([7, 19, 4, 19, 12]) == 12
assert second_largest([5, 3]) == 3
assert second_largest([-8, -1, -3]) == -3

print(f"Second largest: {second_largest([7, 19, 4, 19, 12])}")
print(f"Second largest: {second_largest([5, 3])}")
````

## Exercise 11

### More Practice: Median


````python
def median(numbers):
    ordered = sorted(numbers)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2

assert median([9, 2, 7, 4, 5]) == 5
assert median([8, 3, 6, 1]) == 4.5
assert median([3]) == 3

print(f"Median: {median([9, 2, 7, 4, 5])}")
print(f"Median: {median([8, 3, 6, 1])}")
````

## Exercise 12

### More Practice: Mode


````python
def mode(numbers):
    best = numbers[0]
    best_count = 0
    for value in numbers:
        count = 0
        for other in numbers:
            if other == value:
                count = count + 1
        if count > best_count:
            best = value
            best_count = count
    return best

assert mode([4, 2, 4, 7, 2, 4]) == 4
assert mode([2, 3, 3, 2]) == 2
assert mode([9]) == 9

print(mode([4, 2, 4, 7, 2, 4]))
````

## Exercise 14

### More Practice: Reverse in Place


````python
def reverse_in_place(numbers):
    left = 0
    right = len(numbers) - 1
    while left < right:
        temporary = numbers[left]
        numbers[left] = numbers[right]
        numbers[right] = temporary
        left = left + 1
        right = right - 1
    return numbers

sample = [6, 1, 9, 4]
assert reverse_in_place(sample) == [4, 9, 1, 6]
assert reverse_in_place([7]) == [7]
assert reverse_in_place([1, 2, 3]) == [3, 2, 1]

print(reverse_in_place([6, 1, 9, 4]))
````

````python
def bar_chart(labels, values):
    rows = []
    for i in range(len(labels)):
        rows.append(f"{labels[i]} " + "#" * values[i])
    return "\n".join(rows)

assert bar_chart(["Mon", "Tue", "Wed"], [3, 5, 2]) == "Mon ###\nTue #####\nWed ##"
assert bar_chart(["Fri"], [0]) == "Fri "

print(bar_chart(["Mon", "Tue", "Wed"], [3, 5, 2]))
````

## Exercise 19

### More Practice: Grid Printer


````python
def grid_text(grid):
    rows = []
    for values in grid:
        row = ""
        for value in values:
            if value == 1:
                row = row + "#"
            else:
                row = row + "."
        rows.append(row)
    return "\n".join(rows)

assert grid_text([[1, 0, 1], [0, 1, 0]]) == "#.#\n.#."
assert grid_text([[0]]) == "."

print(grid_text([[1, 0, 1], [0, 1, 0]]))
````

## Exercise 20

### More Practice: Tic-Tac-Toe Winner


````python
def winner(board):
    lines = []
    for row in board:
        lines.append(row)
    for col in range(3):
        line = ""
        for row in range(3):
            line = line + board[row][col]
        lines.append(line)
    diagonal = ""
    other = ""
    for i in range(3):
        diagonal = diagonal + board[i][i]
        other = other + board[i][2 - i]
    lines.append(diagonal)
    lines.append(other)
    for line in lines:
        if line == "XXX":
            return "X"
        elif line == "OOO":
            return "O"
    return "none"

assert winner(["XOX", "OXO", "OOX"]) == "X"
assert winner(["OOO", "XX.", "X.."]) == "O"
assert winner(["XOX", "XOO", "OXX"]) == "none"
assert winner(["X..", "X..", "X.."]) == "X"

print(winner(["XOX", "OXO", "OOX"]))
print(winner(["OOO", "XX.", "X.."]))
print(winner(["XOX", "XOO", "OXX"]))
````

````python
def is_magic(grid):
    target = sum(grid[0])
    for row in grid:
        if sum(row) != target:
            return False
    for col in range(3):
        total = 0
        for row in range(3):
            total = total + grid[row][col]
        if total != target:
            return False
    first = 0
    second = 0
    for i in range(3):
        first = first + grid[i][i]
        second = second + grid[i][2 - i]
    return first == target and second == target

assert is_magic([[2, 7, 6], [9, 5, 1], [4, 3, 8]]) == True
assert is_magic([[1, 2, 3], [4, 5, 6], [7, 8, 9]]) == False
assert is_magic([[5, 5, 5], [5, 5, 5], [5, 5, 5]]) == True

print(is_magic([[2, 7, 6], [9, 5, 1], [4, 3, 8]]))
print(is_magic([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
````

````python
# Return the earliest position containing a longest word.
def position_of_longest(words):
    best_position = 0
    for position in range(1, len(words)):
        if len(words[position]) > len(words[best_position]):
            best_position = position
    return best_position

assert position_of_longest(["owl", "panther", "fox"]) == 1
assert position_of_longest(["kiwi", "apricot", "fig"]) == 1
assert position_of_longest(["sun", "map", "key"]) == 0
assert position_of_longest(["tiny", "enormous", "medium", "enormous"]) == 1

print(position_of_longest(["owl", "panther", "fox"]))
print(position_of_longest(["kiwi", "apricot", "fig"]))
print(position_of_longest(["sun", "map", "key"]))
````

## Exercise 33

### Challenge: Vertical Bar Chart


````python
def vertical_bars(values):
    rows = []
    top = max(values)
    for i in range(top):
        level = top - i
        row = ""
        for value in values:
            if value >= level:
                row = row + "#"
            else:
                row = row + "."
        rows.append(row)
    return "\n".join(rows)

assert vertical_bars([2, 4, 1]) == ".#.\n.#.\n##.\n###"
assert vertical_bars([1]) == "#"

print(vertical_bars([2, 4, 1]))
````

### Challenge: Minesweeper Counts


````python
def mine_counts(board):
    rows = len(board)
    cols = len(board[0])
    result = []
    for row in range(rows):
        line = ""
        for col in range(cols):
            if board[row][col] == "*":
                line = line + "*"
            else:
                count = 0
                for dr in range(-1, 2):
                    for dc in range(-1, 2):
                        r = row + dr
                        c = col + dc
                        if 0 <= r < rows and 0 <= c < cols and board[r][c] == "*":
                            count = count + 1
                line = line + str(count)
        result.append(line)
    return result

assert mine_counts(["*..", "...", ".*."]) == ["*10", "221", "1*1"]
assert mine_counts([".."]) == ["00"]

print(mine_counts(["*..", "...", ".*." ]))
````

## Exercise 35

### Challenge: Game of Life Step


````python
def life_step(board):
    rows = len(board)
    cols = len(board[0])
    result = []
    for row in range(rows):
        line = ""
        for col in range(cols):
            count = 0
            for dr in range(-1, 2):
                for dc in range(-1, 2):
                    r = row + dr
                    c = col + dc
                    if (dr != 0 or dc != 0) and 0 <= r < rows and 0 <= c < cols and board[r][c] == "#":
                        count = count + 1
            if board[row][col] == "#" and (count == 2 or count == 3):
                line = line + "#"
            elif board[row][col] == "." and count == 3:
                line = line + "#"
            else:
                line = line + "."
        result.append(line)
    return result

assert life_step(["...", "###", "..."]) == [".#.", ".#.", ".#."]
assert life_step(["..."]) == ["..."]

print(life_step(["...", "###", "..."]))
````

## Exercise 36

### Challenge: Selection Sort


````python
def selection_sort(numbers):
    for i in range(len(numbers)):
        smallest = i
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[smallest]:
                smallest = j
        temporary = numbers[i]
        numbers[i] = numbers[smallest]
        numbers[smallest] = temporary
    return numbers

sample = [29, 10, 14, 37, 13]
assert selection_sort(sample) == [10, 13, 14, 29, 37]
assert selection_sort([2, 1, 2]) == [1, 2, 2]

print(selection_sort([29, 10, 14, 37, 13]))
````

## Exercise 37

### Challenge: Statistics Report


````python
def median(numbers):
    ordered = sorted(numbers)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2

def mode(numbers):
    best = numbers[0]
    best_count = 0
    for value in numbers:
        count = 0
        for other in numbers:
            if other == value:
                count = count + 1
        if count > best_count:
            best = value
            best_count = count
    return best

def stats_report(numbers):
    mean = sum(numbers) / len(numbers)
    middle = median(numbers)
    common = mode(numbers)
    return f"mean {mean} median {middle} mode {common}"

assert stats_report([6, 2, 9, 2, 5]) == "mean 4.8 median 5 mode 2"
assert stats_report([1, 1, 1]) == "mean 1.0 median 1 mode 1"

print(stats_report([6, 2, 9, 2, 5]))
````
