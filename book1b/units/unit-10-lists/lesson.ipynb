# Lists

A game club has a whole row of scores to organize, compare, and improve. Your challenge is to keep related values together, visit every value, and build useful new collections without losing track of the original data.

By Lesson 3, you will turn a score list into filtered results and running totals with reusable functions.

## Lesson 1: Build and Read

### Start the score board

A **list** keeps several values in order. Square brackets make a list literal, and commas separate its items. An index reaches one position: index `0` is the first position.

```python
def first_score(scores):
    return scores[0]

practice_scores = [3, 1, 2]
practice_scores.append(6)
print(practice_scores)
print(first_score(practice_scores))
```

**Visible payoff:** `practice_scores` becomes `[3, 1, 2, 6]`, and `first_score(practice_scores)` returns `3`. The method call `practice_scores.append(6)` adds one item at the end of that list.

`first_score` expects a list with at least one item. An index asks for an existing position; position `0` would not exist in an empty list.

### Loop over the values

When you need each value but not its position, loop directly over the list. The loop below visits one score at a time. `len(scores)` gives the number of items, and `sum(scores)` gives their total.

```python
def total_with_values(scores):
    total = 0
    for score in scores:
        total = total + score
    return total

def score_report(scores):
    return f"{len(scores)} scores total {sum(scores)}"

print(total_with_values([4, 7, 5]))
print(score_report([4, 7, 5]))
```

The calls return `16` and the report `3 scores total 16`. The hand-written loop exposes the running total, while `sum` performs the same common job in one call. A direct value loop is the clearest choice when the positions do not matter.

### Build a new list with `append`

Start with an empty list when the answer will be built one item at a time. Each loop trip calculates one new value and appends it.

```python
def double_values(numbers):
    doubled = []
    for number in numbers:
        doubled.append(number * 2)
    return doubled

print(double_values([2, 5, 8]))
print(double_values([-3, 0, 4]))
```

The calls return `[4, 10, 16]` and `[-6, 0, 8]`. The input list supplies the values; the empty `doubled` list grows with one `append` call per trip. The function returns a new list.

### Loop over the positions

Sometimes a calculation needs both the current item and another position. `range(len(scores))` supplies every valid index from `0` through one less than the list length. To compare each score with the previous score, begin at index `1`.

```python
def score_changes(scores):
    changes = []
    for i in range(1, len(scores)):
        change = scores[i] - scores[i - 1]
        changes.append(change)
    return changes

print(score_changes([8, 11, 10, 15]))
print(score_changes([6]))
```

The calls return `[3, -1, 5]` and `[]`. The index loop is useful because `scores[i]` and `scores[i - 1]` reach neighboring positions. A one-item list has no pair to compare, so its result stays empty.

## Lesson 2: Order and Choose

A tournament organizer needs a low-to-high score board and the position of the winning score. Sorting can order the values, but choosing a position requires a careful index loop.

### Notice: `sort` changes a list and returns `None`

The list method `sort` rearranges the existing list in place. Its returned value is `None`, which means it does not hand back a useful new list. Predict the two printed lines.

```python
nums = [9, 2, 6]
scores = nums.sort()
print(nums)
print(scores)
```

Python prints `[2, 6, 9]` and then `None`. The call changed `nums`, but `scores = nums.sort()` saved the method's missing return value. This is the `None` trap: do not assign the result of `sort` when you want the changed list.

### Sort in place when changing the input is the job

Call `sort` as its own statement, then return the now-ordered list if the caller needs that value.

```python
def order_in_place(scores):
    scores.sort()
    return scores

print(order_in_place([12, 5, 8]))
print(order_in_place([-2, -9, -4]))
```

The calls return `[5, 8, 12]` and `[-9, -4, -2]`. This function deliberately changes the list it receives. Its name says that in-place change is part of the job.

### Use `sorted` when the original order must stay

The built-in function `sorted(scores)` returns a new ordered list. It leaves the original list in its old order.

```python
def ordered_copy(scores):
    return sorted(scores)

original_scores = [14, 7, 11]
new_order = ordered_copy(original_scores)
print(original_scores)
print(new_order)
```

Python prints `[14, 7, 11]` and `[7, 11, 14]`. Choose `sort` for an intentional in-place change and `sorted` for a separate ordered result.

### Built-in list calculations

For a list with at least one number, `min` returns its smallest item and `max` returns its largest item. `sum` returns the total. These built-ins answer common value questions, but they do not tell you the position of a winning value.

```python
def number_summary(numbers):
    return [min(numbers), max(numbers), sum(numbers)]

print(number_summary([7, 2, 10, 4]))
print(number_summary([-8, -3, -12]))
```

The calls return `[2, 10, 23]` and `[-12, -3, -23]`. The second call is a reminder that the largest negative number is still negative.

### Find an extreme by hand

A hand-written extreme loop remembers the best value seen so far. This function assumes the list has at least one item, so `numbers[0]` is a real starting value. Never start `best` at `0`: a list of negative numbers might not contain any value as large as 0.

```python
def find_largest(numbers):
    best = numbers[0]
    for number in numbers:
        if number > best:
            best = number
    return best

print(find_largest([5, 12, 9]))
print(find_largest([-11, -4, -7]))
```

The calls return `12` and `-4`. Seeding from the first item makes the loop correct for negative values too. The built-in `max` is shorter when only the value matters; the hand-written loop can be adapted when the position matters.

### Find the position of the maximum

An **argmax** is the position of a largest value. Remember a position rather than the value itself, then compare through list indexes. This function assumes the list has at least one item.

```python
def position_of_largest(numbers):
    best_position = 0
    for i in range(1, len(numbers)):
        if numbers[i] > numbers[best_position]:
            best_position = i
    return best_position

print(position_of_largest([13, 21, 18]))
print(position_of_largest([-2, -7, -1]))
```

The calls return `1` and `2`. Because the comparison uses `>` rather than `>=`, equal largest values leave the earliest winning position in place.

### Apply argmax to words

The same position loop can compare calculated qualities instead of number values. To find the longest word, compare the lengths of the words at two positions. The list must have at least one item.

```python
def position_of_longest(words):
    best_position = 0
    for i in range(1, len(words)):
        if len(words[i]) > len(words[best_position]):
            best_position = i
    return best_position

print(position_of_longest(["ant", "butterfly", "moth"]))
print(position_of_longest(["red", "blue", "gold"]))
```

The calls return `1` and `1`. In the second list, `blue` and `gold` tie at length 4, so the earlier position wins. This is why an index loop can solve more than the built-in value questions.

## Lesson 3: Filter and Combine

A coach needs a clean list of qualifying scores and a progress line that shows the total after every round. Both jobs build a new list one decision at a time.

### Count values that pass a test

A counter can record how many items pass a condition. It starts at 0 and increases only when the current score reaches the minimum.

```python
def count_qualifiers(scores, minimum):
    count = 0
    for score in scores:
        if score >= minimum:
            count = count + 1
    return count

print(count_qualifiers([8, 13, 11, 6], 10))
print(count_qualifiers([4, 5, 3], 7))
```

The calls return `2` and `0`. Counting answers how many values pass; filtering keeps the passing values themselves.

### Filter into a new list

To filter, start with an empty result list. Loop over the input, test each item, and call `append` only inside the `if` body. This example also practices membership with `in`.

```python
def keep_allowed(values, allowed):
    kept = []
    for value in values:
        if value in allowed:
            kept.append(value)
    return kept

print(keep_allowed([4, 9, 2, 9, 7], [2, 4, 7]))
print(keep_allowed([5, 6], [1, 2, 3]))
```

The calls return `[4, 2, 7]` and `[]`. The result keeps the input order and includes only values that belong to `allowed`. Building with an explicit loop makes every filtering decision visible.

### Build prefix sums

A **prefix sum** is the total so far after each item. Update the running total first, then append that total once per loop trip.

```python
def prefix_sums(numbers):
    totals = []
    running_total = 0
    for number in numbers:
        running_total = running_total + number
        totals.append(running_total)
    return totals

print(prefix_sums([3, 5, 2, 6]))
print(prefix_sums([-2, 7, -1]))
```

The calls return `[3, 8, 10, 16]` and `[-2, 5, 4]`. Each returned item answers, “What is the total through this position?”

### Put it together: keep above-average scores

This function assumes the list has at least one item. First calculate the average with `sum(scores) / len(scores)`. Then filter every score that is strictly greater than that average into a new list.

```python
def above_average(scores):
    average = sum(scores) / len(scores)
    kept = []
    for score in scores:
        if score > average:
            kept.append(score)
    return kept

print(above_average([10, 16, 12, 18]))
print(above_average([7, 7, 7]))
```

The first average is `14.0`, so the function returns `[16, 18]`. The second average is `7.0`, so no score is strictly above it and the function returns `[]`.

**Put it together:** lists keep ordered data, value loops visit items, index loops connect values to positions, and `append` builds new results. Sorting, finding an extreme, filtering, and prefix sums are different questions built from those same tools.
