## Exercise 1

### Garden Border

The function returns the full distance around two widths and two heights.

````python
# Return the perimeter of a rectangle.
def rectangle_perimeter(width, height):
    return 2 * width + 2 * height

assert rectangle_perimeter(8, 5) == 26
assert rectangle_perimeter(12, 3) == 30
assert rectangle_perimeter(7.5, 2) == 19.0
````

## Exercise 2

### Repair the Ride-Cost Tool

Returning the cost lets the caller save it or use it in another calculation.

````python
# Return the total cost for the requested number of rides.
def ticket_cost(rides):
    return rides * 3

assert ticket_cost(1) == 3
assert ticket_cost(4) == 12
assert ticket_cost(7) == 21
assert ticket_cost(4) + 2 == 14
````

## Exercise 3

### Exact-Multiple Test

A remainder of zero means the number splits into equal groups.

````python
# Report whether divisor divides number evenly.
def is_multiple(number, divisor):
    return number % divisor == 0

assert is_multiple(42, 7) is True
assert is_multiple(43, 7) is False
assert is_multiple(18, 3) is True
````

## Exercise 4

### Score Span

The requested built-ins find the highest and lowest scores before subtraction.

````python
# Return the distance from the lowest score to the highest score.
def score_span(a, b, c):
    higher_first_two = max(a, b)
    highest = max(higher_first_two, c)
    lowest = min(a, b, c)
    return highest - lowest

assert score_span(4, 11, 7) == 7
assert score_span(9, 2, 13) == 11
assert score_span(6, 6, 6) == 0
````

## Exercise 5

### Total through an Endpoint

A local running total grows once for each number through the endpoint.

````python
# Return the total of the integers from 1 through n.
def sum_to_n(n):
    total = 0
    for number in range(1, n + 1):
        total = total + number
    return total

assert sum_to_n(4) == 10
assert sum_to_n(8) == 36
assert sum_to_n(1) == 1
````

## Exercise 6

### Fibonacci Signal

The temporary name protects the next value while the pair moves forward.

````python
# Return the Fibonacci value at position n.
def fib(n):
    a = 0
    b = 1
    for step in range(n):
        nxt = a + b
        a = b
        b = nxt
    return a

assert fib(0) == 0
assert fib(2) == 1
assert fib(5) == 5
assert fib(9) == 34
````

## Exercise 8

### Challenge 1: Euclid's GCD Tool

Each loop trip replaces the pair with the second number and the remainder.

````python
# Return the greatest common divisor of two positive integers.
def gcd(a, b):
    while b != 0:
        remainder = a % b
        a = b
        b = remainder
    return a

assert gcd(54, 24) == 6
assert gcd(35, 15) == 5
assert gcd(81, 27) == 27
````

## Exercise 9

### Challenge 2: Prime Gate

A found divisor returns `False` immediately; reaching the end means the key is prime.

````python
# Report whether n is prime.
def is_prime(n):
    for divisor in range(2, n):
        if n % divisor == 0:
            return False
    return True

assert is_prime(17) is True
assert is_prime(21) is False
assert is_prime(2) is True
````
