## Exercise 4

### Score Span

````python
# Return the distance from the lowest score to the highest score.
def score_span(a, b, c):
    higher_first_two = max(a, b)
    highest = max(higher_first_two, c)
    lowest = min(a, b, c)
    return highest - lowest

assert score_span(4, 11, 7) == 7
assert score_span(9, 2, 13) == 11
assert score_span(6, 6, 6) == 0

print(score_span(4, 11, 7))
print(score_span(9, 2, 13))
print(score_span(6, 6, 6))
````

## Exercise 10

### Least Common Multiple

````python
# Return the greatest common divisor, then use it for the shared cycle.
def gcd(a, b):
    while b != 0:
        remainder = a % b
        a = b
        b = remainder
    return a

def lcm(a, b):
    return a * b // gcd(a, b)

assert lcm(4, 6) == 12
assert lcm(21, 6) == 42
assert lcm(5, 7) == 35

print(lcm(4, 6))
print(lcm(21, 6))
````

### More Practice: Digital Root

````python
# Sum decimal digits, then repeat until one digit remains.
def digit_sum(n):
    total = 0
    while n > 0:
        total = total + n % 10
        n = n // 10
    return total

def digital_root(n):
    while n >= 10:
        n = digit_sum(n)
    return n

assert digit_sum(9875) == 29
assert digit_sum(29) == 11
assert digital_root(9875) == 2
assert digital_root(9) == 9
assert digital_root(99) == 9

print(digital_root(9875))
````

## Exercise 12

### More Practice: Primes Below

````python
# Count prime labels strictly below the limit.
def is_prime(n):
    for divisor in range(2, n):
        if n % divisor == 0:
            return False
    return True

def count_primes_below(limit):
    count = 0
    for number in range(2, limit):
        if is_prime(number):
            count = count + 1
    return count

assert count_primes_below(100) == 25
assert count_primes_below(10) == 4
assert count_primes_below(2) == 0

print(count_primes_below(100))
````

## Exercise 13

### More Practice: Roman Numeral

````python
# Build a Roman numeral by taking the largest available symbol.
def to_roman(n):
    result = ""
    while n > 0:
        if n >= 10:
            result = result + "X"
            n = n - 10
        elif n >= 9:
            result = result + "IX"
            n = n - 9
        elif n >= 5:
            result = result + "V"
            n = n - 5
        elif n >= 4:
            result = result + "IV"
            n = n - 4
        else:
            result = result + "I"
            n = n - 1
    return result

assert to_roman(38) == "XXXVIII"
assert to_roman(14) == "XIV"
assert to_roman(9) == "IX"

print(to_roman(38))
print(to_roman(14))
````

## Exercise 14

### More Practice: Days in a Month

````python
# Apply the leap-year rule when choosing February's length.
def is_leap(year):
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)

def days_in_month(month, year):
    if month == 2:
        if is_leap(year):
            return 29
        else:
            return 28
    elif month == 4 or month == 6 or month == 9 or month == 11:
        return 30
    else:
        return 31

assert days_in_month(2, 2000) == 29
assert days_in_month(2, 2100) == 28
assert days_in_month(4, 2023) == 30
assert days_in_month(1, 2023) == 31
assert days_in_month(2, 2024) == 29

print(days_in_month(2, 2000))
print(days_in_month(2, 2100))
print(days_in_month(4, 2023))
print(days_in_month(1, 2023))
````

## Exercise 32

### Challenge: Longest Collatz

````python
# Count Collatz steps and keep the earliest start with the greatest count.
def collatz_steps(n):
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        steps = steps + 1
    return steps

def longest_collatz_below(limit):
    best_start = 1
    best_steps = 0
    for start in range(1, limit):
        steps = collatz_steps(start)
        if steps > best_steps:
            best_start = start
            best_steps = steps
    return best_start

assert collatz_steps(18) == 20
assert longest_collatz_below(20) == 18
assert collatz_steps(1) == 0
assert longest_collatz_below(2) == 1

print(collatz_steps(18))
print(longest_collatz_below(20))
````
