## Exercise 2

### Supply Label

```python
# Build a two-line label for the shared supply shelf.
item = "markers"
location = "blue bin"
line_1 = f"Item: {item}"
print(line_1)
assert line_1 == "Item: markers"
line_2 = f"Location: {location}"
print(line_2)
assert line_2 == "Location: blue bin"
```
