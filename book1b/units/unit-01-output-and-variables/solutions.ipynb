## Exercise 1

### Welcome Sign

## Exercise 2

### Supply Label

```python
# Build a two-line label for the shared supply shelf.
item = "markers"
location = "blue bin"
item_line = f"Item: {item}"
location_line = f"Location: {location}"
result = item_line + "\n" + location_line
print(item_line)
print(location_line)
assert result == "Item: markers\nLocation: blue bin"
```

## Exercise 3

### Repeat the Motto

```python
motto = "Keep trying"
result = motto + "\n" + motto + "\n" + motto
print(motto)
print(motto)
print(motto)
assert result == "Keep trying\nKeep trying\nKeep trying"
```

## Exercise 11

### Challenge 1: Two Member Badges

```python
first_member = "Inez"
first_color = "gold"
second_member = "Omar"
second_color = "green"
first_badge = f"Member: {first_member}\nColor: {first_color}"
second_badge = f"Member: {second_member}\nColor: {second_color}"
result = first_badge + "\n\n" + second_badge
print(result)
assert result == "Member: Inez\nColor: gold\n\nMember: Omar\nColor: green"
```

## Exercise 12

### Challenge 2: Error Detective Card

```python
activity_name = "Puzzle Lab"
room_name = "Room B"
result = (
    "ACTIVITY CARD\n"
    + f"Activity: {activity_name}\n"
    + "Room: "
    + room_name
)
print("ACTIVITY CARD")
print(f"Activity: {activity_name}")
print("Room: " + room_name)
assert result == "ACTIVITY CARD\nActivity: Puzzle Lab\nRoom: Room B"
```
