## Lesson 1: Reach into a String

### Build badge initials

A **string** is a sequence of characters. Each character has a numbered position called an **index**. Python starts numbering at 0, so `[0]` reaches the first character. The index `-1` reaches the last character.

```python
first_name = "Maya"
last_name = "Chen"
initials = first_name[0] + last_name[0]

print(initials)
print(first_name[-1])
```

The initials are `MC`, and the last character of `Maya` is `a`. The brackets reach one character, so `first_name[0]` is itself a one-character string.

Meaningful names such as `first_name` and `initials` help a reader understand the job.

### Cut out a substring with a slice

A **slice** reaches a section of a string. In `text[a:b]`, position `a` is included and position `b` is excluded. This is the same stop rule used by `range(a, b)`.

```python
signal = "ORBIT-739"
word_part = signal[0:5]
number_part = signal[6:9]

print(word_part)
print(number_part)
```

**Visible payoff:** the two lines print `ORBIT` and then `739`. The first slice includes positions 0 through 4 but stops before 5. The second includes positions 6 through 8 but stops before 9. A slice is still a string, even when it contains only part of the original string.

### Measure the sequence

`len(text)` returns the number of characters as an integer. Spaces and punctuation are characters too.

```python
call_sign = "Blue Fox"
size = len(call_sign)

print(f"Badge text: {call_sign}")
print("Characters: " + str(size))
```

The f-string places the string value directly in a report. The next report uses `str(size)` to convert the integer 8 into text before concatenating with `+`.

## Lesson 2: Clean and Test

### Repair a messy signal

A control panel receives extra spaces and mixed capital letters. String methods are small tools attached to a string value. This unit uses only `upper`, `lower`, `strip`, and `replace`.

```python
raw_signal = "  RED COMET  "
clean_signal = raw_signal.strip().lower()
changed_signal = clean_signal.replace("red", "blue")
loud_signal = changed_signal.upper()

print(clean_signal)
print(changed_signal)
print(loud_signal)
```

**Visible payoff:** the three lines print `red comet`, `blue comet`, and `BLUE COMET`, in that order. `strip()` removes spaces from the two ends. `lower()` and `upper()` make new lowercase or uppercase strings. `replace(old, new)` makes a new string with each matching piece swapped. The original `raw_signal` stays unchanged, so save a returned string when you need it later.

### Test membership

The `in` operator asks whether a character or substring appears in a string. It returns a boolean. Matching is case-sensitive, so cleaning case first can make a test dependable. The two tests below print `True` and then `False`.

```python
message = "Launch at Dawn"
quiet_message = message.lower()

print("dawn" in quiet_message)
print("moon" in quiet_message)
```

### Test a palindrome from both ends

A **palindrome** reads the same from left to right and right to left. Compare matching positions from the two ends. Only half the string needs checking.

```python
def is_palindrome(text):
    cleaned = text.lower().replace(" ", "")
    for i in range(len(cleaned) // 2):
        if cleaned[i] != cleaned[len(cleaned) - 1 - i]:
            return False
    return True

print(is_palindrome("Never odd or even"))
print(is_palindrome("space camp"))
```

For each loop counter `i`, the left position is `i` and the matching right position is `len(cleaned) - 1 - i`. A mismatch returns `False` immediately. `return True` belongs after the loop because every tested pair must match first.

### Count vowels by condition

A counter can grow only when the current character passes a membership test. The report below prints `Vowels: 5`.

```python
def vowel_count(text):
    total = 0
    cleaned = text.lower()
    for ch in cleaned:
        if ch in "aeiou":
            total = total + 1
    return total

amount = vowel_count("Solar eclipse")
print(f"Vowels: {amount}")
```

### Choose among three character kinds

Membership tests also fit an `if` / `elif` / `else` branch. Test the narrow vowel case first, the wider alphabet case second, and everything else last. The three calls below print `vowel`, `consonant`, and `other`.

```python
def character_kind(ch):
    letter = ch.lower()
    if letter in "aeiou":
        return "vowel"
    elif letter in "abcdefghijklmnopqrstuvwxyz":
        return "consonant"
    else:
        return "other"

print(character_kind("E"))
print(character_kind("z"))
print(character_kind("?"))
```

## Lesson 3: Transform and Search

### Build a new string one character at a time

A transmission effect needs every character doubled. Strings can be accumulated with `+`: begin with an empty string, transform each character, and save the growing result.

```python
def double_letters(text):
    transformed = ""
    for ch in text:
        transformed = transformed + ch + ch
    return transformed

print(double_letters("Go!"))
print(double_letters("42"))
```

**Visible payoff:** the calls print `GGoo!!` and then `4422`. The accumulator `transformed` starts empty. Each loop trip adds the current character twice. The function returns a new string; it does not change `text`. This is the **transform-each** pattern.

### Search from left to right

A navigation display needs the first position of one character. Visit positions in order. Return the loop counter as soon as the character matches, or return `-1` only after the whole loop ends.

```python
def position(text, ch):
    for i in range(len(text)):
        if text[i] == ch:
            return i
    return -1

print(position("meteor", "e"))
print(position("meteor", "x"))
```

The first call returns the integer 1. The second returns `-1`, the signal that no position matched. Returning inside the loop preserves the first match. Returning `-1` inside the loop would stop too early. The short report below prints `Saved position: 4`.

```python
saved_position = position("satellite", "l")
report = "Saved position: " + str(saved_position)
print(report)
```

### Count one chosen character

Searching for the first match and counting every match are related but different jobs. For a character count, keep walking after a match and update an integer counter. The two calls below print `2` and then `0`.

```python
def count_char(text, ch):
    matches = 0
    for current in text:
        if current == ch:
            matches = matches + 1
    return matches

print(count_char("balloon", "l"))
print(count_char("balloon", "z"))
```

### Print an alphabet scan table

A report can scan the alphabet one letter at a time. The outer loop chooses a letter, and `count_char` scans the message for that letter. `print` writes one table row per alphabet letter.

```python
def print_letter_table(text):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    cleaned = text.lower()
    print("letter | total")
    for letter in alphabet:
        total = count_char(cleaned, letter)
        print(f"{letter} | {total}")

print_letter_table("Cabana!")
```

The table always has one row for each of the 26 lowercase letters. For `Cabana!`, the nonzero rows are `a | 3`, `b | 1`, `c | 1`, and `n | 1`; all other rows show `0`. It reports every total without choosing a winning letter. The loops make each scan step visible.

### Challenge: Shift a message

A Caesar shift moves each letter forward through the alphabet. Build the search helper first. Then lowercase the message, search the alphabet for each character, and use remainder arithmetic to wrap from `z` to `a`. Nonletters pass through unchanged. Here `k >= 0`.

```python
def position(alphabet, ch):
    for i in range(len(alphabet)):
        if alphabet[i] == ch:
            return i
    return -1

def shift(message, k):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    shifted = ""
    cleaned = message.lower()
    for ch in cleaned:
        pos = position(alphabet, ch)
        if pos == -1:
            shifted = shifted + ch
        else:
            new_pos = (pos + k) % 26
            shifted = shifted + alphabet[new_pos]
    return shifted

print(shift("Meet at 8!", 3))
print(shift("Zebra", 2))
```

**Challenge payoff:** the calls print `phhw dw 8!` and then `bgdtc`.

**Put it together:** indexing reaches one character, slicing reaches a substring, the four taught methods clean text, and `in` tests membership. Counter loops measure matches, transform-each loops build new strings, and a left-to-right search returns a useful position or `-1`.
