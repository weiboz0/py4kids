## Exercise 1

### Save and Read a Labeled Score Report


```python
def score_text_with_header(scores, path):
    with open(path, "w") as f:
        f.write(f"ARCADE SCORES\n")
        for score in scores:
            f.write(f"{score}\n")
    with open(path, "r") as f:
        text = f.read()
    return text

assert score_text_with_header([45, 70, 55], "ex1_scores.txt") == "ARCADE SCORES\n45\n70\n55\n"
assert score_text_with_header([120, 95], "ex1_scores.txt") == "ARCADE SCORES\n120\n95\n"
assert score_text_with_header([300], "ex1_scores.txt") == "ARCADE SCORES\n300\n"
```

## Exercise 2

### Rebuild the Number List


```python
def score_list_round_trip(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    loaded = []
    with open(path, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    return loaded

assert score_list_round_trip([18, 27, 36], "ex2_score_list.txt") == [18, 27, 36]
assert score_list_round_trip([5, 105], "ex2_score_list.txt") == [5, 105]
assert score_list_round_trip([42], "ex2_score_list.txt") == [42]
```

## Exercise 3

### Total and Count the Saved Tickets


```python
def saved_ticket_report(tickets, target, path):
    with open(path, "w") as f:
        for ticket in tickets:
            f.write(f"{ticket}\n")
    total = 0
    count = 0
    with open(path, "r") as f:
        for line in f:
            ticket = int(line.strip())
            total = total + ticket
            if ticket >= target:
                count = count + 1
    return [total, count]

assert saved_ticket_report([20, 35, 15], 25, "ex3_tickets.txt") == [70, 1]
assert saved_ticket_report([100, 50], 50, "ex3_tickets.txt") == [150, 2]
assert saved_ticket_report([9, 12, 7], 20, "ex3_tickets.txt") == [28, 0]
```

## Exercise 4

### Build a Four-Number Summary


```python
def saved_score_summary(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    loaded = []
    with open(path, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    return [min(loaded), max(loaded), sum(loaded), len(loaded)]

assert saved_score_summary([8, 15, 11], "ex4_summary.txt") == [8, 15, 34, 3]
assert saved_score_summary([50, 20], "ex4_summary.txt") == [20, 50, 70, 2]
assert saved_score_summary([99], "ex4_summary.txt") == [99, 99, 99, 1]
```

## Exercise 5

### Find the Saved High Score by Scanning


```python
def highest_saved_score(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    best = scores[0]
    with open(path, "r") as f:
        for line in f:
            score = int(line.strip())
            if score > best:
                best = score
    return best

assert highest_saved_score([44, 81, 63], "ex5_high_score.txt") == 81
assert highest_saved_score([205, 199, 201], "ex5_high_score.txt") == 205
assert highest_saved_score([17], "ex5_high_score.txt") == 17
```

## Exercise 6

### Build a Saved Player Summary


```python
def player_record_summary(name, level, points, path):
    with open(path, "w") as f:
        f.write(f"{name}\n")
        f.write(f"{level}\n")
        f.write(f"{points}\n")
    with open(path, "r") as f:
        saved_name = f.readline().strip()
        saved_level = int(f.readline().strip())
        saved_points = int(f.readline().strip())
    return f"{saved_name}: level {saved_level}, {saved_points} points"

assert player_record_summary("Lina", 3, 480, "ex6_typed_record.txt") == "Lina: level 3, 480 points"
assert player_record_summary("Mateo", 5, 925, "ex6_typed_record.txt") == "Mateo: level 5, 925 points"
assert player_record_summary("Zoe", 1, 75, "ex6_typed_record.txt") == "Zoe: level 1, 75 points"
```

## Exercise 7

### Search a Saved Roster


```python
def save_and_find_player(names, target, path):
    with open(path, "w") as f:
        for name in names:
            f.write(f"{name}\n")
    with open(path, "r") as f:
        for line in f:
            if line.strip() == target:
                return f"Found {target}."
    return f"{target} was not found."

assert save_and_find_player(["Inez", "Kai", "Noor"], "Inez", "ex7_roster.txt") == "Found Inez."
assert save_and_find_player(["Inez", "Kai", "Noor"], "Noor", "ex7_roster.txt") == "Found Noor."
assert save_and_find_player(["Inez", "Kai", "Noor"], "Pia", "ex7_roster.txt") == "Pia was not found."
```

## Exercise 8

### Challenge 1: Prove That a Fresh Save Replaces the Old One


```python
def replace_and_load_scores(first_scores, final_scores, path):
    with open(path, "w") as f:
        for score in first_scores:
            f.write(f"{score}\n")
    with open(path, "w") as f:
        for score in final_scores:
            f.write(f"{score}\n")
    loaded = []
    with open(path, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    return loaded

assert replace_and_load_scores([10, 20], [30, 40], "ex8_replace.txt") == [30, 40]
assert replace_and_load_scores([99], [4, 5, 6], "ex8_replace.txt") == [4, 5, 6]
assert replace_and_load_scores([7, 8, 9], [100], "ex8_replace.txt") == [100]
```

## Exercise 9

### Challenge 2: Find Two Statistics in One Pass


```python
def saved_total_and_best(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    total = 0
    best = scores[0]
    with open(path, "r") as f:
        for line in f:
            score = int(line.strip())
            total = total + score
            if score > best:
                best = score
    return [total, best]

assert saved_total_and_best([12, 30, 18], "ex9_combined_stats.txt") == [60, 30]
assert saved_total_and_best([75, 20], "ex9_combined_stats.txt") == [95, 75]
assert saved_total_and_best([9], "ex9_combined_stats.txt") == [9, 9]
```
