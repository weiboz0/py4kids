## Exercise 1

### Save and Read a Labeled Score Report


````python
def score_text_with_header(scores, path):
    with open(path, "w") as f:
        f.write(f"ARCADE SCORES\n")
        for score in scores:
            f.write(f"{score}\n")
    with open(path, "r") as f:
        text = f.read()
    return text

assert score_text_with_header([45, 70, 55], "ex1_scores.txt") == "ARCADE SCORES\n45\n70\n55\n"
assert score_text_with_header([120, 95], "ex1_scores.txt") == "ARCADE SCORES\n120\n95\n"
assert score_text_with_header([300], "ex1_scores.txt") == "ARCADE SCORES\n300\n"

print(score_text_with_header([45, 70, 55], "ex1_scores.txt"), end="")
````

## Exercise 3

### Total and Count the Saved Tickets


````python
def saved_ticket_report(tickets, target, path):
    with open(path, "w") as f:
        for ticket in tickets:
            f.write(f"{ticket}\n")
    total = 0
    count = 0
    with open(path, "r") as f:
        for line in f:
            ticket = int(line.strip())
            total = total + ticket
            if ticket >= target:
                count = count + 1
    return [total, count]

assert saved_ticket_report([20, 35, 15], 25, "ex3_tickets.txt") == [70, 1]
assert saved_ticket_report([100, 50], 50, "ex3_tickets.txt") == [150, 2]
assert saved_ticket_report([9, 12, 7], 20, "ex3_tickets.txt") == [28, 0]

print(saved_ticket_report([20, 35, 15], 25, "ex3_tickets.txt"))
````

````python
def saved_score_summary(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    loaded = []
    with open(path, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    return [min(loaded), max(loaded), sum(loaded), len(loaded)]

assert saved_score_summary([8, 15, 11], "ex4_summary.txt") == [8, 15, 34, 3]
assert saved_score_summary([50, 20], "ex4_summary.txt") == [20, 50, 70, 2]
assert saved_score_summary([99], "ex4_summary.txt") == [99, 99, 99, 1]

print(saved_score_summary([8, 15, 11], "ex4_summary.txt"))
````

## Exercise 5

### Find the Saved High Score by Scanning


````python
def highest_saved_score(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    best = scores[0]
    with open(path, "r") as f:
        for line in f:
            score = int(line.strip())
            if score > best:
                best = score
    return best

assert highest_saved_score([44, 81, 63], "ex5_high_score.txt") == 81
assert highest_saved_score([205, 199, 201], "ex5_high_score.txt") == 205
assert highest_saved_score([17], "ex5_high_score.txt") == 17

print(highest_saved_score([44, 81, 63], "ex5_high_score.txt"))
````

````python
def player_record_summary(name, level, points, path):
    with open(path, "w") as f:
        f.write(f"{name}\n")
        f.write(f"{level}\n")
        f.write(f"{points}\n")
    with open(path, "r") as f:
        saved_name = f.readline().strip()
        saved_level = int(f.readline().strip())
        saved_points = int(f.readline().strip())
    return f"{saved_name}: level {saved_level}, {saved_points} points"

assert player_record_summary("Lina", 3, 480, "ex6_typed_record.txt") == "Lina: level 3, 480 points"
assert player_record_summary("Mateo", 5, 925, "ex6_typed_record.txt") == "Mateo: level 5, 925 points"
assert player_record_summary("Zoe", 1, 75, "ex6_typed_record.txt") == "Zoe: level 1, 75 points"

print(player_record_summary("Lina", 3, 480, "ex6_typed_record.txt"))
````

## Exercise 7

### Search a Saved Roster


````python
def save_and_find_player(names, target, path):
    with open(path, "w") as f:
        for name in names:
            f.write(f"{name}\n")
    with open(path, "r") as f:
        for line in f:
            if line.strip() == target:
                return f"Found {target}."
    return f"{target} was not found."

assert save_and_find_player(["Inez", "Kai", "Noor"], "Inez", "ex7_roster.txt") == "Found Inez."
assert save_and_find_player(["Inez", "Kai", "Noor"], "Noor", "ex7_roster.txt") == "Found Noor."
assert save_and_find_player(["Inez", "Kai", "Noor"], "Pia", "ex7_roster.txt") == "Pia was not found."

print(save_and_find_player(["Inez", "Kai", "Noor"], "Inez", "ex7_roster.txt"))
````

## Exercise 8

### Append a Late Score


````python
# Save the original scores, append one late score, and reload them.
def append_round_trip(scores, late_score, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    with open(path, "a") as f:
        f.write(f"{late_score}\n")
    loaded = []
    with open(path, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    return loaded

assert append_round_trip([62, 48], 91, "ex8_late_score.txt") == [62, 48, 91]
assert append_round_trip([], 7, "ex8_late_score.txt") == [7]

print(append_round_trip([62, 48], 91, "ex8_late_score.txt"))
````

## Exercise 10

### Best Team from Records


````python
# Save team records and return the first team with the highest points.
def best_team(records, path):
    with open(path, "w") as f:
        for record in records:
            f.write(f"{record}\n")
    best_points = -1
    best_name = ""
    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            points = int(parts[1])
            if points > best_points:
                best_points = points
                best_name = parts[0]
    return best_name

assert best_team(["Ivo,15", "Uma,9", "Teo,22"], "ex10_teams.txt") == "Teo"
assert best_team(["Ari,7", "Bea,7"], "ex10_teams.txt") == "Ari"

print(best_team(["Ivo,15", "Uma,9", "Teo,22"], "ex10_teams.txt"))
````

## Exercise 11

### More Practice: Long Words in a File


````python
def long_word_count(lines, path):
    with open(path, "w") as f:
        for text in lines:
            f.write(f"{text}\n")
    count = 0
    with open(path, "r") as f:
        for line in f:
            for word in line.split():
                if len(word) > 3:
                    count = count + 1
    return count

assert long_word_count(["red sails drift", "past docks"], "ex11_words.txt") == 4
assert long_word_count(["a blue bird"], "ex11_words.txt") == 2

print(long_word_count(["red sails drift", "past docks"], "ex11_words.txt"))
````

## Exercise 13

### More Practice: Numbered Lines


````python
def numbered(lines, path):
    with open(path, "w") as f:
        for text in lines:
            f.write(f"{text}\n")
    rows = []
    number = 1
    with open(path, "r") as f:
        for line in f:
            text = line.strip()
            rows.append(f"{number}: {text}")
            number = number + 1
    return "\n".join(rows)

assert numbered(["whisk batter", "add eggs", "bake"], "ex13_numbered.txt") == "1: whisk batter\n2: add eggs\n3: bake"
assert numbered(["mix"], "ex13_numbered.txt") == "1: mix"

print(numbered(["whisk batter", "add eggs", "bake"], "ex13_numbered.txt"))
````

## Exercise 15

### More Practice: Grade Report


````python
def grade_report(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    a_count = 0
    b_count = 0
    c_count = 0
    d_count = 0
    with open(path, "r") as f:
        for line in f:
            score = int(line.strip())
            if score >= 90:
                a_count = a_count + 1
            elif score >= 80:
                b_count = b_count + 1
            elif score >= 70:
                c_count = c_count + 1
            else:
                d_count = d_count + 1
    rows = [f"A: {a_count}", f"B: {b_count}", f"C: {c_count}", f"D: {d_count}"]
    return "\n".join(rows)

assert grade_report([91, 78, 85, 62, 97], "ex15_grades.txt") == "A: 2\nB: 1\nC: 1\nD: 1"
assert grade_report([90, 80, 70, 69], "ex15_grades.txt") == "A: 1\nB: 1\nC: 1\nD: 1"

print(grade_report([91, 78, 85, 62, 97], "ex15_grades.txt"))
````

## Exercise 16

### More Practice: Shout Copy


````python
def shout_copy(lines, source, target):
    with open(source, "w") as f:
        for text in lines:
            f.write(f"{text}\n")
    with open(source, "r") as f:
        with open(target, "w") as out:
            for line in f:
                out.write(line.upper())
    loaded = []
    with open(target, "r") as f:
        for line in f:
            loaded.append(line.strip())
    return loaded

assert shout_copy(["hush now", "well done"], "ex16_source.txt", "ex16_target.txt") == ["HUSH NOW", "WELL DONE"]
assert shout_copy(["go"], "ex16_source.txt", "ex16_target.txt") == ["GO"]

print(shout_copy(["hush now", "well done"], "ex16_source.txt", "ex16_target.txt"))
````

````python
def keep_todo(lines, source, target):
    with open(source, "w") as f:
        for text in lines:
            f.write(f"{text}\n")
    with open(source, "r") as f:
        with open(target, "w") as out:
            for line in f:
                text = line.strip()
                if text.startswith("TODO"):
                    out.write(f"{text}\n")
    loaded = []
    with open(target, "r") as f:
        for line in f:
            loaded.append(line.strip())
    return loaded

assert keep_todo(["TODO paint", "done sweep", "TODO wash"], "ex17_source.txt", "ex17_target.txt") == ["TODO paint", "TODO wash"]
assert keep_todo(["done rest"], "ex17_source.txt", "ex17_target.txt") == []

print(keep_todo(["TODO paint", "done sweep", "TODO wash"], "ex17_source.txt", "ex17_target.txt"))
````

````python
def first_long_word(words, minimum, path):
    with open(path, "w") as f:
        for word in words:
            f.write(f"{word}\n")
    with open(path, "r") as f:
        for line in f:
            word = line.strip()
            if len(word) >= minimum:
                return word
    return "none"

assert first_long_word(["sky", "meadow", "brook"], 5, "ex19_first.txt") == "meadow"
assert first_long_word(["sun", "sea"], 5, "ex19_first.txt") == "none"

print(first_long_word(["sky", "meadow", "brook"], 5, "ex19_first.txt"))
````

## Exercise 20

### More Practice: Line of a Name


````python
def line_of(names, target, path):
    with open(path, "w") as f:
        for name in names:
            f.write(f"{name}\n")
    number = 0
    with open(path, "r") as f:
        for line in f:
            number = number + 1
            if line.strip() == target:
                return number
    return -1

assert line_of(["Rae", "Tovi", "Wen"], "Wen", "ex20_names.txt") == 3
assert line_of(["Rae", "Tovi", "Wen"], "Zia", "ex20_names.txt") == -1
assert line_of(["Wen", "Wen"], "Wen", "ex20_names.txt") == 1

print(line_of(["Rae", "Tovi", "Wen"], "Wen", "ex20_names.txt"))
````

## Exercise 21

### More Practice: Chart from a File


````python
def chart_from_file(values, path):
    with open(path, "w") as f:
        for value in values:
            f.write(f"{value}\n")
    rows = []
    with open(path, "r") as f:
        for line in f:
            number = int(line.strip())
            rows.append("#" * number)
    return "\n".join(rows)

assert chart_from_file([3, 1, 4], "ex21_chart.txt") == "###\n#\n####"
assert chart_from_file([2], "ex21_chart.txt") == "##"

print(chart_from_file([3, 1, 4], "ex21_chart.txt"))
````

## Exercise 26

### Challenge: Find Two Statistics in One Pass


````python
def saved_total_and_best(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    total = 0
    best = scores[0]
    with open(path, "r") as f:
        for line in f:
            score = int(line.strip())
            total = total + score
            if score > best:
                best = score
    return [total, best]

assert saved_total_and_best([12, 30, 18], "ex26_combined_stats.txt") == [60, 30]
assert saved_total_and_best([75, 20], "ex26_combined_stats.txt") == [95, 75]
assert saved_total_and_best([9], "ex26_combined_stats.txt") == [9, 9]

print(saved_total_and_best([12, 30, 18], "ex26_combined_stats.txt"))
````

## Exercise 27

### Challenge: Merge Two Score Files


````python
def merge_files(first, second, path_a, path_b):
    with open(path_a, "w") as f:
        for score in first:
            f.write(f"{score}\n")
    with open(path_b, "w") as f:
        for score in second:
            f.write(f"{score}\n")
    loaded = []
    with open(path_a, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    with open(path_b, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    loaded.sort()
    return loaded

assert merge_files([12, 40], [25, 3], "ex27_first.txt", "ex27_second.txt") == [3, 12, 25, 40]
assert merge_files([], [5, 1], "ex27_first.txt", "ex27_second.txt") == [1, 5]

print(merge_files([12, 40], [25, 3], "ex27_first.txt", "ex27_second.txt"))
````

## Exercise 28

### Challenge: High-Score Table


````python
def update_high_scores(scores, new_score, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    loaded = []
    with open(path, "r") as f:
        for line in f:
            loaded.append(int(line.strip()))
    loaded.append(new_score)
    ordered = sorted(loaded)
    top = [ordered[-1], ordered[-2], ordered[-3]]
    with open(path, "w") as f:
        for score in top:
            f.write(f"{score}\n")
    saved_top = []
    with open(path, "r") as f:
        for line in f:
            saved_top.append(int(line.strip()))
    return saved_top

assert update_high_scores([900, 850, 700], 880, "ex28_high_scores.txt") == [900, 880, 850]
assert update_high_scores([10, 8, 6], 2, "ex28_high_scores.txt") == [10, 8, 6]

print(update_high_scores([900, 850, 700], 880, "ex28_high_scores.txt"))
````
