### Hours and Minutes

```python
total_minutes = 155
hours = total_minutes // 60
minutes = total_minutes % 60
line_1 = f"{hours} hours {minutes} minutes"
print(line_1)
assert line_1 == "2 hours 35 minutes"
```

### Converted Average

```python
first_text = "8"
second_text = "12"
first_score = int(first_text)
second_score = int(second_text)
average = (first_score + second_score) / 2
line_1 = f"Average: {average}"
print(line_1)
assert line_1 == "Average: 10.0"
```

## Exercise 5

### Ordered Score Ladder

```python
score = 94
if score >= 90:
    line_1 = "Level A"
elif score >= 80:
    line_1 = "Level B"
elif score >= 70:
    line_1 = "Level C"
else:
    line_1 = "Keep practicing"
print(line_1)
assert line_1 == "Level A"
```

### Range and Entry Test

```python
age = 12
has_pass = False
has_invitation = True
if age >= 10 and age <= 13 and (has_pass == True or has_invitation == True):
    line_1 = "Entry approved"
else:
    line_1 = "Entry denied"
print(line_1)
assert line_1 == "Entry approved"
```
