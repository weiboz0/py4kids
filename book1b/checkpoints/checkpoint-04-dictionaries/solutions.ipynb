## Question 1

### Build a Trail Lookup


```python
def trail_distance(name):
    distances = {"Creek": 3, "Hill": 5, "Lake": 8}
    return distances[name]

assert trail_distance("Creek") == 3
assert trail_distance("Lake") == 8
assert trail_distance("Hill") == 5
```

## Question 2

### Default Locker Count


```python
def locker_count(lockers, item):
    return lockers.get(item, 0)

assert locker_count({"balls": 6, "cones": 9}, "cones") == 9
assert locker_count({"balls": 6, "cones": 9}, "bibs") == 0
assert locker_count({}, "balls") == 0
```

## Question 3

### Color Frequency Tally


```python
def color_counts(colors):
    counts = {}
    for color in colors:
        if color in counts:
            counts[color] = counts[color] + 1
        else:
            counts[color] = 1
    return counts

assert color_counts(["red", "blue", "red"]) == {"red": 2, "blue": 1}
assert color_counts(["gold", "gold", "green", "gold"]) == {"gold": 3, "green": 1}
assert color_counts([]) == {}
```

## Question 4

### Inventory Total


```python
def inventory_total(stock):
    total = 0
    for item, amount in stock.items():
        total = total + amount
    return total

assert inventory_total({"notebooks": 4, "pencils": 10, "erasers": 3}) == 17
assert inventory_total({"folders": 8}) == 8
assert inventory_total({}) == 0
```

## Question 5

### Most Common Choice


```python
def most_common(counts):
    best_key = ""
    best_count = -1
    for key, count in counts.items():
        if count > best_count:
            best_key = key
            best_count = count
    return best_key

assert most_common({"robot": 5, "rocket": 8, "maze": 3}) == "rocket"
assert most_common({"sun": 1, "moon": 4}) == "moon"
assert most_common({"solo": 6}) == "solo"
```

## Question 6

### Level Lookup Message


```python
def level_message(level_text):
    level = int(level_text)
    points = {1: 4, 2: 7, 3: 10}
    score = points[level]
    if score == 10:
        label = "Gold"
    elif score == 7:
        label = "Silver"
    else:
        label = "Bronze"
    return f"{label}: {score}"

message = level_message("3")
print(f"Result: {message}")
assert level_message("3") == "Gold: 10"
assert level_message("2") == "Silver: 7"
assert level_message("1") == "Bronze: 4"
```

## Question 7

### First-Word Report


```python
def first_word_report(words):
    counts = {}
    for word in words:
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    first = words[0]
    return f"{first}: {counts[first]} of {len(words)}"

assert first_word_report(["map", "star", "map"]) == "map: 2 of 3"
assert first_word_report(["red", "blue", "green", "red"]) == "red: 2 of 4"
assert first_word_report(["solo"]) == "solo: 1 of 1"
```
