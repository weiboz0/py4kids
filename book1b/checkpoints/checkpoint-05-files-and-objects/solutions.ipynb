## Question 1

### Score File Round Trip


```python
def save_scores(scores, path):
    with open(path, "w") as f:
        for score in scores:
            f.write(f"{score}\n")
    return path

def load_scores(path):
    scores = []
    with open(path, "r") as f:
        for line in f:
            scores.append(int(line.strip()))
    return scores

assert save_scores([14, 27, 31], "q1_scores.txt") == "q1_scores.txt"
assert load_scores("q1_scores.txt") == [14, 27, 31]
assert save_scores([5, 8], "q1_scores.txt") == "q1_scores.txt"
assert load_scores("q1_scores.txt") == [5, 8]
assert save_scores([99], "q1_scores.txt") == "q1_scores.txt"
assert load_scores("q1_scores.txt") == [99]
```

## Question 2

### Load a Score Summary


```python
with open("q2_scores.txt", "w") as f:
    for score in [12, 19, 7, 19]:
        f.write(f"{score}\n")

def score_summary(path):
    scores = []
    with open(path, "r") as f:
        for line in f:
            scores.append(int(line.strip()))
    total = 0
    strong_count = 0
    for score in scores:
        total = total + score
        if score >= 15:
            strong_count = strong_count + 1
    highest = max(scores)
    return [total, highest, strong_count]

assert score_summary("q2_scores.txt") == [57, 19, 2]
with open("q2_scores.txt", "w") as f:
    for score in [3, 8, 10]:
        f.write(f"{score}\n")
assert score_summary("q2_scores.txt") == [21, 10, 0]
with open("q2_scores.txt", "w") as f:
    for score in [15, 30]:
        f.write(f"{score}\n")
assert score_summary("q2_scores.txt") == [45, 30, 2]
```

## Question 3

### Find the First Matching Team


```python
with open("q3_teams.txt", "w") as f:
    for team in ["Comet", "Otter", "Falcon"]:
        f.write(f"{team}\n")

def find_team(path, target):
    with open(path, "r") as f:
        for line in f:
            if line.strip() == target:
                return f"Found {target}"
    return f"No team named {target}"

assert find_team("q3_teams.txt", "Comet") == "Found Comet"
assert find_team("q3_teams.txt", "Otter") == "Found Otter"
assert find_team("q3_teams.txt", "Lynx") == "No team named Lynx"
```

## Question 4

### Make a Point


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

first = Point(6, 2)
second = Point(-3, 8)
assert first.x == 6
assert first.y == 2
assert second.x == -3
assert second.y == 8
first.x = 10
assert first.x == 10
assert second.x == -3
```

## Question 5

### Rectangle Report


```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def describe(self):
        return f"{self.width} by {self.height} has area {self.area()}"

poster = Rectangle(8, 3)
assert poster.width == 8
assert poster.height == 3
assert poster.area() == 24
assert poster.describe() == "8 by 3 has area 24"
assert Rectangle(2.5, 4).area() == 10.0
assert Rectangle(1, 9).describe() == "1 by 9 has area 9"
```

## Question 6

### Counter Award Band


```python
class Counter:
    def __init__(self, count):
        self.count = count

    def award_band(self):
        if self.count >= 10:
            return "gold"
        elif self.count >= 5:
            return "silver"
        else:
            return "bronze"

    def reached_goal(self):
        return self.count >= 10

gold = Counter(12)
silver = Counter(7)
bronze = Counter(2)
assert gold.count == 12
assert gold.award_band() == "gold"
assert silver.award_band() == "silver"
assert bronze.award_band() == "bronze"
assert gold.reached_goal() == True
assert silver.reached_goal() == False
```

## Question 7

### Save a Labeled Rectangle


```python
class Rectangle:
    def __init__(self, code, width, height):
        self.code = code
        self.width = width
        self.height = height

    def short_code(self):
        return self.code[0:2]

    def save(self, path):
        with open(path, "w") as f:
            f.write(f"{self.code}\n")
            f.write(f"{self.width}\n")
            f.write(f"{self.height}\n")
        return path

sign = Rectangle("NW-17", 9, 4)
assert sign.code == "NW-17"
assert sign.short_code() == "NW"
assert sign.save("q7_rectangle.txt") == "q7_rectangle.txt"
with open("q7_rectangle.txt", "r") as f:
    saved_text = f.read()
assert saved_text == "NW-17\n9\n4\n"
second_sign = Rectangle("SE-2", 5, 6)
assert second_sign.short_code() == "SE"
assert second_sign.save("q7_rectangle.txt") == "q7_rectangle.txt"
with open("q7_rectangle.txt", "r") as f:
    second_text = f.read()
assert second_text == "SE-2\n5\n6\n"
```
