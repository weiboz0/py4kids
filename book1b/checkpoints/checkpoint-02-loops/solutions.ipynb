## Question 1

### Ordered Battery Ladder

```python
charge = 67
if charge >= 80:
    line = "Full"
elif charge >= 50:
    line = "Ready"
elif charge >= 20:
    line = "Low"
else:
    line = "Recharge"
print(line)
assert line == "Ready"
```

## Question 2

### Cartons and Leftovers

```python
pencils = 137
carton_size = 12
cartons = pencils // carton_size
leftovers = pencils % carton_size
line = f"{cartons} cartons, {leftovers} left"
print(line)
assert line == "11 cartons, 5 left"
```

## Question 3

### While-Counter Pass Bar

```python
lap_goal = 9
laps = 0
while laps < lap_goal:
    laps = laps + 1
line = f"Count: {laps}"
print(line)
assert laps == 9
assert line == "Count: 9"
```

## Question 4

### Collatz Repeat-Until

```python
n = 6
steps = 0
while True:
    if n == 1:
        break
    if n % 2 == 0:
        n = n // 2
    else:
        n = n * 3 + 1
    steps = steps + 1
line = f"Steps: {steps}"
print(line)
assert n == 1
assert steps == 8
assert line == "Steps: 8"
```

## Question 5

### For-Total Pass Bar

```python
last_day = 20
total = 0
for day in range(1, last_day + 1):
    total = total + day
line = f"Sum: {total}"
print(line)
assert total == 210
assert line == "Sum: 210"
```

## Question 6

### Count by Condition

```python
count = 0
for round_number in range(1, 51):
    if round_number % 7 == 0:
        count = count + 1
line = f"Count: {count}"
print(line)
assert count == 7
assert line == "Count: 7"
```

## Question 7

### Four-by-Four Times Table

```python
table_size = 4
row_count = 0
for row in range(1, table_size + 1):
    row_text = ""
    for col in range(1, table_size + 1):
        if col == 1:
            row_text = f"{row * col}"
        else:
            row_text = row_text + f" {row * col}"
    print(row_text)
    if row == 1:
        assert row_text == "1 2 3 4"
    if row == table_size:
        assert row_text == "4 8 12 16"
    row_count = row_count + 1
assert row_count == 4
```
