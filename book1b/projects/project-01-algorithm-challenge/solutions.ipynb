## Problem 1

```python
def nth_prime(k):
    candidate = 2
    count = 0
    while count < k:
        is_prime = True
        for d in range(2, candidate):
            if candidate % d == 0:
                is_prime = False
                break
        if is_prime:
            count = count + 1
            if count == k:
                return candidate
        candidate = candidate + 1

assert nth_prime(1) == 2
assert nth_prime(5) == 11
assert nth_prime(10) == 29
assert nth_prime(3) == 5
```

## Problem 2

```python
def reverse_digits(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n = n // 10
    return rev

assert reverse_digits(1234) == 4321
assert reverse_digits(1200) == 21
assert reverse_digits(0) == 0
assert reverse_digits(507) == 705
```

## Problem 3

```python
def top_three(scores):
    s = sorted(scores)
    return [s[len(s)-1], s[len(s)-2], s[len(s)-3]]

assert top_three([4, 9, 1, 7, 3]) == [9, 7, 4]
assert top_three([5, 5, 2, 8]) == [8, 5, 5]
assert top_three([10, 20, 30]) == [30, 20, 10]
```

## Problem 4 — Challenge

```python
def merge_sorted(a, b):
    merged = []
    i = 0
    j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            merged.append(a[i])
            i = i + 1
        else:
            merged.append(b[j])
            j = j + 1
    while i < len(a):
        merged.append(a[i])
        i = i + 1
    while j < len(b):
        merged.append(b[j])
        j = j + 1
    return merged

assert merge_sorted([1, 4, 6], [2, 3, 5]) == [1, 2, 3, 4, 5, 6]
assert merge_sorted([], [2, 9]) == [2, 9]
assert merge_sorted([3, 8, 9], []) == [3, 8, 9]
assert merge_sorted([2, 6, 6], [6, 10]) == [2, 6, 6, 6, 10]
```

## Problem 5 — Challenge

```python
def binary_search(nums, target):
    lo = 0
    hi = len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

assert binary_search([1, 3, 5, 7, 9], 7) == 3
assert binary_search([1, 3, 5, 7, 9], 4) == -1
assert binary_search([2, 4, 6, 8, 10, 12], 2) == 0
assert binary_search([2, 4, 6, 8, 10, 12], 12) == 5
```

## Problem 6

```python
def count_substring(text, part):
    count = 0
    for i in range(len(text)-len(part)+1):
        if text[i:i+len(part)] == part:
            count = count + 1
    return count

assert count_substring("cocoon", "co") == 2
assert count_substring("aaaa", "aa") == 3
assert count_substring("possession", "ss") == 2
```

## Problem 7

```python
def word_counts_from_file(path):
    counts = {}
    with open(path, "r") as f:
        for line in f:
            word = line.strip()
            counts[word] = counts.get(word, 0) + 1
    return counts

words = ["fern", "moss", "fern", "larch", "moss", "fern"]
with open("p7_words.txt", "w") as f:
    for word in words:
        f.write(f"{word}\n")
assert word_counts_from_file("p7_words.txt") == {"fern": 3, "moss": 2, "larch": 1}

words = ["maple", "birch", "maple", "cedar"]
with open("p7b_words.txt", "w") as f:
    for word in words:
        f.write(f"{word}\n")
assert word_counts_from_file("p7b_words.txt") == {"maple": 2, "birch": 1, "cedar": 1}

words = ["wren", "wren", "wren", "finch"]
with open("p7c_words.txt", "w") as f:
    for word in words:
        f.write(f"{word}\n")
assert word_counts_from_file("p7c_words.txt") == {"wren": 3, "finch": 1}
```

## Problem 8

```python
def most_common_word(path):
    counts = {}
    with open(path, "r") as f:
        for line in f:
            word = line.strip()
            counts[word] = counts.get(word, 0) + 1
    best_count = 0
    best_word = ""
    for key in counts:
        if counts[key] > best_count:
            best_count = counts[key]
            best_word = key
    return best_word

words = ["kite", "yoyo", "kite", "kite", "yoyo"]
with open("p8_words.txt", "w") as f:
    for word in words:
        f.write(f"{word}\n")
assert most_common_word("p8_words.txt") == "kite"

words = ["seal", "crane", "crane", "crane", "seal"]
with open("p8b_words.txt", "w") as f:
    for word in words:
        f.write(f"{word}\n")
assert most_common_word("p8b_words.txt") == "crane"

words = ["heron", "ibis", "koi", "ibis"]
with open("p8c_words.txt", "w") as f:
    for word in words:
        f.write(f"{word}\n")
assert most_common_word("p8c_words.txt") == "ibis"
```

## Problem 9

```python
def group_by_parity(nums):
    result = {"even": [], "odd": []}
    for number in nums:
        if number % 2 == 0:
            result["even"].append(number)
        else:
            result["odd"].append(number)
    return result

assert group_by_parity([1, 2, 3, 4]) == {"even": [2, 4], "odd": [1, 3]}
assert group_by_parity([0, 7, 10]) == {"even": [0, 10], "odd": [7]}
assert group_by_parity([]) == {"even": [], "odd": []}
```

## Problem 10

```python
class RunningTally:
    def __init__(self):
        self.values = []

    def add(self, value):
        self.values.append(value)
        return len(self.values)

    def total(self):
        return sum(self.values)

    def highest(self):
        return max(self.values)

    def describe(self):
        return f"{len(self.values)} values, total {sum(self.values)}, highest {max(self.values)}"

t = RunningTally()
assert t.add(5) == 1
assert t.add(9) == 2
assert t.add(0) == 3
assert t.total() == 14
assert t.highest() == 9
assert t.describe() == "3 values, total 14, highest 9"

second = RunningTally()
assert second.add(4) == 1
assert second.total() == 4
assert t.total() == 14
```

## Problem 11

```python
def running_totals_to_file(in_path, out_path):
    totals = []
    running = 0
    with open(in_path, "r") as input_file:
        for line in input_file:
            running = running + int(line.strip())
            totals.append(running)
    with open(out_path, "w") as output_file:
        for value in totals:
            output_file.write(f"{value}\n")
    return totals

numbers = [5, 3, 2]
with open("p11_in.txt", "w") as f:
    for number in numbers:
        f.write(f"{number}\n")
assert running_totals_to_file("p11_in.txt", "p11_out.txt") == [5, 8, 10]
with open("p11_out.txt", "r") as f:
    output_text = f.read()
assert output_text == "5\n8\n10\n"

numbers = [10, -4, 6]
with open("p11b_in.txt", "w") as f:
    for number in numbers:
        f.write(f"{number}\n")
assert running_totals_to_file("p11b_in.txt", "p11b_out.txt") == [10, 6, 12]
with open("p11b_out.txt", "r") as f:
    output_text = f.read()
assert output_text == "10\n6\n12\n"

numbers = [0, 12, 8, 1]
with open("p11c_in.txt", "w") as f:
    for number in numbers:
        f.write(f"{number}\n")
assert running_totals_to_file("p11c_in.txt", "p11c_out.txt") == [0, 12, 20, 21]
with open("p11c_out.txt", "r") as f:
    output_text = f.read()
assert output_text == "0\n12\n20\n21\n"
```
